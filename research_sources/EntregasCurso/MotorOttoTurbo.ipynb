# **1. Introducción**
Los motores son máquinas térmicas que permiten convertir la energía química del combustible, liberada mediante procesos de combustión, en trabajo, generalmente en forma de energía mecánica. Dentro de estos, los más comunes en la industria son los motores de combustión interna (MCI), los cuales operan bajo el principio de desplazamiento positivo, utilizando el movimiento lineal de un pistón acoplado a un mecanismo biela-manivela para generar trabajo. Se denominan de combustión interna debido a que la liberación de energía ocurre dentro del propio sistema, específicamente en la cámara de combustión, donde el fluido de trabajo evoluciona térmicamente. (Frías Reséndiz et al., s.f.)

Estos motores tienen una amplia variedad de aplicaciones, desde usos cotidianos hasta industriales. Dependiendo de la aplicación, existen distintas configuraciones como motores de dos o cuatro tiempos, ciclos termodinámicos Otto o Diésel, y diferentes tipos de combustibles como gasolina, diésel o biocombustibles. Sin embargo, más allá de estas diferencias, el funcionamiento del motor está fuertemente condicionado por el combustible utilizado, ya que este constituye la fuente principal de energía del sistema.

En este contexto, el combustible no solo representa una fuente energética, sino también un sistema dinámico cuyo comportamiento depende de reacciones químicas y de la interacción de múltiples variables. En particular, la eficiencia del motor está directamente relacionada con la calidad de la combustión, la cual depende de condiciones adecuadas de estequiometría. Cuando la mezcla aire-combustible se encuentra en condiciones óptimas, las reacciones se desarrollan de manera más completa, permitiendo una mayor liberación de energía y, por tanto, un mejor aprovechamiento del sistema.

No obstante, en la práctica, estas condiciones ideales no siempre se alcanzan. Factores como la variación en la mezcla, las condiciones de operación y la dinámica propia del sistema afectan el desarrollo de las reacciones, generando pérdidas de eficiencia. Por esta razón, resulta fundamental analizar el comportamiento del combustible desde una perspectiva dinámica, considerando aspectos como la cinética de reacción y la evolución temporal del sistema.

Este tipo de análisis cobra especial relevancia en aplicaciones de alto desempeño, como la Fórmula 1, donde los motores operan bajo condiciones extremas y cada detalle en la combustión impacta significativamente el rendimiento. En este contexto, comprender y modelar el comportamiento del combustible permite optimizar el desempeño del motor y maximizar la eficiencia del sistema.

Sin embargo, a pesar de sus ventajas en términos de desempeño, estos motores presentan una alta dependencia de combustibles fósiles, lo cual implica consecuencias negativas a nivel ambiental y de sostenibilidad. El uso continuo de este tipo de combustibles contribuye a la emisión de gases contaminantes y al agotamiento de recursos no renovables, lo que ha impulsado la búsqueda de alternativas más limpias y eficientes.

En este sentido, resulta relevante no solo analizar el comportamiento del combustible tradicional, sino también explorar posibles sustitutos de origen renovable que permitan mantener o incluso mejorar el rendimiento del sistema. Por lo tanto, en el presente proyecto se plantea el estudio del comportamiento dinámico del combustible en motores de combustión interna, así como la evaluación de una alternativa renovable, mediante un modelo basado en ecuaciones diferenciales que permita describir la interacción entre las variables principales del sistema.

## **3.1 Motor de combustión interna (ICE)**

El motor de combustión interna integrado en la unidad de potencia corresponde a un motor alternativo de cuatro tiempos, configuración V6, turboalimentado, con una cilindrada nominal de 1600 cm³ (+0 / −10 cm³) (FIA, 2026). Opera mediante un sistema de encendido por chispa y se rige por un ciclo termodinámico Otto modificado, lo que le permite alcanzar elevadas relaciones de compresión y, en consecuencia, altos niveles de eficiencia térmica.




```python
# @title
import numpy as np
import plotly.graph_objects as go

# --- F1 Engine Realities ---
compression_ratio = 14.0
v_max = 1.6  # Bottom Dead Center (BDC)
v_min = v_max / compression_ratio  # Top Dead Center (TDC)

# Turbo & MGU-H Pressures
p_intake_boost = 350.0  # ~3.5 bar absolute (high boost)
p_exhaust_backpressure = 250.0  # ~2.5 bar (high backpressure from turbine)

# Gammas (Ratio of specific heats)
gamma_comp = 1.35  # Air/Fuel mixture
gamma_exp = 1.30   # Hot exhaust gases (lower gamma)

# --- Define the 7 Stages of the F1 Cycle ---

# 1. Intake (High boost, BDC to TDC) - Blue
v_intake = np.linspace(v_min, v_max, 30)
p_intake = np.full_like(v_intake, p_intake_boost)

# 2. Compression (BDC to TDC) - Orange
v_comp = np.linspace(v_max, v_min, 40)
p_comp = p_intake_boost * (v_max / v_comp)**gamma_comp

# 3. Combustion (Curved, piston moving down slightly) - Red
# F1 engines burn fast, but not instantly. Volume expands by ~8% during burn.
v_comb = np.linspace(v_min, v_min * 1.08, 20)
p_peak = 28000.0  # ~280 bar peak cylinder pressure
# Use a sine curve for a smooth pressure rise
p_comb = p_comp[-1] + (p_peak - p_comp[-1]) * np.sin(np.linspace(0, np.pi/2, 20))

# 4. Power / Expansion (Down to 90% of stroke) - Green
# Exhaust valves open early to feed the MGU-H
v_exp = np.linspace(v_min * 1.08, v_max * 0.90, 40)
p_exp = p_comb[-1] * ((v_min * 1.08) / v_exp)**gamma_exp

# 5. Blowdown (Early valve opening, smooth drop to backpressure) - Purple
v_blow = np.linspace(v_max * 0.90, v_max, 20)
# Cosine curve for a smooth "toe" at the bottom right of the graph
p_blow = p_exhaust_backpressure + (p_exp[-1] - p_exhaust_backpressure) * np.cos(np.linspace(0, np.pi/2, 20))

# 6. Exhaust (High backpressure pushing out, BDC to TDC) - Gray
v_exhaust = np.linspace(v_max, v_min, 30)
p_exhaust = np.full_like(v_exhaust, p_exhaust_backpressure)

# 7. Valve Overlap (Closing the loop from Exhaust to Intake) - Cyan
v_overlap = np.full(10, v_min)
p_overlap = np.linspace(p_exhaust_backpressure, p_intake_boost, 10)


# --- Group Stages for Animation ---
stages = [
    ("1. Intake (Boost)", v_intake, p_intake, "blue"),
    ("2. Compression", v_comp, p_comp, "orange"),
    ("3. Combustion (Ignition)", v_comb, p_comb, "red"),
    ("4. Power Stroke", v_exp, p_exp, "green"),
    ("5. Blowdown (MGU-H)", v_blow, p_blow, "purple"),
    ("6. Exhaust", v_exhaust, p_exhaust, "gray"),
    ("7. Valve Overlap", v_overlap, p_overlap, "cyan")
]

# Calculate total animation frames
lengths = [len(s[1]) for s in stages]
cumulative_lengths = np.cumsum(lengths)
total_frames = cumulative_lengths[-1]

# --- Build the Animation Frames ---
frames = []
for i in range(1, total_frames + 1):
    frame_data = []
    current_x, current_y = 0, 0

    for j, stage in enumerate(stages):
        name, v, p, color = stage
        start_idx = cumulative_lengths[j-1] if j > 0 else 0
        end_idx = cumulative_lengths[j]

        if i <= start_idx:
            cur_v, cur_p = [None], [None]
        elif i > end_idx:
            cur_v, cur_p = v, p
        else:
            local_i = i - start_idx
            cur_v, cur_p = v[:local_i], p[:local_i]
            if len(cur_v) > 0:
                current_x, current_y = cur_v[-1], cur_p[-1]

        frame_data.append(go.Scatter(x=cur_v, y=cur_p, mode='lines',
                                     line=dict(color=color, width=3), name=name))

    # Add tracking dot
    if current_x != 0:
        frame_data.append(go.Scatter(x=[current_x], y=[current_y], mode='markers',
                                     marker=dict(size=12, color="black"), showlegend=False))

    frames.append(go.Frame(data=frame_data))

# --- Setup the Initial Figure ---
fig = go.Figure(
    data=[go.Scatter(x=[], y=[], name=s[0]) for s in stages],
    layout=go.Layout(
        title="F1 V6 Turbo Hybrid: P-V Diagram",
        xaxis=dict(range=[0, 1.7], title="Volume (Normalized)"),
        yaxis=dict(range=[0, 30000], title="Pressure (kPa)"),
        hovermode="closest",
        updatemenus=[dict(
            type="buttons",
            showactive=False,
            buttons=[dict(label="Start F1 Engine", method="animate",
                          args=[None, {"frame": {"duration": 15, "redraw": True},
                                       "fromcurrent": True, "mode": "immediate"}])]
        )]
    ),
    frames=frames
)

fig.show()
```

Los cuatro tiempos observados el la gráfica del ciclo Otto consisten en: admisión (entra a la cámara de combustión la mezcla de aire-combustible), compresión (se comprime el fluido de trabajo dentro de la cámara por medio del émbolo/pistón), combustión (reacciones químicas exotérmicas que impulsan el pistón y se aprovecha ese movimiento para generar trabajo) y escape (los gases de la combustión se extraen de la cámara de combustión mediante válvulas de escape). Estos 4 tiempos constituyen el ciclo termodinámico del ICE de un Fórmula 1.

Estas máquinas generalmente están compuestas por: pistones, bielas, cigüeñales (manivelas), cilíndros, cámaras de combustión, una culata, el bloque del motor y el cárter del aceite.

## **3.2 Cámara de combustión**

La cámara de combustión constituye el volumen de control en el cual se libera la energía química del combustible mediante reacciones de oxidación. En esta región tienen lugar procesos fundamentales como la inyección, atomización, vaporización, formación de la mezcla aire–combustible y su posterior ignición (Heywood, 2018).

Desde el punto de vista constructivo, la selección de materiales representa un aspecto crítico, dado que los componentes deben soportar condiciones extremas de temperatura y presión. En este sentido, los pistones y las superficies internas suelen incorporar recubrimientos cerámicos o barreras térmicas, diseñadas para resistir temperaturas cercanas a 2600 °C, mientras que la geometría de cada componente se optimiza para maximizar el desempeño termodinámico del motor.

El análisis térmico de la cámara de combustión se aborda típicamente como un sistema cerrado transitorio durante la fase de combustión, debido a la rápida variación de las propiedades termodinámicas a lo largo del ciclo.

## **3.3 Sistema de alimentación e inyección de combustible**

El sistema de alimentación e inyección de combustible hacia la cámara de combustión constituye uno de los subsistemas más complejos del ICE, ya que debe garantizar una dosificación precisa y una adecuada atomización del combustible. Para lograrlo, el proceso de inyección se divide en dos etapas: la inyección en la precámara y la inyección directa en la cámara principal.


image.png

*Figura 1: diagrama de la pre-cámara y la cámara de combustión de un ICE de F1. (Fuente: Gómez Ruiz, 2014)*


Esta arquitectura fue introducida en la Fórmula 1 a partir de 2014, en conjunto con la implementación de las unidades de potencia híbridas con el objetivo de mejorar la eficiencia de combustión y el desempeño energético del motor.

El propósito fundamental de esta tecnología es generar un flujo altamente turbulento que incremente la velocidad de combustión, promoviendo una rápida vaporización del combustible y su posterior atomización en una mezcla lo más homogénea posible. Dicha mezcla es inyectada ~500 bar dentro de la cámara de combustión (Gómez Ruiz, 2014). Este comportamiento se logra mediante la optimización de la geometría de la precámara y del diseño de los orificios de inyección.

Los factores termodinamicos más importantes del sistema son: la relación aire-combustible (AFR), la temperatura máxima teórica de combustion y la presión y temperatura del combustible a la entrada de la cámara.

## **3.4 Sistema de admisión de aire y el turbocompresor**

El sistema de admisión de aire y el turbocompresor conforman un subsistema acoplado cuya función principal es regular y acondicionar el flujo de aire que ingresa al motor, determinando así la cantidad de oxígeno disponible para la combustión. Este sistema es fundamental para maximizar la densidad de carga y, en consecuencia, la potencia específica de la unidad de potencia. Como se muestra en la figura X, estos se cuentran a la entrada del ICE.


image.png

*Figura 1: diagrama de la ubiocación de un turbo compresor de un ICE de F1. (Fuente: Honda, s.f)*


El turbocompresor está compuesto por una turbina y un compresor montados sobre un mismo eje. La turbina extrae energía de los gases de escape mediante un proceso de expansión, mientras que el compresor utiliza esta energía para aumentar la presión del aire de admisión. Este proceso puede analizarse bajo relaciones de flujo compresible y eficiencias isentrópicas, donde el aumento de presión viene acompañado de un incremento en la temperatura del aire.

El aire comprimido presenta una mayor densidad, lo que incrementa el flujo másico de aire ($\dot{m}_{air}$) hacia la cámara de combustión. Este aumento en la masa de oxígeno disponible permite inyectar una mayor cantidad de combustible manteniendo condiciones de combustión eficientes, lo que se traduce en un incremento de la potencia generada.

## **3.5 Sistema de refrigeración**

El sistema de refrigeración tiene como función controlar el régimen térmico del motor, garantizando condiciones de operación estables y evitando fallas estructurales asociadas a sobrecalentamiento. En el ICE de Fórmula 1, donde las temperaturas de combustión pueden llegar alrededor 2600 °C, este sistema es fundamental para gestionar el calor que no es convertido en trabajo útil (Heywood, 2018). El proceso de transferencia de calor en este sistema involucra múltiples mecanismos (Çengel & Ghajar, 2015):

- Convección interna: entre los gases de combustión y las paredes del cilindro.
- Conducción: a través del material del bloque del motor.
- Convección forzada: entre las superficies del motor y el fluido refrigerante.
- Radiación: emitida por componentes críticos de la PU como el bloque del motor, el turbocompresor y el sistema de gases de escape.

La disipación de calor tiene un efecto directo sobre la temperatura de la mezcla reactiva, lo que a su vez impacta la cinética de reacción, la velocidad de combustión y la tasa de liberación de calor (HRR). Una mayor pérdida térmica reduce la temperatura pico de combustión, disminuyendo la eficiencia térmica pero también contribuyendo a la estabilidad del sistema y a la protección de sus componentes (Turns, 2012). Es por esto que la correcta estimación del sistema de refrigeración es esencial para predecir de manera realista la evolución de la temperatura, la energía liberada y el desempeño global del motor bajo diferentes condiciones de operación y tipos de combustible.



## **3.6 Sistemas de recuperación de energía**

Los sistemas de recuperación de energía (ERS) constituyen un subsistema clave dentro de la unidad de potencia de Fórmula 1, estos sisteman son los encargados de capturar, almacenar y reutilizar parte de la energía que, en un motor convencional, se disiparía como pérdidas térmicas o mecánicas.

En la normativa vigente hasta 2025, el ERS está compuesto por dos dispositivos principales (FIA, 2025):

- MGU-K (Motor Generator Unit - Kinetic):
Recupera energía cinética durante el frenado, funcionando como generador eléctrico.
- MGU-H (Motor Generator Unit - Heat):
Acoplado al turbocompresor, donde recupera energía de los gases de escape y puede convertirla en energía eléctrica o utilizarla para controlar la velocidad del turbo.

El flujo de energía entre estos dos componentes y el ICE, es decir, el flujo total de energía de la unidad de potencia de un monoplaza de F1 se puede entender a partir del siguiente diagrama de flujo.

A partir de 2026, el ERS experimenta una reconfiguración significativa, eliminando la unidad MGU-H e incrementando sustancialmente la capacidad de la MGU-K. Este cambio implica un aumento considerable en la potencia eléctrica disponible, hasta el punto en que el sistema eléctrico puede aportar aproximadamente el 50% de la energía total requerida por el monoplaza durante carrera.

La eliminación de la MGU-H responde a la necesidad de simplificar la arquitectura de la unidad de potencia y mejorar su relevancia tecnológica para aplicaciones en la industria automotriz.

En contraste, la MGU-K incrementa significativamente su capacidad tanto de recuperación como de entrega de potencia, pasando de aproximadamente 120 kW a valores cercanos a 350 kW, lo que representa casi una triplicación de su contribución energética dentro del sistema.

Adicionalmente, la nueva normativa amplía las condiciones de operación bajo las cuales es posible recuperar energía, permitiendo no solo la regeneración durante el frenado, sino también en fases de alta velocidad y desaceleración. Esto se traduce en un aumento considerable de la energía recuperada por vuelta y, por ende, en una mejora del rendimiento energético global del vehículo (FIA, 2026).

# **5. Combustible fósil**
La gasolina corriente es el combustible principal para los vehículos automóviles de uso cotidiano debido a su octanaje. El octanaje representa la medida de la resistencia a la autoignición del combustible (Instituto Tecnológico de Santo Domingo [INTEC], s. f.). Es decir, la resistencia del combustible para alcanzar su temperatura de autoencendido y evitar que esta haga combustión antes de tiempo. Un mayor octanaje representa una mayor resistencia, por lo que el motor trabaja de manera más eficientemente, por una garantía hacia el cumplimiento de los ciclos del motor. Según la literatura, la gasolina corriente tiene un octanaje cercano a los 87 octanos (Renting Colombia, s. f.), lo que la hace ideal para motores de compresión baja, mientras que combustibles de mayor octanaje son ideales para motores de alta compresión y desempeño más exigente.

La gasolina está compuesta por varios componentes, donde priman los hidrocarburos. Al no ser una sustancia pura, esta puede tener isoalcanos, aromáticos, alcanos, alquenos y cicloalcanos (entre otros), e incluso puede llegar a tener otros elementos químicos y aditivos que permiten modificar su comportamiento con el fin de lograr una mayor eficiencia en la combustión (Instituto Tecnológico de Santo Domingo [INTEC], s. f.).  La gasolina puede estar conformada por cadenas desde 5 hasta 12 carbonos, donde la cantidad de estos es determinada según sean los requermientos energéticos del combustible. Con una menor cantidad de carbonos se logra una menor volatilidad (facilidad para evaporarse a una temperatura determinada) y viceversa. Como bien se sabe, es esencial que la gasolina esté en estado de vapor para garantizar una reacción más estequiométrica y por ende, más eficiente.

Usualmente en química, la gasolina al ser un alcano sigue la fórmula general para este grupo:

$C_2H_{2n + 2}$

Ya que este combustible está compuesto por varios componentes distintos, se toma como referencia su fórmula: $C_8H_{18}$. Esta fórmula permite tener un valor intermedio para la volatilidad y también se utiliza porque esta composición tiene propiedades intermedias comparadas con otros hidrocarburos, por lo que esta formulación es adecuada para simplificar cálculos que puedan surgir a partir de la misma. Además, el octanaje se basa en compuestos derivados del octano, lo que lo convierte en una referencia estándar; por eso, aunque la gasolina es una mezcla compleja, usar $C_8H_{18}$ permite hacer análisis teóricos de forma más sencilla y consistente.

## 5.1 **Composición última**

El combustible **E10** es un tipo de gasolina, la cuál consta de gasolina corriente (isooctano) mezclado con etanol en proporciones 90% gasolina y 10% etanol (por eso su nombre). Esta mezcla permite aumentar su octanaje en un 3.16%, pasando de 95 octanos a 98, para gasolina extra (Awad et al., 2023). Adicionalmente, otro beneficio de esta mezcla es la reducción de gases contamintantes, residuos de la combustión, gracias a que el etanol contiene oxígeno en su composición, por lo que permite una reacción más eficiente en la combustión. Finalmente, el etanol puede ser producido a partir de biomasa, por lo que también puede ser una alternativa un poco más limpia para estos combustibles convencionales.

Según (Awad et al., 2023), la composición última, obtenida por medio de la prueba M1394 de laboratorio, para el E10 es la siguiente:

| Elemento     | Gasolina (%) | Etanol (%) | E10 (%) |
|--------------|:-----------:|:----------:|:-------:|
| Carbono (C)  |   87.50     |   52.35    |  81.27  |
| Hidrógeno (H)|   12.50     |   13.25    |  13.74  |
| Oxígeno (O)  |   0.00      |   34.40    |  5.60   |


Se observa que la gasolina no contiene oxígeno en su composición, mientras que el etanol presenta un alto contenido de este elemento. Como resultado, la mezcla E10 incorpora una fracción de oxígeno intermedia, lo cual influye directamente en la reducción del oxígeno externo requerido para la combustión. Para entender un poco mejor esta reacción y el balance másico del compuesto E10, se presenta el siguiente código, que requiere un input de la masa deseada de E10 y retorna la cantidad de masa de cada elemento dentro del combustible, y la cantidad de oxígeno requerida para una reacción estequiométrica. El programa también da la cantidad de dióxido de carbono producida y de agua, debido a la fórmula general de reacción:

$
aC_xH_y + bO_2 \rightarrow cCO_2 + dH_2O
$

Donde a, b, c y d son las cantidades (en kg) de cada elemento.

Para ilustrar adecuadamente la descomposición del E10 (90% gasolina y 10% etanol), el resultado del programa permitirá saber las cantidades de gasolina y etanol individualmente, siguiendo esa proporción.


```python
from IPython.display import display, Math
import numpy as np

def balance_e10_desglosado(masa_total_e10):
    """
    Calcula el balance de masa desglosando el E10 en Isooctano y Etanol
    según la proporción molar 0.9 / 0.1.
    """
    # --- Pesos Atómicos y Moleculares ---
    aw_c, aw_h, aw_o = 12.011, 1.0078, 15.999
    pm_iso = (8 * aw_c) + (18 * aw_h)          # C8H18
    pm_eta = (2 * aw_c) + (6 * aw_h) + (1 * aw_o) # C2H5OH

    # --- 1. Separación de Masas (Isooctano vs Etanol) ---
    # Sistema: m_iso + m_eta = masa_total
    #          (m_iso/pm_iso) / (m_eta/pm_eta) = 0.9 / 0.1
    relacion_molar = 0.9 / 0.1

    # Resolviendo el sistema para obtener masas individuales
    m_eta = masa_total_e10 / (1 + (relacion_molar * (pm_iso / pm_eta)))
    m_iso = masa_total_e10 - m_eta

    # --- 2. Balance de Oxígeno Externo ---
    # Usamos los porcentajes de tu tabla para mantener la fidelidad al paper
    m_c = masa_total_e10 * 0.8127
    m_h = masa_total_e10 * 0.1374
    m_o_propio = masa_total_e10 * 0.056

    pm_o2 = 31.998
    o2_necesario = (m_c * (pm_o2 / aw_c)) + (m_h * (pm_o2 / (4 * aw_h))) - m_o_propio

    # --- 3. Productos ---
    m_co2 = m_c * (44.009 / aw_c)
    m_h2o = m_h * (18.015 / (2 * aw_h))

    # --- 4. Salida Visual ---
    print(f"{'='*65}")
    print(f"   BALANCE DESGLOSADO PARA {masa_total_e10} kg DE MEZCLA E10")
    print(f"{'='*65}")
    print(f"COMPONENTES DE LA MEZCLA:")
    print(f"   - Isooctano (C8H18): {m_iso:10.4f} kg")
    print(f"   - Etanol (C2H5OH):   {m_eta:10.4f} kg")
    print(f"\nPRODUCTOS Y REACTIVOS:")
    print(f"   - O2 Externo:        {o2_necesario:10.4f} kg")
    print(f"   - CO2 Producido:     {m_co2:10.4f} kg")
    print(f"   - H2O Producida:     {m_h2o:10.4f} kg")
    print("-" * 65)

    # --- 5. Ecuación en LaTeX ---
    ecuacion = (
        rf"{m_iso:.3f} \, C_8H_{{18}} + {m_eta:.3f} \, C_2H_5OH + "
        rf"{o2_necesario:.3f} \, O_2 \rightarrow "
        rf"{m_co2:.3f} \, CO_2 + {m_h2o:.3f} \, H_2O"
    )

    print("\nEcuación de balance de masa (kg):")
    display(Math(ecuacion))

# Ejecución
try:
    m_usr = float(input("Ingrese masa total de combustible (kg): "))
    balance_e10_desglosado(m_usr)
except:
    print("Error en el dato.")
```

## 5.2 **Relación aire-combustible (AFR)**
La relación aire-combustible (Air to Fuel Ratio) es la relación existente entre las cantidades de aire y combustible suministradas al sistema para una combustión estequiométrica. Este parámetro permite identificar si la mezcla está debajo, igual o encima del valor estequiométrico. Esto con el fin de conocer si las proporciones de reactivos están siendo suministradas de manera adecuada o si se debe aumentar la cantidad de alguno de estos.

Para condiciones estequiométricas, este valor representa todo el combustible reaccionó con el oxígeno disponible en el sistema y de esta forma evitando la existencia de hidrocarburos no quemados. Una relación estequiométrica además de ser esencial para la eficiencia del sistema, también tiene un impacto ambiental, ya que para mezclas donde hay más cantidades de combustible que oxígeno (mezcla rica) habrá cantidades de hidrocarburo que no se queman, volviendo el sistema ineficiente y contaminante para el medio ambiente. Si en cambio hay una mayor cantidad de oxígeno que de combustible (mezcla pobre), la cámara de combustión puede enfriarse, reduciendo la eficiencia del sistema ya que se requiere de trabajo de entrada mayor para compensar la pérdida de temperatura y así lograr una combustión adecuada.

A partir de la ecuación química balanceada para la mezcla que compone el combustible sostenible:$C_xH_yO_z$:

$$C_xH_yO_z + a(O_2 + 3.76N_2) \rightarrow xCO_2 + \frac{y}{2}H_2O + 3.76aN_2$$

La fórmula para el $AFR_{st}$ en términos de masa es:$$AFR_{st} = \frac{m_{aire}}{m_{fuel}} = \frac{4.76a \cdot MW_{aire}}{MW_{fuel}}$$

Donde:
- $MW_{aire}$: Masa molar del aire ($\approx 28.97 \text{ g/mol}$).
- $MW_{fuel}$: Masa molar de la mezcla de combustible.
- aire = $a(O2​+3.76N2​) → n_{aire}​=a(1+3.76)=4.76a$

En condiciones reales de carrera, el motor no siempre opera en el punto estequiométrico. Es por esto que se utiliza el factor $\lambda$ para medir la desviación:
$$\lambda = \frac{AFR_{real}}{AFR_{st}}$$

Donde:

- $\lambda = 1.0$: Mezcla estequiométrica.
- $\lambda < 1.0$: Mezcla Rica (exceso de combustible). Se usa para enfriar la cámara y proteger componentes térmicos, aunque aumenta el consumo.
- $\lambda > 1.0$: Mezcla Pobre (exceso de aire). Maximiza la eficiencia del combustible, pero eleva la temperatura de combustión y genera la producción de $NO_x$.

Para el caso del E10 estequiométrico se utiliza la composición última del combustible, por lo que no se tiene en cuenta el nitrógeno en el aire. Para esto, se muestra el cálculo en el siguiente código:


```python
def calcular_afr_raw_e10(lambda_op=1.0):
    # 1. Datos de composición directos de tu tabla (Base masa)
    # Nota: Suman 1.0061, se calculan tal cual ("RAW")
    mC, mH, mO = 0.8127, 0.1374, 0.056

    # 2. Masas atómicas y constante del aire
    ma_O2, ma_C, ma_H = 31.998, 12.011, 1.0078
    y_o2_aire = 0.232  # Fracción de masa de O2 en el aire

    # 3. Cálculo de Oxígeno requerido (Estequiometría de masas)
    # kg de O2 necesarios por cada kg de "mezcla de combustible"
    o2_req_kg = (mC * (ma_O2 / ma_C)) + (mH * (ma_O2 / (4 * ma_H))) - mO

    # 4. AFR Estequiométrico (Teórico)
    afr_st = o2_req_kg / y_o2_aire

    # 5. AFR Real (Basado en Lambda de operación)
    afr_real = afr_st * lambda_op

    # --- Salida de Resultados ---
    print(f"--- CÁLCULO AFR RAW (E10) ---")
    print(f"Masa total fuel (entrada): {mC + mH + mO:.4f} kg")
    print(f"O2 requerido:              {o2_req_kg:.4f} kg_O2")
    print(f"AFR Estequiométrico:       {afr_st:.4f} kg_aire/kg_fuel")
    #print(f"AFR Real (λ = {lambda_op}):        {afr_real:.4f} kg_aire/kg_fuel")

    return afr_st, afr_real

# Ejemplo de uso:
st, real = calcular_afr_raw_e10(lambda_op=1.0)
```

Según los valores reportados en (Demirbas, 2007), el AFR del E10 es cercano a 13.81, por lo que a comparación del calculado previamente, hay un error relativo cercano al 0.14%. Este valor puede diferir un poco del "real" estequiométrico, ya que la composición última de la referencia mencionada anteriormente suma un poco más de 1 kg exacto de combustible para la entrada, sin embargo, se adopta este valor como correcto.

## 5.3 **Lower Heating Value (LHV)**
El poder calorífico inferior (LHV) es una propiedad que representa la cantidad de energía liberada durante la combustión completa de un combustible, sin considerar la recuperación del calor asociado a la condensación del vapor de agua formado. Cuando no se dispone de datos experimentales, el LHV puede estimarse a partir de la composición elemental del combustible mediante correlaciones empíricas, como la ecuación de Dulong. Esta expresión relaciona el contenido de carbono, hidrógeno y oxígeno con la energía liberada, teniendo en cuenta que el carbono y el hidrógeno son los principales aportantes energéticos, mientras que el oxígeno reduce el potencial energético al encontrarse parcialmente oxidado. En su forma modificada, la ecuación permite calcular directamente el LHV como la suma de las contribuciones del carbono y del hidrógeno efectivo (H−O/8), menos una corrección asociada al calor de vaporización del agua producida durante la combustión. De esta manera, a partir de las fracciones másicas de los elementos obtenidas en el análisis último, es posible estimar el contenido energético del combustible sin necesidad de mediciones experimentales directas. (Elfasakhany, 2016)

Por ende, la ecuación queda de la siguiente manera:

$LHV = 38.2 m_c + 84.9 (m_h - m_o / 8.0) - \Delta h_l$

Donde $m_c$ es la masa de carbono, $m_h$ la masa de hidrógeno y $m_o$ la de oxígeno. El término $\Delta h_l$ es la corrección mencionada previamente asociada al calor de vaporización de agua producida en la combustión.

A continuación se muestra el cálculo de este parámetro para el E10:


```python
def calcular_lhv_e10(delta_hl=0):
    """
    Calcula el Lower Heating Value (LHV) usando la composición última del E10.
    LHV [kJ/g] = 38.2*mC + 84.9*(mH – mO/8) – ΔHl
    """
    # 1. Fracciones de masa (m/m) del paper
    m_c = 0.8127
    m_h = 0.1374
    m_o = 0.056

    # 2. Aplicación de la fórmula
    # mH - mO/8 representa el hidrógeno disponible para quemar
    # (restando el que se combina con el oxígeno interno para formar agua)
    lhv_kj_g = 38.2 * m_c + 84.9 * (m_h - m_o / 8.0) - delta_hl

    # 3. Conversiones útiles
    lhv_mj_kg = lhv_kj_g  # kJ/g es numéricamente igual a MJ/kg

    # 4. Mostrar resultados
    print(f"{'='*45}")
    print(f"   CÁLCULO DEL PODER CALORÍFICO (LHV)")
    print(f"{'='*45}")
    print(f"Composición usada: C:{m_c*100}% | H:{m_h*100}% | O:{m_o*100}%")
    print(f"Entalpía de vaporización (ΔHl): {delta_hl} kJ/g")
    print(f"{'-'*45}")
    print(f"LHV Resultante: {lhv_kj_g:.4f} kJ/g")
    print(f"LHV Resultante: {lhv_mj_kg:.4f} MJ/kg")
    print(f"{'='*45}")

    return lhv_mj_kg

# Ejecución (puedes pasarle el valor de delta_hl si lo tienes)
# Por defecto lo dejamos en 0 si no se especifica.
resultado_lhv = calcular_lhv_e10()
```

Según los valores reportados en (Awad et al., 2023), el LHV del E10 es de 41.13 MJ/kg, lo que da un error relativo del 2.40%. Esto permite suponer que la aproximación realizada por la ecuación modificada de Dulong es adecuada y permite tener un valor cercano a los reportados por empresas productoras de este combustible.

## 5.4 **Energía liberada**

La energía liberada se expresa matemáticamente de la siguiente manera:

$Q = m_f \cdot LHV$

Donde Q es el calor liberado, $m_f$ es la masa de combustible utilizada y LHV es el lower heating value del combustible.

Este valor permite conocer la cantidad de energía que se libera durante la combustión por unidad de masa de combustible. Esta medición de energía está en unidades de J/g. La mayor cantidad de energía será obtenida cuando la mezcla esté en condiciones estequiométricas, por ende, la energía liberada dependiente del AFR y LHV queda:

```python
import numpy as np
import matplotlib.pyplot as plt

# --- CONSTANTES RAW DE TU TABLA (E10) ---
mC, mH, mO = 0.8127, 0.1374, 0.056
LHV_fuel = 42.1161  # MJ/kg

# 1. CÁLCULO ESTEQUIOMÉTRICO RAW
ma_O2, ma_C, ma_H = 31.998, 12.011, 1.0078
# Oxígeno para 1.0061 kg de combustible
o2_req_g = (mC * (ma_O2 / ma_C)) + (mH * (ma_O2 / (4 * ma_H))) - mO
afr_st = o2_req_g / 0.232

# 2. SIMULACIÓN
afr_range = np.linspace(10, 18, 1000)
q_mezcla_vector = []

for afr in afr_range:
    if afr >= afr_st:
        q_total = LHV_fuel
    else:
        q_total = LHV_fuel * ((afr * 0.232) / o2_req_g)

    q_mezcla_vector.append(q_total / (1 + afr))

# 3. MÁXIMOS
idx_max = np.argmax(q_mezcla_vector)
x_max = afr_range[idx_max]
y_max = q_mezcla_vector[idx_max]

# 4. GRÁFICA
plt.rcParams['font.family'] = 'serif'
fig, ax = plt.subplots(figsize=(12, 7), facecolor='w')

ax.plot(afr_range, q_mezcla_vector, color='#2c3e50', lw=3, label='Energía de Mezcla')
ax.axhline(y=y_max, color='orange', linestyle=':', alpha=0.6)
ax.axvline(x=afr_st, color='red', linestyle='--', alpha=0.5, label='AFR Estequiométrico')
ax.scatter(x_max, y_max, color='red', s=80, zorder=5)

ax.set_title('Calor Específico de la Mezcla E10 (Datos RAW)', fontsize=14, fontweight='bold')
ax.set_xlabel('AFR (kg aire / kg fuel)', fontsize=12)
ax.set_ylabel('Energía de la Mezcla (MJ / kg total)', fontsize=12)
ax.grid(True, ls=':', alpha=0.6)
ax.set_ylim(y_max * 0.85, y_max * 1.05)

# --- LEYENDA Y TEXTO FUERA ---
ax.legend(loc='upper left', bbox_to_anchor=(1.02, 1), frameon=True)

texto_resultados = (
    f"--- Resultados RAW ---\n"
    f"AFR Estequiométrico: {afr_st:.4f}\n"
    f"Q Máximo: {y_max:.4f} MJ/kg\n"
    f"LHV: {LHV_fuel} MJ/kg\n"
    f"Suma de masa: {mC+mH+mO:.4f}"
)

plt.text(1.02, 0.5, texto_resultados, transform=ax.transAxes,
         bbox=dict(boxstyle='round', facecolor='#f9f9f9', edgecolor='gray'))

plt.tight_layout()
plt.show()

print(f"AFR Estequiométrico RAW: {afr_st:.4f}")
```

# **6. Cinética de reacción del combustible**

A partir de la caracterización de combustible hecha previemente, es posible abordar el análisis de los mecanismos de reacción y la cinética de combustión, los cuales determinan la tasa de liberación de energía dentro de la cámara.

## **6.1 Velocidad de reacción**

### **6.1.1 Modelo Lotka Volterra**

Existen modelos que permiten describir la velocidad y comportamiento de este tipo de sistemas, como por ejemplo el modelo biológico Lotka Volterra (SIR). Este permite modelar el comportamiento de un sistema en el cual se tiene una población dividida en 3 categorías principales:

- $S(t)$ que son las personas suceptibles a infectarse por una enfermedad presente en la población.

- $I(t)$ son las personas infectadas por dicha enfermedad.

- $R(t)$ son aquellos que se han recuperado por completo.

Este modelo se definió por Kermack y McKendrick mediante las siguientes relaciones:

- $\frac{dS}{dt} = \frac{-βSI}{N}$

- $\frac{dI}{dt} = \frac{βSI}{N} - γI$

- $\frac{dR}{dt} = γI$

Donde se modela la taza de cambio para cada una de estas variables ($S, I, R$) en función del tiempo. Estas variables están directamente relacionadas a los parámetros $β$ y $γ$, siendo $β$ la tasa de contacto efectiva de la enfermedad; un individuo infectado interactúa con $\beta N$ personas adicionales por unidad de tiempo; de ellas, la proporción susceptible de contagiarse es $S/N$. Por otro lado, $\gamma$ representa la tasa promedio de recuperación, lo que implica que $1/\gamma$ corresponde al tiempo medio durante el cual un individuo infectado puede transmitir la enfermedad.

### **6.1.2 Velocidad de combustión**

La combustión no ocurre de forma instantánea al alcanzar la temperatura de ignición, es un proceso dependiente del tiempo y de la frecuencia de colisiones moleculares eficientes. Es por esto que se emplea un mecanismo cinético global de un solo paso, que simplifica las multiples reacciones intermedias de cada compuesto en una única ecuación de tasa de consumo.

La velocidad a la cual el combustible equivalente ($C_xH_yO_z$) reacciona en la cámara de combustión está definida por la concentración molar de los reactivos y la constante de velocidad específica. La velocidad de reacción se define como:

$$\frac{d[C_xH_yO_z]}{dt} = -k \cdot [C_xH_yO_z]^n \cdot [O_2]^m$$

Donde:
- $[C_xH_yO_z]$ y $[O_2]$: Son las concentraciones molares instantáneas ($\text{mol/cm}^3$).
- $n, m$: Son los órdenes de reacción empíricos, que para hidrocarburos pesados y se aproximan a $0.25$ y $1.5$ respectivamente (Westbrook & Dryer, 1981).
- $k$: Es el coeficiente específico de velocidad, que depende de la temperatura.

En este modelo el nitrógeno presente en la mezcla no se considera en el cálculo de la velocidad de reacción ya que este se toma como un componente inerte y no afecta la oxidación del combustible.

## **6.1.3 Cálculo de K, reacción en función de la temperatura**

La reactividad de la mezcla no es constante durante el tiempo de combustión, esta aumenta exponencialmente a medida que la temperatura de la llama aumenta. Esta dependencia se expresa mediante la ecuación de Arrhenius:

$$k = A e^{(-\frac{E_a}{R T})}$$

donde:
- $K$ es la constante de la rapidez
- $A$ es el factor de frecuencia
- $E_{\rm a}$ es la energía de activación necesaria
- $T$ es la temperatura
- $R$ es la constante universal de gases, $R = 1.987$ cal/K$⋅$mol

El valor de la energía de activación ($E_a$) utilizado en este modelo se deriva de la caracterización cinética de combustibles sustitutos (surrogates). Dado que la composición última del combustible presenta comportamientos similares al isooctano y mezclas de aromáticos, se adoptan los parámetros encontrados en la literatura (Westbrook & Dryer), los cuales han sido obtenidos mediante el ajuste de curvas experimentales en reactores de flujo y tubos de choque, asegurando que la tasa de liberación de calor en la simulación coincida con el comportamiento real del frente de llama.

## **6.1.4 Parámetros cinéticos del modelo**

El modelo cinético propuesto se basa en un mecanismo global de un solo paso para la combustión de un combustible equivalente del tipo $C_xH_yO_z$. Los parámetros utilizados corresponden a correlaciones empíricas ampliamente reportadas en la literatura para la oxidación de hidrocarburos en fase gaseosa.

Los parámetros empleados en el modelo son:

### **Constante cinética ($k$)**

Se modela mediante la ecuación de Arrhenius:

$$
k = A \, e^{-\frac{E_a}{RT}}
$$

donde:

- $A$: factor pre-exponencial  
- $E_a$: energía de activación  
- $R$: constante universal de los gases ($8.314 \, \text{J/mol·K}$)  
- $T$: temperatura absoluta  

Valores típicos adoptados (Westbrook & Dryer, 1981):

- $A = 5.0 \times 10^8 \, \text{(cm}^3/\text{mol)}^{(m+n-1)} \text{s}^{-1}$  
- $E_a = 125 \, \text{kJ/mol}$  



### **Órdenes de reacción ($n$, $m$)**

Los órdenes de reacción se consideran empíricos y se obtienen a partir de ajustes experimentales para mecanismos globales:

- $n = 0.25$ (dependencia con el combustible)  
- $m = 1.5$ (dependencia con el oxígeno)  

Estos valores han sido reportados para la combustión de hidrocarburos pesados y mezclas representativas de combustibles líquidos.



### **Coeficiente estequiométrico ($a$)**

El coeficiente estequiométrico del oxígeno se determina a partir del balance de masa del combustible equivalente:

$$
C_xH_yO_z + a O_2 \rightarrow b CO_2 + c H_2O
$$

donde:

$$
a = x + \frac{y}{4} - \frac{z}{2}
$$

Este parámetro depende directamente de la composición elemental del combustible.



### **Poder calorífico inferior ($LHV$)**

El LHV se estima a partir de la composición elemental del combustible mediante la ecuación de Dulong modificada:

$$
LHV = 38.2\,m_C + 84.9(m_H - \frac{m_O}{8}) - \Delta H_l
$$

Para el caso del E10, se obtiene:

- $LHV \approx 42 \, \text{MJ/kg}$  



##**6.2 Modelo cinético**

Teniendo en cuenta lo anterior, se plantea el siguiente modelo cinético haciendo la relación entre la velocidad de reacción y el modelo biológico Lotka Volterra:

Se definen las siguientes variables dependientes del tiempo:

- $F(t)$: concentración molar del combustible equivalente $C_xH_yO_z$  
- $O(t)$: concentración molar de oxígeno  
- $P(t)$: concentración de productos de combustión  

El sistema de ecuaciones diferenciales que describe la dinámica del proceso es:

$$
\frac{dF}{dt} = -k \, F^n \, O^m
$$

$$
\frac{dO}{dt} = -a \, k \, F^n \, O^m
$$

$$
\frac{dP}{dt} = k \, F^n \, O^m
$$

donde:

- $k$: constante cinética dependiente de la temperatura  
- $n, m$: órdenes de reacción empíricos  
- $a$: coeficiente estequiométrico del oxígeno  

La primera ecuación describe el consumo del combustible, el cual disminuye a una tasa proporcional a la frecuencia de colisiones efectivas con el oxígeno. La segunda ecuación representa el consumo de oxígeno, ajustado por la estequiometría de la reacción, mientras que la tercera describe la formación de productos como resultado de la combustión.

La tasa de liberación de energía del sistema se modela como:

$$
\frac{dQ}{dt} = LHV \cdot k \, F^n \, O^m
$$

donde $LHV$ corresponde al poder calorífico inferior del combustible. Esta expresión permite relacionar directamente la cinética química con la energía generada durante el proceso de combustión.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

def simular_cinetica_e10_final_v2():
    # 1. DATOS EXACTOS DE LA TABLA (Imagen del paper)
    mC_p, mH_p, mO_p = 0.8127, 0.1374, 0.0560
    LHV_mass = 42.1161  # MJ/kg (Valor exacto de la tabla)

    mw_c, mw_h, mw_o = 12.011, 1.0078, 15.999
    C_val = (mC_p * 100) / mw_c
    H_val = (mH_p * 100) / mw_h
    O_val = (mO_p * 100) / mw_o

    # 2. PARÁMETROS WESTBROOK & DRYER
    A, Ea, R, T = 5.0e8, 125000, 8.314, 1200
    n_order, m_order = 0.25, 1.5
    k_arrhenius = A * np.exp(-Ea / (R * T))

    # 3. TERMODINÁMICA
    a_req = C_val + (H_val/4) - (O_val/2)
    MW_fuel = (C_val * mw_c) + (H_val * mw_h) + (O_val * mw_o)
    LHV_molar = LHV_mass * (MW_fuel / 1000)

    # 4. SISTEMA DE ECUACIONES
    def dY_dt(t, y):
        Fuel, O2, CO2, H2O, Q = y
        # Estricta positividad para evitar errores con n=0.25
        Fuel_safe = max(Fuel, 0.0)
        O2_safe = max(O2, 0.0)

        rate = k_arrhenius * (Fuel_safe**n_order) * (O2_safe**m_order)

        return [-rate, -a_req*rate, C_val*rate, (H_val/2)*rate, LHV_molar*rate]

    # 5. RESOLUCIÓN
    t_final_ms = 0.6
    t_span = (0, t_final_ms / 1000)
    y0 = [1.0, a_req, 0.0, 0.0, 0.0] # Fuel empieza en 1.0 exacto

    sol = solve_ivp(dY_dt, t_span, y0, method='BDF', t_eval=np.linspace(0, t_span[1], 1000), atol=1e-8, rtol=1e-6)

    # 6. VISUALIZACIÓN
    plt.rcParams['font.family'] = 'serif'
    # Aumentamos el ancho para que la leyenda quepa a la derecha
    fig, ax1 = plt.subplots(figsize=(12, 6))

    ax1.plot(sol.t * 1000, sol.y[0], color='#1f77b4', lw=3, label='Combustible')
    ax1.plot(sol.t * 1000, sol.y[1], color='#2ca02c', lw=3, label='Oxígeno ($O_2$)')
    ax1.plot(sol.t * 1000, sol.y[2], color='#d62728', ls='--', lw=2, label='$CO_2$')
    ax1.plot(sol.t * 1000, sol.y[3], color='#17becf', ls='--', lw=2, label='$H_2O$')

    ax1.set_xlabel('Tiempo (ms)', fontsize=12, fontweight='bold')
    ax1.set_ylabel('Concentración (moles)', fontsize=12, fontweight='bold')
    ax1.set_xlim(0, t_final_ms)
    ax1.set_ylim(0, None) # <--- Evita que se vea negativo
    ax1.grid(True, linestyle=':', alpha=0.6)

    ax2 = ax1.twinx()
    ax2.plot(sol.t * 1000, sol.y[4], color='#ff7f0e', lw=4, label='Energía $Q$')
    ax2.set_ylabel('Energía Liberada (MJ / mol)', color='#ff7f0e', fontsize=12, fontweight='bold')
    ax2.tick_params(axis='y', labelcolor='#ff7f0e')
    ax2.set_ylim(0, None)

    plt.title(f'Dinámica de Combustión Global E10 (LHV={LHV_mass} MJ/kg)\nModelo Westbrook & Dryer @ {T}K', fontsize=14, pad=15)

    # --- AJUSTE DE LEYENDA (Fuera del gráfico) ---
    lines, labels = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax1.legend(lines + lines2, labels + labels2,
               loc='upper left', bbox_to_anchor=(1.15, 1), # La mueve a la derecha
               frameon=True, edgecolor='black')

    plt.tight_layout() # Ajusta el espacio para que no se corte la leyenda
    plt.show()

simular_cinetica_e10_final_v2()
```

# **7. Combustible Sostenible Avanzado**

## **7.1 Nueva normativa**

Con la reciente actualización normativa, se introduce la implementación de combustibles sostenibles avanzados, definidos como aquellos derivados de biomasa no alimentaria, residuos municipales o combustibles renovables de origen no biológico (RFNBO) (Mercedes-AMG PETRONAS F1 Team, s. f.).

La biomasa no alimentaria se obtiene a partir de materia orgánica no destinada al consumo humano, generalmente subproductos agrícolas como cáscaras de banano, residuos de arroz o bagazo de caña, entre otros. Por otra parte, los residuos municipales corresponden a desechos domésticos previamente utilizados, tales como aceites de cocina usados, papel e incluso materiales provenientes de muebles y electrodomésticos. En cuanto a los RFNBO, estos consisten en combustibles sintéticos producidos mediante el uso de energía renovable para generar hidrógeno ($H^+$), el cual posteriormente se combina con dióxido de carbono capturado para formar hidrocarburos sintéticos.

El objetivo principal de la implementación de estos combustibles es lograr una reducción mínima del 65% en las emisiones de gases de efecto invernadero asociadas al ciclo de vida del combustible, en concordancia con los estándares establecidos en la Directiva de Energías Renovables de la Unión Europea (Comisión Europea, s. f.).

Dado que la composición específica del combustible utilizado por cada equipo de Fórmula 1 es confidencial, no es posible establecer una reacción química global representativa. No obstante, la normativa FIA 2026 clasifica los componentes del combustible según su estructura química y su función en el proceso de combustión, estableciendo restricciones específicas para cada categoría con el fin de garantizar condiciones de seguridad, eficiencia y sostenibilidad (FIA, 2026). Las restricciones y categorias se muestran a continuación:

### 1. Parafinas (Alcanos)
- **Ejemplos:** Iso-octano, N-heptano  
- **Función:** Constituyen la base energética del combustible, proporcionando estabilidad química y una combustión controlada.  
- **Restricción:** No presentan un límite directo; sin embargo, su proporción influye en propiedades globales como la densidad y el número de octano.


### 2. Aromáticos
- **Ejemplos:** Tolueno, Xileno.  
- **Función:** Incrementan el número de octano, reduciendo la tendencia a la detonación durante la compresión.
- **Restricción:** ≤ 0.40 en fracción másica.

### 3. Olefinas (Alquenos)
- **Ejemplos:** Propileno, Buteno.  
- **Función:** Aumentan la reactividad y velocidad de llama en el proceso de combustion.
- **Restricción:** ≤ 0.17 en fracción másica.

### 4. Diolefinas
- **Ejemplo:** Isopreno, Butadieno.
- **Función:** Contribuyen al incremento del octanaje, aunque presentan alta inestabilidad química.
- **Restricción:** ≤ 0.005 en fracción másica.

### 5. Compuestos Oxigenados (Alcoholes)
- **Ejemplos:** Etanol, Butanol, MTBE  
- **Función:** Mejoran la calidad de la combustión al aportar oxígeno a la mezcla, incrementando el octanaje y favoreciendo una oxidación más completa.
- **Restricción:** 0.067 – 0.071 en fracción másica de contenido de oxígeno.

### 6. Aditivos
- **Ejemplos:** Antioxidantes, detergentes, inhibidores de corrosión.  
- **Función:** Mejoran la estabilidad química del combustible y protegen los componentes del motor.  
- **Restricción en el reglamento:** ≤ 0.01 en fracción másica.


De acuerdo con la normativa vigente, todos los componentes del combustible deben ser de origen sostenible. El cumplimiento de los límites establecidos se evalúa principalmente en términos de fracción másica o volumétrica, dependiendo del parámetro en cuestión. Entre estos, el contenido de oxígeno constituye uno de los factores más críticos y restrictivos en la formulación, debido a su influencia directa sobre la combustión y la eficiencia energética del motor.

## **7.2 Balance de masa**

Dado que estos combustibles están compuesto por una mezcla compleja de múltiples componentes, se plantea modelarlo como un combustible equivalente único que reacciona con aire. Bajo estas suposiciones, la reacción global de combustión puede expresarse de la siguiente forma:

$$C_xH_yO_z + a(O_2 + 3.76N_2) → xCO_2 + \frac{y}{2}H_2O + 3.76aN_2$$

Donde el coeficiente estequiométrico de oxígeno
a:

 $$a = x + \frac{y}{4} - \frac{z}{2}$$.

Y las fracciones atómicas ($x,y,z$) se calculan como el promedio ponderado de los átomos presentes en cada componente de la mezcla.

- Átomos de carbono totales: $ x = ∑(C_iX_i)$
- Átomos de hidrógeno totales: $ y = ∑(H_iX_i)$
- Átomos de oxígeno totales: $ y = ∑(O_iX_i)$

Donde $X_i$ es la fracción molar de cada componente en la mezcla.


## **7.3 Cálculo de la fórmula global de reacción**

Considerando que no existe una composición única para estos combustibles, se propone el desarrollo de un modelo computacional en Python que permita al usuario definir la composición del combustible dentro de los límites establecidos por la normativa FIA. El programa presenta los compuestos disponibles, su clasificación y sus restricciones, modificar sus porcentajes por medio de sliders. A partir de estos, el modelo verifica el cumplimiento normativo de la mezcla y calcula la reacción estequiométrica correspondiente para su posterior análisis termodinámico.

El propósito principal del modelo es permitir la exploración de diferentes composiciones vaiando los parámetros de composición del combustible, con el fin de identificar configuraciones óptimas que maximicen la eficiencia energética y el rendimiento del motor bajo las restricciones impuestas por la normativa.


```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# ==============================
# BASE DE DATOS Y CONFIGURACIÓN
# ==============================
fuel_db = {
    "iso-octano": {"C":8,"H":18,"O":0,"M":114.23,"type":"parafina"},
    "n-heptano": {"C":7,"H":16,"O":0,"M":100.2,"type":"parafina"},
    "tolueno": {"C":7,"H":8,"O":0,"M":92.14,"type":"aromatico"},
    "benceno": {"C":6,"H":6,"O":0,"M":78.11,"type":"aromatico"},
    "xileno": {"C":8,"H":10,"O":0,"M":106.16,"type":"aromatico"},
    "etileno": {"C":2,"H":4,"O":0,"M":28.05,"type":"olefina"},
    "propileno": {"C":3,"H":6,"O":0,"M":42.08,"type":"olefina"},
    "buteno": {"C":4,"H":8,"O":0,"M":56.11,"type":"olefina"},
    "etanol": {"C":2,"H":6,"O":1,"M":46.07,"type":"oxigenado"},
    "butanol": {"C":4,"H":10,"O":1,"M":74.12,"type":"oxigenado"},
    "mtbe": {"C":5,"H":12,"O":1,"M":88.15,"type":"eter"},
    "isopreno": {"C":5,"H":8,"O":0,"M":68.12,"type":"diolefina"},
    "aditivo": {"C":0,"H":0,"O":0,"M":100.0,"type":"aditivo"}
}

cat_styles = {"aromatico": "#FFCDD2", "olefina": "#FFF9C4", "oxigenado": "#C8E6C9", "eter": "#E1BEE7", "diolefina": "#FFCCBC", "aditivo": "#F5F5F5"}
max_limits = {"aromatico": 40.0, "olefina": 17.0, "diolefina": 0.5, "aditivo": 1.0, "oxigenado": 30.0, "eter": 30.0}

# Variables Globales de Comunicación
global_composition = {"C": 0, "H": 0, "O": 0, "M_avg": 0}
global_mix = {}

def fuel_simulator(base_comp, **kwargs):
    global global_composition, global_mix
    other_components = kwargs
    sum_others = sum(other_components.values())

    # El componente base (parafina) absorbe el resto
    base_val = max(0, 100.0 - sum_others)
    mix_pct = {base_comp: base_val}
    mix_pct.update(other_components)

    mix = {k: v/100.0 for k, v in mix_pct.items() if v > 0}

    aromaticos = olefinas = diolefinas = oxigenados = aditivos = 0
    C_tot = H_tot = O_tot = masa_total = 0

    for comp, frac in mix.items():
        tipo = fuel_db[comp]["type"]
        if tipo == "aromatico": aromaticos += frac
        elif tipo == "olefina": olefinas += frac
        elif tipo == "diolefina": diolefinas += frac
        elif tipo in ["oxigenado", "eter"]: oxigenados += frac
        elif tipo == "aditivo": aditivos += frac

        masa_total += fuel_db[comp]["M"] * frac
        C_tot += fuel_db[comp]["C"] * frac
        H_tot += fuel_db[comp]["H"] * frac
        O_tot += fuel_db[comp]["O"] * frac

    O_wt = (16 * O_tot) / masa_total if masa_total > 0 else 0

    print(f"\n{'='*60}")
    print(f" Simulador de composición de combustible | Base: {base_comp.upper()} ({base_val:.2f}%) Jardero")
    print(f"{'='*60}")

    print("\n Composición final:")
    for c, v in mix_pct.items():
        if v > 0: print(f"  - {c.capitalize():12}: {v:6.2f}%")

    # Combustión con Aire (O2 + 3.76 N2)
    b = C_tot
    c = H_tot / 2
    a = (2*b + c - O_tot) / 2
    n2 = a * 3.76

    # Salida en consola (Texto plano)
    print("\n" + "="*60)
    print("\n Reacción exotérmica):")
    print(f"  C{C_tot:.2f}H{H_tot:.2f}O{O_tot:.2f} + {a:.2f}(O2 + 3.76N2) →")
    print(f"  → {b:.2f}CO2 + {c:.2f}H2O + {n2:.2f}N2")
    print("="*60)

    print(f"\n Validación de la normativa:")
    status = lambda cond: "✅" if cond else "❌"
    print(f" {status(aromaticos <= 0.40)} Aromáticos: {aromaticos*100:5.2f}% (Máx 40%)")
    print(f" {status(olefinas <= 0.17)} Olefinas:   {olefinas*100:5.2f}% (Máx 17%)")
    print(f" {status(diolefinas <= 0.005)} Diolefinas: {diolefinas*100:5.3f}% (Máx 0.5%)")
    print(f" {status(0.066 <= O_wt <= 0.071)} Oxígeno wt:  {O_wt:.4f} (Rango 0.067-0.071)")


    # ACTUALIZACIÓN GLOBAL
    global_composition = {"C": C_tot, "H": H_tot, "O": O_tot, "M_avg": masa_total}
    global_mix = mix


def build_interface(change):
    clear_output(wait=True)
    display(base_selector)
    current_base = base_selector.value

    sliders = {}
    grouped_widgets = []

    # Crear grupos de sliders por categoría
    for cat, color in cat_styles.items():
        cat_components = [k for k, v in fuel_db.items() if v["type"] == cat]
        if not cat_components: continue

        cat_sliders = []
        cat_label = widgets.HTML(f"<b style='color:#333'>{cat.upper()}</b>")
        cat_sliders.append(cat_label)

        for comp in cat_components:
            max_v = max_limits.get(cat, 100.0)
            s = widgets.FloatSlider(
                value=0, min=0, max=max_v, step=0.01 if max_v <= 1 else 0.1,
                description=comp.capitalize(),
                continuous_update=True,
                style={'slider_color': 'white'} # El color del track se maneja por el Box
            )
            sliders[comp] = s
            cat_sliders.append(s)

        # Agrupar en una caja con color de fondo
        box = widgets.VBox(cat_sliders, layout=widgets.Layout(
            background_color=color, padding='10px', margin='5px', border='1px solid #ddd', border_radius='8px'
        ))
        grouped_widgets.append(box)

    # Organizar en 2 columnas
    ui_layout = widgets.GridBox(grouped_widgets, layout=widgets.Layout(
        grid_template_columns='repeat(2, 1fr)'
    ))

    out = widgets.interactive_output(lambda **kwargs: fuel_simulator(current_base, **kwargs), sliders)
    display(ui_layout, out)

# Define base_selector before its first use
paraffin_options = [k for k, v in fuel_db.items() if v["type"] == "parafina"]
base_selector = widgets.Dropdown(
    options=paraffin_options,
    value=paraffin_options[0] if paraffin_options else None,
    description='Base Fuel:',
    disabled=False,
)

base_selector.observe(build_interface, names='value')
build_interface(None)
```

## **8. Modelo cinético**

Utilizando un modelo SIR se planta el modelo de la reacción de oxidación combustible sostenible empleado en el carro de Fórmula 1. Sabiendo que la reacción global del combustible es:

$$C_xH_yO_z + a(O_2 + 3.76N_2) → xCO_2 + \frac{y}{2}H_2O + 3.76aN_2$$

Se plantea que las cantidades S, I y R representarán las concentraciónes de
$C_xH_yO_z$, $O_2 + N_2$ y $CO_2 + H_2O + N_2$ respectivamente. Las ecuaciónes diferenciales que describem este modelo SIR son:

$$\frac{d[C_xH_yO_z]}{dt} = -\beta \cdot \frac{[C_xH_yO_z] \cdot [O_2 + N_2]}{N}$$
$$\frac{d[O_2 + N_2]}{dt} = \beta \cdot \frac{[C_xH_yO_z] \cdot [O_2 + N_2]}{N} - \gamma \cdot [O_2 + N_2]$$
$$\frac{d[CO_2 + H_2O + N_2]}{dt} = \gamma \cdot [O_2 + N_2]$$

Donde:
- $\beta$ es la tasa de reacción entre $C_xH_yO_z$ y $O_2 + N_2$.
- $\gamma$ es la tasa de formación de $CO_2 + H_2O + N_2$.
- N es la concentración total de moléculas.

A continuación se presenta la gráfica del modelo dinámico del proceso de combustion, la fórmula equivalente que se modela es la obtenida a partir de la sección 8.3 del programa de variación paramétrica.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

def graficar_cinetica():
    C, H, O = global_composition["C"], global_composition["H"], global_composition["O"]
    if C == 0: return print("⚠️ Define el combustible en la Celda 1.")

    a_req = C + (H/4) - (O/2)
    def odes(t, y):
        rate = 8000 * max(y[0], 0) * (max(y[1], 0)/a_req) if a_req > 0 else 0
        return [-rate, -a_req*rate, C*rate, (H/2)*rate]

    sol = solve_ivp(odes, (0, 0.002), [1.0, a_req, 0, 0], method='BDF', t_eval=np.linspace(0, 0.002, 500))

    plt.figure(figsize=(10, 4))
    plt.plot(sol.t*1000, sol.y[0], label='Combustible', lw=2)
    plt.plot(sol.t*1000, sol.y[1], label='Oxígeno', lw=2)
    plt.plot(sol.t*1000, sol.y[2], '--', label='CO2')
    plt.plot(sol.t*1000, sol.y[3], '--', label='H2O')
    plt.title("Cinética Global de la Combustión", fontweight='bold')
    plt.xlabel("Tiempo (ms)"); plt.ylabel("Moles Relativos"); plt.legend(); plt.grid(alpha=0.3)
    plt.show()

graficar_cinetica()
```

## **8.1 Balance de energía térmica**

El balance de energía permite cuantificar la transformación de la energía química almacenada en los enlaces moleculares del combustible en energía térmica (calor). Para un motor de F1, este cálculo es el punto de partida para el diseño de los sistemas de refrigeración y el estudio de la fatiga térmica en componentes.

### **8.1.1 Entalpía de formación y calor de reacción**

La liberación de energía en la combustión es consecuencia del reordenamiento de los enlaces atómicos. Pasamos de moléculas con alta energía potencial (combustibles) a moléculas muy estables y de baja energía potencial ($CO_2$ y $H_2O$).Para calcular el calor total de la reacción se emplea la siguiente ecuación:

$$ΔH_r^° = ΣΔH^°_f (Productos) - ΣΔH^°_f (Reactivos)$$
Donde:
- O significa estándar
- $_r$ de reacción
- $_f$ de formación

Como se menciona en el cuadernillo de curso número 5, la energía térmica neta se calcula como la diferencia de entalpias entre los productos y los reactantes. Dado que los oxidantes ($O_2$ y $N_2$) se encuentran en su estado estándar, sus entalpías de formación son 0. Por lo tanto, el calor de reacción depende depende únicamente del combustible y de los productos, la energía térmica neta se defina como:

$$ΔH_r^∘​=[x⋅ΔH^°_{f,CO2}​+\frac{y}{2}ΔH^°_{f,H2​O}​]−[∑​X_i​⋅ΔH^°_{f,i}​]$$

Con: $$x=∑​X_i​⋅n_{C,i​}$$ $$y=∑​X_i​⋅n_{H,i​} $$

Donde:
- $x$ es el número total de átomos de carbono.
- $y$ es el número total de átomos de hidrógeno.
- $i$ es el indice que identifica que componente se esta sumando.
- $X_i$ es la fracción del componente.
- $∑​X_i​⋅ΔH^°_{f,i}$ es la entalpía de formación del componente del combustible.
Por ejemplo: $X_{etanol}​⋅ΔH^°_{f,etanol}​$

### **8.1.2 Poder calorífico y LHV**

El poder calorífico se obtiene directamente a partir del calor de reacción calculado previamente:

$$LHV_{mezcla} = -\frac{\Delta H_{r,mezcla}^\circ}{\sum_i X_i \cdot M_i} $$

Donde:
- $ΔH_r^∘$ es el calor de reacción (KJ/mol).
- $(X_i)$ es la fracción del componente $(i)$.
- $(M_i)$ es la masa molar del componente $(i)$.

Para una mexcla con múltiples componentes como lo es el combustible sostenible utilizado, el LHV también se puede expresar como:

$$LHV_{mezcla} = \sum_i X_i \cdot LHV_i$$

Donde:

- $(X_i)$ es la fracción del componente $(i)$.
- $(LHV_i)$ es el poder calorífico de cada componente.

### **8.1.3 Tasa de liberación de Calor (HRR)**

La potencia térmica instantánea $\dot{Q}(t)$ no es constante; sigue la curva dinámica del modelo SIR. Se define como el producto de la velocidad de consumo de combustible por la energía específica:

$$\dot{Q}(t) = \left( -\frac{dm_{fuel}}{dt} \right) \cdot LHV_{mix}$$

Donde:
- ($\frac{dm_{fuel}}{dt}$) es la tasa de consumo de combustible.
- ($LHV_{mix}$) es el poder calorífico inferior de la mezcla.

Integrando la tasa de liberación de calor en el tiempo, se obtiene la energía total liberada durante el proceso de combustión:

$$Q_{total} = \int \dot{Q}(t) dt$$

Esta energía representa la energía química convertida durante la combustión.

### **8.1.4 Implementación computacional del balance térmico**

El bloque de código a continuación toma la fracción de la mezcla generada en la sección 8.3 y calcula la entalpía de reacción teórica, determina el LHV exacto y utiliza la curva de consumo de combustible del modelo SIR de la sección 9.2 para graficar la Tasa de Liberación de Calor (HRR).




```python
# ====================================================
# 9.3.4 BALANCE TÉRMICO (Q = MASA * LHV + GRÁFICAS)
# ====================================================
import numpy as np
import matplotlib.pyplot as plt

def simular_q_liberado_validado():
    try:
        mix = global_mix
        C_val = global_composition["C"]
        H_val = global_composition["H"]
        O_val = global_composition["O"]
        MW_mix = global_composition["M_avg"]
    except:
        print("⚠️ Error: Ejecuta la Celda 1 para cargar la mezcla.")
        return

    # 1. --- TERMODINÁMICA ---
    hf_db = {"iso-octano":-224.1, "n-heptano":-187.8, "tolueno":12.0, "etanol":-235.3, "mtbe":-313.6, "buteno":-0.1, "isopreno":75.8}
    hf_reac = sum(hf_db.get(c, 0) * frac for c, frac in mix.items())
    hf_prod = (C_val * -393.5) + ((H_val / 2) * -241.8)
    lhv_mj_kg = (abs(hf_prod - hf_reac) / (MW_mix / 1000)) / 1000

    # --- CÁLCULO DEL AFR ESTEQUIOMÉTRICO ---
    n_o2_st = C_val + H_val/4 - O_val/2
    afr_st = (n_o2_st * (1/0.21) * 28.97) / MW_mix
    q_mix_max = lhv_mj_kg / (1 + afr_st)

    # 2. --- CINEMÁTICA Y MASA ---
    flujo_fia_kg_s = 100 / 3600
    rpm = 12000
    ciclos_por_seg = rpm / 120
    masa_ciclo_kg = flujo_fia_kg_s / (ciclos_por_seg * 6)

    q_ciclo_j = masa_ciclo_kg * (lhv_mj_kg * 1e6)
    q_dot_mw = flujo_fia_kg_s * lhv_mj_kg
    potencia_mecanica_hp = (q_dot_mw * 0.50) * 1341.02

    # 3. --- GENERACIÓN DE DATOS PARA GRÁFICAS ---
    t_ms = np.linspace(0, 2, 1000)
    k_hrr = 3000
    hrr_mw = (q_ciclo_j * k_hrr * np.exp(-k_hrr * (t_ms/1000))) / 1e6

    afr_range = np.linspace(8, 20, 200)
    # Energía de mezcla: LHV*(AFR/AFRst)/(1+AFR) en rica, LHV/(1+AFR) en pobre
    q_mix_vector = [((lhv_mj_kg * (a/afr_st))/(1+a) if a <= afr_st else lhv_mj_kg/(1+a)) for a in afr_range]

    # 4. --- RESULTADOS IMPRESOS ---
    print(f"{'='*55}")
    print(f"       BALANCE TÉRMICO DIRECTO (Q = m * LHV)")
    print(f"{'='*55}")
    print(f"🔹 LHV de la Mezcla:               {lhv_mj_kg:.4f} MJ/kg_fuel")
    print(f"🔹 AFR Estequiométrico:            {afr_st:.4f}")
    print(f"🔥 Q MÁXIMO MEZCLA (Pico):         {q_mix_max:.4f} MJ/kg_mix")
    print(f"{'-'*55}")
    print(f"🔹 Masa de fuel (1 ciclo/cil):     {masa_ciclo_kg*1000:.4f} g")
    print(f"🔥 Calor Liberado por Ciclo (Q):   {q_ciclo_j:.2f} Joules")
    print(f"⚙️ Potencia Estimada (50% ef.):    {potencia_mecanica_hp:.0f} HP")
    print(f"{'='*55}")

    # 5. --- VISUALIZACIÓN ---
    fig, ax = plt.subplots(1, 2, figsize=(15, 5))

    # Gráfica 1: Perfil de Liberación de Calor
    ax[0].plot(t_ms, hrr_mw, color='#D90429', lw=2.5)
    ax[0].fill_between(t_ms, hrr_mw, color='#D90429', alpha=0.1)
    ax[0].set_title("Perfil Térmico (HRR)", fontweight='bold')
    ax[0].set_xlabel("Tiempo (ms)"); ax[0].set_ylabel("Potencia Térmica [MW]")
    ax[0].grid(alpha=0.3)

    # Gráfica 2: Energía de Mezcla vs AFR
    ax[1].plot(afr_range, q_mix_vector, color='#2B2D42', lw=2.5)
    ax[1].axvline(afr_st, color='red', ls='--', label=f'ST: {afr_st:.2f}')
    ax[1].scatter(afr_st, q_mix_max, color='red', zorder=5)
    ax[1].set_title("Energía Específica de la Mezcla", fontweight='bold')
    ax[1].set_xlabel("Air-Fuel Ratio (AFR)"); ax[1].set_ylabel("MJ / kg_total")
    ax[1].legend(); ax[1].grid(alpha=0.3)

    plt.tight_layout()
    plt.show()

simular_q_liberado_validado()
```

```python
# ==============================================================
# SECCIÓN 2: SIMULACIÓN DE MEZCLA ÓPTIMA (RESTURADA + Q_MIX)
# Basado en: C6.91 H15.56 O0.32
# ==============================================================
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

def simulacion_optima_f1():
    # 1. --- DEFINICIÓN DE LA MEZCLA ÓPTIMA ---
    C, H, O = 6.42, 14.59, 0.44
    M_fuel = (C*12.011 + H*1.008 + O*15.999) # g/mol

    # Parámetros de la reacción (Moles Estequiométricos)
    n_o2_st = C + H/4 - O/2
    afr_st = (n_o2_st * (1/0.21) * 28.97) / M_fuel

    # 2. --- BALANCE TÉRMICO (LHV & POTENCIA) ---
    hf_reac = -240.5
    hf_prod = (C * -393.5) + ((H/2) * -241.8)
    lhv = (abs(hf_prod - hf_reac) / (M_fuel/1000)) / 1000 # MJ/kg

    # Cálculo del Q máximo de la mezcla (LHV / (AFR+1))
    q_mix_max = lhv / (1 + afr_st)

    # Potencia bajo límite FIA (100 kg/h @ 50% eficiencia)
    flujo_kg_s = 100 / 3600
    potencia_mw_mec = (flujo_kg_s * lhv) * 0.50
    hp_final = potencia_mw_mec * 1341.02

    # 3. --- CINÉTICA QUÍMICA (ODE) - ORIGINAL ---
    def cinetica(t, y):
        fuel, o2, co2, h2o = y
        k = 3500 # Constante de velocidad (ajustada a 2ms)
        r = k * fuel * (o2 / n_o2_st)
        return [-r, -n_o2_st*r, C*r, (H/2)*r]

    t_span = (0, 0.002)
    t_eval = np.linspace(0, 0.002, 1000)
    y0 = [1.0, n_o2_st, 0.0, 0.0]
    sol = solve_ivp(cinetica, t_span, y0, t_eval=t_eval)

    # 4. --- CÁLCULO DE HRR (Tasa de Liberación de Calor) ---
    energia_ciclo_j = (flujo_kg_s / ((12000/120) * 6)) * (lhv * 1e6)
    k_hrr = 3000
    hrr_mw = (energia_ciclo_j * k_hrr * np.exp(-k_hrr * t_eval)) / 1e6

    # 5. --- VISUALIZACIÓN TRIPLE ---
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    plt.subplots_adjust(wspace=0.3)

    # Gráfica A: Cinética (Original)
    axes[0].plot(sol.t*1000, sol.y[0], label='Combustible', lw=2)
    axes[0].plot(sol.t*1000, sol.y[1], label='Oxígeno', lw=2)
    axes[0].plot(sol.t*1000, sol.y[2], '--', label='CO2')
    axes[0].plot(sol.t*1000, sol.y[3], '--', label='H2O')
    axes[0].set_title("Cinética Global de Combustión", fontweight='bold')
    axes[0].set_xlabel("Tiempo (ms)"); axes[0].set_ylabel("Moles Relativos"); axes[0].legend(); axes[0].grid(alpha=0.3)

    # Gráfica B: Perfil Térmico (Original)
    axes[1].plot(t_eval*1000, hrr_mw, color='#D90429', lw=3)
    axes[1].fill_between(t_eval*1000, hrr_mw, color='#D90429', alpha=0.1)
    axes[1].set_title(f"Perfil Térmico (Pico: {np.max(hrr_mw):.2f} MW)", fontweight='bold')
    axes[1].set_xlabel("Tiempo (ms)"); axes[1].set_ylabel("Potencia [MW]"); axes[1].grid(alpha=0.3)

    # Gráfica C: Barrido de AFR (Energía Mezcla)
    afr_range = np.linspace(8, 20, 200)
    q_mix = [((lhv * (a/afr_st))/(1+a) if a <= afr_st else lhv/(1+a)) for a in afr_range]
    axes[2].plot(afr_range, q_mix, color='#2B2D42', lw=3)
    axes[2].axvline(afr_st, color='red', ls='--')
    axes[2].set_title(f"Energía de Mezcla (AFR: {afr_st:.2f})", fontweight='bold')
    axes[2].set_xlabel("AFR"); axes[2].set_ylabel("MJ / kg_total"); axes[2].grid(alpha=0.3)

    print(f"--- RESULTADOS FINALES ---")
    print(f"LHV: {lhv:.2f} MJ/kg | AFR st: {afr_st:.2f} | Potencia: {hp_final:.1f} HP")
    print(f"Q MÁXIMO DE LA MEZCLA: {q_mix_max:.4f} MJ/kg_mezcla")
    plt.show()

simulacion_optima_f1()
```

# **9. Comparación entre combustibles**

Los combustibles a comparar en este trabajo para el contexto de la fórmula 1 son el E10 y el compuesto definido por la fórmula: $C_{6.42}H_{14.59}O_{0.44}+9.85(O_2+3.76N_2)→6.42CO_2+7.29H_2O+37.05N_2$.


La comparación se basa en parámetros como el LHV, el calor liberado y la mezcla de AFR. Además, se evaluará un parámetro adicional; la emisión de gases contaminantes en la combustión. Para poder analizar este último es necesario establecer el criterio de evaluación de estas emisiones. Para esto se plantea la siguiente metodología:

## **Metodología para el cálculo de emisiones específicas**

En esta sección se realiza una comparación entre el combustible E10 y un biocombustible de composición ($C_{6.42}H_{14.59}O_{0.44}$), evaluando sus emisiones de gases en función de la energía liberada durante la combustión.

El análisis se basa en el cálculo de índices de emisión específicos, definidos como la masa de contaminante emitida por unidad de energía generada:

$
EI = \frac{\text{masa de contaminante}}{\text{energía liberada}} \quad [g/MJ]
$

Este enfoque permite comparar de manera consistente combustibles con diferentes poderes caloríficos.



### 1. **Definición de parámetros de entrada**

Para el caso del E10, se utilizan fracciones másicas promedio de carbono, hidrógeno y oxígeno, junto con un valor reportado de poder calorífico inferior (LHV). Además, se asume una eficiencia de combustión cercana al 98.5%.

Para el biocombustible, se parte de su fórmula química ($C_{6.42}H_{14.59}O_{0.44}$), a partir de la cual se calcula su masa molar:

$
M_{fuel} = 6.42(12.011) + 14.59(1.008) + 0.44(15.999)
$

El poder calorífico inferior se estima mediante el balance de entalpías de formación:

$
LHV = \frac{|\Delta H_{productos} - \Delta H_{reactivos}|}{M_{fuel}}
$

donde los productos considerados son ($CO_2$) y ($H_2O$), utilizando valores tabulados de entalpía de formación. Este procedimiento permite aproximar la energía liberada por unidad de masa de combustible.


### **2. Cálculo de emisiones de ($CO_2$)**

#### **Para el E10**

Se asume que todo el carbono presente en el combustible se convierte en ($CO_2$). Por lo tanto, la masa de ($CO_2$) generada por unidad de masa de combustible se calcula como:

$
m_{CO_2} = m_C \cdot \frac{M_{CO_2}}{M_C}
$

donde:

* ($m_C$) es la fracción másica de carbono,
* ($M_{CO_2}$ = 44.01) g/mol,
* ($M_C = 12.011$) g/mol.

El índice de emisión se obtiene normalizando por el poder calorífico:

$
EI_{CO_2} = \frac{m_{CO_2}}{LHV}
$

#### **Para el biocombustible**

En este caso, el cálculo se realiza a partir de la estequiometría:

* 1 mol de combustible produce 6.42 mol de ($CO_2$)

Por lo tanto, la masa de ($CO_2$) generada por mol es:

$
m_{CO_2} = 6.42 \cdot 44.01
$

y la masa de combustible es ($M_{fuel}$), por lo que:

$
EI_{CO_2} = \frac{m_{CO_2}/M_{fuel}}{LHV}
$

### **3. Cálculo de hidrocarburos no quemados (UHC)**

Las emisiones de hidrocarburos no quemados se modelan a partir de la eficiencia de combustión. Se asume que la fracción de combustible que no reacciona completamente es:

$
1 - \eta
$

Por lo tanto, el índice de emisión de UHC se aproxima como:

$
EI_{UHC} = \frac{1 - \eta}{LHV}
$

Este modelo simplificado permite estimar la cantidad de combustible no oxidado que contribuye a las emisiones contaminantes.

Finalmente, los resultados se presentan mediante gráficos de barras que comparan:

Adicionalmente, se imprime una tabla resumen con:

### **5. Consideraciones del modelo**

El análisis realizado se basa en las siguientes simplificaciones:

* Se asume combustión completa para el cálculo de ($CO_2$)
* No se modela explícitamente la formación de CO ni NOx
* El LHV del biocombustible se estima mediante entalpías de formación
* Las emisiones de UHC se relacionan únicamente con la eficiencia de combustión


```python
import numpy as np
import matplotlib.pyplot as plt

def simulacion_comparativa_emisiones():
    # ==============================================================
    # 1. PARÁMETROS DE ENTRADA (E10 vs BIOCOMBUSTIBLE)
    # ==============================================================

    # --- Datos E10 (Referencia) ---
    mC_e10, mH_e10, mO_e10 = 0.8127, 0.1374, 0.056
    lhv_e10 = 42.1161      # MJ/kg
    eta_e10 = 0.985     # Eficiencia de combustión (98.5%)

    # --- Datos Biocombustible (C6.42 H14.59 O0.44) ---
    C_bio, H_bio, O_bio = 6.42, 14.59, 0.44
    M_fuel_bio = (C_bio*12.011 + H_bio*1.008 + O_bio*15.999) # g/mol

    # Termodinámica Biocombustible
    hf_reac_bio = -220.0 # kJ/mol (estimado para oxigenado)
    hf_prod_bio = (C_bio * -393.5) + ((H_bio/2) * -241.8)
    lhv_bio = (abs(hf_prod_bio - hf_reac_bio) / (M_fuel_bio/1000)) / 1000 # MJ/kg
    eta_bio = 0.991     # Eficiencia mejorada por presencia de O2 molecular

    # ==============================================================
    # 2. CÁLCULOS DE EMISIONES ESPECÍFICAS
    # ==============================================================

    # --- CO2 (Gramos por cada Mega Joule de energía) ---
    # E10: masa_C * (MW_CO2 / MW_C) / LHV
    co2_kg_e10 = mC_e10 * (44.01 / 12.011)
    ei_co2_e10 = (co2_kg_e10 / lhv_e10) * 1000

    # Bio: (moles_C * MW_CO2) / (masa_fuel * LHV)
    co2_kg_bio = (C_bio * 44.01) / M_fuel_bio
    ei_co2_bio = (co2_kg_bio / lhv_bio) * 1000

    # --- UHC (Gramos por cada Mega Joule de energía) ---
    # EI = (1 - eta) / LHV
    ei_uhc_e10 = ((1 - eta_e10) / lhv_e10) * 1000
    ei_uhc_bio = ((1 - eta_bio) / lhv_bio) * 1000

    # ==============================================================
    # 3. VISUALIZACIÓN Y RESULTADOS
    # ==============================================================

    fuels = ['E10 (Estándar)', 'Biocombustible']
    data_co2 = [ei_co2_e10, ei_co2_bio]
    data_uhc = [ei_uhc_e10, ei_uhc_bio]

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    plt.subplots_adjust(wspace=0.3)

    # Gráfica A: CO2
    bars1 = ax1.bar(fuels, data_co2, color=['#8D99AE', '#2B2D42'], width=0.5)
    ax1.set_title("Huella de Carbono Específica ($CO_2$)", fontweight='bold')
    ax1.set_ylabel("g $CO_2$ / MJ")
    ax1.grid(axis='y', ls='--', alpha=0.5)

    # Gráfica B: UHC
    bars2 = ax2.bar(fuels, data_uhc, color=['#8D99AE', '#D90429'], width=0.5)
    ax2.set_title("Hidrocarburos No Quemados (UHC)", fontweight='bold')
    ax2.set_ylabel("g UHC / MJ")
    ax2.grid(axis='y', ls='--', alpha=0.5)

    # Etiquetas de valor en barras
    for bars in [bars1, bars2]:
        for bar in bars:
            height = bar.get_height()
            ax = bar.axes
            ax.text(bar.get_x() + bar.get_width()/2., height,
                    f'{height:.2f}' if height > 1 else f'{height:.4f}',
                    ha='center', va='bottom', fontweight='bold')

    print(f"{'='*60}")
    print(f"{'PARÁMETRO':<25} | {'E10':<12} | {'BIOCOMBUSTIBLE':<12}")
    print(f"{'-'*60}")
    print(f"{'LHV [MJ/kg]':<25} | {lhv_e10:<12.2f} | {lhv_bio:<12.2f}")
    print(f"{'Eficiencia Comb. [%]':<25} | {eta_e10*100:<12.1f} | {eta_bio*100:<12.1f}")
    print(f"{'Índice CO2 [g/MJ]':<25} | {ei_co2_e10:<12.2f} | {ei_co2_bio:<12.2f}")
    print(f"{'Índice UHC [g/MJ]':<25} | {ei_uhc_e10:<12.4f} | {ei_uhc_bio:<12.4f}")
    print(f"{'='*60}")

    plt.show()

simulacion_comparativa_emisiones()
```

# **1. Motores turbocargados**

Los motores de combustión interna trabajan utilizando aire y combustible para convertir la energía térmica que disipa su reacción en energía mecánica mediante el recorrido que hace el pistón ubicado en el cigüeñal. La presión y temperatura de estos son fundamentales para una combustión adecuada, por ende, un correcto funcionamiento del motor. La potencia que un motor puede suministrar depende de la cilindrada del motor, sin embargo, la sobrealimentación permite aumentar la potencia que este otorga sin la necesidad de aumentar el volumen del mismo.

El origen de los turbocompresores data a inicios del siglo XX, sin embargo, hay distintas versiones sobre su creador. La invención de este dispositivo mecánico fue revolucionaria para la humanidad, generando un alto impacto en el sector automovilístico. Actualmente, el turbocompresor es uno de los sistemas de sobrealimentación más utilizados en los carros. Estos sistemas permiten aumentar la masa de aire que ingresa en los cilindros del motor durante la admisión, lo que ocasiona una mayor fuerza ejercida sobre el pistón, aumentando la potencia que genera el sistema.

En particular, el turbocompresor "aprovecha la fuerza de los gases de salida del escape del motor para girar una turbina acoplada al múltiple de escape y a su vez un compresor ubicado en la admisión" (Coello Salcedo et al., 2015, Campoverde Marín y Velepucha Velepucha, 2014). De esta manera, el compresor aspira el aire del exterior, lo comprime y lo suministra a la admisión del motor con una mayor presión que la del aire exterior (atmosférica).

En la figura anterior se pueden ver los componentes de un turbocompresor.

Como se mencionó anteriormente, el funcionamiento del turbocompresor se fundamenta en el aprovechamiento de la energía térmica residual contenida en los gases de escape del motor para posteriormente incrementar la cantidad de aire suministrada durante la admisión. Cuando ocurre la combustión dentro del pistón, se generan gases a alta presión y temperatura que, después de realizar trabajo sobre el pistón, salen del motor a través del múltiple de escape (Heywood, 2018). En lugar de desperdiciar completamente esta energía térmica y cinética, los gases son dirigidos hacia el turbocompresor.

Estos gases entran directamente a una turbina, donde experimentan un proceso de expansión que transforma parte de su energía en energía mecánica rotacional. Como se muestra en la figura **1**, esta turbina se encuentra conectada mediante un eje a un compresor centrífugo ubicado en la admisión, de modo que la rotación producida por la expansión de los gases de escape en la turbina acciona directamente el compresor.

En el compresor se tiene otro proceso pasando en simultáneo, se aspira aire atmosférico dentro del compresor y este lo acelera radialmente mediante el movimiento de sus álabes, aumentando inicialmente su energía cinética; posteriormente, el flujo pasa por un difusor donde parte de esta energía cinética se convierte en energía de presión, produciendo así la compresión del aire antes de ingresar al motor. Como consecuencia del proceso de compresión, tanto la presión como la densidad del aire aumentan, esto para incrementar la cantidad de oxígeno que entra a el motor resultando en una combustión más eficiente y por ende, mayor potencia en el motor (Watson & Janota, 1982).

Estos triángulos de velocidades describen el comportamiento dinámico del impulsor de una máquina hidráulica a la entrada y a la salida. El triángulo se compone por las siguientes variables:

Independiente del fluido de trabajo o del tipo de máquina hidráulica, su comportamiento dinámico se describe mediante el triángulo de velocidades.
El triángulo de velocidades permite describir el comportamiento de la máquina hidráulica sin importar su fluido de trabajo e incluso el tipo de máquina que sea (compresor, bomba, turbina, etc). - rpms máximos de la turbina: 125000 rpm


Para resolver el comportamiento termodinámico y cinemático del turbocompresor, se adoptó un modelo de flujo fundamentado en la Ecuación de Euler para Turbomáquinas. Este enfoque permite derivar los saltos de temperatura y presión estrictamente a partir del intercambio de cantidad de movimiento en los triángulos de velocidades de los álabes.

**1. Parámetros Termodinámicos Asumidos**
Las propiedades del fluido de trabajo se establecieron considerando la naturaleza del ciclo de Fórmula 1:
* **Aire de admisión (Compresor):** Se asumió un estado estándar a la entrada con $T_{1c}$ = 25°C (298.15 K) y una presión de $P_{1c}$ = 1.013 bar (1 atm). Al ser un gas ideal diatómico, sus calores específicos son $c_{p,a}$ = 1005 J/kg·K y $\gamma_a$ = 1.4. Además, se asume que la densidad de este aire será la obtenida en estas condiciones de presión y temperatura (P = 1 atm y T = 298.15 K), resultando en 1.292 $kg/m^3$ aproximadamente.
* **Gases de escape (Turbina):** Producto de la combustión, el fluido altera sus propiedades. Se asumió una temperatura de entrada representativa de alta carga $T_{1t}$ = 1000°C (1273.15 K). El calor específico aumenta a $c_{p,g} \approx$ 1150 J/kg·K y la relación de calores específicos desciende a $\gamma_g$ = 1.33.

**2. Interconexión entre Geometría, Cinemática y Termodinámica**
El modelo analítico calcula las variables de estado operando en secuencia lógica:

* **1: Trabajo Específico**
A partir de los vectores tangenciales del fluido ($C_u$) y de las puntas del álabe ($U$), obtenidos trigonométricamente de la geometría asumida, se aplicó la Ecuación de Trabajo de Euler. Como no se trata de una bomba hidráulica incompresible, la energía no se evalúa en metros de cabeza hidroestática ($H_u = \frac{U_2 C_{2u} - U_1 C_{1u}}{g}$), sino como el trabajo específico directamente impartido o extraído de la masa de gas en Joules por kilogramo:
$$w = U_2 C_{2u} - U_1 C_{1u}$$

Posteriormente, la potencia mecánica se calcula multiplicando este trabajo por el caudal másico correspondiente:

$$\dot{W} = \dot{m} \cdot w$$

* **2: Cálculo de Temperaturas (Primera Ley de la Termodinámica)**
El trabajo específico ($w$) calculado de las velocidades se iguala al cambio de entalpía total del flujo. Asumiendo el calor específico constante, la temperatura real de salida se deduce analíticamente mediante la solución directa de la ecuación:
$$T_{2,comp} = T_{1,comp} + \frac{w_{comp}}{c_{p,a}}$$
$$T_{2,turb} = T_{1,turb} - \frac{w_{turb}}{c_{p,g}}$$

- **3: Relación de compresión, relación de expansiones y presiones** La relación de compresión del turbocargador ($r_c$) está relacionado directamente con la relación de temperaturas calculadas en el paso 2. Para esto, se supone una eficiencia isentrópica estimada de $η_c$ = 0.80 y $η_t$ = 0.85.

$$r_c = \left( 1 + \eta_c \frac{T_2 - T_1}{T_1} \right)^{\frac{\gamma}{\gamma-1}}$$

La relación de expansión de la turbina será $$r_t = \frac{1}{r_c}$$


- **4: Geometría de los álabes**
Más allá de los parámetros de entrada, el modelo calcula los ángulos cinemáticos necesarios para el diseño físico de los rodetes. Utilizando las componentes de velocidad obtenidas ($U, C_u, C_m$), se determinan los ángulos relativos ($\beta$) y absolutos ($\alpha$) mediante relaciones trigonométricas inversas:

- **En el Compresor:** Se calcula el ángulo de inducción ($\beta_1$) necesario para que el flujo axial entre sin choque al rodete en rotación, y el ángulo de salida del flujo absoluto ($\alpha_2$) hacia el difusor:

  $$\beta_1 = \arctan \left(\frac{C_{m1}}{U_1}\right) \quad ; \quad \alpha_2 = \arctan \left(\frac{C_{m2}}{C_{u2}}\right)$$



- **En la Turbina:** Se determina el ángulo de incidencia relativo ($\beta_1$) con el que el gas impacta las paletas tras pasar por el estator, y el ángulo constructivo de salida ($\beta_2$) para garantizar una descarga axial eficiente:
   $$\beta_1 = \arctan\left(\frac{C_{m1}}{C_{u1} - U_1}\right) \quad ; \quad \beta_2 = \arctan\left(\frac{C_{m2}}{U_2}\right)$$

Adicionalmente, para calcular el ancho de los álabes ($b$) se toma la ecuación de conservación de masa que establece que:

$$\dot{m} = ρ ⋅ A ⋅ C_m$$

Ahora, sabiendo que el área $A$ está determinada por la siguiente fórumla:

$$ A = π ⋅ D \cdot b$$

Se despeja el ancho $b$ de la ecuación, resultando en:

$$ b = \frac{\dot{m}}{ρ \cdot π \cdot D \cdot C_m}$$

- **5: Eficiencia**: Se asumió un valor de eficiencia de 80% para el compresor y 85% para la turbina para poder cerrar el sistema de ecuaciones y hallar las presiones. Para calcular la eficiencia y no suponerla como un parámetro, se necesita tener los valores exactos de presión y temperatura obtenidas por telemetría, y despejar la ecuación:

$$\eta_c = \frac{\left(\frac{P_2}{P_1}\right)^{\frac{\gamma-1}{\gamma}} - 1}{\frac{T_2}{T_1} - 1}$$

Al no contar con los sensores internos del motor Mercedes, lo correcto en el diseño 1D es suponer estos valores como parámetros.

De esta forma, todos los valores arrojados en la tabla de resultados (Caudales, Presiones, Temperaturas, Potencia mecánica en el eje y las componentes de Velocidad $C$, $W$, $U$ y $C_m$) se encuentran acoplados matemáticamente a través de las leyes de conservación de masa, momento y energía.

Ya que no existen valores públicos sobre la geometría exacta del turbocargador, se asumieron valores para las velocidades $C_m$ en el caso de la turbina y del compresor. De igual manera, se asumen los valores para los ángulos β$_2$ del compresor y α$_1$ de la turbina.


```python
# @title
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# ==============================================================================
# 1. PARÁMETROS DE OPERACIÓN (F1 W15 - 2024)
# ==============================================================================
m_dot_f_kg_h = 100.0
AFR = 13.7918
cp_aire, cp_gas = 1005.0, 1150.0
gamma_a, gamma_g = 1.4, 1.33

# Constantes de los gases (R = cp * (gamma - 1) / gamma)
R_a = cp_aire * (gamma_a - 1.0) / gamma_a
R_g = cp_gas * (gamma_g - 1.0) / gamma_g

T1_c, T1_t = 298.15, 1273.15
P1_c, P2_t = 101325.0, 105000.0
eta_c, eta_t = 0.80, 0.85
RPM = 100000.0
omega = RPM * np.pi / 30.0

# GEOMETRÍA ASUMIDA
D1_c, D2_c = 0.050, 0.090
beta2_c_deg = 45.0            # Ángulo constructivo de salida (Backswept)
Cm1_c, Cm2_c = 120.0, 80.0

D1_t, D2_t = 0.085, 0.065
alpha1_t_deg = 70.0           # Ángulo de nozzle (estator)
Cm1_t, Cm2_t = 150.0, 200.0

# ==============================================================================
# 2. CÁLCULOS CINEMÁTICOS Y TERMODINÁMICOS
# ==============================================================================
m_dot_a = (m_dot_f_kg_h / 3600.0) * AFR
m_dot_t = m_dot_a + (m_dot_f_kg_h / 3600.0)

# --- COMPRESOR ---
U1_c, U2_c = omega * (D1_c / 2.0), omega * (D2_c / 2.0)
Cu1_c = 0.0
alpha1_c = 90.0
beta1_c = np.degrees(np.arctan2(Cm1_c, U1_c))

beta2_c_rad = np.radians(beta2_c_deg)
Wu2_c = Cm2_c / np.tan(beta2_c_rad)
Cu2_c = U2_c - Wu2_c
alpha2_c = np.degrees(np.arctan2(Cm2_c, Cu2_c))

# Trabajo, Temperatura y Relación de Compresión (rc)
w_comp = U2_c * Cu2_c - U1_c * Cu1_c
T2_c = T1_c + (w_comp / cp_aire)
rc = (1.0 + eta_c * ((T2_c - T1_c) / T1_c)) ** (gamma_a / (gamma_a - 1.0))
P2_c = P1_c * rc

# Densidades y Ancho del álabe (b) [Convertido a mm]
rho1_c = P1_c / (R_a * T1_c)
rho2_c = P2_c / (R_a * T2_c)
b1_c = (m_dot_a / (rho1_c * np.pi * D1_c * Cm1_c)) * 1000
b2_c = (m_dot_a / (rho2_c * np.pi * D2_c * Cm2_c)) * 1000

# --- TURBINA ---
U1_t, U2_t = omega * (D1_t / 2.0), omega * (D2_t / 2.0)
alpha1_t_rad = np.radians(alpha1_t_deg)
Cu1_t = Cm1_t * np.tan(alpha1_t_rad)
beta1_t = np.degrees(np.arctan2(Cm1_t, (Cu1_t - U1_t)))

Cu2_t = 0.0
alpha2_t = 90.0
beta2_t = np.degrees(np.arctan2(Cm2_t, U2_t))

# Trabajo, Temperatura y Relación de Expansión (rt_exp)
w_turb = U1_t * Cu1_t - U2_t * Cu2_t
T2_t = T1_t - (w_turb / cp_gas)
rt_inversa = (1.0 - (1.0 / eta_t) * ((T1_t - T2_t) / T1_t)) ** (gamma_g / (gamma_g - 1.0))
P1_t = P2_t / rt_inversa
rt_expansión = 1.0 / rt_inversa # Relación de expansión P1/P2 (> 1)

# Densidades y Ancho del álabe (b) [Convertido a mm]
rho1_t = P1_t / (R_g * T1_t)
rho2_t = P2_t / (R_g * T2_t)
b1_t = (m_dot_t / (rho1_t * np.pi * D1_t * Cm1_t)) * 1000
b2_t = (m_dot_t / (rho2_t * np.pi * D2_t * Cm2_t)) * 1000

# ==============================================================================
# 3. VISUALIZACIÓN DE TRIÁNGULOS
# ==============================================================================
def plot_turbo_triangle(ax, U, Cu, Cm, a_deg, b_deg, title):
    ax.axis('off')
    ax.set_title(title, fontweight='bold', pad=30, fontsize=14)

    ax.quiver(0, 0, U, 0, angles='xy', scale_units='xy', scale=1, color='black', width=0.008)
    ax.quiver(0, 0, Cu, Cm, angles='xy', scale_units='xy', scale=1, color='blue', width=0.008)
    ax.quiver(U, 0, Cu-U, Cm, angles='xy', scale_units='xy', scale=1, color='green', width=0.008)

    bbox_props = dict(boxstyle="round,pad=0.3", fc="white", ec="none", alpha=0.85)

    ax.text(U/2, -40, f"U = {U:.0f}", ha='center', color='black', fontsize=10, fontweight='bold', bbox=bbox_props)

    C_mag = np.sqrt(Cu**2+Cm**2)
    offset_c_x = -45 if Cu >= 0 else 45
    ax.text(Cu/2 + offset_c_x, Cm/2 + 25, f"C = {C_mag:.0f}", ha='center', color='blue', fontsize=10, fontweight='bold', bbox=bbox_props)

    W_mag = np.sqrt((Cu-U)**2 + Cm**2)
    offset_w_x = 45 if (Cu-U) <= 0 else -45
    ax.text((U+Cu)/2 + offset_w_x, Cm/2 + 25, f"W = {W_mag:.0f}", ha='center', color='green', fontsize=10, fontweight='bold', bbox=bbox_props)

    if abs(Cu) > 1.0:
        ax.plot([Cu, Cu], [0, Cm], 'k--', alpha=0.3)
        ax.text(Cu, Cm/2, f"Cm = {Cm:.0f}", color='gray', ha='center', fontsize=9, bbox=bbox_props)
    else:
        ax.text(Cu + 35, Cm/2, f"Cm = {Cm:.0f}", color='gray', ha='center', fontsize=9, bbox=bbox_props)

    R_alpha, R_beta = 100, 100
    y_alpha = max(R_alpha * np.sin(np.radians(a_deg / 2)), 35)
    x_alpha = R_alpha * np.cos(np.radians(a_deg / 2))

    if a_deg >= 85.0:
        x_alpha, y_alpha = -45, Cm / 3

    ax.text(x_alpha, y_alpha, f"α = {a_deg:.1f}°", color='blue', fontweight='bold', ha='center', bbox=bbox_props)

    y_beta = max(R_beta * np.sin(np.radians(b_deg / 2)), 35)
    x_beta = U - R_beta * np.cos(np.radians(b_deg / 2))

    if b_deg > 95.0:
        x_beta, y_beta = U + 50, Cm / 3
    elif 85.0 <= b_deg <= 95.0:
        x_beta, y_beta = U + 40, Cm / 3

    ax.text(x_beta, y_beta, f"β = {b_deg:.1f}°", color='green', fontweight='bold', ha='center', bbox=bbox_props)

    ax.set_aspect('equal')
    ax.set_xlim(min(-80, Cu - 80), max(U + 110, Cu + 80))
    ax.set_ylim(-70, Cm + 70)

fig, axs = plt.subplots(2, 2, figsize=(16, 12), facecolor='white')
plot_turbo_triangle(axs[0, 0], U1_c, Cu1_c, Cm1_c, alpha1_c, beta1_c, "Compresor - Entrada")
plot_turbo_triangle(axs[0, 1], U2_c, Cu2_c, Cm2_c, alpha2_c, beta2_c_deg, "Compresor - Salida")
plot_turbo_triangle(axs[1, 0], U1_t, Cu1_t, Cm1_t, alpha1_t_deg, beta1_t, "Turbina - Entrada")
plot_turbo_triangle(axs[1, 1], U2_t, Cu2_t, Cm2_t, alpha2_t, beta2_t, "Turbina - Salida")
plt.tight_layout()
plt.show()

# ==============================================================================
# 4. TABLA INTEGRAL DE RESULTADOS
# ==============================================================================
res = {
    "Variable": ["Caudal [kg/s]", "Relación de Compresión (rc)", "Relación de Expansión (rt)", "P_ent [bar]", "P_sal [bar]", "T_ent [°C]", "T_sal [°C]",
                 "α1 [deg]", "α2 [deg]", "β1 [deg]", "β2 [deg]", "Ancho b1 [mm]", "Ancho b2 [mm]", "Potencia [kW]"],
    "Compresor": [round(m_dot_a,3), round(rc, 2), "-", round(P1_c/1e5,2), round(P2_c/1e5,2), round(T1_c-273.15,1), round(T2_c-273.15,1),
                  round(alpha1_c,1), round(alpha2_c,1), round(beta1_c,1), round(beta2_c_deg,1), round(b1_c,2), round(b2_c,2), round(m_dot_a*w_comp/1e3,1)],
    "Turbina": [round(m_dot_t,3), "-", round(rt_expansión, 2), round(P1_t/1e5,2), round(P2_t/1e5,2), round(T1_t-273.15,1), round(T2_t-273.15,1),
                round(alpha1_t_deg,1), round(alpha2_t,1), round(beta1_t,1), round(beta2_t,1), round(b1_t,2), round(b2_t,2), round(m_dot_t*w_turb/1e3,1)]
}
display(pd.DataFrame(res))
```


# **3. Impacto termodinámico en el ciclo del motor**

Generalmente, los motores de combustión interna de encendido por chispa trabajan con el ciclo termodinámico Otto. Este ciclo consta de 4 tiempos: **admisión, compresión, combustión/expansión y escape**. Un motor de estos debe realizar 2 revoluciones completas para poder completar 1 ciclo Otto. El ciclo ideal se caracteriza por 4 estados, donde se realiza lo siguiente:

- 1 → 2: Compresión isentrópica. ⇑ P y ⇑ V

- 2 → 3: Adición de calor a volumen constante. ⇑ P y = V

- 3 → 4: Expansión isentrópica. ↓ P y ↓ V

- 4 → 1: Rechazo de calor a volumen constante. ↓ P y = V



image.png


El ciclo real funciona distinto, ya que este tiene en cuenta otros componentes del motor, como lo son las válvulas de admisión y escape. En este caso, el ciclo real consta de los siguientes procesos:

- Admisión de aire y gasolina a la cámara de combustión.

- Compresión de la mezcla.

- Encendido de la mezcla y combustión.

- Expansión del pistón.

- Escape de los gases de la combustión.

Ahora bien, como se mencionó anteriormente, un turbocompresor funciona para sobrealimentar al sistema de aire con el fin de generar potencia adicional. En el caso del ciclo Otto ideal, lo que ocurre con y sin turbocompresor es lo siguiente:

1. **Admisión:** El pistón se traslada del punto muerto superior al punto muerto inferior, se admite aire al sistema y este ingresa a la cámara de combustión.
    - **Sin turbo:** El aire entra a presión atmosférica (aproximadamente 1 bar).
    - **Con turbo:** El aire entra a presión mayor a la atmosférica, aproximadamente de 3 a 5 bar absolutos.


2. **Compresión:** El pistón sube del punto muerto inferior al punto muerto superior.
    - **Sin turbo:** Las válvulas se cierran y el pistón sube comprimiendo el aire hasta el volumen de la cámara de combustión ($V_2 = V_c$). La presión sube según la relación isentrópica $P_2 = P_1 \cdot r_c^\gamma$.

    - **Con turbo:** Las presiones altas del aire generadas por el turbo pueden hacer que se alcance la temperatura de autoencendido del combustible, causando una detonación temprana, lo que genera riesgos de fala del motor.

3. **Combustión y expansión:** Se enciende la mezcla y el pistón se traslada nuevamente del punto muerto superior al punto muerto inferior. Hay una expansión en el volumen.

    - **Sin turbo:** Se enciende la mezcla de aire combustible y aumenta la presión y el volumen. La carrera del pistón hace que se produzca trabajo en el sistema.

    - **Con turbo:** Se enciende la mezcla de aire combustible, pero, como el turbo aumentó la presión del aire y de igual forma aumentó la masa de este dentro de la cámara de combustión, se puede inyectar una mayor cantidad de combustible, manteniendo el AFR ideal. Por lo que se inyectó una mayor cantidad de mezcla a la cámara, la presión resultante en este punto fue mucho mayor. Al ser mayor la presión, se aumenta también el trabajo que genera el sistema.

4. **Escape:** Los gases de la combustión salen de la cámara de combustión por el escape del vehículo.

    - **Sin turbo:** Los gases salen del motor hacia el entorno.

    - **Con turbo:** Los gases de salida son aprovechados para hacer girar la turbina del turbocompresor, quien está conectada al mismo eje del compresor, haciendo que haya una mayor admisión de aire presurizado al sistema.

El diagrama P-V del ciclo otto con y sin turbo se presenta a continuación:


```python
# @title
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# ==============================================================================
# 1. PARÁMETROS GEOMÉTRICOS Y TERMODINÁMICOS BASE
# ==============================================================================
V_total = 1.6e-3              # Cilindrada total [m^3] (1.6 Litros)
V_d = V_total / 6.0           # Volumen desplazado por cilindro [m^3]
r_c = 11.0                    # Relación de compresión geométrica

V_c = V_d / (r_c - 1.0)       # Volumen de la cámara de combustión (Clearance)
V_1 = V_c + V_d               # Volumen máximo (Punto Muerto Inferior)

T_in = 323.15                 # Temperatura de admisión [K] (50°C)
gamma = 1.3                   # Relación de calores específicos
cv = 750.0                    # Calor específico a volumen constante [J/kg.K]
R = 287.0                     # Constante del gas ideal [J/kg.K]

# ==============================================================================
# 2. MODELADO: CICLO OTTO ATMOSFÉRICO (NA - Natural Aspirated)
# ==============================================================================
P1_na = 1.01325e5             # Presión de admisión atmosférica (1 atm ~ 1 bar)
m_na = (P1_na * V_1) / (R * T_in) # Masa de aire atrapada

Q_in_na = 275.0               # Calor inyectado base [Joules por ciclo]

# Estados termodinámicos NA
V1_na, T1_na = V_1, T_in
V2_na = V_c
P2_na = P1_na * (V1_na / V2_na)**gamma
T2_na = T1_na * (V1_na / V2_na)**(gamma - 1)

V3_na = V_c
T3_na = T2_na + Q_in_na / (m_na * cv)
P3_na = P2_na * (T3_na / T2_na)

V4_na = V_1
P4_na = P3_na * (V3_na / V4_na)**gamma
T4_na = T3_na * (V3_na / V4_na)**(gamma - 1)

# Vectores para la curva NA
v_comp_na = np.linspace(V1_na, V2_na, 100)
p_comp_na = P1_na * (V1_na / v_comp_na)**gamma
v_exp_na = np.linspace(V3_na, V4_na, 100)
p_exp_na = P3_na * (V3_na / v_exp_na)**gamma

# ==============================================================================
# 3. MODELADO: CICLO OTTO TURBO (4 bar de presión de admisión)
# ==============================================================================
P1_t = 4.0e5                  # Presión de admisión turbo (4 bar absolutos)
m_t = (P1_t * V_1) / (R * T_in)   # Masa atrapada (Es casi 4 veces mayor)

# Como hay 4 veces más aire, podemos inyectar 4 veces más combustible
# manteniendo la misma mezcla estequiométrica:
Q_in_t = Q_in_na * (m_t / m_na)

# Estados termodinámicos Turbo
V1_t, T1_t = V_1, T_in
V2_t = V_c
P2_t = P1_t * (V1_t / V2_t)**gamma
T2_t = T1_t * (V1_t / V2_t)**(gamma - 1)

V3_t = V_c
T3_t = T2_t + Q_in_t / (m_t * cv)
P3_t = P2_t * (T3_t / T2_t)

V4_t = V_1
P4_t = P3_t * (V3_t / V4_t)**gamma

# Vectores para la curva Turbo
v_comp_t = np.linspace(V1_t, V2_t, 100)
p_comp_t = P1_t * (V1_t / v_comp_t)**gamma
v_exp_t = np.linspace(V3_t, V4_t, 100)
p_exp_t = P3_t * (V3_t / v_exp_t)**gamma

# ==============================================================================
# 4. CÁLCULO DE TRABAJO Y EFICIENCIA (Usando la nueva función np.trapezoid)
# ==============================================================================
W_net_na = np.trapezoid(p_exp_na, v_exp_na) - np.trapezoid(p_comp_na[::-1], v_comp_na[::-1])
eta_na = (W_net_na / Q_in_na) * 100

W_net_t = np.trapezoid(p_exp_t, v_exp_t) - np.trapezoid(p_comp_t[::-1], v_comp_t[::-1])
eta_t = (W_net_t / Q_in_t) * 100

# ==============================================================================
# 5. VISUALIZACIÓN GRÁFICA (3 Paneles)
# ==============================================================================
fig, axs = plt.subplots(1, 3, figsize=(20, 6), facecolor='white')

# --- Panel 1: Motor Atmosférico ---
axs[0].plot(v_comp_na * 1e6, p_comp_na / 1e5, 'b--', linewidth=2, label='Compresión')
axs[0].plot(v_exp_na * 1e6, p_exp_na / 1e5, 'b-', linewidth=2, label='Expansión')
axs[0].plot([V2_na*1e6, V3_na*1e6], [P2_na/1e5, P3_na/1e5], 'b-', linewidth=2)
axs[0].plot([V4_na*1e6, V1_na*1e6], [P4_na/1e5, P1_na/1e5], 'b-', linewidth=2)
axs[0].fill_between(v_exp_na * 1e6, p_comp_na / 1e5, p_exp_na / 1e5, color='blue', alpha=0.1)

axs[0].set_title("1. Ciclo Otto Atmosférico (1 bar)", fontweight='bold', fontsize=13)
axs[0].set_xlabel("Volumen [cm³]", fontsize=11)
axs[0].set_ylabel("Presión [bar]", fontsize=11)
# Límite ajustado a la presión real NA
axs[0].set_ylim(0, (P3_na/1e5) * 1.1)
axs[0].grid(True, linestyle=':', alpha=0.7)
axs[0].legend()

# --- Panel 2: Motor Turbo ---
axs[1].plot(v_comp_t * 1e6, p_comp_t / 1e5, 'r--', linewidth=2, label='Compresión')
axs[1].plot(v_exp_t * 1e6, p_exp_t / 1e5, 'r-', linewidth=2, label='Expansión')
axs[1].plot([V2_t*1e6, V3_t*1e6], [P2_t/1e5, P3_t/1e5], 'r-', linewidth=2)
axs[1].plot([V4_t*1e6, V1_t*1e6], [P4_t/1e5, P1_t/1e5], 'r-', linewidth=2)
axs[1].fill_between(v_exp_t * 1e6, p_comp_t / 1e5, p_exp_t / 1e5, color='red', alpha=0.1)

axs[1].set_title("2. Ciclo Otto Turbo (4 bar)", fontweight='bold', fontsize=13)
axs[1].set_xlabel("Volumen [cm³]", fontsize=11)
# Límite enorme para abarcar el pico del turbo
axs[1].set_ylim(0, (P3_t/1e5) * 1.1)
axs[1].grid(True, linestyle=':', alpha=0.7)
axs[1].legend()

# --- Panel 3: Comparación Superpuesta a la Misma Escala ---
axs[2].plot(v_comp_na * 1e6, p_comp_na / 1e5, 'b-', linewidth=2, label='Atmosférico')
axs[2].plot(v_exp_na * 1e6, p_exp_na / 1e5, 'b-', linewidth=2)
axs[2].plot([V2_na*1e6, V3_na*1e6], [P2_na/1e5, P3_na/1e5], 'b-', linewidth=2)
axs[2].plot([V4_na*1e6, V1_na*1e6], [P4_na/1e5, P1_na/1e5], 'b-', linewidth=2)
axs[2].fill_between(v_exp_na * 1e6, p_comp_na / 1e5, p_exp_na / 1e5, color='blue', alpha=0.3)

axs[2].plot(v_comp_t * 1e6, p_comp_t / 1e5, 'r-', linewidth=2, label='Turbo (4 bar)')
axs[2].plot(v_exp_t * 1e6, p_exp_t / 1e5, 'r-', linewidth=2)
axs[2].plot([V2_t*1e6, V3_t*1e6], [P2_t/1e5, P3_t/1e5], 'r-', linewidth=2)
axs[2].plot([V4_t*1e6, V1_t*1e6], [P4_t/1e5, P1_t/1e5], 'r-', linewidth=2)
axs[2].fill_between(v_exp_t * 1e6, p_comp_t / 1e5, p_exp_t / 1e5, color='red', alpha=0.1)

axs[2].set_title("3. Comparación Directa", fontweight='bold', fontsize=13)
axs[2].set_xlabel("Volumen [cm³]", fontsize=11)
axs[2].set_ylim(0, (P3_t/1e5) * 1.1) # Usa la escala del turbo para evidenciar cómo el NA se ve enano
axs[2].grid(True, linestyle=':', alpha=0.7)
axs[2].legend()

plt.tight_layout()
plt.show()

# ==============================================================================
# 6. TABLA DE RESULTADOS Y VERIFICACIÓN TERMODINÁMICA
# ==============================================================================
res = {
    "Variable Operativa": [
        "Presión de Admisión (P1) [bar]",
        "Masa de Aire Atrapada [gramos/ciclo]",
        "Calor Inyectado [Joules/ciclo]",
        "Presión de Compresión (P2) [bar]",
        "Pico de Combustión (P3) [bar]",
        "TRABAJO NETO POR CILINDRO [Joules]",
        "Eficiencia Térmica Ideal [%]"
    ],
    "Motor Atmosférico (NA)": [
        round(P1_na / 1e5, 2),
        round(m_na * 1000, 3),
        round(Q_in_na, 1),
        round(P2_na / 1e5, 1),
        round(P3_na / 1e5, 1),
        round(W_net_na, 1),
        round(eta_na, 2)
    ],
    "Motor Turbo (F1)": [
        round(P1_t / 1e5, 2),
        round(m_t * 1000, 3),
        round(Q_in_t, 1),
        round(P2_t / 1e5, 1),
        round(P3_t / 1e5, 1),
        round(W_net_t, 1),
        round(eta_t, 2)
    ]
}

df_turbo = pd.DataFrame(res)
display(df_turbo)
```

Se puede ver que el ciclo Otto es esencialmente el mismo, sin embargo, el cambio es que la gráfica se traslada verticalmente hacia arriba, significando que hay mayores presiones involucradas en el ciclo. Esto se traduce a un mayor trabajo de salida en el sistema, aumentando la potencia del vehículo.

Sin embargo, debido a que la eficiencia de este ciclo es baja (entre 30% y 35%), los equipos de fórmula 1 han implementado una modificación particular en el motor para trabajar en un ciclo termodinámico distinto. Desde el 2014 se vienen haciendo estas implementaciones en los motores. Dicha modificación consta de la alteración de la sincronización del motor, ya que han alterado el árbol de levas con el fin de cerrar las válvulas de admisión antes de que el pistón llegue al punto muerto inferior (PMI). Esto se conoce como Early Intake Valve Closing (EIVC), mientras que el ciclo Otto adapta el nombre de ciclo Miller (Wen et al., 2022). El efecto de esta modificación es expandir ligeramente la mezcla antes de comprimirla, lo que hace que la relación de compresión efectiva sea menor, evitando la detonación a pesar de las presiones del turbo. Esto trae como ventaja poder aprovechar más la carrera de expansión del pistón (termodinámicamente), brindando un mayor trabajo mecánico.


```python
# @title
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# ==============================================================================
# 1. PARÁMETROS GEOMÉTRICOS Y TERMODINÁMICOS BASE (Coherentes con F1 Turbo)
# ==============================================================================
V_total = 1.6e-3              # Cilindrada total [m^3] (1.6 Litros)
V_d = V_total / 6.0           # Volumen desplazado por cilindro [m^3]

# Relaciones del motor
r_geom = 15.0                 # Relación de compresión/expansión geométrica (Otto simétrico)
r_comp_eff = 11.0             # Relación de compresión efectiva (Ciclo Miller EIVC)

V_c = V_d / (r_geom - 1.0)    # Volumen de la cámara de combustión (Clearance)
V_1 = V_c + V_d               # Volumen máximo (Punto Muerto Inferior)
V_eff = V_c * r_comp_eff      # Volumen donde inicia la compresión efectiva en Miller

# Condiciones de Admisión Turbo
P1_in = 4.0e5                 # Presión de admisión [Pa] (4.0 bar absolutos)
T_in = 323.15                 # Temperatura de admisión [K] (50°C)
gamma = 1.3                   # Relación de calores específicos
cv = 750.0                    # Calor específico a volumen constante [J/kg.K]
R = 287.0                     # Constante del gas ideal [J/kg.K]

# Energía del Combustible
Q_in = 1100.0                 # Calor inyectado por cilindro por ciclo [J]

# ==============================================================================
# 2. MODELADO TERMODINÁMICO: CICLO OTTO TURBO (Simétrico rc = 15.0)
# ==============================================================================
m_otto = (P1_in * V_1) / (R * T_in) # Masa de aire total atrapada

# Estados Otto
P1_o, V1_o, T1_o = P1_in, V_1, T_in
V2_o = V_c
P2_o = P1_o * (V1_o / V2_o)**gamma
T2_o = T1_o * (V1_o / V2_o)**(gamma - 1)

V3_o = V_c
T3_o = T2_o + Q_in / (m_otto * cv)
P3_o = P2_o * (T3_o / T2_o)

V4_o = V_1
P4_o = P3_o * (V3_o / V4_o)**gamma

# Vectores Otto
v_comp_o = np.linspace(V1_o, V2_o, 100)
p_comp_o = P1_o * (V1_o / v_comp_o)**gamma
v_exp_o = np.linspace(V3_o, V4_o, 100)
p_exp_o = P3_o * (V3_o / v_exp_o)**gamma

# ==============================================================================
# 3. MODELADO TERMODINÁMICO: CICLO MILLER TURBO (Asimétrico F1)
# ==============================================================================
# Menor masa atrapada por el cierre temprano de válvula (EIVC)
m_miller = (P1_in * V_eff) / (R * T_in)

# Estados Miller
P1_m, V1_m, T1_m = P1_in, V_eff, T_in
V2_m = V_c
P2_m = P1_m * (V1_m / V2_m)**gamma
T2_m = T1_m * (V1_m / V2_m)**(gamma - 1)

V3_m = V_c
T3_m = T2_m + Q_in / (m_miller * cv)
P3_m = P2_m * (T3_m / T2_m)

V4_m = V_1
P4_m = P3_m * (V3_m / V4_m)**gamma

# Vectores Miller
v_comp_m = np.linspace(V1_m, V2_m, 100)
p_comp_m = P1_m * (V1_m / v_comp_m)**gamma
v_exp_m = np.linspace(V3_m, V4_m, 100)
p_exp_m = P3_m * (V3_m / v_exp_m)**gamma

# ==============================================================================
# 4. CÁLCULO DE TRABAJO Y EFICIENCIA (Usando np.trapezoid)
# ==============================================================================
W_exp_o = np.trapezoid(p_exp_o, v_exp_o)
W_comp_o = np.trapezoid(p_comp_o[::-1], v_comp_o[::-1])
W_net_o = W_exp_o - W_comp_o
eta_o = (W_net_o / Q_in) * 100

W_exp_m = np.trapezoid(p_exp_m, v_exp_m)
W_comp_m = np.trapezoid(p_comp_m[::-1], v_comp_m[::-1])
W_net_m = W_exp_m - W_comp_m
eta_m = (W_net_m / Q_in) * 100

# ==============================================================================
# 5. VISUALIZACIÓN GRÁFICA MEJORADA (1 Fila, 3 Columnas)
# ==============================================================================
fig, axs = plt.subplots(1, 3, figsize=(20, 6), facecolor='white')

# Configuración compartida para los 3 gráficos
for ax in axs:
    ax.set_xlim(0, (V_1*1e6) * 1.1)  # <-- CORREGIDO: Se quitó el max()
    ax.set_ylim(0, (P3_o/1e5) * 1.1) # <-- CORREGIDO: Se quitó el max()
    ax.set_xlabel("Volumen del Cilindro [cm³]", fontsize=11)
    ax.set_ylabel("Presión Absoluta [bar]", fontsize=11)
    ax.grid(True, linestyle=':', alpha=0.7)

# --- GRÁFICO 1: CICLO OTTO TURBO ---
axs[0].plot(v_comp_o * 1e6, p_comp_o / 1e5, 'b--', linewidth=2, label='Compresión (rc=15)')
axs[0].plot(v_exp_o * 1e6, p_exp_o / 1e5, 'b-', linewidth=2, label='Expansión')
axs[0].plot([V2_o*1e6, V3_o*1e6], [P2_o/1e5, P3_o/1e5], 'b-', linewidth=2)
axs[0].plot([V4_o*1e6, V1_o*1e6], [P4_o/1e5, P1_o/1e5], 'b-', linewidth=2)
axs[0].fill_between(v_exp_o * 1e6, p_comp_o / 1e5, p_exp_o / 1e5, color='blue', alpha=0.1)
axs[0].set_title("1. Otto Turbo (Simétrico rc=15.0)", fontweight='bold', fontsize=13)
axs[0].legend(loc='upper right')

# --- GRÁFICO 2: CICLO MILLER TURBO ---
axs[1].plot(v_comp_m * 1e6, p_comp_m / 1e5, 'g--', linewidth=2, label='Compresión Efectiva (rc=11)')
axs[1].plot(v_exp_m * 1e6, p_exp_m / 1e5, 'g-', linewidth=2, label='Expansión')
axs[1].plot([V2_m*1e6, V3_m*1e6], [P2_m/1e5, P3_m/1e5], 'g-', linewidth=2)
axs[1].plot([V4_m*1e6, V1_m*1e6], [P4_m/1e5, P1_m/1e5], 'g-', linewidth=2)
axs[1].plot([V1_m*1e6, V1_o*1e6], [P1_m/1e5, P1_o/1e5], 'g:', linewidth=2, label='EIVC (Válvula Abierta)')
axs[1].fill_between(v_exp_m * 1e6, p_comp_m / 1e5, p_exp_m / 1e5, color='green', alpha=0.1)
axs[1].set_title("2. Miller Turbo F1 (Asimétrico)", fontweight='bold', fontsize=13)
axs[1].legend(loc='upper right')

# --- GRÁFICO 3: COMPARACIÓN DIRECTA ---
# Otto en fondo semi-transparente
axs[2].plot(v_comp_o * 1e6, p_comp_o / 1e5, 'b--', linewidth=1.5, alpha=0.4)
axs[2].plot(v_exp_o * 1e6, p_exp_o / 1e5, 'b-', linewidth=1.5, alpha=0.4)
axs[2].plot([V2_o*1e6, V3_o*1e6], [P2_o/1e5, P3_o/1e5], 'b-', linewidth=1.5, alpha=0.4)
axs[2].plot([V4_o*1e6, V1_o*1e6], [P4_o/1e5, P1_o/1e5], 'b-', linewidth=1.5, alpha=0.4, label='Perfil Otto')

# Miller resaltado
axs[2].plot(v_comp_m * 1e6, p_comp_m / 1e5, 'g--', linewidth=2.5)
axs[2].plot(v_exp_m * 1e6, p_exp_m / 1e5, 'g-', linewidth=2.5)
axs[2].plot([V2_m*1e6, V3_m*1e6], [P2_m/1e5, P3_m/1e5], 'g-', linewidth=2.5)
axs[2].plot([V4_m*1e6, V1_m*1e6], [P4_m/1e5, P1_m/1e5], 'g-', linewidth=2.5, label='Perfil Miller')

# ¡Relleno del Trabajo Ahorrado! (Diferencia entre las dos curvas de compresión)
p_comp_o_interp = P1_o * (V1_o / v_comp_m)**gamma
axs[2].fill_between(v_comp_m * 1e6, p_comp_o_interp / 1e5, p_comp_m / 1e5, color='orange', alpha=0.5, label='TRABAJO AHORRADO')

axs[2].set_title("3. Comparación y Trabajo Ahorrado", fontweight='bold', fontsize=13)
axs[2].legend(loc='upper right')

plt.tight_layout()
plt.show()

# ==============================================================================
# 6. TABLA DE RESULTADOS
# ==============================================================================
res = {
    "Parámetro": [
        "Presión de Admisión Turbo (P1) [bar]",
        "Relación de Compresión Efectiva",
        "Relación de Expansión Geométrica",
        "Calor Inyectado [J/ciclo]",
        "Presión Máxima de Combustión [bar]",
        "Trabajo de Compresión [J]",
        "Trabajo de Expansión [J]",
        "TRABAJO NETO POR CICLO [J]",
        "Eficiencia Térmica Ideal [%]"
    ],
    "Ciclo Otto Turbo": [
        round(P1_in / 1e5, 2),
        round(r_geom, 1),
        round(r_geom, 1),
        round(Q_in, 1),
        round(P3_o / 1e5, 1),
        round(W_comp_o, 1),
        round(W_exp_o, 1),
        round(W_net_o, 1),
        round(eta_o, 2)
    ],
    "Ciclo Miller Turbo (F1)": [
        round(P1_in / 1e5, 2),
        round(r_comp_eff, 1),
        round(r_geom, 1),
        round(Q_in, 1),
        round(P3_m / 1e5, 1),
        round(W_comp_m, 1),
        round(W_exp_m, 1),
        round(W_net_m, 1),
        round(eta_m, 2)
    ]
}

df_comparacion = pd.DataFrame(res)
display(df_comparacion)
```

Como el ciclo Miller maneja una relación de compresión menor a su relación de expansión, el trabajo de entrada que requiere el sistema es menor al de un ciclo Otto que mantiene esta relación de compresión. Es por esto que, aunque ambos ciclos involucren turbocompresores para su funcionamiento, las relaciones de compresión del ciclo Miller son quienes se encargan de proveer al sistema un mayor trabajo de salida. Al ofrecer un mayor trabajo de salida, el sistema se vuelve más eficiente ya que, a las mismas condiciones de operación, el sistema requirió menos gasto energético para poder comprimir todos los gases de la admisión, y se aprovechó mejor la expansión del pistón, a comparación del ciclo Otto con turbo.

#**4. Análisis del funcionamiento de la máquina**
Para comprender mejor el funcionamiento del turbocompresor como máquina, es esencial analizar su desempeño mediante la variación de algunos parámetros. Estos parámetros pueden ser tanto geométricos como de funcionamiento. Para este caso en particular, se variarán: la geometría del turbocompresor, la velocidad a la que éste opera y las presiones de funcionamiento, con el fin de analizar variables como: la producción de trabajo, el consumo energético y la eficiencia térmica. Para esto, se realizarán los siguientes casos:


| Caso de Estudio | Parámetro a Variar (Independiente) | Rango de Variación Estimado | Variables a Observar (Dependientes) |
| :--- | :--- | :--- | :--- |
| **Caso 1: Régimen de Giro del Eje** | Velocidad angular ($\omega$ / RPM) | $80,000$ a $120,000$ RPM | Trabajo de Euler ($w_c, w_t$), Relaciones de Presión ($r_c, r_t$) |
| **Caso 2: Geometría del Compresor** | Ángulo constructivo de salida ($\beta_{2c}$) | $30^\circ$ a $70^\circ$ | Componente tangencial ($C_{u2}$), Potencia consumida, Relación de presión ($r_c$) |
| **Caso 3: Carga Térmica (Boost)** | Presión de Admisión ($P_{1,in}$) | $2.0$ a $5.0$ bar absolutos | Calor inyectado ($Q_{in}$), Presión pico de combustión ($P_3$), Trabajo Neto ($W_{net}$) |
| **Caso 4: Estrategia de Válvulas** | Relación de compresión efectiva ($r_{comp,eff}$) | $8:1$ a $14:1$ (Ciclo Miller) | Trabajo de compresión interno, Eficiencia Térmica Ideal ($\eta_{térmica}$) |


Así mismo, también se podrán variar parámetros como el diámetro del turbo, el ancho de los álabes a la entrada y salida de este y el ángulo de estos.


```python
# @title
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import ipywidgets as widgets
from IPython.display import display

# ==============================================================================
# CONSTANTES DEL SISTEMA F1 W15 Y REGLAMENTO FIA
# ==============================================================================
V_total = 1.6e-3              # Cilindrada 1.6L [m^3]
V_d = V_total / 6.0
r_geom = 15.0                 # Expansión geométrica fija
V_c = V_d / (r_geom - 1.0)
cv_m, gamma_m = 750.0, 1.3
R_m = 287.0
T1_in = 323.15                # Temp admisión 50°C
Q_in_nominal = 1100.0         # Joules de combustible base
RPM_ENGINE = 10500.0          # RPM del motor de combustión para flujo de masa

# Geometría base del compresor
D1_c = 0.050                  # Diámetro entrada compresor [m] (Fijo para el ojo del rodete)
P_atm = 101325.0              # Presión atmosférica [Pa]
T_atm = 298.15                # Temperatura ambiente [K]
eta_c = 0.80                  # Eficiencia isentrópica del compresor

# Normativa FIA
FIA_MAX_TURBO_RPM = 125000

# ==============================================================================
# FUNCIÓN PRINCIPAL DEL SIMULADOR INTERACTIVO
# ==============================================================================
def simulador_interactivo_f1(rpm_turbo, beta2_deg, p_boost_bar, r_eff, b1_mm, b2_mm, d2_mm):
    # Crear el lienzo
    fig, axs = plt.subplots(1, 3, figsize=(22, 7), facecolor='white')

    # Verificación Normativa FIA
    if rpm_turbo > FIA_MAX_TURBO_RPM:
        warning_text = f"⚠️ ALERTA FIA: Límite excedido. Turbocompresor a {rpm_turbo} RPM (Máx permitido: 125,000 RPM)"
        fig.text(0.5, 0.93, warning_text, ha='center', va='center', color='red', fontsize=14, fontweight='bold',
                 bbox=dict(facecolor='yellow', alpha=0.8, edgecolor='red', boxstyle='round,pad=0.5'))
    else:
        fig.text(0.5, 0.93, "Simulador de Unidad de Potencia (Verificado FIA ✓)",
                 ha='center', va='center', color='green', fontsize=14, fontweight='bold')

    # --------------------------------------------------------------------------
    # 1. ACOPLAMIENTO DE FLUJO MÁSICO (Motor -> Turbo)
    # --------------------------------------------------------------------------
    P_in = p_boost_bar * 1e5
    V_eff = V_c * r_eff
    V_1 = V_c + V_d

    # Masa de aire por cilindro por ciclo
    m_m = (P_in * V_eff) / (R_m * T1_in)

    # Flujo másico total del motor [kg/s] (6 cilindros, 4 tiempos)
    m_dot_total = m_m * 6.0 * (RPM_ENGINE / 120.0)

    # --------------------------------------------------------------------------
    # 2. CÁLCULOS DEL TURBOCOMPRESOR (Con Espesores y Diámetro Variable)
    # --------------------------------------------------------------------------
    b1 = b1_mm / 1000.0
    b2 = b2_mm / 1000.0
    D2_c = d2_mm / 1000.0

    # Estado 1 (Entrada compresor)
    rho1 = P_atm / (R_m * T_atm)
    Cm1_c = m_dot_total / (rho1 * np.pi * D1_c * b1)

    # Estado 2 (Salida compresor asumiendo compresión hacia P_in)
    T2_ideal = T_atm * (P_in / P_atm)**((1.4 - 1)/1.4)
    T2_real = T_atm + (T2_ideal - T_atm) / eta_c
    rho2 = P_in / (R_m * T2_real)
    Cm2_c = m_dot_total / (rho2 * np.pi * D2_c * b2)

    # Cinemática Salida
    omega = rpm_turbo * np.pi / 30.0
    U2 = omega * (D2_c / 2.0)
    beta2_rad = np.radians(beta2_deg)
    Cu2 = U2 - (Cm2_c / np.tan(beta2_rad))

    C2 = np.sqrt(Cu2**2 + Cm2_c**2)
    W2 = np.sqrt((U2 - Cu2)**2 + Cm2_c**2)
    alpha2_deg = np.degrees(np.arctan2(Cm2_c, Cu2))
    w_c_especifico = U2 * Cu2 # J/kg

    # --------------------------------------------------------------------------
    # 3. CÁLCULOS DEL MOTOR (CICLO MILLER)
    # --------------------------------------------------------------------------
    m_base = (4.0e5 * (V_c * 11.0)) / (R_m * T1_in)
    Q_in = Q_in_nominal * (m_m / m_base)

    P1_m, V1_m, T1_m = P_in, V_eff, T1_in
    V2_m = V_c
    P2_m = P1_m * (V1_m / V2_m)**gamma_m
    T2_m = T1_m * (V1_m / V2_m)**(gamma_m - 1.0)
    V3_m = V_c
    T3_m = T2_m + Q_in / (m_m * cv_m)
    P3_m = P2_m * (T3_m / T2_m)
    V4_m = V_1
    P4_m = P3_m * (V3_m / V4_m)**gamma_m

    v_comp = np.linspace(V1_m, V2_m, 100)
    p_comp = P1_m * (V1_m / v_comp)**gamma_m
    v_exp = np.linspace(V3_m, V4_m, 100)
    p_exp = P3_m * (V3_m / v_exp)**gamma_m

    W_exp_ciclo = np.trapezoid(p_exp, v_exp)
    W_comp_ciclo = np.trapezoid(p_comp[::-1], v_comp[::-1])
    W_neto = W_exp_ciclo - W_comp_ciclo
    eta = (W_neto / Q_in) * 100.0
    W_turbo_ciclo = w_c_especifico * m_m

    # --------------------------------------------------------------------------
    # GRÁFICO 1: TRIÁNGULO DE VELOCIDADES ROTULADO
    # --------------------------------------------------------------------------
    ax1 = axs[0]
    ax1.set_title(f"1. Triángulo de Salida Compresor ($C_{{m2}}$ = {Cm2_c:.1f} m/s)", fontweight='bold', fontsize=12, pad=15)

    ax1.set_aspect('equal', adjustable='datalim')
    ax1.axis('off')

    # Dibujar Vectores
    ax1.quiver(0, 0, U2, 0, angles='xy', scale_units='xy', scale=1, color='black', width=0.008)
    ax1.quiver(0, 0, Cu2, Cm2_c, angles='xy', scale_units='xy', scale=1, color='blue', width=0.008)
    ax1.quiver(U2, 0, Cu2-U2, Cm2_c, angles='xy', scale_units='xy', scale=1, color='red', width=0.008)
    ax1.plot([Cu2, Cu2], [0, Cm2_c], 'k--', alpha=0.3)

    bbox_props = dict(boxstyle="round,pad=0.3", fc="white", ec="none", alpha=0.85)

    # Textos de Velocidades
    ax1.text(U2/2, -max(15, Cm2_c*0.1), f"U = {U2:.0f} m/s", ha='center', fontweight='bold', bbox=bbox_props)
    ax1.text(Cu2/2 - 25, Cm2_c/2 + 15, f"C = {C2:.0f} m/s", ha='center', color='blue', fontweight='bold', bbox=bbox_props)
    ax1.text((U2+Cu2)/2 + 25, Cm2_c/2 + 15, f"W = {W2:.0f} m/s", ha='center', color='red', fontweight='bold', bbox=bbox_props)
    ax1.text(Cu2 + 35, Cm2_c/2, f"Cm2 = {Cm2_c:.0f}", color='gray', ha='center', bbox=bbox_props)

    # Arcos
    arc_alpha = patches.Arc((0,0), U2*0.3, U2*0.3, angle=0, theta1=0, theta2=alpha2_deg, color='blue', lw=2)
    ax1.add_patch(arc_alpha)
    ax1.text(U2*0.15 + 10, Cm2_c*0.15, f"α2={alpha2_deg:.1f}°", color='blue', fontweight='bold', bbox=bbox_props)

    arc_beta = patches.Arc((U2,0), U2*0.3, U2*0.3, angle=0, theta1=180-beta2_deg, theta2=180, color='red', lw=2)
    ax1.add_patch(arc_beta)
    ax1.text(U2 - U2*0.15 - 10, Cm2_c*0.15, f"β2={beta2_deg:.1f}°", color='red', fontweight='bold', ha='right', bbox=bbox_props)

    # --------------------------------------------------------------------------
    # GRÁFICO 2: DIAGRAMA P-V DINÁMICO
    # --------------------------------------------------------------------------
    ax2 = axs[1]
    ax2.set_title(f"2. Termodinámica: Ciclo P-V (η = {eta:.2f}%)", fontweight='bold', fontsize=12, pad=15)

    ax2.plot(v_comp * 1e6, p_comp / 1e5, 'g--', linewidth=2.5, label=f'Compresión (rc={r_eff:.1f})')
    ax2.plot(v_exp * 1e6, p_exp / 1e5, 'g-', linewidth=2.5, label='Expansión')
    ax2.plot([V2_m*1e6, V3_m*1e6], [P2_m/1e5, P3_m/1e5], 'g-', linewidth=2.5)
    ax2.plot([V4_m*1e6, V1_m*1e6], [P4_m/1e5, P1_m/1e5], 'g-', linewidth=2.5)
    ax2.fill_between(v_exp * 1e6, p_comp / 1e5, p_exp / 1e5, color='green', alpha=0.15)

    ax2.set_xlabel("Volumen del Cilindro [cm³]")
    ax2.set_ylabel("Presión Absoluta [bar]")
    ax2.set_xlim(0, (V_1*1e6) * 1.1)
    ax2.set_ylim(0, 450)
    ax2.grid(True, linestyle=':', alpha=0.7)
    ax2.legend(loc='upper right')

    # --------------------------------------------------------------------------
    # GRÁFICO 3: CONSUMO ENERGÉTICO (TRABAJO VS TRABAJO)
    # --------------------------------------------------------------------------
    ax3 = axs[2]
    ax3.set_title(f"3. Balance Energético (Flujo: {m_dot_total:.3f} kg/s)", fontweight='bold', fontsize=12, pad=15)

    labels = ['Produce\nExpansión', 'Consume\nMotor', 'Trabajo\nNeto', 'Consume\nTurbo']
    valores = [W_exp_ciclo, -W_comp_ciclo, W_neto, -W_turbo_ciclo]
    colores = ['#2ca02c', '#d62728', '#1f77b4', '#ff7f0e']

    barras = ax3.bar(labels, valores, color=colores, edgecolor='black', width=0.6)
    ax3.axhline(0, color='black', linewidth=1.5)

    for bar in barras:
        yval = bar.get_height()
        offset = 150 if yval >= 0 else -250
        ax3.text(bar.get_x() + bar.get_width()/2, yval + offset, f"{int(abs(yval))} J",
                 ha='center', va='center', fontweight='bold', color='black', fontsize=10)

    ax3.set_ylabel("Energía Transmitida [Joules]")
    ax3.set_ylim(-1500, 2500)
    ax3.grid(axis='y', linestyle=':', alpha=0.7)

    plt.tight_layout(rect=[0, 0, 1, 0.90])
    plt.show()

# ==============================================================================
# EJECUCIÓN DE WIDGETS INTERACTIVOS (SLIDERS CONFIGURADOS)
# ==============================================================================
# Definir estilo y tamaño para que el texto no se corte
estilo_texto = {'description_width': 'initial'}
tamano_slider = widgets.Layout(width='400px')

widgets.interact(simulador_interactivo_f1,
                 rpm_turbo=widgets.IntSlider(min=80000, max=135000, step=1000, value=100000,
                                             description='RPM Turbo:', style=estilo_texto, layout=tamano_slider),
                 beta2_deg=widgets.FloatSlider(min=30.0, max=90.0, step=1.0, value=45.0,
                                               description='Ángulo Salida β2 [°]:', style=estilo_texto, layout=tamano_slider),
                 p_boost_bar=widgets.FloatSlider(min=1.0, max=5.0, step=0.1, value=4.0,
                                                 description='Presión Boost [bar]:', style=estilo_texto, layout=tamano_slider),
                 r_eff=widgets.FloatSlider(min=8.0, max=14.0, step=0.1, value=11.0,
                                           description='Relación Compresión Miller (r_eff):', style=estilo_texto, layout=tamano_slider),
                 b1_mm=widgets.FloatSlider(min=10.0, max=40.0, step=0.5, value=25.0,
                                           description='Espesor Entrada b1 [mm]:', style=estilo_texto, layout=tamano_slider),
                 b2_mm=widgets.FloatSlider(min=2.0, max=15.0, step=0.1, value=5.0,
                                           description='Espesor Salida b2 [mm]:', style=estilo_texto, layout=tamano_slider),
                 d2_mm=widgets.FloatSlider(min=60.0, max=130.0, step=1.0, value=90.0,
                                           description='Diámetro Salida D2 [mm]:', style=estilo_texto, layout=tamano_slider))
```

Para comprender mejor el funcionamiento del turbocompresor como máquina, es esencial analizar su desempeño mediante la variación de algunos parámetros. Estos parámetros pueden ser tanto geométricos como de funcionamiento. Para este caso en particular, se enfoca en la variación paramétrica de las condiciones de operación del turbo: presión de entrada de la turbina y de sobrealimentación del compresor, caudal del aire de entrada al compresor, temperatura de entrada a la turbina y de salida del compresor y velocidad a la que rota la turbina. Esto con el fin de analizar variables como: la producción de trabajo, el consumo energético y la eficiencia térmica.

A continuación, se presenta una tabla que resume los parámetros a variar para el análisis del turbocompresor:

| Parámetro a Variar (Independiente) | Rango de Variación Estimado | Variables a Observar (Dependientes) |
| :--- | :--- | :--- |
| Presión de entrada a la turbina ($P_{1t}$) | $1.5$ a $3.0$ bar absolutos | Potencia de la turbina ($\dot{W}_t$), Relación de expansión ($r_t$), $P_2/P_1$ del compresor |
 | Presión de sobrealimentación del compresor ($P_{2c}$) | $2.0$ a $5.0$ bar absolutos | Potencia consumida del compresor ($\dot{W}_c$), Relación de compresión ($r_c$), Temperatura de salida del compresor ($T_{2c}$) |
 | Caudal másico de aire ($\dot{m}_a$) | $0.1$ a $0.5$ kg/s | Velocidad tangencial ($C_u$), Diámetro de salida ($D_{2c}$), Ancho del álabe ($b$) |
| Temperatura de entrada a la turbina ($T_{1t}$) | $900$ a $1300$ K | Potencia de la turbina ($\dot{W}_t$), Relación de expansión ($r_t$), Eficiencia térmica del ciclo |
 | Temperatura de salida del compresor ($T_{2c}$) | $350$ a $500$ K | Densidad del aire de admisión ($\rho_{2c}$), Masa de aire atrapada ($m_a$), Potencia del motor ($\dot{W}_{motor}$) |
 | Velocidad angular ($\omega$ / RPM) | $80,000$ a $120,000$ RPM | Trabajo de Euler ($w_c, w_t$), Relaciones de Presión ($r_c, r_t$), Potencia del sistema |


```python
# @title
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import ipywidgets as widgets
from IPython.display import display

# ==============================================================================
# CONSTANTES DEL SISTEMA F1 W15 Y REGLAMENTO FIA (Fixed values)
# ==============================================================================
V_total = 1.6e-3              # Cilindrada 1.6L [m^3]
V_d = V_total / 6.0
r_geom = 15.0                 # Relación de expansión geométrica fija para Miller
V_c = V_d / (r_geom - 1.0)    # Volumen de la cámara de combustión (Clearance)
cv_m, gamma_m = 750.0, 1.3    # Calor específico y gamma para gases en motor
R_m = 287.0                   # Constante del gas ideal para gases en motor
T1_in_base = 323.15           # Temperatura de admisión base [K] (50°C)
Q_in_nominal = 1100.0         # Joules de combustible base por ciclo por cilindro
RPM_ENGINE = 10500.0          # RPM del motor de combustión para flujo de masa

P_atm = 101325.0              # Presión atmosférica [Pa] (1.01325 bar)
T_atm = 298.15                # Temperatura ambiente [K] (25°C)

FIA_MAX_TURBO_RPM = 125000

# Specific heats and gas constants for compressor (air) and turbine (gas)
cp_aire, cp_gas = 1005.0, 1150.0
gamma_a, gamma_g = 1.4, 1.33
R_a = cp_aire * (gamma_a - 1.0) / gamma_a
R_g = cp_gas * (gamma_g - 1.0) / gamma_g

# Assumed Fixed Geometries and efficiencies (from 7r8zLNOzPVJW or original code's defaults)
D1_c_fixed = 0.050 # m (Compressor inlet eye diameter)
D2_c_fixed = 0.090 # m (Compressor outlet diameter - from old slider default)
b1_c_fixed = 0.025 # m (Compressor inlet blade width - from old slider default)
b2_c_fixed = 0.005 # m (Compressor outlet blade width - from old slider default)
beta2_c_deg_fixed = 45.0 # deg (Compressor outlet blade angle - from old slider default)

D1_t_fixed = 0.085 # m (Turbine inlet diameter)
D2_t_fixed = 0.065 # m (Turbine outlet diameter)
alpha1_t_deg_fixed = 70.0 # deg (Turbine inlet nozzle angle)
Cm1_t_fixed = 150.0 # m/s (Turbine inlet meridional velocity)
Cm2_t_fixed = 200.0 # m/s (Turbine outlet meridional velocity)
Cu2_t_fixed = 0.0 # m/s (Turbine outlet tangential velocity, axial exit assumption)

eta_t_fixed = 0.85 # Turbine isentropic efficiency
r_eff_miller_fixed = 11.0 # Effective compression ratio for Miller cycle

AFR_fixed = 13.7918 # Fixed Air-Fuel Ratio (from E10 analysis, for consistency)
LHV_fuel_MJ_kg = 40.97 # MJ/kg (from biocombustible analysis, for consistency)

# ==============================================================================
# FUNCIÓN PRINCIPAL DEL SIMULADOR INTERACTIVO (Parámetros Actualizados)
# ==============================================================================
def simulador_interactivo_parametric(
    p1_t_bar,           # Presión de entrada a la turbina (input)
    p_boost_bar,        # Presión de sobrealimentación del compresor (input)
    m_dot_a_kg_s,       # Caudal másico de aire (input)
    t1_t_k,             # Temperatura de entrada a la turbina (input)
    t2_c_k,             # Temperatura de salida del compresor (input)
    rpm_turbo           # Velocidad angular del turbocompresor (input)
):
    # Crear el lienzo
    fig, axs = plt.subplots(1, 3, figsize=(22, 7), facecolor='white')

    # Verificación Normativa FIA
    if rpm_turbo > FIA_MAX_TURBO_RPM:
        warning_text = f"⚠️ ALERTA FIA: Límite excedido. Turbocompresor a {rpm_turbo} RPM (Máx permitido: 125,000 RPM)"
        fig.text(0.5, 0.93, warning_text, ha='center', va='center', color='red', fontsize=14, fontweight='bold',
                 bbox=dict(facecolor='yellow', alpha=0.8, edgecolor='red', boxstyle='round,pad=0.5'))
    else:
        fig.text(0.5, 0.93, "Simulador de Unidad de Potencia (Verificado FIA ✓)",
                 ha='center', va='center', color='green', fontsize=14, fontweight='bold')

    # --------------------------------------------------------------------------
    # 1. FLUJOS MÁSICOS Y ACOPLAMIENTO (Motor <-> Turbo)
    # --------------------------------------------------------------------------
    m_dot_air_comp = m_dot_a_kg_s  # Input caudal másico de aire
    m_dot_fuel_kg_s = m_dot_air_comp / AFR_fixed # Derivado de AFR fijo
    m_dot_gas_turb = m_dot_air_comp + m_dot_fuel_kg_s # Masa total a la turbina

    # Masa por ciclo y cilindro para el motor
    m_m_cycle_air = m_dot_air_comp / (6.0 * (RPM_ENGINE / 120.0))
    m_m_cycle_fuel = m_dot_fuel_kg_s / (6.0 * (RPM_ENGINE / 120.0))

    # --------------------------------------------------------------------------
    # 2. CÁLCULOS DEL TURBOCOMPRESOR
    # --------------------------------------------------------------------------
    omega = rpm_turbo * np.pi / 30.0 # Velocidad angular del turbo

    # --- COMPRESOR ---
    P_comp_in = P_atm
    T_comp_in = T_atm
    P_comp_out = p_boost_bar * 1e5
    T_comp_out = t2_c_k  # Input temperatura de salida del compresor

    # Eficiencia isentrópica del compresor (derivada de los inputs)
    eta_c_derived = ((P_comp_out / P_comp_in)**((gamma_a - 1)/gamma_a) - 1) / ((T_comp_out / T_comp_in) - 1)

    rho1_c = P_comp_in / (R_a * T_comp_in)
    Cm1_c = m_dot_air_comp / (rho1_c * np.pi * D1_c_fixed * b1_c_fixed)

    rho2_c = P_comp_out / (R_a * T_comp_out)
    Cm2_c = m_dot_air_comp / (rho2_c * np.pi * D2_c_fixed * b2_c_fixed)

    U2_c = omega * (D2_c_fixed / 2.0)
    Cu2_c = U2_c - (Cm2_c / np.tan(np.radians(beta2_c_deg_fixed))) # Asumiendo beta2_c_deg_fixed
    w_c_especifico = U2_c * Cu2_c # Trabajo específico del compresor

    # --- TURBINA ---
    P_turb_in = p1_t_bar * 1e5  # Input presión de entrada a la turbina
    T_turb_in = t1_t_k         # Input temperatura de entrada a la turbina

    U1_t = omega * (D1_t_fixed / 2.0)
    U2_t = omega * (D2_t_fixed / 2.0)
    Cu1_t_val = Cm1_t_fixed * np.tan(np.radians(alpha1_t_deg_fixed)) # De geometría fija
    # Cu2_t_fixed = 0.0 (already defined as fixed for axial exit)
    w_turb_specific = U1_t * Cu1_t_val - U2_t * Cu2_t_fixed # Trabajo específico de la turbina

    T_turb_out = T_turb_in - (w_turb_specific / cp_gas)
    # Calcular P_turb_out usando la eficiencia isentrópica fija de la turbina
    T_turb_out_isentropic = T_turb_in - (T_turb_in - T_turb_out) / eta_t_fixed
    P_turb_out = P_turb_in * (T_turb_out_isentropic / T_turb_in)**(gamma_g / (gamma_g - 1))

    # --------------------------------------------------------------------------
    # 3. CÁLCULOS DEL MOTOR (CICLO MILLER)
    # --------------------------------------------------------------------------
    P_engine_in = P_comp_out # Presión de admisión del motor es la salida del compresor
    T_engine_in = T_comp_out # Temperatura de admisión del motor es la salida del compresor

    V_eff = V_c * r_eff_miller_fixed # Volumen efectivo para el inicio de la compresión Miller
    Q_in_cycle = m_m_cycle_fuel * LHV_fuel_MJ_kg * 1e6 # Calor inyectado por ciclo por cilindro (J)

    # Estados Miller
    P1_m, V1_m, T1_m = P_engine_in, V_eff, T_engine_in
    V2_m = V_c
    P2_m = P1_m * (V1_m / V2_m)**gamma_m
    T2_m = T1_m * (V1_m / V2_m)**(gamma_m - 1.0)
    V3_m = V_c
    T3_m = T2_m + Q_in_cycle / (m_m_cycle_air * cv_m)
    P3_m = P2_m * (T3_m / T2_m)
    V4_m = V_total / 6.0 + V_c # V_1 = V_c + V_d
    P4_m = P3_m * (V3_m / V4_m)**gamma_m

    # Vectores para la curva Miller
    v_comp = np.linspace(V1_m, V2_m, 100)
    p_comp = P1_m * (V1_m / v_comp)**gamma_m
    v_exp = np.linspace(V3_m, V4_m, 100)
    p_exp = P3_m * (V3_m / v_exp)**gamma_m

    W_exp_ciclo = np.trapezoid(p_exp, v_exp)
    W_comp_ciclo = np.trapezoid(p_comp[::-1], v_comp[::-1])
    W_neto = W_exp_ciclo - W_comp_ciclo
    eta_miller = (W_neto / Q_in_cycle) * 100.0

    # --------------------------------------------------------------------------
    # GRÁFICO 1: TRIÁNGULO DE VELOCIDADES ROTULADO (Salida del Compresor)
    # --------------------------------------------------------------------------
    ax1 = axs[0]
    C2_c = np.sqrt(Cu2_c**2 + Cm2_c**2)
    W2_c = np.sqrt((U2_c - Cu2_c)**2 + Cm2_c**2)
    alpha2_c_deg = np.degrees(np.arctan2(Cm2_c, Cu2_c))

    ax1.set_title("1. Compresor: Salida (" + r"$\eta_c$ = " + f"{eta_c_derived:.2f}" + ")", fontweight='bold', fontsize=12, pad=15)
    ax1.set_aspect('equal', adjustable='datalim')
    ax1.axis('off')

    ax1.quiver(0, 0, U2_c, 0, angles='xy', scale_units='xy', scale=1, color='black', width=0.008)
    ax1.quiver(0, 0, Cu2_c, Cm2_c, angles='xy', scale_units='xy', scale=1, color='blue', width=0.008)
    ax1.quiver(U2_c, 0, Cu2_c - U2_c, Cm2_c, angles='xy', scale_units='xy', scale=1, color='red', width=0.008)
    ax1.plot([Cu2_c, Cu2_c], [0, Cm2_c], 'k--', alpha=0.3)

    bbox_props = dict(boxstyle="round,pad=0.3", fc="white", ec="none", alpha=0.85)

    ax1.text(U2_c/2, -max(15, Cm2_c*0.1), f"U = {U2_c:.0f} m/s", ha='center', fontweight='bold', bbox=bbox_props)
    ax1.text(Cu2_c/2 - 25, Cm2_c/2 + 15, f"C = {C2_c:.0f} m/s", ha='center', color='blue', fontweight='bold', bbox=bbox_props)
    ax1.text((U2_c+Cu2_c)/2 + 25, Cm2_c/2 + 15, f"W = {W2_c:.0f} m/s", ha='center', color='red', fontweight='bold', bbox=bbox_props)
    ax1.text(Cu2_c + 35, Cm2_c/2, f"Cm2 = {Cm2_c:.0f}", color='gray', ha='center', bbox=bbox_props)

    arc_alpha = patches.Arc((0,0), U2_c*0.3, U2_c*0.3, angle=0, theta1=0, theta2=alpha2_c_deg, color='blue', lw=2)
    ax1.add_patch(arc_alpha)
    ax1.text(U2_c*0.15 + 10, Cm2_c*0.15, f"α2={alpha2_c_deg:.1f}°", color='blue', fontweight='bold', bbox=bbox_props)

    arc_beta = patches.Arc((U2_c,0), U2_c*0.3, U2_c*0.3, angle=0, theta1=180-beta2_c_deg_fixed, theta2=180, color='red', lw=2)
    ax1.add_patch(arc_beta)
    ax1.text(U2_c - U2_c*0.15 - 10, Cm2_c*0.15, f"β2={beta2_c_deg_fixed:.1f}°", color='red', fontweight='bold', ha='right', bbox=bbox_props)

    # --------------------------------------------------------------------------
    # GRÁFICO 2: DIAGRAMA P-V DINÁMICO (Ciclo Miller)
    # --------------------------------------------------------------------------
    ax2 = axs[1]
    ax2.set_title("2. Termodinámica: Ciclo P-V (" + r"$\eta$ = " + f"{eta_miller:.2f}" + "%) ", fontweight='bold', fontsize=12, pad=15)

    ax2.plot(v_comp * 1e6, p_comp / 1e5, 'g--', linewidth=2.5, label=f'Compresión (rc={r_eff_miller_fixed:.1f})')
    ax2.plot(v_exp * 1e6, p_exp / 1e5, 'g-', linewidth=2.5, label='Expansión')
    ax2.plot([V2_m*1e6, V3_m*1e6], [P2_m/1e5, P3_m/1e5], 'g-', linewidth=2.5)
    ax2.plot([V4_m*1e6, V1_m*1e6], [P4_m/1e5, P1_m/1e5], 'g-', linewidth=2.5)
    ax2.fill_between(v_exp * 1e6, p_comp / 1e5, p_exp / 1e5, color='green', alpha=0.15)

    ax2.set_xlabel("Volumen del Cilindro [cm³]")
    ax2.set_ylabel("Presión Absoluta [bar]")
    ax2.set_xlim(0, (V_total / 6.0 + V_c) * 1e6 * 1.1) # Adjusted xlim
    ax2.set_ylim(0, 450) # Keep max pressure limit for F1 engines
    ax2.grid(True, linestyle=':', alpha=0.7)
    ax2.legend(loc='upper right')

    # --------------------------------------------------------------------------
    # GRÁFICO 3: BALANCE ENERGÉTICO
    # --------------------------------------------------------------------------
    ax3 = axs[2]
    ax3.set_title(f"3. Balance Energético (Turbina: {P_turb_in/1e5:.1f} bar, Compresor: {P_comp_out/1e5:.1f} bar)", fontweight='bold', fontsize=12, pad=15)

    # Work consumed by compressor (per engine cycle)
    W_turbo_consumed_by_engine_cycle = w_c_especifico * m_m_cycle_air

    labels = ['Produce\nExpansión', 'Consume\nCompresión', 'Trabajo\nNeto', 'Consume\nCompresor']
    valores = [W_exp_ciclo, -W_comp_ciclo, W_neto, -W_turbo_consumed_by_engine_cycle]
    colores = ['#2ca02c', '#d62728', '#1f77b4', '#ff7f0e']

    barras = ax3.bar(labels, valores, color=colores, edgecolor='black', width=0.6)
    ax3.axhline(0, color='black', linewidth=1.5)

    for bar in barras:
        yval = bar.get_height()
        offset = 150 if yval >= 0 else -250
        ax3.text(bar.get_x() + bar.get_width()/2, yval + offset, f"{int(abs(yval))} J",
                 ha='center', va='center', fontweight='bold', color='black', fontsize=10)

    ax3.set_ylabel("Energía Transmitida [Joules/Ciclo]")
    ax3.set_ylim(-1500, 2500)
    ax3.grid(axis='y', linestyle=':', alpha=0.7)

    plt.tight_layout(rect=[0, 0, 1, 0.90])
    plt.show()

# ==============================================================================
# EJECUCIÓN DE WIDGETS INTERACTIVOS (SLIDERS CONFIGURADOS CON NUEVOS PARÁMETROS)
# ==============================================================================
estilo_texto = {'description_width': 'initial'}
tamano_slider = widgets.Layout(width='400px')

widgets.interact(simulador_interactivo_parametric,
                 p1_t_bar=widgets.FloatSlider(min=1.5, max=3.0, step=0.05, value=2.0,
                                             description='Presión Entrada Turbina (bar):', style=estilo_texto, layout=tamano_slider),
                 p_boost_bar=widgets.FloatSlider(min=2.0, max=5.0, step=0.1, value=4.0,
                                                 description='Presión Boost Compresor (bar):', style=estilo_texto, layout=tamano_slider),
                 m_dot_a_kg_s=widgets.FloatSlider(min=0.1, max=0.5, step=0.01, value=0.3,
                                                 description='Caudal Aire Compresor (kg/s):', style=estilo_texto, layout=tamano_slider),
                 t1_t_k=widgets.IntSlider(min=900, max=1300, step=10, value=1100,
                                         description='Temp Entrada Turbina (K):', style=estilo_texto, layout=tamano_slider),
                 t2_c_k=widgets.IntSlider(min=350, max=500, step=5, value=400,
                                         description='Temp Salida Compresor (K):', style=estilo_texto, layout=tamano_slider),
                 rpm_turbo=widgets.IntSlider(min=80000, max=120000, step=1000, value=100000,
                                             description='RPM Turbo:', style=estilo_texto, layout=tamano_slider))
```

Utilizando la herramienta anterior se pueden ver los efectos que tienen estos parámetros en el turbo y como la variación de estos afecta su comportamiento. A continuación se presenta el caso que maximiza la cantidad de trabajo neto generado por la máquina y sus condiciones de operación.

```python
# @title
import numpy as np
import pandas as pd

# Helper function to extract key engine metrics for optimization
def get_engine_metrics_for_optimization(
    p1_t_bar, p_boost_bar, m_dot_a_kg_s, t1_t_k, t2_c_k, rpm_turbo
):
    # Re-use all fixed constants from the previous cell's global scope
    # Global constants are accessed directly within this function

    # 1. FLOW RATES
    m_dot_air_comp = m_dot_a_kg_s
    m_dot_fuel_kg_s = m_dot_air_comp / AFR_fixed
    m_dot_gas_turb = m_dot_air_comp + m_dot_fuel_kg_s

    # Avoid division by zero if RPM_ENGINE is 0
    if RPM_ENGINE == 0:
        m_m_cycle_air = 0
        m_m_cycle_fuel = 0
    else:
        m_m_cycle_air = m_dot_air_comp / (6.0 * (RPM_ENGINE / 120.0))
        m_m_cycle_fuel = m_dot_fuel_kg_s / (6.0 * (RPM_ENGINE / 120.0))

    # 2. TURBOCOMPRESSOR CALCULATIONS
    omega = rpm_turbo * np.pi / 30.0

    # --- COMPRESSOR ---
    P_comp_in = P_atm
    T_comp_in = T_atm
    P_comp_out = p_boost_bar * 1e5
    T_comp_out = t2_c_k

    # Derived compressor efficiency. Handle potential division by zero.
    if (T_comp_out / T_comp_in) - 1 == 0:
        eta_c_derived = 0
    else:
        eta_c_derived = ((P_comp_out / P_comp_in)**((gamma_a - 1)/gamma_a) - 1) / ((T_comp_out / T_comp_in) - 1)

    # Filter out physically impossible or unrealistic efficiencies
    if eta_c_derived <= 0.01 or eta_c_derived > 1.0:
        return -np.inf, 0, eta_c_derived # Return a very low work value for infeasible points

    rho1_c = P_comp_in / (R_a * T_comp_in)
    if D1_c_fixed == 0 or b1_c_fixed == 0 or rho1_c == 0: # Avoid division by zero
        Cm1_c = 0
    else:
        Cm1_c = m_dot_air_comp / (rho1_c * np.pi * D1_c_fixed * b1_c_fixed)

    rho2_c = P_comp_out / (R_a * T_comp_out)
    if D2_c_fixed == 0 or b2_c_fixed == 0 or rho2_c == 0: # Avoid division by zero
        Cm2_c = 0
    else:
        Cm2_c = m_dot_air_comp / (rho2_c * np.pi * D2_c_fixed * b2_c_fixed)

    U2_c = omega * (D2_c_fixed / 2.0)
    # Check for division by zero if tan(beta2_c_deg_fixed) is zero or very close
    if np.abs(np.tan(np.radians(beta2_c_deg_fixed))) < 1e-6:
        Cu2_c = U2_c # Assuming purely radial blade for simplicity if tan is near zero
    else:
        Cu2_c = U2_c - (Cm2_c / np.tan(np.radians(beta2_c_deg_fixed)))

    w_c_especifico = U2_c * Cu2_c

    # --- TURBINE ---
    P_turb_in = p1_t_bar * 1e5
    T_turb_in = t1_t_k

    U1_t = omega * (D1_t_fixed / 2.0)
    U2_t = omega * (D2_t_fixed / 2.0)
    Cu1_t_val = Cm1_t_fixed * np.tan(np.radians(alpha1_t_deg_fixed))
    w_turb_specific = U1_t * Cu1_t_val - U2_t * Cu2_t_fixed

    T_turb_out = T_turb_in - (w_turb_specific / cp_gas)
    if T_turb_out <= 0: # Physically impossible temperature
        return -np.inf, 0, eta_c_derived

    # Handle potential division by zero for T_turb_in and eta_t_fixed
    if eta_t_fixed == 0 or T_turb_in == 0:
        T_turb_out_isentropic = 0
    else:
        T_turb_out_isentropic = T_turb_in - (T_turb_in - T_turb_out) / eta_t_fixed

    if T_turb_out_isentropic <= 0 or T_turb_in == 0 or (gamma_g / (gamma_g - 1)) == 0:
        P_turb_out = 0 # Infeasible pressure
    else:
        P_turb_out = P_turb_in * (T_turb_out_isentropic / T_turb_in)**(gamma_g / (gamma_g - 1))

    # Constraint check: Turbine must power compressor
    # If the compressor work is positive (compression happens), then turbine work should be at least that
    if w_c_especifico > 0 and (m_dot_gas_turb * w_turb_specific < m_dot_air_comp * w_c_especifico * 0.95): # Adding a small buffer for losses
        return -np.inf, 0, eta_c_derived

    # 3. MILLER CYCLE CALCULATIONS
    P_engine_in = P_comp_out
    T_engine_in = T_comp_out

    V_eff = V_c * r_eff_miller_fixed
    Q_in_cycle = m_m_cycle_fuel * LHV_fuel_MJ_kg * 1e6

    # Avoid division by zero for m_m_cycle_air or cv_m
    if m_m_cycle_air == 0 or cv_m == 0:
        T3_m = 0
    else:
        T3_m = T_engine_in + Q_in_cycle / (m_m_cycle_air * cv_m)

    P1_m, V1_m, T1_m = P_engine_in, V_eff, T_engine_in
    V2_m = V_c

    # Handle potential division by zero for V2_m or V1_m
    if V2_m == 0:
        P2_m = 0 # Infeasible pressure
    else:
        P2_m = P1_m * (V1_m / V2_m)**gamma_m

    V3_m = V_c
    # Handle potential division by zero for T_engine_in
    if T_engine_in == 0:
        P3_m = 0
    else:
        P3_m = P2_m * (T3_m / T_engine_in)

    V4_m = V_total / 6.0 + V_c
    # Handle potential division by zero for V4_m or V3_m
    if V4_m == 0:
        P4_m = 0
    else:
        P4_m = P3_m * (V3_m / V4_m)**gamma_m

    # Work calculation
    v_comp = np.linspace(V1_m, V2_m, 100)
    p_comp = P1_m * (V1_m / v_comp)**gamma_m
    v_exp = np.linspace(V3_m, V4_m, 100)
    p_exp = P3_m * (V3_m / v_exp)**gamma_m

    W_exp_ciclo = np.trapezoid(p_exp, v_exp)
    W_comp_ciclo = np.trapezoid(p_comp[::-1], v_comp[::-1])
    W_neto = W_exp_ciclo - W_comp_ciclo

    if Q_in_cycle == 0:
        eta_miller = 0
    else:
        eta_miller = (W_neto / Q_in_cycle) * 100.0

    return W_neto, eta_miller, eta_c_derived

# --- OPTIMIZATION SETUP ---

# Define parameter ranges for the grid search (coarser steps for demonstration)
p1_t_bar_range = np.linspace(1.5, 3.0, 5) # 5 steps for turbine inlet pressure
p_boost_bar_range = np.linspace(2.0, 5.0, 5) # 5 steps for compressor boost pressure
m_dot_a_kg_s_range = np.linspace(0.1, 0.5, 5) # 5 steps for air mass flow
t1_t_k_range = np.linspace(900, 1300, 5) # 5 steps for turbine inlet temp
t2_c_k_range = np.linspace(350, 500, 5) # 5 steps for compressor outlet temp
rpm_turbo_range = np.linspace(80000, FIA_MAX_TURBO_RPM, 5) # 5 steps up to FIA limit

max_w_neto = -np.inf
best_params = {}

print("Starting optimization search...")

# Perform Grid Search
for p1_t in p1_t_bar_range:
    for p_boost in p_boost_bar_range:
        for m_dot_a in m_dot_a_kg_s_range:
            for t1_t in t1_t_k_range:
                for t2_c in t2_c_k_range:
                    for rpm_t in rpm_turbo_range:
                        # Apply FIA RPM limit as a hard constraint
                        if rpm_t > FIA_MAX_TURBO_RPM:
                            continue

                        w_neto, eta_miller, eta_c_derived = get_engine_metrics_for_optimization(p1_t, p_boost, m_dot_a, t1_t, t2_c, rpm_t)

                        if w_neto > max_w_neto:
                            max_w_neto = w_neto
                            best_params = {
                                'p1_t_bar': p1_t,
                                'p_boost_bar': p_boost,
                                'm_dot_a_kg_s': m_dot_a,
                                't1_t_k': t1_t,
                                't2_c_k': t2_c,
                                'rpm_turbo': rpm_t,
                                'eta_miller': eta_miller,
                                'eta_c_derived': eta_c_derived
                            }

print("Optimization complete.\n")

if best_params:
    print("==========================================")
    print("   Optimal Operating Point (Max W_neto)   ")
    print("==========================================")
    print(f"Net Work per Cycle (W_neto): {max_w_neto:.2f} J")
    print(f"Miller Cycle Efficiency:     {best_params['eta_miller']:.2f} %")
    print(f"Compressor Efficiency:       {best_params['eta_c_derived']:.2f}")
    print("------------------------------------------")
    print("Optimal Parameters:")
    print(f"  Turbine Inlet Pressure: {best_params['p1_t_bar']:.2f} bar")
    print(f"  Compressor Boost Pressure: {best_params['p_boost_bar']:.2f} bar")
    print(f"  Air Mass Flow Rate:     {best_params['m_dot_a_kg_s']:.3f} kg/s")
    print(f"  Turbine Inlet Temperature: {best_params['t1_t_k']:.0f} K")
    print(f"  Compressor Outlet Temperature: {best_params['t2_c_k']:.0f} K")
    print(f"  Turbo RPM:              {best_params['rpm_turbo']:.0f}")
    print("==========================================")

    # Visualize the optimal point
    print("\nGenerating plots for the optimal operating point...")
    simulador_interactivo_parametric(
        p1_t_bar=best_params['p1_t_bar'],
        p_boost_bar=best_params['p_boost_bar'],
        m_dot_a_kg_s=best_params['m_dot_a_kg_s'],
        t1_t_k=best_params['t1_t_k'],
        t2_c_k=best_params['t2_c_k'],
        rpm_turbo=best_params['rpm_turbo']
    )

else:
    print("No feasible optimal point found within the defined ranges and constraints.")
```

```python
# @title
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import ipywidgets as widgets
from IPython.display import display, clear_output

# ==============================================================================
# CONSTANTES DEL SISTEMA F1 W15 Y REGLAMENTO FIA
# ==============================================================================
V_total = 1.6e-3              # Cilindrada 1.6L [m^3]
V_d = V_total / 6.0
r_geom = 15.0                 # Relación de expansión geométrica fija para Miller
V_c = V_d / (r_geom - 1.0)    # Volumen de la cámara de combustión (Clearance)
cv_m, gamma_m = 750.0, 1.3    # Calor específico y gamma para gases en motor
R_m = 287.0                   # Constante del gas ideal para gases en motor
T1_in_base = 323.15           # Temperatura de admisión base [K] (50°C)
Q_in_nominal = 1100.0         # Joules de combustible base por ciclo por cilindro
RPM_ENGINE = 10500.0          # RPM del motor de combustión para flujo de masa

P_atm = 101325.0              # Presión atmosférica [Pa] (1.01325 bar)
T_atm = 298.15                # Temperatura ambiente [K] (25°C)

FIA_MAX_TURBO_RPM = 125000

# Specific heats and gas constants for compressor (air) and turbine (gas)
cp_aire, cp_gas = 1005.0, 1150.0
gamma_a, gamma_g = 1.4, 1.33
R_a = cp_aire * (gamma_a - 1.0) / gamma_a
R_g = cp_gas * (gamma_g - 1.0) / gamma_g

# Fixed efficiencies
eta_c_fixed = 0.80 # Compressor isentropic efficiency
eta_t_fixed = 0.85 # Turbine isentropic efficiency
r_eff_miller_fixed = 11.0 # Effective compression ratio for Miller cycle

AFR_fixed = 13.7918 # Fixed Air-Fuel Ratio
LHV_fuel_MJ_kg = 40.97 # MJ/kg

D1_c_fixed = 0.050 # m (Compressor inlet eye diameter)
b1_t_fixed = 0.020 # m (Assumed turbine inlet blade width)
b2_t_fixed = 0.015 # m (Assumed turbine outlet blade width)
Cu2_t_fixed = 0.0  # m/s (Turbine outlet tangential velocity)

# Fixed operational parameters
p1_t_bar_optimal = 1.50
p_boost_bar_optimal = 5.00
m_dot_a_kg_s_optimal = 0.100
t1_t_k_optimal = 900
t2_c_k_optimal = 500
rpm_turbo_optimal = 80000

# ==============================================================================
# HELPER FUNCTION: PLOT SINGLE VELOCITY TRIANGLE (ANTI-SOLAPAMIENTO)
# ==============================================================================
def plot_single_velocity_triangle(ax, U, Cu, Cm, alpha_text, beta_text, title):
    ax.set_title(title, fontweight='bold', fontsize=10, pad=15)

    max_val = max(abs(U), abs(Cu), abs(Cm), np.sqrt(Cu**2+Cm**2), np.sqrt((U-Cu)**2+Cm**2))
    if max_val == 0: max_val = 1

    # Límites dinámicos con margen amplio para los textos
    margin = max_val * 0.35
    ax.set_xlim(min(-margin, U-margin, Cu-margin), max(margin, U+margin, Cu+margin))
    ax.set_ylim(-max_val*0.2, max_val*0.8)
    ax.set_aspect('equal', adjustable='box')
    ax.axis('off')

    # Vectores principales
    ax.quiver(0, 0, U, 0, angles='xy', scale_units='xy', scale=1, color='black', width=0.003)
    ax.quiver(0, 0, Cu, Cm, angles='xy', scale_units='xy', scale=1, color='blue', width=0.003)
    ax.quiver(U, 0, Cu - U, Cm, angles='xy', scale_units='xy', scale=1, color='red', width=0.003)

    bbox_props = dict(boxstyle="round,pad=0.2", fc="white", ec="none", alpha=0.9)

    # U (Abajo en el centro)
    ax.text(U/2, -max_val*0.1, f"U={U:.0f}m/s", ha='center', color='black', fontsize=8, fontweight='bold', bbox=bbox_props)

    # C (Empujado hacia arriba y afuera de la mitad)
    C_mag = np.sqrt(Cu**2+Cm**2)
    offset_c_x = -max_val * 0.15 if Cu >= U/2 else max_val * 0.15
    ax.text(Cu/2 + offset_c_x, Cm/2 + max_val * 0.15, f"C={C_mag:.0f}m/s", ha='center', color='blue', fontsize=8, fontweight='bold', bbox=bbox_props)

    # W (Empujado hacia arriba y afuera de la mitad)
    W_mag = np.sqrt((Cu-U)**2 + Cm**2)
    offset_w_x = max_val * 0.15 if (Cu-U) <= 0 else -max_val * 0.15
    ax.text((U+Cu)/2 + offset_w_x, Cm/2 + max_val * 0.15, f"W={W_mag:.0f}m/s", ha='center', color='red', fontsize=8, fontweight='bold', bbox=bbox_props)

    # Cm (Desplazado a la derecha para no chocar con vértices)
    if abs(Cu) > 1.0:
        ax.plot([Cu, Cu], [0, Cm], 'k--', alpha=0.3, lw=0.5)
    ax.text(Cu + max_val*0.06, Cm/2, f"Cm={Cm:.0f}", color='gray', ha='left', va='center', fontsize=8, bbox=bbox_props)

    # TEXTO ALFA (Anclado siempre a la esquina izquierda baja)
    if alpha_text > 80.0:
        x_alpha = -max_val * 0.08
        y_alpha = max_val * 0.08
    else:
        x_alpha = max_val * 0.12
        y_alpha = max_val * 0.04
        if Cu < 0: x_alpha = -x_alpha

    ax.text(x_alpha, y_alpha, f"α={alpha_text:.1f}°", color='blue', fontweight='bold', ha='center', fontsize=8, bbox=bbox_props)

    # TEXTO BETA (Anclado siempre a la esquina derecha baja)
    if 80.0 < beta_text < 100.0:
        x_beta = U + max_val * 0.08
        y_beta = max_val * 0.08
    elif beta_text >= 100.0:
        x_beta = U + max_val * 0.12
        y_beta = max_val * 0.04
    else:
        x_beta = U - max_val * 0.12
        y_beta = max_val * 0.04

    ax.text(x_beta, y_beta, f"β={beta_text:.1f}°", color='red', fontweight='bold', ha='center', fontsize=8, bbox=bbox_props)

# ==============================================================================
# FUNCIÓN PRINCIPAL DEL SIMULADOR INTERACTIVO
# ==============================================================================
def simulador_interactivo_geometric(
    beta2_c_deg, D2_c_mm, b1_c_mm, b2_c_mm,
    D1_t_mm, D2_t_mm, alpha1_t_deg, beta2_t_deg
):
    clear_output(wait=True)

    p1_t_bar = p1_t_bar_optimal
    m_dot_a_kg_s = m_dot_a_kg_s_optimal
    t1_t_k = t1_t_k_optimal
    rpm_turbo = rpm_turbo_optimal

    fig, axs = plt.subplots(2, 3, figsize=(22, 12), facecolor='white')

    if rpm_turbo > FIA_MAX_TURBO_RPM:
        warning_text = f"⚠️ ALERTA FIA: Límite excedido. Turbocompresor a {rpm_turbo} RPM (Máx permitido: 125,000 RPM)"
        fig.text(0.5, 0.93, warning_text, ha='center', va='center', color='red', fontsize=14, fontweight='bold',
                 bbox=dict(facecolor='yellow', alpha=0.8, edgecolor='red', boxstyle='round,pad=0.5'))
    else:
        fig.text(0.5, 0.93, "Simulador de Unidad de Potencia (Acoplado Geométricamente)",
                 ha='center', va='center', color='green', fontsize=14, fontweight='bold')

    # 1. FLUJOS MÁSICOS
    m_dot_air_comp = m_dot_a_kg_s
    m_dot_fuel_kg_s = m_dot_air_comp / AFR_fixed
    m_dot_gas_turb = m_dot_air_comp + m_dot_fuel_kg_s

    m_m_cycle_air = m_dot_air_comp / (6.0 * (RPM_ENGINE / 120.0))
    m_m_cycle_fuel = m_dot_fuel_kg_s / (6.0 * (RPM_ENGINE / 120.0))
    m_m_cycle_total = m_m_cycle_air + m_m_cycle_fuel
    m_t_cycle_gas = m_dot_gas_turb / (6.0 * (RPM_ENGINE / 120.0))

    # 2. CÁLCULOS DEL TURBOCOMPRESOR
    omega = rpm_turbo * np.pi / 30.0

    b1_c = b1_c_mm / 1000.0
    b2_c = b2_c_mm / 1000.0
    D2_c = D2_c_mm / 1000.0

    # --- COMPRESOR (Iterative solver) ---
    P_comp_in = P_atm
    T_comp_in = T_atm
    P_comp_out = p_boost_bar_optimal * 1e5
    T_comp_out = t2_c_k_optimal

    for _ in range(10):
        rho2_c = P_comp_out / (R_a * T_comp_out) if T_comp_out > 0 else 1.2
        Cm2_c = m_dot_air_comp / (rho2_c * np.pi * D2_c * b2_c) if (rho2_c * np.pi * D2_c * b2_c) != 0 else 0
        U2_c = omega * (D2_c / 2.0)

        tan_beta2_c = np.tan(np.radians(beta2_c_deg))
        if np.abs(tan_beta2_c) < 1e-9:
            Cu2_c = U2_c
        else:
            Cu2_c = U2_c - (Cm2_c / tan_beta2_c)

        w_c_especifico = U2_c * Cu2_c
        if w_c_especifico < 0: w_c_especifico = 0.0

        T_comp_out = T_comp_in + (w_c_especifico / cp_aire)
        P_comp_out = P_comp_in * (1.0 + eta_c_fixed * (w_c_especifico / (cp_aire * T_comp_in)))**(gamma_a / (gamma_a - 1.0))

    eta_c_derived = eta_c_fixed

    # Estado de entrada del compresor
    rho1_c = P_comp_in / (R_a * T_comp_in) if T_comp_in > 0 else 0
    Cm1_c = m_dot_air_comp / (rho1_c * np.pi * D1_c_fixed * b1_c) if (rho1_c * np.pi * D1_c_fixed * b1_c) != 0 else 0
    U1_c = omega * (D1_c_fixed / 2.0)
    Cu1_c = 0.0
    alpha1_c_deg = 90.0
    beta1_c_deg = np.degrees(np.arctan2(Cm1_c, abs(Cu1_c - U1_c))) if abs(Cu1_c - U1_c) > 1e-9 else 90.0

    if Cm2_c == 0 and Cu2_c == 0:
        alpha2_c_deg = 0
    elif Cu2_c == 0:
        alpha2_c_deg = 90.0
    else:
        alpha2_c_deg = np.degrees(np.arctan2(Cm2_c, abs(Cu2_c)))

    # --- TURBINA ---
    P_turb_in = p1_t_bar * 1e5
    T_turb_in = t1_t_k
    P_turb_out = P_atm

    D1_t = D1_t_mm / 1000.0
    D2_t = D2_t_mm / 1000.0

    U1_t = omega * (D1_t / 2.0)
    rho1_t = P_turb_in / (R_g * T_turb_in) if T_turb_in > 0 else 0
    Cm1_t = m_dot_gas_turb / (rho1_t * np.pi * D1_t * b1_t_fixed) if (rho1_t * np.pi * D1_t * b1_t_fixed) != 0 else 0
    Cu1_t_val = Cm1_t * np.tan(np.radians(alpha1_t_deg))

    if (Cu1_t_val - U1_t) == 0:
        beta1_t_deg = 90.0
    else:
        beta1_t_deg = np.degrees(np.arctan2(Cm1_t, abs(Cu1_t_val - U1_t)))

    U2_t = omega * (D2_t / 2.0)
    T_turb_out_temp = T_turb_in - (U1_t * Cu1_t_val / cp_gas)
    rho2_t = P_turb_out / (R_g * T_turb_out_temp) if T_turb_out_temp > 0 else 0
    Cm2_t = m_dot_gas_turb / (rho2_t * np.pi * D2_t * b2_t_fixed) if (rho2_t * np.pi * D2_t * b2_t_fixed) != 0 else 0

    tan_beta2_t = np.tan(np.radians(beta2_t_deg))
    if np.abs(tan_beta2_t) < 1e-9:
        Cu2_t_work = U2_t
    else:
        Cu2_t_work = U2_t - (Cm2_t / tan_beta2_t)

    alpha2_t_deg = np.degrees(np.arctan2(Cm2_t, abs(Cu2_t_work))) if abs(Cu2_t_work) > 1e-9 else 90.0

    w_turb_specific = U1_t * Cu1_t_val - U2_t * Cu2_t_work

    T_turb_out = T_turb_in - (w_turb_specific / cp_gas)
    if T_turb_out <= 0:
        P_turb_out = 0
    else:
        if T_turb_in == 0 or eta_t_fixed == 0 or (gamma_g / (gamma_g - 1)) == 0:
            P_turb_out = 0
        else:
            T_turb_out_isentropic = T_turb_in - (T_turb_in - T_turb_out) / eta_t_fixed
            if T_turb_out_isentropic > 0:
                P_turb_out = P_turb_in * (T_turb_out_isentropic / T_turb_in)**(gamma_g / (gamma_g - 1))
            else:
                P_turb_out = 0

    # 3. CÁLCULOS DEL MOTOR (CICLO MILLER CORREGIDO)
    P_engine_in = P_comp_out
    T_engine_in = T_comp_out

    V_eff = V_c * r_eff_miller_fixed
    Q_in_cycle = m_m_cycle_fuel * LHV_fuel_MJ_kg * 1e6

    # Estado 1: Entrada (después del compresor)
    P1_m, V1_m, T1_m = P_engine_in, V_eff, T_engine_in

    # Estado 2: Fin de compresión
    V2_m = V_c
    if V2_m == 0:
        P2_m, T2_m = 0, 0
    else:
        P2_m = P1_m * (V1_m / V2_m)**gamma_m
        T2_m = T1_m * (V1_m / V2_m)**(gamma_m - 1.0) # CORRECCIÓN: T2 calculada correctamente

    # Estado 3: Fin de Combustión (volumen constante)
    V3_m = V_c
    if m_m_cycle_total == 0 or cv_m == 0:
        T3_m = T2_m
    else:
        T3_m = T2_m + Q_in_cycle / (m_m_cycle_total * cv_m) # CORRECCIÓN: El calor se suma a T2, no a T1

    if T2_m == 0:
        P3_m = 0
    else:
        P3_m = P2_m * (T3_m / T2_m) # CORRECCIÓN: Multiplicador respecto a T2

    # Estado 4: Fin de expansión
    V4_m = V_total / 6.0 + V_c
    if V4_m == 0:
        P4_m = 0
    else:
        P4_m = P3_m * (V3_m / V4_m)**gamma_m

    # Interpolación para gráficos P-V
    v_comp = np.linspace(V1_m, V2_m, 100) if V1_m != V2_m else np.array([V1_m, V2_m])
    p_comp = P1_m * (V1_m / v_comp)**gamma_m if v_comp.any() else np.array([0,0])

    v_exp = np.linspace(V3_m, V4_m, 100) if V3_m != V4_m else np.array([V3_m, V4_m])
    p_exp = P3_m * (V3_m / v_exp)**gamma_m if v_exp.any() else np.array([0,0])

    # Cálculo de Trabajo Neto y Eficiencia
    W_exp_ciclo = np.trapezoid(p_exp, v_exp)
    W_comp_ciclo = np.trapezoid(p_comp[::-1], v_comp[::-1])
    W_neto = W_exp_ciclo - W_comp_ciclo

    if Q_in_cycle == 0:
        eta_miller = 0
    else:
        eta_miller = (W_neto / Q_in_cycle) * 100.0

    # GRÁFICOS
    plot_single_velocity_triangle(axs[0,0], U1_c, Cu1_c, Cm1_c, alpha1_c_deg, beta1_c_deg, "Compresor - Entrada")
    plot_single_velocity_triangle(axs[0,1], U2_c, Cu2_c, Cm2_c, alpha2_c_deg, beta2_c_deg, "Compresor - Salida")
    plot_single_velocity_triangle(axs[0,2], U1_t, Cu1_t_val, Cm1_t, alpha1_t_deg, beta1_t_deg, "Turbina - Entrada")
    plot_single_velocity_triangle(axs[1,0], U2_t, Cu2_t_work, Cm2_t, alpha2_t_deg, beta2_t_deg, "Turbina - Salida")

    # GRÁFICO 5: DIAGRAMA P-V DINÁMICO
    ax2 = axs[1,1]
    ax2.set_title("Termodinámica: Ciclo P-V (" + r"$\eta$ = " + f"{eta_miller:.2f}" + "%) ", fontweight='bold', fontsize=12, pad=15)

    ax2.plot(v_comp * 1e6, p_comp / 1e5, 'g--', linewidth=2.5, label=f'Compresión (rc={r_eff_miller_fixed:.1f})')
    ax2.plot(v_exp * 1e6, p_exp / 1e5, 'g-', linewidth=2.5, label='Expansión')
    ax2.plot([V2_m*1e6, V3_m*1e6], [P2_m/1e5, P3_m/1e5], 'g-', linewidth=2.5)
    ax2.plot([V4_m*1e6, V1_m*1e6], [P4_m/1e5, P1_m/1e5], 'g-', linewidth=2.5)
    ax2.fill_between(v_exp * 1e6, p_comp / 1e5, p_exp / 1e5, color='green', alpha=0.15)

    ax2.set_xlabel("Volumen del Cilindro [cm³]")
    ax2.set_ylabel("Presión Absoluta [bar]")
    ax2.set_xlim(0, (V_total / 6.0 + V_c) * 1e6 * 1.1)
    ax2.set_ylim(0, max(P3_m/1e5 * 1.1, 150)) # Ajustado límite max para no comprimir si hay poca presión
    ax2.grid(True, linestyle=':', alpha=0.7)
    ax2.legend(loc='upper right')

    # GRÁFICO 6: BALANCE ENERGÉTICO (Anti-Solapamiento Arreglado)
    ax3 = axs[1,2]
    ax3.set_title(f"Balance Energético (W_neto={W_neto:.0f} J)", fontweight='bold', fontsize=12, pad=15)

    W_turbo_consumed_by_engine_cycle = w_c_especifico * m_m_cycle_air
    W_turb_produced_cycle = w_turb_specific * m_t_cycle_gas

    labels = ['Produce\nExpansión', 'Consume\nCompresión', 'Trabajo\nNeto', 'Produce\nTurbina', 'Consume\nCompresor']
    valores = [W_exp_ciclo, -W_comp_ciclo, W_neto, W_turb_produced_cycle, -W_turbo_consumed_by_engine_cycle]
    colores = ['#2ca02c', '#d62728', '#1f77b4', '#9467bd', '#ff7f0e']

    barras = ax3.bar(labels, valores, color=colores, edgecolor='black', width=0.6)
    ax3.axhline(0, color='black', linewidth=1.5)

    # Calculamos una distancia dinámica basada en el valor más grande del chart
    max_val_chart = max([abs(v) for v in valores]) if any(valores) else 1000

    for bar in barras:
        yval = bar.get_height()

        # Distanciamiento estricto para que los textos nunca toquen el eje cero
        if yval >= 0:
            y_pos = yval + (max_val_chart * 0.08)
        else:
            y_pos = yval - (max_val_chart * 0.08)

        ax3.text(bar.get_x() + bar.get_width()/2, y_pos, f"{int(yval)} J",
                 ha='center', va='center', fontweight='bold', color='black', fontsize=9)

    ax3.set_ylabel("Energía Transmitida [Joules/Ciclo]")
    # Ajustar dinámicamente los límites para asegurar que quepan las etiquetas
    ax3.set_ylim(min(valores) - (max_val_chart * 0.15), max(valores) + (max_val_chart * 0.15))
    ax3.grid(axis='y', linestyle=':', alpha=0.7)

    plt.tight_layout(rect=[0, 0, 1, 0.90])
    plt.show()

# ==============================================================================
# EJECUCIÓN DE WIDGETS INTERACTIVOS
# ==============================================================================
estilo_texto = {'description_width': 'initial'}
tamano_slider = widgets.Layout(width='400px')

widgets.interact(simulador_interactivo_geometric,
                 beta2_c_deg=widgets.FloatSlider(min=30.0, max=70.0, step=1.0, value=45.0,
                                                description='Comp. Outlet Blade β2 [°]:', style=estilo_texto, layout=tamano_slider),
                 D2_c_mm=widgets.FloatSlider(min=60.0, max=130.0, step=1.0, value=90.0,
                                           description='Comp. Outlet Diameter D2 [mm]:', style=estilo_texto, layout=tamano_slider),
                 b1_c_mm=widgets.FloatSlider(min=10.0, max=40.0, step=0.5, value=25.0,
                                           description='Comp. Inlet Blade Width b1 [mm]:', style=estilo_texto, layout=tamano_slider),
                 b2_c_mm=widgets.FloatSlider(min=2.0, max=15.0, step=0.1, value=5.0,
                                           description='Comp. Outlet Blade Width b2 [mm]:', style=estilo_texto, layout=tamano_slider),
                 D1_t_mm=widgets.FloatSlider(min=70.0, max=100.0, step=1.0, value=85.0,
                                           description='Turb. Inlet Diameter D1 [mm]:', style=estilo_texto, layout=tamano_slider),
                 D2_t_mm=widgets.FloatSlider(min=50.0, max=80.0, step=1.0, value=65.0,
                                           description='Turb. Outlet Diameter D2 [mm]:', style=estilo_texto, layout=tamano_slider),
                 alpha1_t_deg=widgets.FloatSlider(min=60.0, max=80.0, step=1.0, value=70.0,
                                                description='Turb. Inlet Nozzle α1 [°]:', style=estilo_texto, layout=tamano_slider),
                 beta2_t_deg=widgets.FloatSlider(min=80.0, max=100.0, step=1.0, value=90.0,
                                                description='Turb. Outlet Blade β2 [°]:', style=estilo_texto, layout=tamano_slider));
```

Como se muestra en la simulación, el comportamiento del sistema modelado depende fuertemente de los parámetros geométricos del turbocompresor, ya que estos modifican directamente los triángulos de velocidad y la transferencia de energía entre el fluido y los rotores. En el modelo desarrollado, las variaciones geométricas afectan las velocidades periféricas $U$, tangenciales $C_u$ y meridionales $C_m$, las cuales determinan el trabajo específico transmitido mediante la ecuación de Euler para turbomáquinas. Debido a este acoplamiento, cualquier cambio en la geometría del compresor o la turbina altera simultáneamente el comportamiento hidráulico, el balance energético y el desempeño termodinámico del motor.

En el compresor, el diámetro de salida $D_{2,c}$ es uno de los parámetros más importantes. Al incrementarlo, aumenta la velocidad periférica y el trabajo transmitido al flujo, generando mayores niveles de presión de admisión. Esto permite introducir más masa de aire al motor y aumentar el trabajo neto del ciclo Miller. Sin embargo, también se incrementa la temperatura del aire comprimido, el consumo energético del compresor y la inercia rotacional, lo cual puede generar mayor turbo lag. De forma similar, el ancho de salida $b_{2,c}$ modifica el área de paso del flujo, tener valores mayores reducen la velocidad meridional y eincrementan la componente tangencial, aumentando la capacidad de compresión.

El ángulo de salida del álabe del compresor $β_{2,c}$ controla la cantidad de swirl impartida al flujo. Ángulos mayores producen configuraciones más agresivas, capaces de generar mayores niveles de presión y potencia, pero también incrementan las temperaturas, el consumo energético y el riesgo de inestabilidad aerodinámica. En la turbina, el diámetro de entrada $D_{1,t}$ y el ángulo de entrada $α_{1,t}$ afectan directamente la capacidad de extracción energética de los gases de escape. Valores mayores permiten recuperar más energía y transmitir mayor potencia al compresor, aunque incrementan la inercia del sistema y pueden reducir la respuesta transitoria del turbo.

Por otra parte, el diámetro y ángulo de salida de la turbina ($D_{2,t}, β_{2,t}$) controlan la energía residual presente en el flujo de escape. Configuraciones que generan demasiado swirl residual reducen la eficiencia de extracción energética, ya que parte de la energía cinética del flujo no se transforma en trabajo útil. Este fenómeno es especialmente importante en sistemas de Fórmula 1, donde la recuperación energética y la eficiencia global son aspectos fundamentales del diseño.

Las simulaciones muestran que geometrías más agresivas producen mayores niveles de boost y potencia, pero también aumentan las temperaturas, el consumo energético y las pérdidas aerodinámicas. Por el contrario, configuraciones más conservadoras ofrecen mejor estabilidad operativa y menor lag, aunque con menor potencia máxima. Esto demuestra que el diseño óptimo de un turbocompresor requiere un equilibrio entre presión de compresión, eficiencia térmica, recuperación energética y respuesta dinámica del sistema para un mejor desempeño global de la máquina.

Debido a ello, resulta un componente adecuado para estudiar fenómenos transitorios, permitiendo analizar cómo diferentes configuraciones afectan la evolución temporal de la temperatura del aire de admisión y, en consecuencia, el comportamiento general del sistema.

## **2. Modelado del intercooler**

Con el fin de analizar el efecto de la gestión térmica sobre el desempeño del motor basado en ciclo Miller, se propone la incorporación de un intercooler entre la salida del compresor y el múltiple de admisión. Este dispositivo tiene como objetivo reducir la temperatura del aire comprimido proveniente del turbocompresor, aumentando su densidad antes de ingresar a los cilindros y mejorando así la eficiencia volumétrica del motor. De acuerdo con Heywood (2018), el enfriamiento de la carga de admisión constituye uno de los mecanismos más efectivos para incrementar la masa de aire atrapada en el cilindro y mejorar el desempeño de motores sobrealimentados.

Desde el punto de vista termodinámico, el intercooler puede modelarse como un intercambiador de calor de flujo cruzado, configuración ampliamente utilizada en aplicaciones automotrices debido a su elevada compacidad y capacidad de transferencia térmica (Incropera et al., 2017). Su comportamiento depende principalmente de la diferencia de temperatura entre el aire comprimido y el medio de enfriamiento, el área disponible para la transferencia de calor, el coeficiente global de transferencia de calor y los caudales másicos de los fluidos involucrados. Bromnick, Pearson y Winterbone (1998) desarrollaron uno de los modelos dinámicos más utilizados para intercoolers automotrices, describiendo su desempeño mediante relaciones entre la efectividad térmica y el número de unidades de transferencia (NTU), metodología ampliamente empleada en simulaciones transitorias de motores turboalimentados.

Para el presente trabajo se adopta una aproximación basada en la efectividad del intercambiador, definida como la razón entre la transferencia de calor real y la transferencia de calor máxima teóricamente posible. Bajo esta formulación, la temperatura de salida del aire puede estimarse mediante:

$T_{sal}=T_{amb}+(T_{ent}-T_{amb})(1-\varepsilon)$


donde ($T_{ent}$) corresponde a la temperatura del aire a la salida del compresor, ($T_{amb}$) representa la temperatura del medio de enfriamiento y ($\varepsilon$) es la efectividad térmica del intercooler. Este parámetro constituye una medida global del desempeño térmico del dispositivo y es ampliamente utilizado en el análisis de intercambiadores compactos (Kays, London y Shah, 1984).

El calor transferido puede calcularse mediante:

$
Q=\dot m C_p (T_{ent}-T_{sal})
$

siendo ($\dot m$) el caudal másico del aire y ($C_p$) su calor específico a presión constante. Esta expresión se obtiene directamente del balance de energía aplicado al flujo de aire a través del intercooler (Incropera et al., 2017).

Con el objetivo de estudiar la respuesta transitoria del sistema, se considera además la capacidad térmica del intercooler mediante un modelo de parámetros concentrados. Bajo esta aproximación, la evolución temporal de la temperatura característica del dispositivo puede describirse mediante un balance de energía de primer orden:

$
mC_p\frac{dT}{dt}=Q_{in}-Q_{out}
$

donde ($m$) representa la masa térmicamente activa asociada al núcleo del intercooler. Este enfoque ha sido ampliamente utilizado para representar la dinámica térmica de intercambiadores compactos en aplicaciones automotrices debido a su bajo costo computacional y buena capacidad para reproducir fenómenos transitorios (Bromnick et al., 1998).

Para la evaluación de diferentes diseños se propone modificar parámetros geométricos y térmicos asociados al intercooler, tales como el área efectiva de transferencia de calor, la efectividad térmica global o el coeficiente global de transferencia de calor. Estas variables están directamente relacionadas con la geometría interna del núcleo, la densidad de aletas y la configuración de flujo del intercambiador. Algunos estudios numéricos recientes muestran que incrementos en el área de intercambio y mejoras en la distribución del flujo pueden aumentar significativamente la efectividad térmica del dispositivo, aunque generalmente a costa de mayores pérdidas de presión y requerimientos de espacio (Aghili et al., 2022).

En motores turboalimentados que emplean estrategias de operación similares al ciclo Miller, el intercooler constituye un componente fundamental debido a que la reducción de la temperatura del aire de admisión permite compensar parcialmente los efectos asociados a la sobrealimentación y maximizar la masa de aire disponible para la combustión. Diversas investigaciones sobre motores Miller turboalimentados han demostrado que la combinación de sobrealimentación e intercooling resulta esencial para alcanzar mejoras simultáneas en eficiencia térmica, potencia específica y consumo de combustible (He et al., 2017; Oh et al., 2021).

Para caracterizar el comportamiento térmico del intercooler y evaluar distintas configuraciones de diseño, se consideran como variables principales de entrada:
- La temperatura del aire a la salida del compresor $(T_{ent})$
- La temperatura del fluido refrigerante o del ambiente $(T_{amb})$
- La temperatura de salida del aire $(T_{sal})$
- El calor específico del aire $(C_p)$
- El caudal másico de aire $(\dot m_{air})$
- El caudal másico del fluido refrigerante $(\dot m_{water})$

Adicionalmente, se incluyen parámetros geométricos asociados al intercambiador, tales como:
- El número de aletas $(N_f)$
- El área transversal de los tubos $(A_t)$
- El área total de transferencia de calor $(A)$
- La longitud característica de los canales de flujo
- La densidad de aletas por unidad de longitud.

Desde el punto de vista de la transferencia de calor, también se consideran:
- El coeficiente convectivo del lado del aire $(h_{air})$
- El coeficiente convectivo del lado del refrigerante $(h_{water})$
- La conductividad térmica del material del intercooler $(k)$
- El coeficiente global de transferencia de calor $(U)$

Estos parámetros determinan directamente la capacidad del intercambiador para remover calor del aire comprimido y, por consiguiente, afectan la efectividad térmica del sistema.

La selección de estas variables permite estudiar el impacto de modificaciones geométricas y operativas sobre el desempeño del intercooler, evaluando indicadores como la temperatura de salida del aire, la potencia térmica transferida, la efectividad térmica y la respuesta transitoria del sistema. A continuación, se presenta el código desarrollado para implementar este modelo y permitir la modificación de dichos parámetros con el fin de analizar su influencia sobre la eficiencia y el comportamiento dinámico del sistema.

```python
# @title
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

# ==============================================================================
# CONSTANTES BASE (MOTOR Y FLUIDOS)
# ==============================================================================
V_total = 1.6e-3              # Cilindrada 1.6L [m^3]
V_d = V_total / 6.0           # Volumen desplazado por cilindro
r_geom = 15.0                 # Relación de compresión geométrica
V_c = V_d / (r_geom - 1.0)    # Volumen de la cámara (Clearance)
r_eff_miller = 11.0           # Compresión efectiva Miller

cv_m, gamma_m = 750.0, 1.3    # Propiedades termodinámicas en combustión
cp_a, gamma_a = 1005.0, 1.4   # Aire
cp_w = 4184.0                 # Agua/Refrigerante
R_a = cp_a * (gamma_a - 1.0) / gamma_a

AFR = 13.7918                 # Relación Aire/Combustible E10
LHV_fuel = 40.97e6            # Poder calorífico [J/kg]
RPM_ENGINE = 10500.0

# ==============================================================================
# FUNCIÓN DEL SIMULADOR INTERACTIVO DEL INTERCOOLER
# ==============================================================================
def simulador_intercooler(
    T_ent_c, P_boost_bar, m_dot_a,
    T_amb_c, m_dot_w,
    Area_IC, h_air, h_water
):
    clear_output(wait=True)

    # 1. CONVERSIONES Y ESTADO INICIAL
    T_ent = T_ent_c + 273.15
    T_amb = T_amb_c + 273.15
    P_boost = P_boost_bar * 1e5

    # 2. CÁLCULO DE TRANSFERENCIA DE CALOR (Método e-NTU estricto)
    C_air = m_dot_a * cp_a
    C_water = m_dot_w * cp_w
    C_min = min(C_air, C_water)
    C_max = max(C_air, C_water)
    C_r = C_min / C_max if C_max > 0 else 0

    # Coeficiente Global U
    if h_air > 0 and h_water > 0:
        U = 1.0 / (1.0/h_air + 1.0/h_water)
    else:
        U = 0.0

    NTU = (U * Area_IC) / C_min if C_min > 0 else 0

    # Efectividad para flujo cruzado
    if NTU > 0:
        epsilon = 1 - np.exp((1/C_r) * (NTU**0.22) * (np.exp(-C_r * (NTU**0.78)) - 1))
        epsilon = max(0.0, min(epsilon, 1.0)) # Matemáticamente delimitado
    else:
        epsilon = 0.0

    # Temperaturas Estacionarias
    Q_max = C_min * (T_ent - T_amb)
    Q_real = epsilon * Q_max

    T_sal_ic = T_ent - (Q_real / C_air) if C_air > 0 else T_ent
    # Seguridad matemática: El aire nunca sale más frío que el refrigerante
    T_sal_ic = max(T_sal_ic, T_amb)

    # 3. IMPACTO EN EL CICLO MILLER (SIN IC vs CON IC)
    # -- Caso 1: SIN Intercooler --
    rho_no_ic = P_boost / (R_a * T_ent)
    m_cycle_a_no_ic = rho_no_ic * (V_c * r_eff_miller)
    m_cycle_f_no_ic = m_cycle_a_no_ic / AFR
    Q_in_no_ic = m_cycle_f_no_ic * LHV_fuel

    T3_no_ic = T_ent + Q_in_no_ic / ((m_cycle_a_no_ic + m_cycle_f_no_ic) * cv_m)
    P2_no_ic = P_boost * (r_eff_miller)**gamma_m
    P3_no_ic = P2_no_ic * (T3_no_ic / (T_ent * (r_eff_miller)**(gamma_m-1)))

    # -- Caso 2: CON Intercooler --
    rho_ic = P_boost / (R_a * T_sal_ic)
    m_cycle_a_ic = rho_ic * (V_c * r_eff_miller)
    m_cycle_f_ic = m_cycle_a_ic / AFR
    Q_in_ic = m_cycle_f_ic * LHV_fuel

    T3_ic = T_sal_ic + Q_in_ic / ((m_cycle_a_ic + m_cycle_f_ic) * cv_m)
    P2_ic = P_boost * (r_eff_miller)**gamma_m
    P3_ic = P2_ic * (T3_ic / (T_sal_ic * (r_eff_miller)**(gamma_m-1)))

    # Cálculos de trabajo P-V
    V1_m, V2_m, V4_m = V_c * r_eff_miller, V_c, V_total/6.0 + V_c

    v_comp = np.linspace(V1_m, V2_m, 100)
    p_comp = P_boost * (V1_m / v_comp)**gamma_m # Misma compresión en ambos casos teóricamente

    v_exp = np.linspace(V2_m, V4_m, 100)
    p_exp_no_ic = P3_no_ic * (V2_m / v_exp)**gamma_m
    p_exp_ic = P3_ic * (V2_m / v_exp)**gamma_m

    W_net_no_ic = np.trapezoid(p_exp_no_ic, v_exp) - np.trapezoid(p_comp[::-1], v_comp[::-1])
    W_net_ic = np.trapezoid(p_exp_ic, v_exp) - np.trapezoid(p_comp[::-1], v_comp[::-1])

    # 4. ANÁLISIS TRANSITORIO (Efecto "Heat Soak" real y estable)
    time = np.linspace(0, 10, 200) # 10 segundos
    tau = 2.0 # Inercia térmica [s]

    # Curva transitoria: de un sobre-enfriamiento inicial a la temperatura estacionaria
    T_sal_dinamica = T_sal_ic + (T_amb - T_sal_ic) * np.exp(-time / tau)

    # ==============================================================================
    # GRÁFICOS
    # ==============================================================================
    fig, axs = plt.subplots(2, 2, figsize=(18, 12), facecolor='white')

    # --- 1. DIAGRAMA P-V COMPARATIVO ---
    ax1 = axs[0,0]
    ax1.set_title("Efecto del Intercooler en el Ciclo Miller P-V", fontweight='bold', fontsize=12)

    ax1.plot(v_comp*1e6, p_comp/1e5, 'k--', label='Compresión Efectiva')
    ax1.plot(v_exp*1e6, p_exp_no_ic/1e5, 'r-', lw=2, label='Expansión (Sin IC)')
    ax1.plot([V2_m*1e6, V2_m*1e6], [p_comp[-1]/1e5, P3_no_ic/1e5], 'r-')

    ax1.plot(v_exp*1e6, p_exp_ic/1e5, 'b-', lw=2, label='Expansión (Con IC)')
    ax1.plot([V2_m*1e6, V2_m*1e6], [p_comp[-1]/1e5, P3_ic/1e5], 'b-')

    ax1.fill_between(v_exp*1e6, p_exp_ic/1e5, p_exp_no_ic/1e5, color='green', alpha=0.15, label='Trabajo Extra Ganado')

    ax1.set_xlabel("Volumen del Cilindro [cm³]")
    ax1.set_ylabel("Presión Absoluta [bar]")
    ax1.grid(True, linestyle=':', alpha=0.6)
    ax1.legend()

    # --- 2. RESPUESTA TRANSITORIA TÉRMICA ("Heat Soak") ---
    ax2 = axs[0,1]
    ax2.set_title("Dinámica del Intercooler (Heat Soak al Acelerar)", fontweight='bold', fontsize=12)

    ax2.plot(time, T_sal_dinamica - 273.15, 'b-', lw=3, label='T Aire Admisión (Dinámica)')
    ax2.axhline(T_sal_ic - 273.15, color='black', linestyle=':', lw=2, label='Límite Estacionario')

    ax2.fill_between(time, T_sal_dinamica - 273.15, T_sal_ic - 273.15, color='cyan', alpha=0.3, label='Ventaja térmica inercial')

    # Escala Dinámica Inteligente ajustada
    margen_y = max(5, (T_sal_ic - 273.15 - T_amb_c) * 0.5)
    ax2.set_ylim(T_amb_c - 2, (T_sal_ic - 273.15) + margen_y)

    ax2.set_xlabel("Tiempo tras acelerador a fondo [s]")
    ax2.set_ylabel("Temperatura [°C]")
    ax2.grid(True, linestyle=':', alpha=0.6)
    ax2.legend(loc='lower right')

    # --- 3. RENDIMIENTO DEL INTERCAMBIADOR (BARRAS) ---
    ax3 = axs[1,0]
    ax3.set_title(f"Métricas Térmicas ($\epsilon$ = {epsilon*100:.1f}%, U = {U:.0f} W/m²K)", fontweight='bold', fontsize=12)

    labels_ic = ['Densidad\nSin IC', 'Densidad\nCon IC', 'T Salida\nSin IC', 'T Salida\nCon IC']
    vals_ic = [rho_no_ic, rho_ic, T_ent_c, T_sal_ic - 273.15]
    colors_ic = ['#ff9896', '#1f77b4', '#d62728', '#aec7e8']

    barras1 = ax3.bar(labels_ic, vals_ic, color=colors_ic, edgecolor='black', width=0.5)

    ax3.set_ylabel("Magnitud (kg/m³ o °C)")
    for i, bar in enumerate(barras1):
        yval = bar.get_height()
        unit = "kg/m³" if i < 2 else "°C"
        ax3.text(bar.get_x() + bar.get_width()/2, yval + (max(vals_ic)*0.03), f"{yval:.2f} {unit}",
                 ha='center', va='bottom', fontweight='bold', fontsize=10)

    ax3.set_ylim(0, max(vals_ic) * 1.15)
    ax3.grid(axis='y', linestyle=':', alpha=0.6)

    # --- 4. BALANCE DE TRABAJO DEL MOTOR ---
    ax4 = axs[1,1]
    aumento_trabajo = ((W_net_ic/W_net_no_ic)-1)*100 if W_net_no_ic > 0 else 0
    ax4.set_title(f"Aumento Teórico de Potencia Neta: +{aumento_trabajo:.1f}%", fontweight='bold', fontsize=12)

    labels_w = ['Trabajo Neto\nSin IC', 'Trabajo Neto\nCon IC']
    vals_w = [W_net_no_ic, W_net_ic]

    barras2 = ax4.bar(labels_w, vals_w, color=['#d62728', '#1f77b4'], edgecolor='black', width=0.4)
    ax4.set_ylabel("Energía por Ciclo [Joules]")

    for bar in barras2:
        yval = bar.get_height()
        ax4.text(bar.get_x() + bar.get_width()/2, yval + (max(vals_w)*0.03), f"{int(yval)} J",
                 ha='center', va='bottom', fontweight='bold', fontsize=11)

    ax4.set_ylim(0, max(vals_w) * 1.2)
    ax4.grid(axis='y', linestyle=':', alpha=0.6)

    plt.tight_layout()
    plt.show()

# ==============================================================================
# WIDGETS INTERACTIVOS (Con Defaults Realistas)
# ==============================================================================
style = {'description_width': '220px'}
layout = widgets.Layout(width='600px')

widgets.interact(simulador_intercooler,
    T_ent_c=widgets.FloatSlider(min=100.0, max=250.0, step=5.0, value=160.0,
                                description='Temp Aire Entrada (del Turbo) [°C]:', style=style, layout=layout),
    P_boost_bar=widgets.FloatSlider(min=2.0, max=6.0, step=0.1, value=4.5,
                                description='Presión de Sobrealimentación [bar]:', style=style, layout=layout),
    m_dot_a=widgets.FloatSlider(min=0.05, max=0.25, step=0.01, value=0.12,
                                description='Caudal Másico de Aire [kg/s]:', style=style, layout=layout),
    T_amb_c=widgets.FloatSlider(min=10.0, max=60.0, step=2.0, value=25.0,
                                description='Temp Refrigerante / Ambiente [°C]:', style=style, layout=layout),
    m_dot_w=widgets.FloatSlider(min=0.1, max=1.0, step=0.05, value=0.5,
                                description='Caudal de Agua del IC [kg/s]:', style=style, layout=layout),
    Area_IC=widgets.FloatSlider(min=0.5, max=5.0, step=0.1, value=1.2,
                                description='Área de Transferencia de Calor [m²]:', style=style, layout=layout),
    h_air=widgets.FloatSlider(min=50.0, max=300.0, step=10.0, value=100.0,
                                description='Coef. Convección Aire [W/m²K]:', style=style, layout=layout),
    h_water=widgets.FloatSlider(min=1000.0, max=5000.0, step=100.0, value=3000.0,
                                description='Coef. Convección Agua [W/m²K]:', style=style, layout=layout)
);
```

Las simulaciones muestran que la incorporación del intercooler produce una disminución significativa de la temperatura del aire de admisión, pasando de aproximadamente 160 °C a 78 °C para las condiciones analizadas. Como consecuencia, la densidad del aire aumenta de 3.62 kg/m³ a 4.46 kg/m³, permitiendo una mayor masa de aire disponible para la combustión dentro del cilindro.

Este incremento en la densidad se refleja directamente en el diagrama P-V, donde la curva de expansión correspondiente al caso con intercooler alcanza presiones más elevadas que el caso sin enfriamiento. El área adicional encerrada por la curva representa un aumento del trabajo neto desarrollado por ciclo, estimado en aproximadamente un 26.1 % para los parámetros seleccionados.

La respuesta transitoria del intercooler evidencia además la presencia de inercia térmica. Inmediatamente después de una aceleración, el sistema presenta una capacidad temporal de enfriamiento superior a la condición estacionaria, fenómeno conocido como heat soak. Con el transcurso del tiempo, la temperatura de admisión converge gradualmente hacia su valor de equilibrio, determinado por las condiciones de transferencia de calor del intercambiador.

Los resultados también muestran una fuerte dependencia respecto a los parámetros de diseño y operación. Un aumento en el área de transferencia de calor $(A)$, en los coeficientes convectivos del aire y del refrigerante $(h_{air}$) y $(h_{water})$, o en el caudal másico del refrigerante $(\dot m_{water})$, incrementa el coeficiente global de transferencia de calor y la efectividad térmica del intercooler, reduciendo la temperatura de salida del aire. De manera similar, temperaturas de entrada más elevadas provenientes del turbocompresor generan una mayor carga térmica sobre el sistema, mientras que mayores presiones de sobrealimentación incrementan la densidad del aire y amplifican el beneficio potencial asociado al enfriamiento.
