## Objetivo General

Desarrollar y analizar los principales equipos hidráulicos y termodinámicos que conforman una planta termoeléctrica de cogeneración, integrando máquinas rotacionales, dispositivos de intercambio de calor y el ciclo Rankine, con el fin de evaluar su comportamiento termofluídico, hidráulico, energético y su influencia sobre el desempeño global del sistema.

## Objetivos Especificos

1. Calcular los principales parámetros hidráulicos y energéticos de las máquinas rotacionales, considerando las condiciones de operación del fluido, la relación de compresión o expansión y los datos geométricos asociados al diseño de los álabes.


2. Implementar un dispositivo termodinámico asociado al ciclo Rankine, como un intercambiador de calor, evaluando su respuesta transitoria y su efecto sobre la transferencia de energía dentro del sistema térmico.

3. Acoplar las máquinas hidráulicas y los dispositivos termodinámicos al ciclo completo de la planta de cogeneración, analizando cómo modifican las condiciones de operación, la producción de trabajo, el consumo energético y la eficiencia global del sistema.

4. Comparar diferentes configuraciones de operación y diseño, evaluando el efecto de variar las condiciones de intercambio de calor, el consumo energético, la producción de trabajo, la eficiencia térmica y otras variables relevantes de la planta.

5. Concluir sobre la viabilidad técnica de las intervenciones tecnológicas propuestas, estableciendo los ajustes necesarios para la implementación de los equipos hidráulicos y termodinámicos dentro del sistema de cogeneración.

#**1. Ciclo Termodinámico**

## Descripción general de la planta de cogeneración y ciclo Rankine

En la planta de cogeneración propuesta, el proceso inicia con el ingreso de gases calientes a la turbina del ciclo Brayton. En esta etapa, la energía térmica del flujo de gas se transforma en trabajo mecánico rotacional mediante la expansión en la turbina. Dicho trabajo es transmitido a un generador eléctrico, donde se convierte en energía eléctrica útil. De esta manera, el ciclo Brayton constituye la primera etapa de generación de potencia dentro del sistema.


CICLO TERMODINÁMICO.jpg


Después de la expansión en la turbina Brayton, los gases de salida aún conservan una cantidad significativa de energía térmica. En lugar de ser descargados directamente al ambiente, estos gases se conducen hacia un intercambiador de calor, donde transfieren parte de su energía al fluido de trabajo del ciclo Rankine. Esta integración permite aprovechar el calor residual del ciclo Brayton para calentar, evaporar o aumentar la temperatura del fluido de trabajo del ciclo de vapor, mejorando así el aprovechamiento energético global de la planta.

El ciclo Rankine opera de manera convencional, utilizando el calor recuperado para generar vapor y producir trabajo adicional en una turbina de vapor. Posteriormente, el fluido de trabajo pasa por los procesos de condensación y bombeo para completar el ciclo. Gracias a esta configuración, la planta puede generar energía eléctrica tanto a partir de la turbina Brayton como de la turbina Rankine, aumentando la eficiencia global del sistema. A partir de esta descripción, el trabajo se enfocará en el modelado individual de los principales equipos de la planta, incluyendo las turbinas, el generador eléctrico, los intercambiadores de calor y los componentes auxiliares del ciclo Rankine.


# **1.1 Turbina de Gas**

## 1.1.1. Turbina de Gas Industrial SGT-800

La turbina de gas industrial Siemens SGT-800 es una máquina térmica rotacional empleada en generación eléctrica en ciclo simple, ciclo combinado y aplicaciones de cogeneración. Esta máquina alcanza una potencia eléctrica nominal de $62.5\ \text{MW}$, opera a una velocidad de eje de $6600\ \text{rpm}$ y presenta una relación de presión de $21.1:1$. Su arquitectura está compuesta principalmente por una admisión de aire, un compresor axial de 15 etapas, una cámara de combustión, una turbina axial de 3 etapas, el eje principal y el sistema de escape.

Durante su operación, el aire ambiente ingresa por la admisión y pasa hacia el compresor axial, donde su presión y temperatura aumentan progresivamente antes de entrar a la cámara de combustión. En esta zona, el aire comprimido se mezcla con el combustible, liberando energía térmica y generando gases a alta temperatura. Posteriormente, estos gases se expanden en la turbina axial, produciendo torque sobre el eje. Parte de la potencia generada se emplea para accionar el compresor, mientras que la potencia neta restante se entrega al generador eléctrico. Finalmente, los gases de escape salen con un flujo másico aproximado de $135.5\ \text{kg/s}$ y una temperatura cercana a $596^\circ\text{C}$, lo cual permite su aprovechamiento térmico dentro del sistema de cogeneración.

## 1.1.2. Etapas de la Turbina de Gas

### **Admisión**

La admisión constituye la entrada principal del fluido de trabajo al sistema, que en este caso corresponde al aire ambiente. Su función es acondicionar y conducir el flujo hacia el compresor axial con condiciones adecuadas de presión, temperatura, limpieza y dirección. Para ello, esta sección puede incorporar filtros, ductos, silenciadores, elementos de acondicionamiento aerodinámico y álabes guía de entrada, conocidos como IGV (*Inlet Guide Vanes*) o VIGV cuando son variables. Estos últimos permiten modificar la orientación del flujo y controlar el ángulo de incidencia del aire sobre las primeras etapas del compresor. De esta manera, la admisión reduce impurezas, pérdidas de presión y perturbaciones aerodinámicas, favoreciendo una operación estable y eficiente del compresor.

### **Compresión**

La compresión es la etapa encargada de elevar la presión del aire antes de su ingreso a la cámara de combustión. En la Siemens SGT-800, este proceso se realiza mediante un compresor axial de 15 etapas, compuesto por filas sucesivas de álabes móviles y fijos. Los álabes del rotor transfieren energía mecánica al flujo, aumentando su energía total, mientras que los álabes del estator redirigen el aire y contribuyen a la recuperación de presión. Como resultado, el aire incrementa progresivamente su presión y temperatura, aumentando la energía disponible del flujo antes de la adición de calor. Debido a que el compresor consume una fracción importante de la potencia producida por la turbina, su desempeño influye directamente en la potencia neta disponible en el eje.

### **Cámara de combustión**

La cámara de combustión mezcla el aire comprimido con el combustible, que en la Siemens SGT-800 puede ser gas natural o Diésel No. 2. En esta sección, la energía química del combustible se transforma en energía térmica mediante una reacción de oxidación, elevando la temperatura del flujo y generando los gases calientes que ingresan a la turbina. Para proteger los componentes aguas abajo, especialmente las primeras etapas de expansión, este proceso debe mantener pérdidas de presión controladas y una distribución térmica adecuada. Además, la turbina opera con mezcla pobre y excesos de aire que pueden acercarse al 300% respecto al AFR estequiométrico, con el fin de limitar la temperatura máxima de los gases y favorecer la dilución del flujo antes de la expansión.

### **Turbina**

La turbina axial recibe los gases calientes provenientes de la cámara de combustión y convierte parte de su energía térmica y cinética en trabajo mecánico rotacional. En la Siemens SGT-800, esta sección está compuesta por 3 etapas axiales, donde el flujo se expande progresivamente a través de álabes fijos y móviles. Los álabes fijos orientan y aceleran los gases hacia el rotor, mientras que los álabes móviles extraen energía del flujo y generan torque sobre el eje. Durante este proceso, la presión y temperatura de los gases disminuyen, y la potencia producida se utiliza tanto para accionar el compresor axial como para entregar potencia neta al generador eléctrico.

### **Escape**

El escape corresponde a la sección final del flujo de gases después de la expansión en la turbina. En esta etapa, los gases salen con menor presión y temperatura que a la entrada de la turbina, pero aún conservan una cantidad importante de energía térmica debido a su alta temperatura de descarga. En una planta de cogeneración, esta energía residual puede aprovecharse mediante una caldera de recuperación o intercambiador de calor para producir vapor, calentar un fluido de proceso o alimentar un ciclo Rankine. Por ello, el escape no solo representa la salida del ciclo Brayton, sino también el punto de integración térmica con el sistema de recuperación de calor.

## 1.1.3. Modelo Matemático de la Turbina de Gas

### **Admisión**

Se empleó un modelo 0D simplificado, adecuado para una turbina de gas fija en tierra. En este caso, el aire ambiente constituye el fluido de trabajo que ingresa al sistema y se conduce hacia el compresor axial. Debido a que no se consideran efectos asociados al movimiento de la máquina, como velocidad de avance o número de Mach, las condiciones de entrada dependen principalmente de la presión ambiente, la temperatura ambiente y las pérdidas del sistema de admisión. Estas pérdidas se representan mediante una eficiencia de entrada $\eta_{\text{inlet}}$, aplicada directamente sobre la presión ambiente:

$$
p_1=\eta_{\text{inlet}}p_{\text{amb}}
\tag{1}
$$

$$
T_1=T_{\text{amb}}
\tag{2}
$$

donde $p_1$ y $T_1$ corresponden a la presión y temperatura del aire a la entrada del compresor. Para definir los flujos de entrada, se parte del flujo másico total de gases de escape reportado para la turbina, el cual se asume como la suma del flujo másico de aire y del flujo másico de combustible:

$$
\dot{m}_{\text{gas}}=\dot{m}_{\text{air}}+\dot{m}_{\text{fuel}}
\tag{3}
$$

La relación aire-combustible global se define como:

$$
AFR=\frac{\dot{m}_{\text{air}}}{\dot{m}_{\text{fuel}}}
\tag{4}
$$

por lo tanto, el flujo másico de combustible y el flujo másico de aire se calculan como:

$$
\dot{m}_{\text{fuel}}=\frac{\dot{m}_{\text{gas}}}{AFR+1}
\tag{5}
$$

$$
\dot{m}_{\text{air}}=AFR\,\dot{m}_{\text{fuel}}
\tag{6}
$$

Finalmente, considerando una densidad constante del aire, el caudal volumétrico admitido hacia el compresor se calcula mediante:

$$
\dot{V}_{\text{air}}=\frac{\dot{m}_{\text{air}}}{\rho_{\text{air}}}
\tag{7}
$$

### **Compresión axial**

La compresión axial se modeló desde el estado de entrada al compresor, definido por $(p_1,T_1)$, hasta el estado de salida, definido por $(p_2,T_2)$, a partir de una etapa genérica rotor-estator. El rotor transfiere energía mecánica al aire, modificando su velocidad absoluta y aumentando su entalpía, mientras que el estator redirige el flujo y favorece la recuperación de presión antes de la siguiente etapa. Para describir este proceso se emplean triángulos de velocidad, donde $C$ es la velocidad absoluta del flujo, $W$ la velocidad relativa respecto al rotor y $U$ la velocidad periférica del álabe.

En la entrada del rotor, el flujo llega con velocidad absoluta $C_1$, mientras los álabes se mueven con velocidad periférica $U$. Desde el punto de vista del rotor, este movimiento relativo define la velocidad $W_1$, cuya dirección se ajusta al ángulo de entrada $\beta_1$. A la salida, la geometría del álabe guía el flujo con una nueva velocidad relativa $W_2$, asociada al ángulo de salida $\beta_2$, dando lugar a la velocidad absoluta de escape $C_2$. La diferencia entre las componentes tangenciales de $C_1$ y $C_2$ determina el intercambio de energía según la ecuación de Euler.

El modelo considera flujo unidimensional promedio, propiedades constantes del aire, velocidad axial uniforme y análisis en el radio medio de cada etapa. Además, se desprecia la variación radial detallada del flujo, por lo que cada etapa se representa de forma simplificada mediante sus radios de punta y cubo, sus ángulos relativos de entrada y salida, y el caudal volumétrico de aire.


<div align ="left">
 
 </div>

 *Figura 1.3. Triángulos de velocidad en una etapa de compresión axial.*


Para una etapa genérica $i$, el área anular de paso del flujo se calcula con el radio de punta $r_{\text{tip},i}$ y el radio de cubo $r_{\text{hub},i}$:

$$
A_i=\pi\left(r_{\text{tip},i}^{2}-r_{\text{hub},i}^{2}\right)
\tag{8}
$$

A partir de esta geometría se define el radio medio de la etapa, sobre el cual se evalúan las velocidades características del rotor:

$$
r_{m,i}=\frac{r_{\text{tip},i}+r_{\text{hub},i}}{2}
\tag{9}
$$

Con el área anular y el caudal volumétrico de aire admitido, la velocidad axial promedio en la etapa se obtiene como:

$$
V_{a,i}=\frac{\dot{V}_{\text{air}}}{A_i}
\tag{10}
$$

La velocidad periférica del álabe en el radio medio se calcula mediante:

$$
U_i=\omega r_{m,i}
\tag{11}
$$

La velocidad angular del eje se obtiene a partir de la velocidad de rotación de la turbina:

$$
\omega=\frac{2\pi N}{60}
\tag{12}
$$

donde $N$ es la velocidad de giro en rpm y $\omega$ queda expresada en rad/s. Una vez definidas las velocidades principales, se emplean los triángulos de velocidad del rotor. A partir de los ángulos relativos de entrada y salida, $\beta_{1,i}$ y $\beta_{2,i}$, se estiman las componentes tangenciales de la velocidad absoluta:

$$
C_{u1,i}=U_i - V_{a,i}\tan(\beta_{1,i})
\tag{13}
$$

$$
C_{u2,i}=U_i - V_{a,i}\tan(\beta_{2,i})
\tag{14}
$$

La variación de estas componentes tangenciales permite aplicar la ecuación de Euler para turbomáquinas. Así, el trabajo específico entregado al aire en la etapa se calcula como:

$$
\Delta h_i=U_i\left(C_{u2,i}-C_{u1,i}\right)
\tag{15}
$$

Este incremento de entalpía se relaciona con el aumento de temperatura del aire durante la compresión:

$$
\Delta T_i=\frac{\Delta h_i}{c_{p,\text{air}}}
\tag{16}
$$

Conocido el trabajo específico, la potencia requerida por cada etapa se obtiene multiplicando por el flujo másico de aire:

$$
\dot{W}_{c,i}=\dot{m}_{\text{air}}\Delta h_i
\tag{17}
$$

y el torque correspondiente sobre el eje se calcula como:

$$
\tau_{c,i}=\frac{\dot{W}_{c,i}}{\omega}
\tag{18}
$$

Para actualizar el estado termodinámico del aire, se aproxima la relación de presión de cada etapa mediante el incremento de temperatura y la eficiencia de compresión:

$$
\Pi_{c,i}=
\left(
1+\eta_{c,i}\frac{\Delta T_i}{T_{\text{in},i}}
\right)^{\frac{\gamma_{\text{air}}}{\gamma_{\text{air}}-1}}
\tag{19}
$$

De esta forma, la presión y temperatura a la salida de cada etapa quedan definidas como:

$$
p_{\text{out},i}=p_{\text{in},i}\Pi_{c,i}
\tag{20}
$$

$$
T_{\text{out},i}=T_{\text{in},i}+\Delta T_i
\tag{21}
$$

El procedimiento se aplica de forma secuencial a las 15 etapas del compresor, usando la salida de una etapa como entrada de la siguiente. Al finalizar el proceso, la relación de presión total del compresor se obtiene como:

$$
\Pi_c=\frac{p_2}{p_1}
\tag{22}
$$

mientras que la potencia total consumida por el compresor corresponde a la suma de las potencias requeridas en todas las etapas:

$$
\dot{W}_{c,\text{total}}=\sum_{i=1}^{15}\dot{W}_{c,i}
\tag{23}
$$

### **Cámara de combustión**

La cámara de combustión se modeló desde el estado de salida del compresor, definido por $(p_2,T_2)$, hasta el estado de entrada a la turbina, definido por $(p_3,T_3)$. En este modelo no se resuelve la cinética química de la combustión ni la dinámica transitoria de la cámara, sino que se aplica un balance global de energía en régimen estacionario. El combustible se caracteriza mediante su flujo másico y su poder calorífico inferior, mientras que la eficiencia de combustión representa la fracción de energía química efectivamente transferida al flujo de gases.

Primero, el poder calorífico inferior del combustible se expresa en unidades de energía por unidad de masa:

$$
LHV_{J/kg}=LHV_{MJ/kg}\times 10^6
\tag{24}
$$

La tasa de calor agregada por combustión se calcula como:

$$
\dot{Q}_{\text{added}}=\dot{m}_{\text{fuel}}LHV_{J/kg}\eta_{\text{combustion}}
\tag{25}
$$

donde $\dot{m}_{\text{fuel}}$ es el flujo másico de combustible y $\eta_{\text{combustion}}$ es la eficiencia de combustión. Posteriormente, se realiza el balance de energía entre el aire comprimido que entra desde el compresor y la energía térmica liberada por el combustible. Así, la temperatura de salida de la cámara se obtiene mediante:

$$
T_3=
\frac{
\dot{m}_{\text{air}}c_{p,\text{air}}T_2+\dot{Q}_{\text{added}}
}{
\dot{m}_{\text{gas}}c_{p,\text{gas}}
}
\tag{26}
$$

En esta expresión, $\dot{m}_{\text{gas}}$ corresponde al flujo másico total que ingresa a la turbina, compuesto por el aire comprimido y el combustible agregado. Para la presión de salida, se considera una caída de presión simplificada dentro de la cámara de combustión mediante una eficiencia de presión:

$$
p_3=\eta_{\text{combustor,pressure}}p_2
\tag{27}
$$

### **Expansión en la turbina axial**

La expansión en la turbina se modeló desde el estado de salida de la cámara de combustión, definido por $(p_3,T_3)$, hasta el estado de escape, definido por $(p_4,T_4)$. El análisis es análogo al del compresor axial, ya que también se emplean triángulos de velocidad y la ecuación de Euler para turbomáquinas. Sin embargo, la configuración física cambia: en la turbina el flujo pasa primero por el estator y luego por el rotor. El estator acelera y orienta los gases calientes hacia los álabes móviles, mientras que el rotor extrae energía del flujo y la convierte en torque sobre el eje.

En los triángulos de velocidad de la turbina, el flujo entra al rotor con velocidad absoluta $C_2$ y velocidad relativa $W_2$, asociada al ángulo relativo de entrada $\beta_{2,j}$. A la salida, la geometría del álabe guía el flujo relativo con dirección $W_3$, asociada al ángulo $\beta_{3,j}$, dando lugar a la velocidad absoluta de salida $C_3$. A diferencia del compresor, en esta etapa el fluido entrega energía al eje, por lo que la presión y la temperatura disminuyen durante la expansión.


<div style="text-align: left;">

</div>

*Figura 1.4. Triángulos de velocidad en una etapa de turbina axial.*


Para cada etapa genérica $j$ de la turbina se utilizan las mismas relaciones geométricas definidas previamente para el compresor: área anular, radio medio, velocidad axial y velocidad periférica. Por tanto, el análisis parte de $A_j$, $r_{m,j}$, $V_{a,j}$ y $U_j$, calculados de forma equivalente, pero aplicados ahora al flujo de gases de combustión. Con los ángulos relativos de entrada y salida del rotor, las componentes tangenciales de la velocidad absoluta se calculan como:

$$
C_{u,\text{in},j}=U_j-V_{a,j}\tan(\beta_{2,j})
\tag{28}
$$

$$
C_{u,\text{out},j}=U_j-V_{a,j}\tan(\beta_{3,j})
\tag{29}
$$

La diferencia entre estas componentes representa el cambio de momento angular del flujo a través del rotor. Por ello, el torque generado por la etapa se calcula mediante:

$$
\tau_{t,j}=r_{m,j}\dot{m}_{\text{gas}}
\left(C_{u,\text{in},j}-C_{u,\text{out},j}\right)
\tag{30}
$$

A partir del torque, la potencia euleriana entregada al eje es:

$$
\dot{W}_{t,j}^{\text{Euler}}=\omega\tau_{t,j}
\tag{31}
$$

Considerando la eficiencia de la etapa de turbina, la potencia útil extraída del flujo se define como:

$$
\dot{W}_{t,j}=\eta_{t,j}\dot{W}_{t,j}^{\text{Euler}}
\tag{32}
$$

La caída de temperatura asociada a la expansión se obtiene mediante el balance de energía sobre los gases:

$$
\Delta T_j=
\frac{\dot{W}_{t,j}}{\dot{m}_{\text{gas}}c_{p,\text{gas}}}
\tag{33}
$$

Por tanto, la temperatura de salida de cada etapa se actualiza como:

$$
T_{\text{out},j}=T_{\text{in},j}-\Delta T_j
\tag{34}
$$

Para la presión, se emplea una relación termodinámica simplificada en función de la caída de temperatura de la etapa:

$$
\Pi_{t,j}=
\left(
1-\frac{\Delta T_j}{T_{\text{in},j}}
\right)^{\frac{\gamma_{\text{gas}}}{\gamma_{\text{gas}}-1}}
\tag{35}
$$

$$
p_{\text{out},j}=p_{\text{in},j}\Pi_{t,j}
\tag{36}
$$

El procedimiento se aplica secuencialmente a las 3 etapas de la turbina axial. Finalmente, la potencia total útil producida por la turbina se obtiene como:

$$
\dot{W}_{t,\text{total}}=\sum_{j=1}^{3}\dot{W}_{t,j}
\tag{37}
$$

y la relación de presión global de la turbina queda definida por:

$$
\Pi_t=\frac{p_4}{p_3}
\tag{38}
$$

### **Eje rotativo y balance de potencia**

El eje principal se modeló bajo una condición de operación estacionaria, considerando una velocidad angular fija para todo el conjunto rotativo. Bajo esta condición, la compatibilidad mecánica entre compresor, turbina y generador se evaluó mediante un balance de potencia en el eje. Así, la potencia producida por la turbina axial debe compensar la potencia consumida por el compresor axial, y la diferencia corresponde a la potencia neta disponible para generación eléctrica.

A partir de las potencias calculadas en las etapas de compresión y expansión, la potencia neta disponible en el eje se obtuvo como:

$$
\dot{W}_{\text{shaft}}=
\dot{W}_{t,\text{total}}-\dot{W}_{c,\text{total}}
\tag{39}
$$

donde $\dot{W}_{t,\text{total}}$ corresponde a la potencia útil producida por la turbina axial y $\dot{W}_{c,\text{total}}$ a la potencia consumida por el compresor axial. Posteriormente, la potencia eléctrica entregada por el generador se calculó aplicando una eficiencia de conversión electromecánica:

$$
\dot{W}_{\text{elec}}=
\eta_{\text{gen}}\dot{W}_{\text{shaft}}
\tag{40}
$$

## 1.1.4. Simulación de la Turbina de Gas

Con base en el desarrollo matemático anterior, se implementó un modelo estacionario de la turbina de gas Siemens SGT-800. El código integra las ecuaciones de admisión, compresión axial, cámara de combustión, expansión en turbina y escape. A partir de ello, el modelo permite estimar los estados termodinámicos principales, la potencia consumida por el compresor, la potencia producida por la turbina, la potencia neta de eje y la potencia eléctrica entregada al generador.

Debido a que no se dispone de información abierta sobre la geometría interna detallada del compresor y la turbina, los radios y ángulos de los álabes se definieron como una geometría equivalente calibrada. Esta calibración se realizó para reproducir los principales datos de operación de la configuración de $62.5\ \text{MW}$ de la SGT-800, incluyendo la relación de presión de $21.1:1$, la temperatura de escape de $596^\circ\text{C}$, el flujo másico de gases de $135.5\ \text{kg/s}$ y la velocidad de eje de $6600\ \text{rpm}$. Por tanto, la geometría empleada no representa necesariamente las dimensiones reales exactas de la máquina, sino una aproximación funcional coherente con su comportamiento global.

Para simplificar el modelo, las propiedades del aire y de los gases de combustión se consideraron constantes, empleando valores promedio representativos de operación y sin incluir su dependencia con la temperatura o la presión. Como combustible de referencia se seleccionó Diésel No. 2, caracterizado mediante su poder calorífico inferior. Asimismo, se utilizó un AFR global efectivo superior al estequiométrico de este combustible, de acuerdo con la operación típica de turbinas de gas con mezclas pobres y exceso de aire.


```python
# @title
# ============================================================
# MODELO SIMPLIFICADO PARA TURBINA DE GAS
# Siemens SGT-800 - Modelo 0D
# ============================================================

import math
import pandas as pd
from IPython.display import display

# ============================================================
# PASO 1: ENTRADAS, ADMISIÓN Y CÁLCULO AIRE-COMBUSTIBLE
# ============================================================

# -----------------------------
# Entradas principales
# -----------------------------

p_amb = 101325.0              # [Pa]
T_amb = 288.15                # [K]

gamma_air = 1.4               # [-]
rho_air = 1.225               # [kg/m3]
cp_air = 1005.0               # [J/(kg K)]

gamma_gas = 1.33              # [-]
cp_gas = 1100.0               # [J/(kg K)]

fuel_type = "Diesel No. 2"
LHV_fuel_MJ_kg = 42.6         # [MJ/kg]
rho_fuel_kg_L = 0.846         # [kg/L]

m_exhaust_kg_s = 135.5        # [kg/s]
AFR_mass = 35.30352783        # [kg aire/kg combustible]
shaft_rpm = 6600.0            # [rpm]

eta_inlet = 0.98              # [-]
eta_combustion = 0.98         # [-]
eta_combustor_pressure = 0.98 # [-]
eta_compressor_stage = 0.88   # [-]
eta_turbine_stage = 0.88      # [-]
eta_generator = 0.97          # [-]

# -----------------------------
# Cálculos preliminares
# -----------------------------

omega = shaft_rpm * 2.0 * math.pi / 60.0     # [rad/s]

p1 = eta_inlet * p_amb                       # [Pa]
T1 = T_amb                                   # [K]

m_fuel_kg_s = m_exhaust_kg_s / (AFR_mass + 1.0)   # [kg/s]
m_air_kg_s = AFR_mass * m_fuel_kg_s               # [kg/s]
m_gas_kg_s = m_air_kg_s + m_fuel_kg_s             # [kg/s]

Q_fuel_W = m_fuel_kg_s * LHV_fuel_MJ_kg * 1e6     # [W]
Q_fuel_kW = Q_fuel_W / 1000.0                     # [kW]
Q_fuel_MW = Q_fuel_W / 1e6                        # [MW]

dotV_air_m3_s = m_air_kg_s / rho_air              # [m3/s]

dotV_fuel_L_s = m_fuel_kg_s / rho_fuel_kg_L       # [L/s]
dotV_fuel_L_h = dotV_fuel_L_s * 3600.0            # [L/h]
dotV_fuel_m3_h = dotV_fuel_L_h / 1000.0           # [m3/h]


# ============================================================
# PASO 2: COMPRESOR AXIAL MULTIETAPA
# ============================================================

compressor_geometry = [
    {"stage": 1,  "r_tip_m": 0.720000, "r_hub_m": 0.380000, "beta1_deg": 58.750000, "beta2_deg": 41.856912},
    {"stage": 2,  "r_tip_m": 0.718571, "r_hub_m": 0.387143, "beta1_deg": 58.928571, "beta2_deg": 42.214055},
    {"stage": 3,  "r_tip_m": 0.717143, "r_hub_m": 0.394286, "beta1_deg": 59.107143, "beta2_deg": 42.571198},
    {"stage": 4,  "r_tip_m": 0.715714, "r_hub_m": 0.401429, "beta1_deg": 59.285714, "beta2_deg": 42.928341},
    {"stage": 5,  "r_tip_m": 0.714286, "r_hub_m": 0.408571, "beta1_deg": 59.464286, "beta2_deg": 43.285483},
    {"stage": 6,  "r_tip_m": 0.712857, "r_hub_m": 0.415714, "beta1_deg": 59.642857, "beta2_deg": 43.642626},
    {"stage": 7,  "r_tip_m": 0.711429, "r_hub_m": 0.422857, "beta1_deg": 59.821429, "beta2_deg": 43.999769},
    {"stage": 8,  "r_tip_m": 0.710000, "r_hub_m": 0.430000, "beta1_deg": 60.000000, "beta2_deg": 44.356912},
    {"stage": 9,  "r_tip_m": 0.708571, "r_hub_m": 0.437143, "beta1_deg": 60.178571, "beta2_deg": 44.714055},
    {"stage": 10, "r_tip_m": 0.707143, "r_hub_m": 0.444286, "beta1_deg": 60.357143, "beta2_deg": 45.071198},
    {"stage": 11, "r_tip_m": 0.705714, "r_hub_m": 0.451429, "beta1_deg": 60.535714, "beta2_deg": 45.428341},
    {"stage": 12, "r_tip_m": 0.704286, "r_hub_m": 0.458571, "beta1_deg": 60.714286, "beta2_deg": 45.785483},
    {"stage": 13, "r_tip_m": 0.702857, "r_hub_m": 0.465714, "beta1_deg": 60.892857, "beta2_deg": 46.142626},
    {"stage": 14, "r_tip_m": 0.701429, "r_hub_m": 0.472857, "beta1_deg": 61.071429, "beta2_deg": 46.499769},
    {"stage": 15, "r_tip_m": 0.700000, "r_hub_m": 0.480000, "beta1_deg": 61.250000, "beta2_deg": 46.856912},
]

def axial_area(r_tip, r_hub):
    return math.pi * (r_tip**2 - r_hub**2)

compressor_results = []

p_comp_in = p1
T_comp_in = T1
W_comp_total_W = 0.0

for stage_data in compressor_geometry:

    stage_number = stage_data["stage"]

    r_tip_m = stage_data["r_tip_m"]
    r_hub_m = stage_data["r_hub_m"]
    beta1_deg = stage_data["beta1_deg"]
    beta2_deg = stage_data["beta2_deg"]

    r_mean_m = 0.5 * (r_tip_m + r_hub_m)
    A_m2 = axial_area(r_tip_m, r_hub_m)

    beta1_rad = math.radians(beta1_deg)
    beta2_rad = math.radians(beta2_deg)

    Va_m_s = dotV_air_m3_s / A_m2
    U_m_s = omega * r_mean_m

    Cu1_m_s = U_m_s - Va_m_s * math.tan(beta1_rad)
    Cu2_m_s = U_m_s - Va_m_s * math.tan(beta2_rad)

    delta_h_J_kg = U_m_s * (Cu2_m_s - Cu1_m_s)
    delta_T_K = delta_h_J_kg / cp_air

    if delta_h_J_kg <= 0.0:
        raise ValueError("Trabajo de compresión no positivo. Revisar beta1 y beta2.")

    W_stage_W = m_air_kg_s * delta_h_J_kg

    pressure_ratio_stage = (
        1.0 + eta_compressor_stage * delta_T_K / T_comp_in
    ) ** (gamma_air / (gamma_air - 1.0))

    p_comp_out = p_comp_in * pressure_ratio_stage
    T_comp_out = T_comp_in + delta_T_K

    torque_stage_Nm = W_stage_W / omega

    compressor_results.append({
        "Etapa": stage_number,
        "r_tip [m]": r_tip_m,
        "r_hub [m]": r_hub_m,
        "Área [m2]": A_m2,
        "Va [m/s]": Va_m_s,
        "Delta T [K]": delta_T_K,
        "Relación presión etapa [-]": pressure_ratio_stage,
        "p_salida [Pa]": p_comp_out,
        "T_salida [K]": T_comp_out,
        "Potencia etapa [W]": W_stage_W,
        "Torque etapa [N m]": torque_stage_Nm
    })

    W_comp_total_W += W_stage_W

    p_comp_in = p_comp_out
    T_comp_in = T_comp_out

p2 = p_comp_out
T2 = T_comp_out

compressor_PR = p2 / p1

W_comp_total_kW = W_comp_total_W / 1000.0
W_comp_total_MW = W_comp_total_W / 1e6

specific_work_comp_kJ_kg = W_comp_total_W / m_air_kg_s / 1000.0


# ============================================================
# PASO 3: CÁMARA DE COMBUSTIÓN
# ============================================================

LHV_fuel_J_kg = LHV_fuel_MJ_kg * 1e6          # [J/kg]

Q_added_W = m_fuel_kg_s * LHV_fuel_J_kg * eta_combustion
Q_added_kW = Q_added_W / 1000.0
Q_added_MW = Q_added_W / 1e6

T3 = (
    m_air_kg_s * cp_air * T2 + Q_added_W
) / (m_gas_kg_s * cp_gas)

p3 = eta_combustor_pressure * p2


# ============================================================
# PASO 4: TURBINA AXIAL MULTIETAPA
# ============================================================

turbine_geometry = [
    {"stage": 1, "r_tip_m": 0.760000, "r_hub_m": 0.430000, "beta_l_deg": 30.000000, "beta_t_deg": 82.726736},
    {"stage": 2, "r_tip_m": 0.780000, "r_hub_m": 0.410000, "beta_l_deg": 29.000000, "beta_t_deg": 84.726736},
    {"stage": 3, "r_tip_m": 0.800000, "r_hub_m": 0.390000, "beta_l_deg": 28.000000, "beta_t_deg": 86.726736},
]

turbine_results = []

p_turb_in = p3
T_turb_in = T3
W_turb_euler_total_W = 0.0
W_turb_total_W = 0.0

for stage_data in turbine_geometry:

    stage_number = stage_data["stage"]

    r_tip_m = stage_data["r_tip_m"]
    r_hub_m = stage_data["r_hub_m"]
    beta_l_deg = stage_data["beta_l_deg"]
    beta_t_deg = stage_data["beta_t_deg"]

    r_mean_m = 0.5 * (r_tip_m + r_hub_m)
    A_m2 = axial_area(r_tip_m, r_hub_m)

    beta_l_rad = math.radians(beta_l_deg)
    beta_t_rad = math.radians(beta_t_deg)

    Va_m_s = dotV_air_m3_s / A_m2
    U_m_s = omega * r_mean_m

    v_t_in_m_s = U_m_s - Va_m_s * math.tan(beta_l_rad)
    v_t_out_m_s = U_m_s - Va_m_s * math.tan(beta_t_rad)

    torque_stage_Nm = r_mean_m * m_gas_kg_s * (v_t_in_m_s - v_t_out_m_s)

    W_stage_euler_W = omega * torque_stage_Nm
    W_stage_useful_W = W_stage_euler_W * eta_turbine_stage

    if W_stage_useful_W <= 0.0:
        raise ValueError("Trabajo de turbina no positivo. Revisar beta_l y beta_t.")

    delta_T_stage_K = W_stage_useful_W / (m_gas_kg_s * cp_gas)

    T_turb_out = T_turb_in - delta_T_stage_K

    pressure_base = 1.0 - delta_T_stage_K / T_turb_in

    if pressure_base <= 0.0:
        raise ValueError("Caída de temperatura excesiva en la turbina.")

    pressure_ratio_stage = pressure_base ** (gamma_gas / (gamma_gas - 1.0))

    p_turb_out = p_turb_in * pressure_ratio_stage

    turbine_results.append({
        "Etapa": stage_number,
        "r_tip [m]": r_tip_m,
        "r_hub [m]": r_hub_m,
        "Área [m2]": A_m2,
        "Va [m/s]": Va_m_s,
        "Delta T [K]": delta_T_stage_K,
        "Relación presión etapa [-]": pressure_ratio_stage,
        "p_salida [Pa]": p_turb_out,
        "T_salida [K]": T_turb_out,
        "Potencia etapa [W]": W_stage_useful_W,
        "Torque etapa [N m]": torque_stage_Nm
    })

    W_turb_euler_total_W += W_stage_euler_W
    W_turb_total_W += W_stage_useful_W

    p_turb_in = p_turb_out
    T_turb_in = T_turb_out

p4 = p_turb_out
T4 = T_turb_out

turbine_PR = p4 / p3

W_turb_euler_total_kW = W_turb_euler_total_W / 1000.0
W_turb_euler_total_MW = W_turb_euler_total_W / 1e6

W_turb_total_kW = W_turb_total_W / 1000.0
W_turb_total_MW = W_turb_total_W / 1e6

specific_work_turb_kJ_kg = W_turb_total_W / m_gas_kg_s / 1000.0


# ============================================================
# PASO 5: BALANCE DE POTENCIA Y ESCAPE
# ============================================================

W_net_shaft_W = W_turb_total_W - W_comp_total_W
W_net_shaft_MW = W_net_shaft_W / 1e6

W_electric_W = W_net_shaft_W * eta_generator
W_electric_MW = W_electric_W / 1e6

Q_exhaust_W = m_gas_kg_s * cp_gas * (T4 - T_amb)
Q_exhaust_MW = Q_exhaust_W / 1e6

eta_electric = W_electric_W / Q_fuel_W

consumo_termico_especifico_kJ_kWh = (
    Q_fuel_kW / (W_electric_W / 1000.0)
) * 3600.0


# ============================================================
# PASO 6: RESULTADOS
# ============================================================

df_inputs = pd.DataFrame([{
    "Combustible": fuel_type,
    "Presión ambiente [kPa]": p_amb / 1000.0,
    "Temperatura ambiente [K]": T_amb,
    "Densidad del aire [kg/m3]": rho_air,
    "AFR [-]": AFR_mass,
    "Flujo másico de escape [kg/s]": m_exhaust_kg_s,
    "Velocidad del eje [rpm]": shaft_rpm,
    "Velocidad angular [rad/s]": omega,
    "LHV [MJ/kg]": LHV_fuel_MJ_kg,
    "Eficiencia de admisión [-]": eta_inlet,
    "Eficiencia de combustión [-]": eta_combustion,
    "Eficiencia del generador [-]": eta_generator
}])

print("=== Entradas principales del modelo ===")
display(df_inputs.T)


df_flow = pd.DataFrame([{
    "Flujo másico de aire [kg/s]": m_air_kg_s,
    "Flujo másico de combustible [kg/s]": m_fuel_kg_s,
    "Flujo másico de gases [kg/s]": m_gas_kg_s,
    "Caudal volumétrico de aire [m3/s]": dotV_air_m3_s,
    "Caudal de combustible [m3/h]": dotV_fuel_m3_h,
    "Potencia térmica del combustible [MW]": Q_fuel_MW,
    "Calor agregado en combustión [MW]": Q_added_MW
}])

print("\n=== Aire, combustible y energía de entrada ===")
display(df_flow.T)


df_states = pd.DataFrame([
    {"Estado": "1 - Entrada del compresor", "Presión [kPa]": p1/1000.0, "Temperatura [°C]": T1 - 273.15},
    {"Estado": "2 - Salida del compresor",  "Presión [kPa]": p2/1000.0, "Temperatura [°C]": T2 - 273.15},
    {"Estado": "3 - Salida de combustión",  "Presión [kPa]": p3/1000.0, "Temperatura [°C]": T3 - 273.15},
    {"Estado": "4 - Escape de turbina",     "Presión [kPa]": p4/1000.0, "Temperatura [°C]": T4 - 273.15},
])

print("\n=== Estados principales del ciclo ===")
display(df_states)


df_comp = pd.DataFrame(compressor_results)
df_comp_show = df_comp.copy()

for col in df_comp_show.columns:
    if col != "Etapa":
        df_comp_show[col] = df_comp_show[col].astype(float).round(4)

df_comp_show["p_salida [kPa]"] = (df_comp["p_salida [Pa]"] / 1000.0).round(3)
df_comp_show["Potencia etapa [MW]"] = (df_comp["Potencia etapa [W]"] / 1e6).round(4)

print("\n=== Resultados por etapa del compresor axial ===")
display(df_comp_show[[
    "Etapa",
    "r_tip [m]",
    "r_hub [m]",
    "Área [m2]",
    "Va [m/s]",
    "Delta T [K]",
    "Relación presión etapa [-]",
    "p_salida [kPa]",
    "T_salida [K]",
    "Potencia etapa [MW]",
    "Torque etapa [N m]"
]])


df_turb = pd.DataFrame(turbine_results)
df_turb_show = df_turb.copy()

for col in df_turb_show.columns:
    if col != "Etapa":
        df_turb_show[col] = df_turb_show[col].astype(float).round(4)

df_turb_show["p_salida [kPa]"] = (df_turb["p_salida [Pa]"] / 1000.0).round(3)
df_turb_show["Potencia etapa [MW]"] = (df_turb["Potencia etapa [W]"] / 1e6).round(4)

print("\n=== Resultados por etapa de la turbina axial ===")
display(df_turb_show[[
    "Etapa",
    "r_tip [m]",
    "r_hub [m]",
    "Área [m2]",
    "Va [m/s]",
    "Delta T [K]",
    "Relación presión etapa [-]",
    "p_salida [kPa]",
    "T_salida [K]",
    "Potencia etapa [MW]",
    "Torque etapa [N m]"
]])


df_power = pd.DataFrame([{
    "Potencia total del compresor [MW]": W_comp_total_MW,
    "Potencia útil de la turbina [MW]": W_turb_total_MW,
    "Potencia neta de eje [MW]": W_net_shaft_MW,
    "Potencia eléctrica estimada [MW]": W_electric_MW,
    "Calor disponible en el escape [MW]": Q_exhaust_MW,
    "Eficiencia del ciclo [-]": eta_electric,
    "Consumo térmico específico [kJ/kWh]": consumo_termico_especifico_kJ_kWh
}])

print("\n=== Balance de potencia y escape ===")
display(df_power.T)
```

# **1.2 Bomba de Agua**


## **1.2.1 Arquitectura de la Bomba Centrífuga**

En el centro de la bomba se localiza el eje de rotación, encargado de transmitir potencia mecánica al impulsor mediante una velocidad angular $\omega$ y un torque $T_{\text{shaft}}$.

A partir de este movimiento, cada punto del rodete adquiere una velocidad periférica $u=\omega r$, la cual depende directamente del radio considerado. En la zona de entrada, ubicada cerca del centro del impulsor, el fluido ingresa al canal entre álabes con una velocidad absoluta $V_1$, mientras que el álabe se mueve con una velocidad periférica $u_1=\omega r_1$. La diferencia entre el movimiento del fluido y el movimiento del álabe define la velocidad relativa $W_1$, que representa la velocidad con la que el fluido es observado desde un sistema de referencia solidario al rotor. Por esta razón, en la entrada se cumple la relación vectorial:

$$\vec{V}_1=\vec{u}_1+\vec{W}_1$$.

El análisis de este primer triángulo de velocidades permite verificar si el fluido entra de forma adecuada al impulsor, evitando pérdidas por incidencia y garantizando una interacción suave con la geometría de los álabes.

A medida que el fluido avanza desde el centro hacia la periferia del impulsor, recibe energía mecánica del rotor y aumenta su momento angular. En la zona exterior de la bomba, correspondiente al radio de salida $r_2$, la velocidad periférica del álabe es $u_2=\omega r_2$, generalmente mayor que la velocidad periférica de entrada debido al incremento del radio.

En esta región se construye el triángulo de velocidades de salida, formado por la velocidad absoluta $V_2$, la velocidad periférica $u_2$ y la velocidad relativa $W_2$. La velocidad absoluta de salida puede descomponerse en una componente radial $V_{r2}$ y una componente tangencial $V_{t2}$. La componente radial está asociada principalmente al caudal que atraviesa la bomba, mientras que la componente tangencial está directamente relacionada con la transferencia de energía. En consecuencia, el cambio entre la componente tangencial de entrada $V_{t1}$ y la componente tangencial de salida $V_{t2}$ permite cuantificar el aumento de energía que el impulsor entrega al fluido.



image.png


## **1.2.3 Modelo Matemático de la Bomba Centrífuga**

Las ecuaciones mostradas en la figura se obtienen a partir del balance de momento angular aplicado al volumen de control que contiene el impulsor. El caudal másico del fluido puede expresarse como:

$$\dot{m}=\rho Q$$

Donde $\rho$ es la densidad y $Q$ el caudal volumétrico. El torque transmitido al fluido se calcula a partir de la diferencia entre el momento angular de salida y el de entrada, por lo que se obtiene:

$$T_{\text{shaft}}=\rho Q(r_2V_{t2}-r_1V_{t1})$$

Al multiplicar este torque por la velocidad angular del eje se obtiene la potencia hidráulica transferida al fluido:

$$P_w=\omega T_{\text{shaft}}=\rho Q(\omega r_2V_{t2}-\omega r_1V_{t1})$$

y como

$$u_1=\omega r_1$ y $u_2=\omega r_2$$

La expresión anterior se transforma en

$$P_w=\rho Q(u_2V_{t2}-u_1V_{t1})$$

Finalmente, al dividir la potencia hidráulica entre el peso específico por caudal, se obtiene la cabeza hidráulica de la bomba:

$$H=P_w/(\rho g Q)$$ de donde finalmente obtenemos: $$H=(u_2V_{t2}-u_1V_{t1})/g$$ Esta expresión corresponde a la ecuación de Euler para turbomáquinas aplicada a una bomba, y demuestra que la energía entregada al fluido depende del cambio en la componente tangencial de la velocidad absoluta entre la entrada y la salida del impulsor.


## **1.2.4 Simulación de la Bomba Centrífuga**

Con base en el desarrollo matemático propuesto, se implementó un modelo estacionario para una bomba centrífuga del sistema de cogeneración. El código utiliza las ecuaciones de continuidad, conservación de energía y la ecuación de Euler para turbomáquinas, permitiendo estimar la cabeza hidráulica requerida, la cabeza real entregada, la potencia hidráulica, la potencia al eje y el torque necesario para la operación. Además, se construyen los triángulos de velocidad de entrada y salida del impulsor, considerando una entrada sin remolino y una corrección por deslizamiento en la salida.

Para la simulación se consideró agua como fluido de trabajo, con densidad del agua, rho, de 997 kg/m³ y viscosidad dinámica, mu, de 8.9e-4 Pa·s. Las condiciones principales de operación fueron un caudal volumétrico, Q, de 0.020 m³/s, una relación de presión, pi_p, de 4:1 y una velocidad de rotación, N, de 2200 rpm. El impulsor se modeló con radio de entrada, r1, de 0.040 m; radio de salida, r2, de 0.120 m; ángulos geométricos de entrada y salida de 20° y 30°, respectivamente; y ocho álabes. Como aproximaciones del modelo, se asumió flujo estacionario, agua incompresible, entrada radial sin componente tangencial, factores de bloqueo unitarios, factor de deslizamiento de 0.90 y eficiencias hidráulica, mecánica y volumétrica de 0.78, 0.95 y 0.98, respectivamente.




```python
# @title
# =============================================================================
# CÁLCULO DE BOMBA HIDRÁULICA ROTACIONAL - BOMBA CENTRÍFUGA
# =============================================================================

import numpy as np
import pandas as pd
from IPython.display import display, Markdown

# =============================================================================
# 1. DATOS DE ENTRADA
# =============================================================================
# Modifica estos valores según los datos de tu proyecto.

# -----------------------------------------------------------------------------
# Propiedades del fluido
# -----------------------------------------------------------------------------
rho = 997.0          # Densidad del agua [kg/m^3]
mu = 8.9e-4         # Viscosidad dinámica del agua [Pa·s]
g = 9.81            # Gravedad [m/s^2]

# -----------------------------------------------------------------------------
# Condiciones de operación
# -----------------------------------------------------------------------------
Q = 0.020           # Caudal volumétrico [m^3/s]
p1 = 101325.0       # Presión de entrada [Pa]
pi_p = 4.0          # Relación de presión de la bomba: pi_p = p2 / p1
N = 2200.0          # Velocidad de rotación [rpm]

# -----------------------------------------------------------------------------
# Geometría del impulsor
# -----------------------------------------------------------------------------
r1 = 0.040          # Radio de entrada del impulsor [m]
r2 = 0.120          # Radio de salida del impulsor [m]

b1 = 0.025          # Ancho del canal en la entrada [m]
b2 = 0.018          # Ancho del canal en la salida [m]

beta1_geo_deg = 20.0    # Ángulo geométrico del álabe en la entrada [grados]
beta2_geo_deg = 30.0    # Ángulo geométrico del álabe en la salida [grados]

z = 8               # Número de álabes [-]

# -----------------------------------------------------------------------------
# Factores de corrección
# -----------------------------------------------------------------------------
kb1 = 1.0           # Factor de bloqueo en la entrada [-]
kb2 = 1.0           # Factor de bloqueo en la salida [-]

sigma = 0.90        # Factor de deslizamiento [-]

# -----------------------------------------------------------------------------
# Eficiencias
# -----------------------------------------------------------------------------
eta_h = 0.78        # Eficiencia hidráulica [-]
eta_m = 0.95        # Eficiencia mecánica [-]
eta_v = 0.98        # Eficiencia volumétrica [-]

eta_o = eta_h * eta_m * eta_v   # Eficiencia global aproximada [-]


# =============================================================================
# 2. FUNCIONES AUXILIARES
# =============================================================================

def deg_to_rad(angle_deg):
    """Convierte grados a radianes."""
    return np.deg2rad(angle_deg)

def rad_to_deg(angle_rad):
    """Convierte radianes a grados."""
    return np.rad2deg(angle_rad)

def area_anular(r, b, kb=1.0):
    """
    Calcula el área efectiva de paso del fluido en el impulsor.

    A = kb * 2*pi*r*b

    donde:
    kb = factor de bloqueo por espesor de álabes.
    """
    return kb * 2.0 * np.pi * r * b

def magnitud(x, y):
    """Calcula la magnitud de un vector 2D."""
    return np.sqrt(x**2 + y**2)

def crear_tabla(diccionario, nombre_columna="Valor"):
    """
    Convierte un diccionario de resultados en una tabla de pandas.
    """
    return pd.DataFrame(
        list(diccionario.items()),
        columns=["Parámetro", nombre_columna]
    )


# =============================================================================
# 3. CÁLCULOS DE PRESIÓN Y CABEZA HIDRÁULICA REQUERIDA
# =============================================================================

p2 = pi_p * p1                      # Presión de salida [Pa]
delta_p = p2 - p1                   # Aumento de presión [Pa]
H_req = delta_p / (rho * g)         # Cabeza hidráulica requerida [m]

mdot = rho * Q                      # Flujo másico [kg/s]


# =============================================================================
# 4. VELOCIDAD ANGULAR Y VELOCIDADES PERIFÉRICAS
# =============================================================================

omega = 2.0 * np.pi * N / 60.0      # Velocidad angular [rad/s]

U1 = omega * r1                     # Velocidad periférica en la entrada [m/s]
U2 = omega * r2                     # Velocidad periférica en la salida [m/s]


# =============================================================================
# 5. VELOCIDADES MERIDIONALES
# =============================================================================

A1 = area_anular(r1, b1, kb1)       # Área efectiva de entrada [m^2]
A2 = area_anular(r2, b2, kb2)       # Área efectiva de salida [m^2]

Cm1 = Q / A1                       # Velocidad meridional en la entrada [m/s]
Cm2 = Q / A2                       # Velocidad meridional en la salida [m/s]


# =============================================================================
# 6. TRIÁNGULO DE VELOCIDADES EN LA ENTRADA
# =============================================================================
# Se asume entrada sin remolino:
#
# Cu1 = 0
#
# Esto significa que la velocidad absoluta del fluido entra radialmente.

Cu1 = 0.0                           # Componente tangencial absoluta en entrada [m/s]

C1 = magnitud(Cu1, Cm1)             # Magnitud de velocidad absoluta en entrada [m/s]
W1 = magnitud(Cu1 - U1, Cm1)        # Magnitud de velocidad relativa en entrada [m/s]

alpha1_deg = rad_to_deg(np.arctan2(Cm1, Cu1))       # Ángulo absoluto entrada [grados]
beta1_calc_deg = rad_to_deg(np.arctan2(Cm1, U1 - Cu1))  # Ángulo relativo ideal entrada [grados]

incidencia_beta1_deg = beta1_geo_deg - beta1_calc_deg   # Diferencia geométrica vs ideal


# =============================================================================
# 7. TRIÁNGULO DE VELOCIDADES EN LA SALIDA
# =============================================================================
# Para una bomba centrífuga con álabes curvados hacia atrás:
#
# Cu2_ideal = U2 - Cm2 / tan(beta2)
#
# Luego se corrige usando el factor de deslizamiento:
#
# Cu2_real = sigma * Cu2_ideal

beta2_geo_rad = deg_to_rad(beta2_geo_deg)

Cu2_ideal = U2 - Cm2 / np.tan(beta2_geo_rad)   # Componente tangencial ideal [m/s]
Cu2_real = sigma * Cu2_ideal                   # Componente tangencial corregida [m/s]

C2_ideal = magnitud(Cu2_ideal, Cm2)            # Velocidad absoluta ideal salida [m/s]
C2_real = magnitud(Cu2_real, Cm2)              # Velocidad absoluta real salida [m/s]

W2_ideal = magnitud(Cu2_ideal - U2, Cm2)       # Velocidad relativa ideal salida [m/s]
W2_real = magnitud(Cu2_real - U2, Cm2)         # Velocidad relativa corregida salida [m/s]

alpha2_ideal_deg = rad_to_deg(np.arctan2(Cm2, Cu2_ideal))
alpha2_real_deg = rad_to_deg(np.arctan2(Cm2, Cu2_real))

beta2_real_calc_deg = rad_to_deg(np.arctan2(Cm2, U2 - Cu2_real))


# =============================================================================
# 8. ECUACIÓN DE EULER PARA TURBOMÁQUINAS
# =============================================================================
# Energía específica ideal:
#
# Delta h_E = U2*Cu2 - U1*Cu1
#
# Cabeza de Euler:
#
# H_E = (U2*Cu2 - U1*Cu1)/g

delta_h_E_ideal = U2 * Cu2_ideal - U1 * Cu1     # Energía específica ideal [J/kg]
delta_h_E_corr = U2 * Cu2_real - U1 * Cu1       # Energía específica con deslizamiento [J/kg]

H_E_ideal = delta_h_E_ideal / g                 # Cabeza de Euler ideal [m]
H_E_corr = delta_h_E_corr / g                   # Cabeza de Euler corregida [m]

H_real = eta_h * H_E_corr                       # Cabeza hidráulica real entregada [m]


# =============================================================================
# 9. POTENCIAS Y TORQUE
# =============================================================================

# Potencia hidráulica requerida por la relación de presión
P_h_req = rho * g * Q * H_req                   # [W]

# Potencia hidráulica realmente entregada según la geometría calculada
P_h_real = rho * g * Q * H_real                 # [W]

# Potencia al eje necesaria para alcanzar la cabeza requerida
P_eje_req = P_h_req / eta_o                     # [W]

# Potencia al eje estimada para la cabeza real calculada
P_eje_real = P_h_real / eta_o                   # [W]

# Torque asociado a la potencia al eje
T_req = P_eje_req / omega                       # [N·m]
T_real = P_eje_real / omega                     # [N·m]


# =============================================================================
# 10. NÚMEROS DE REYNOLDS APROXIMADOS
# =============================================================================
# Se usa una longitud característica aproximada:
#
# Dh ≈ 2b
#
# Esto es una aproximación sencilla para el canal del impulsor.

Dh1 = 2.0 * b1
Dh2 = 2.0 * b2

Re1 = rho * C1 * Dh1 / mu
Re2 = rho * C2_real * Dh2 / mu


# =============================================================================
# 11. VERIFICACIONES BÁSICAS DE DISEÑO
# =============================================================================

cumple_cabeza = H_real >= H_req
margen_H = H_real - H_req
margen_porcentual = (margen_H / H_req) * 100.0

if Cu2_real <= 0:
    advertencia_Cu2 = "Advertencia: Cu2_real <= 0. La bomba no está entregando remolino positivo suficiente."
else:
    advertencia_Cu2 = "Cu2_real positivo. La transferencia de energía al fluido es físicamente consistente."

if cumple_cabeza:
    estado_diseno = "La bomba SÍ alcanza la cabeza requerida."
else:
    estado_diseno = "La bomba NO alcanza la cabeza requerida. Se debe modificar N, r2, beta2, Q o eficiencias."


# =============================================================================
# 12. TABLAS DE RESULTADOS
# =============================================================================

tabla_operacion = crear_tabla({
    "Caudal volumétrico Q [m^3/s]": Q,
    "Flujo másico mdot [kg/s]": mdot,
    "Aumento de presión Delta p [Pa]": delta_p,
    "Velocidad de rotación N [rpm]": N
})

tabla_geometria = crear_tabla({
    "Radio de entrada r1 [m]": r1,
    "Radio de salida r2 [m]": r2,
    "Número de álabes z [-]": z,
    "Ángulo geométrico beta1 [deg]": beta1_geo_deg,
    "Ángulo geométrico beta2 [deg]": beta2_geo_deg,
})

tabla_triangulos = crear_tabla({
    "Velocidad absoluta C1 [m/s]": C1,
    "Velocidad relativa W1 [m/s]": W1,
})

tabla_energia = crear_tabla({
    "Cabeza Hidraulica real entregada H_real [m]": H_real,
    "Eficiencia global eta_o [-]": eta_o,
    "Potencia al eje requerida P_eje_req [W]": P_eje_req,
    "Torque requerido T_req [N·m]": T_req,
    "Reynolds aproximado en entrada Re1 [-]": Re1,
    "Reynolds aproximado en salida Re2 [-]": Re2
})


# =============================================================================
# 13. FORMATO DE IMPRESIÓN
# =============================================================================

pd.options.display.float_format = "{:,.4f}".format

display(Markdown("## Resultados de operación"))
display(tabla_operacion)

display(Markdown("## Parámetros geométricos del impulsor"))
display(tabla_geometria)

display(Markdown("## Triángulos de velocidad"))
display(tabla_triangulos)

display(Markdown("## Resultados hidráulicos y energéticos"))
display(tabla_energia)



# =============================================================================
# 14. DICCIONARIO PARA USAR EN LA GRÁFICA
# =============================================================================
# Este diccionario se usa en el segundo bloque de código para graficar
# los triángulos de velocidad.

vel = {
    "entrada": {
        "U": U1,
        "Cm": Cm1,
        "Cu": Cu1,
        "C": C1,
        "W": W1,
        "alpha_deg": alpha1_deg,
        "beta_deg": beta1_calc_deg,
        "titulo": "Triángulo de velocidades - Entrada del impulsor"
    },
    "salida": {
        "U": U2,
        "Cm": Cm2,
        "Cu": Cu2_real,
        "C": C2_real,
        "W": W2_real,
        "alpha_deg": alpha2_real_deg,
        "beta_deg": beta2_real_calc_deg,
        "titulo": "Triángulo de velocidades - Salida del impulsor"
    }
}
```

```python
# @title
# =============================================================================
# GRÁFICA TRIÁNGULOS DE VELOCIDAD BOMBA CENTRIFUGA
# =============================================================================

import numpy as np
import matplotlib.pyplot as plt


def punto_sobre_vector(p0, p1, fraccion=0.55):
    """
    Retorna un punto ubicado en cierta fracción del vector p0 -> p1.
    """
    return p0 + fraccion * (p1 - p0)


def dibujar_vector(ax, inicio, fin, color, etiqueta=None,
                   text_shift=None, lw=2.5, fs=11):
    """
    Dibuja un vector desde 'inicio' hasta 'fin' y coloca una etiqueta.
    """

    ax.annotate(
        "",
        xy=(fin[0], fin[1]),
        xytext=(inicio[0], inicio[1]),
        arrowprops=dict(
            arrowstyle="->",
            linewidth=lw,
            color=color,
            shrinkA=0,
            shrinkB=0
        )
    )

    if etiqueta is not None:
        pm = punto_sobre_vector(inicio, fin, fraccion=0.55)

        if text_shift is None:
            text_shift = np.array([0.0, 0.0])

        pos = pm + text_shift

        ax.text(
            pos[0],
            pos[1],
            etiqueta,
            fontsize=fs,
            color=color,
            fontweight="bold",
            bbox=dict(
                boxstyle="round,pad=0.25",
                fc="white",
                ec="none",
                alpha=0.85
            )
        )


def graficar_triangulo_velocidad_mejorado(
    datos,
    nombre_archivo=None,
    config_textos=None,
    mostrar_lineas_auxiliares=True,
    figsize=(9, 6),
    usar_aspecto_igual=True,
    margen_x=0.35,
    margen_y=0.35
):
    """
    Grafica el triángulo de velocidades sin mostrar etiquetas Cu ni Cm.

    Parámetros:
    -----------
    datos : dict
        Debe contener:
        - U
        - Cm
        - Cu
        - C
        - W
        - alpha_deg
        - beta_deg
        - titulo

    nombre_archivo : str o None
        Si se da, guarda la figura.

    config_textos : dict o None
        Ajustes manuales de posición para textos.

    mostrar_lineas_auxiliares : bool
        Si True, muestra las líneas punteadas de proyección.

    figsize : tuple
        Tamaño de figura.

    usar_aspecto_igual : bool
        Si True usa aspect='equal'.
        Si False usa aspect='auto' para mejorar la visualización.

    margen_x, margen_y : float
        Factores de margen para los límites en x y y.
    """

    if config_textos is None:
        config_textos = {}

    U = datos["U"]
    Cm = datos["Cm"]
    Cu = datos["Cu"]
    C = datos["C"]
    W = datos["W"]
    alpha_deg = datos["alpha_deg"]
    beta_deg = datos["beta_deg"]
    titulo = datos["titulo"]

    # -------------------------------------------------------------------------
    # Puntos principales del triángulo
    # -------------------------------------------------------------------------
    O = np.array([0.0, 0.0])       # Origen
    A = np.array([U, 0.0])         # Punta del vector U
    B = np.array([Cu, Cm])         # Punta del vector C

    # -------------------------------------------------------------------------
    # Desplazamientos de etiquetas
    # -------------------------------------------------------------------------
    shift_U = config_textos.get("shift_U", np.array([0.0, -0.55]))
    shift_C = config_textos.get("shift_C", np.array([0.30, 0.20]))
    shift_W = config_textos.get("shift_W", np.array([0.25, 0.30]))
    pos_angulos = config_textos.get("pos_angulos", (0.03, 0.96))

    # -------------------------------------------------------------------------
    # Crear figura
    # -------------------------------------------------------------------------
    fig, ax = plt.subplots(figsize=figsize)

    # -------------------------------------------------------------------------
    # Dibujar vectores
    # -------------------------------------------------------------------------
    dibujar_vector(
        ax,
        inicio=O,
        fin=A,
        color="tab:blue",
        etiqueta=f"U = {U:.2f} m/s",
        text_shift=shift_U
    )

    dibujar_vector(
        ax,
        inicio=O,
        fin=B,
        color="tab:green",
        etiqueta=f"C = {C:.2f} m/s",
        text_shift=shift_C
    )

    dibujar_vector(
        ax,
        inicio=A,
        fin=B,
        color="tab:red",
        etiqueta=f"W = {W:.2f} m/s",
        text_shift=shift_W
    )

    # -------------------------------------------------------------------------
    # Líneas auxiliares (sin etiquetas Cu y Cm)
    # -------------------------------------------------------------------------
    if mostrar_lineas_auxiliares:
        ax.plot(
            [0, Cu],
            [Cm, Cm],
            linestyle="--",
            color="gray",
            linewidth=1.2,
            alpha=0.65
        )

        ax.plot(
            [Cu, Cu],
            [0, Cm],
            linestyle="--",
            color="gray",
            linewidth=1.2,
            alpha=0.65
        )

    # -------------------------------------------------------------------------
    # Puntos
    # -------------------------------------------------------------------------
    ax.scatter(
        [O[0], A[0], B[0]],
        [O[1], A[1], B[1]],
        color="black",
        s=25,
        zorder=5
    )

    # -------------------------------------------------------------------------
    # Recuadro de ángulos
    # -------------------------------------------------------------------------
    texto_angulos = (
        f"Ángulo absoluto α = {alpha_deg:.2f}°\n"
        f"Ángulo relativo β = {beta_deg:.2f}°"
    )

    ax.text(
        pos_angulos[0],
        pos_angulos[1],
        texto_angulos,
        transform=ax.transAxes,
        fontsize=12,
        va="top",
        ha="left",
        bbox=dict(
            boxstyle="round,pad=0.3",
            fc="white",
            ec="0.3",
            alpha=0.9
        )
    )

    # -------------------------------------------------------------------------
    # Ejes y formato
    # -------------------------------------------------------------------------
    ax.axhline(0, color="black", linewidth=1.0)
    ax.axvline(0, color="black", linewidth=1.0)

    ax.set_xlabel("Componente tangencial [m/s]", fontsize=12)
    ax.set_ylabel("Componente meridional / radial [m/s]", fontsize=12)
    ax.set_title(titulo, fontsize=16, fontweight="bold")

    ax.grid(True, linestyle="--", alpha=0.5)

    # -------------------------------------------------------------------------
    # Límites automáticos con márgenes independientes
    # -------------------------------------------------------------------------
    xs = np.array([O[0], A[0], B[0]])
    ys = np.array([O[1], A[1], B[1]])

    dx = max(xs) - min(xs)
    dy = max(ys) - min(ys)

    if dx < 1e-8:
        dx = 1.0
    if dy < 1e-8:
        dy = 1.0

    ax.set_xlim(min(xs) - margen_x * dx, max(xs) + margen_x * dx)
    ax.set_ylim(min(ys) - margen_y * dy, max(ys) + margen_y * dy)

    # -------------------------------------------------------------------------
    # Control del aspecto
    # -------------------------------------------------------------------------
    if usar_aspecto_igual:
        ax.set_aspect("equal", adjustable="box")
    else:
        ax.set_aspect("auto")

    plt.tight_layout()

    if nombre_archivo is not None:
        plt.savefig(nombre_archivo, dpi=300, bbox_inches="tight")

    plt.show()


# =============================================================================
# GRAFICAR TRIÁNGULO DE ENTRADA
# =============================================================================
# Aquí sí conviene aspecto igual.

graficar_triangulo_velocidad_mejorado(
    vel["entrada"],
    nombre_archivo="triangulo_velocidades_entrada_sin_Cu_Cm.png",
    config_textos={
        "shift_U": np.array([0.0, -0.75]),
        "shift_C": np.array([0.45, 0.20]),
        "shift_W": np.array([0.25, 0.35]),
        "pos_angulos": (0.03, 0.97)
    },
    mostrar_lineas_auxiliares=True,
    figsize=(9, 6),
    usar_aspecto_igual=True,
    margen_x=0.35,
    margen_y=0.35
)


# =============================================================================
# GRAFICAR TRIÁNGULO DE SALIDA
# =============================================================================
# Aquí se usa aspecto automático para que no se vea "aplastada".

graficar_triangulo_velocidad_mejorado(
    vel["salida"],
    nombre_archivo="triangulo_velocidades_salida_sin_Cu_Cm.png",
    config_textos={
        "shift_U": np.array([0.0, -0.20]),
        "shift_C": np.array([-2.9, 0.20]),
        "shift_W": np.array([0.40, 0.45]),
        "pos_angulos": (0.03, 0.95)
    },
    mostrar_lineas_auxiliares=True,
    figsize=(10, 5.5),
    usar_aspecto_igual=False,
    margen_x=0.12,
    margen_y=0.90
)
```

# **1.3 Turbina Rankine**

## **1.3.1 Arquitectura de la Turbina Rankine**

La turbina Rankine corresponde a una turbomáquina de vapor diseñada para transformar la energía térmica y de presión del vapor de agua en trabajo mecánico de eje. A diferencia de la turbina de gas, en la cual el fluido de trabajo es aire o gases de combustión, esta turbina opera con vapor proveniente de una caldera o generador de vapor. Su arquitectura se compone principalmente de una carcasa, un eje rotativo, filas de toberas o álabes fijos y filas de álabes móviles. Las toberas se encargan de expandir y acelerar el vapor, orientándolo hacia el rotor con el ángulo adecuado, mientras que los álabes móviles extraen energía del flujo mediante el cambio de cantidad de movimiento, generando torque sobre el eje. En turbinas reales, este proceso se realiza en múltiples etapas para aprovechar de forma progresiva la caída de presión del vapor y evitar una expansión demasiado brusca en una sola fila de álabes.

## **1.3.2 Modelo Matemático de la Turbina Rankine**

El modelamiento matemático de la turbina Rankine se basa en los mismos principios generales de turbomaquinaria presentados previamente en la **Sección 1.1.** para la turbina de gas. En este sentido, se emplean los balances de masa, energía y cantidad de movimiento, junto con el análisis de triángulos de velocidad, trabajo específico, potencia de eje y eficiencia isentrópica. La principal diferencia no se encuentra en la estructura del modelo de turbomáquina, sino en el fluido de trabajo y en sus propiedades termodinámicas, ya que en este caso se analiza vapor de agua en lugar de aire o gases de combustión.


## **1.3.3 Simulacion Turbina Rankine**

Con base en el desarrollo matemático propuesto, se implementó un modelo estacionario 0D/semi-1D para una turbina Rankine axial multietapa del sistema de cogeneración. El código modela la expansión progresiva del vapor a través de 30 etapas, utilizando relaciones de presión por etapa, conservación de masa, estimación de densidad del vapor y la ecuación de Euler para turbomáquinas. A partir de este procedimiento, el modelo permite calcular los estados principales de entrada y salida, el trabajo específico desarrollado, la potencia útil de la turbina, la potencia al eje y la potencia eléctrica generada.

Para la simulación se consideró vapor de agua como fluido de trabajo, con presión de entrada del vapor, p_steam_in, de 12 MPa; temperatura de entrada del vapor, T_steam_in, de 813.15 K; presión de salida del vapor, p_steam_out, de 9 kPa; y flujo másico de vapor, m_steam_kg_s, de 100 kg/s. La turbina se modeló con velocidad de rotación del eje, shaft_rpm, de 4000 rpm, eficiencia por etapa, eta_stage, de 0.86, eficiencia mecánica, eta_mechanical, de 0.98 y eficiencia del generador, eta_generator, de 0.97. Como aproximaciones principales, se usaron propiedades promedio del vapor, una densidad calculada mediante gas ideal, una distribución geométrica fija de radios y ángulos de álabes por etapa, y una expansión secuencial desde la presión de entrada hasta la presión de condensación.


```python
# @title
# ============================================================
# MODELO SIMPLIFICADO PARA TURBINA DE VAPOR RANKINE
# Turbina axial multietapa - Modelo 0D / semi-1D
# ============================================================

import math
import pandas as pd
from IPython.display import display

# ============================================================
# PASO 1: ENTRADAS PRINCIPALES DE LA TURBINA RANKINE
# ============================================================

# -----------------------------
# Estados principales
# -----------------------------

p_steam_in = 12.0e6           # [Pa]
T_steam_in = 813.15           # [K]

p_steam_out = 9.0e3           # [Pa]

# -----------------------------
# Flujo másico de vapor
# -----------------------------

m_steam_kg_s = 100.0          # [kg/s]

# -----------------------------
# Propiedades promedio del vapor
# -----------------------------

cp_steam = 2300.0             # [J/(kg K)]
gamma_steam_eff = 1.18        # [-]
R_steam = 461.5               # [J/(kg K)]

# -----------------------------
# Parámetros mecánicos
# -----------------------------

shaft_rpm = 4000.0            # [rpm]
eta_mechanical = 0.98         # [-]
eta_generator = 0.97          # [-]

# -----------------------------
# Cálculos preliminares
# -----------------------------

omega = shaft_rpm * 2.0 * math.pi / 60.0     # [rad/s]

p3 = p_steam_in
T3 = T_steam_in


# ============================================================
# PASO 2: FUNCIONES AUXILIARES
# ============================================================

def axial_area(r_tip, r_hub):
    return math.pi * (r_tip**2 - r_hub**2)


def steam_density_ideal(p, T, R):
    return p / (R * T)


def saturation_temperature_water_antoine(p_kPa):
    p_mmHg = p_kPa * 7.50062

    A = 8.07131
    B = 1730.63
    C = 233.426

    return B / (A - math.log10(p_mmHg)) - C


# ============================================================
# PASO 3: GEOMETRÍA DE LA TURBINA DE VAPOR MULTIETAPA
# ============================================================

steam_turbine_geometry = [
    {"Etapa": 1,  "r_tip_m": 0.428465, "r_hub_m": 0.411535, "beta_l_deg": 22.000000, "beta_t_deg": 80.315770, "eta_stage": 0.860000},
    {"Etapa": 2,  "r_tip_m": 0.474535, "r_hub_m": 0.458271, "beta_l_deg": 22.275862, "beta_t_deg": 77.561010, "eta_stage": 0.860000},
    {"Etapa": 3,  "r_tip_m": 0.531102, "r_hub_m": 0.514848, "beta_l_deg": 22.551724, "beta_t_deg": 74.709480, "eta_stage": 0.860000},
    {"Etapa": 4,  "r_tip_m": 0.592438, "r_hub_m": 0.575859, "beta_l_deg": 22.827586, "beta_t_deg": 71.694650, "eta_stage": 0.860000},
    {"Etapa": 5,  "r_tip_m": 0.657110, "r_hub_m": 0.639922, "beta_l_deg": 23.103448, "beta_t_deg": 68.578100, "eta_stage": 0.860000},
    {"Etapa": 6,  "r_tip_m": 0.724401, "r_hub_m": 0.706334, "beta_l_deg": 23.379310, "beta_t_deg": 65.431170, "eta_stage": 0.860000},
    {"Etapa": 7,  "r_tip_m": 0.793876, "r_hub_m": 0.774660, "beta_l_deg": 23.655172, "beta_t_deg": 62.323504, "eta_stage": 0.860000},
    {"Etapa": 8,  "r_tip_m": 0.865244, "r_hub_m": 0.844597, "beta_l_deg": 23.931034, "beta_t_deg": 59.316603, "eta_stage": 0.860000},
    {"Etapa": 9,  "r_tip_m": 0.938301, "r_hub_m": 0.915915, "beta_l_deg": 24.206897, "beta_t_deg": 56.459846, "eta_stage": 0.860000},
    {"Etapa": 10, "r_tip_m": 1.012896, "r_hub_m": 0.988435, "beta_l_deg": 24.482759, "beta_t_deg": 53.788792, "eta_stage": 0.860000},
    {"Etapa": 11, "r_tip_m": 1.088920, "r_hub_m": 1.062003, "beta_l_deg": 24.758621, "beta_t_deg": 51.325454, "eta_stage": 0.860000},
    {"Etapa": 12, "r_tip_m": 1.166291, "r_hub_m": 1.136488, "beta_l_deg": 25.034483, "beta_t_deg": 49.079953, "eta_stage": 0.860000},
    {"Etapa": 13, "r_tip_m": 1.244954, "r_hub_m": 1.211769, "beta_l_deg": 25.310345, "beta_t_deg": 47.052891, "eta_stage": 0.860000},
    {"Etapa": 14, "r_tip_m": 1.324870, "r_hub_m": 1.287735, "beta_l_deg": 25.586207, "beta_t_deg": 45.237874, "eta_stage": 0.860000},
    {"Etapa": 15, "r_tip_m": 1.406022, "r_hub_m": 1.364276, "beta_l_deg": 25.862069, "beta_t_deg": 43.623815, "eta_stage": 0.860000},
    {"Etapa": 16, "r_tip_m": 1.488408, "r_hub_m": 1.441283, "beta_l_deg": 26.137931, "beta_t_deg": 42.196835, "eta_stage": 0.860000},
    {"Etapa": 17, "r_tip_m": 1.572043, "r_hub_m": 1.518644, "beta_l_deg": 26.413793, "beta_t_deg": 40.941712, "eta_stage": 0.860000},
    {"Etapa": 18, "r_tip_m": 1.656960, "r_hub_m": 1.596240, "beta_l_deg": 26.689655, "beta_t_deg": 39.842913, "eta_stage": 0.860000},
    {"Etapa": 19, "r_tip_m": 1.743210, "r_hub_m": 1.673945, "beta_l_deg": 26.965517, "beta_t_deg": 38.885289, "eta_stage": 0.860000},
    {"Etapa": 20, "r_tip_m": 1.830865, "r_hub_m": 1.751617, "beta_l_deg": 27.241379, "beta_t_deg": 38.054498, "eta_stage": 0.860000},
    {"Etapa": 21, "r_tip_m": 1.920019, "r_hub_m": 1.829101, "beta_l_deg": 27.517241, "beta_t_deg": 37.337243, "eta_stage": 0.860000},
    {"Etapa": 22, "r_tip_m": 2.010792, "r_hub_m": 1.906221, "beta_l_deg": 27.793103, "beta_t_deg": 36.721376, "eta_stage": 0.860000},
    {"Etapa": 23, "r_tip_m": 2.103335, "r_hub_m": 1.982775, "beta_l_deg": 28.068966, "beta_t_deg": 36.195915, "eta_stage": 0.860000},
    {"Etapa": 24, "r_tip_m": 2.197833, "r_hub_m": 2.058531, "beta_l_deg": 28.344828, "beta_t_deg": 35.751006, "eta_stage": 0.860000},
    {"Etapa": 25, "r_tip_m": 2.294511, "r_hub_m": 2.133222, "beta_l_deg": 28.620690, "beta_t_deg": 35.377852, "eta_stage": 0.860000},
    {"Etapa": 26, "r_tip_m": 2.393643, "r_hub_m": 2.206533, "beta_l_deg": 28.896552, "beta_t_deg": 35.068631, "eta_stage": 0.860000},
    {"Etapa": 27, "r_tip_m": 2.495558, "r_hub_m": 2.278099, "beta_l_deg": 29.172414, "beta_t_deg": 34.816403, "eta_stage": 0.860000},
    {"Etapa": 28, "r_tip_m": 2.600654, "r_hub_m": 2.347488, "beta_l_deg": 29.448276, "beta_t_deg": 34.615023, "eta_stage": 0.860000},
    {"Etapa": 29, "r_tip_m": 2.709405, "r_hub_m": 2.414195, "beta_l_deg": 29.724138, "beta_t_deg": 34.459049, "eta_stage": 0.860000},
    {"Etapa": 30, "r_tip_m": 2.822382, "r_hub_m": 2.477618, "beta_l_deg": 30.000000, "beta_t_deg": 34.343666, "eta_stage": 0.860000},
]

n_turbine_stages = len(steam_turbine_geometry)


# ============================================================
# PASO 4: EXPANSIÓN EN LA TURBINA DE VAPOR
# ============================================================

steam_turbine_results = []

p_turb_in = p3
T_turb_in = T3

W_turb_euler_total_W = 0.0
W_turb_total_W = 0.0

for stage_data in steam_turbine_geometry:

    stage_number = stage_data["Etapa"]

    r_tip_m = stage_data["r_tip_m"]
    r_hub_m = stage_data["r_hub_m"]
    beta_l_deg = stage_data["beta_l_deg"]
    beta_t_deg = stage_data["beta_t_deg"]
    eta_stage = stage_data["eta_stage"]

    r_mean_m = 0.5 * (r_tip_m + r_hub_m)
    A_m2 = axial_area(r_tip_m, r_hub_m)

    p_turb_out = p_steam_in * (
        p_steam_out / p_steam_in
    ) ** (stage_number / n_turbine_stages)

    pressure_ratio_stage = p_turb_out / p_turb_in

    rho_steam_kg_m3 = steam_density_ideal(p_turb_in, T_turb_in, R_steam)
    dotV_steam_m3_s = m_steam_kg_s / rho_steam_kg_m3

    Va_m_s = dotV_steam_m3_s / A_m2
    U_m_s = omega * r_mean_m

    beta_l_rad = math.radians(beta_l_deg)
    beta_t_rad = math.radians(beta_t_deg)

    v_t_in_m_s = U_m_s - Va_m_s * math.tan(beta_l_rad)
    v_t_out_m_s = U_m_s - Va_m_s * math.tan(beta_t_rad)

    torque_stage_Nm = r_mean_m * m_steam_kg_s * (v_t_in_m_s - v_t_out_m_s)

    W_stage_euler_W = omega * torque_stage_Nm
    W_stage_useful_W = W_stage_euler_W * eta_stage

    if W_stage_useful_W <= 0.0:
        raise ValueError(
            f"Trabajo de turbina no positivo en la etapa {stage_number}. "
            "Revisar geometría, presión, temperatura o eficiencia."
        )

    delta_h_actual_J_kg = W_stage_useful_W / m_steam_kg_s
    delta_T_actual_K = delta_h_actual_J_kg / cp_steam

    T_turb_out = T_turb_in - delta_T_actual_K

    steam_turbine_results.append({
        "Etapa": stage_number,
        "r_tip [m]": r_tip_m,
        "r_hub [m]": r_hub_m,
        "Área [m2]": A_m2,
        "Va [m/s]": Va_m_s,
        "Delta h [J/kg]": delta_h_actual_J_kg,
        "Delta T [K]": delta_T_actual_K,
        "Relación presión etapa [-]": pressure_ratio_stage,
        "p_salida [Pa]": p_turb_out,
        "T_salida [K]": T_turb_out,
        "Potencia etapa [W]": W_stage_useful_W,
        "Torque etapa [N m]": torque_stage_Nm
    })

    W_turb_euler_total_W += W_stage_euler_W
    W_turb_total_W += W_stage_useful_W

    p_turb_in = p_turb_out
    T_turb_in = T_turb_out


# ============================================================
# PASO 5: ESTADOS FINALES Y BALANCE DE POTENCIA
# ============================================================

p4 = p_turb_out
T4 = T_turb_out

turbine_PR = p4 / p3

W_turb_euler_total_kW = W_turb_euler_total_W / 1000.0
W_turb_euler_total_MW = W_turb_euler_total_W / 1e6

W_turb_total_kW = W_turb_total_W / 1000.0
W_turb_total_MW = W_turb_total_W / 1e6

W_shaft_W = W_turb_total_W * eta_mechanical
W_shaft_MW = W_shaft_W / 1e6

W_electric_W = W_shaft_W * eta_generator
W_electric_MW = W_electric_W / 1e6

specific_work_turb_kJ_kg = W_turb_total_W / m_steam_kg_s / 1000.0

T_sat_out_C = saturation_temperature_water_antoine(p4 / 1000.0)
T4_C = T4 - 273.15
outlet_superheat_margin_C = T4_C - T_sat_out_C

# ============================================================
# PASO 6: RESULTADOS
# ============================================================

# -----------------------------
# Tabla 1: Entradas principales
# -----------------------------

df_inputs = pd.DataFrame([{
    "Fluido de trabajo": "Vapor de agua",
    "Presión entrada turbina [MPa]": p_steam_in / 1e6,
    "Temperatura entrada turbina [°C]": T_steam_in - 273.15,
    "Presión salida turbina [kPa]": p_steam_out / 1000.0,
    "Número de etapas [-]": n_turbine_stages,
    "Flujo másico de vapor [kg/s]": m_steam_kg_s,
    "Velocidad del eje [rpm]": shaft_rpm,
    "Eficiencia mecánica [-]": eta_mechanical,
    "Eficiencia generador [-]": eta_generator
}])

print("=== Entradas principales del modelo Rankine ===")
display(df_inputs.T)


# -----------------------------
# Tabla 2: Estados principales
# -----------------------------

df_states = pd.DataFrame([
    {
        "Estado": "Entrada turbina / salida caldera",
        "Presión [kPa]": p3 / 1000.0,
        "Temperatura [°C]": T3 - 273.15,
        "T saturación [°C]": None,
        "Sobrecalentamiento [°C]": None
    },
    {
        "Estado": "Salida turbina / entrada condensador",
        "Presión [kPa]": p4 / 1000.0,
        "Temperatura [°C]": T4_C,
        "T saturación [°C]": T_sat_out_C,
        "Sobrecalentamiento [°C]": outlet_superheat_margin_C
    },
])

print("\n=== Estados principales del ciclo Rankine ===")
display(df_states)


# -----------------------------
# Tabla 3: Resultados por etapa
# -----------------------------

df_turb = pd.DataFrame(steam_turbine_results)
df_turb_show = df_turb.copy()

for col in df_turb_show.columns:
    if col != "Etapa":
        df_turb_show[col] = df_turb_show[col].astype(float).round(4)

df_turb_show["p_salida [kPa]"] = (df_turb["p_salida [Pa]"] / 1000.0).round(3)
df_turb_show["T_salida [°C]"] = (df_turb["T_salida [K]"] - 273.15).round(3)
df_turb_show["Delta h [kJ/kg]"] = (df_turb["Delta h [J/kg]"] / 1000.0).round(4)
df_turb_show["Potencia etapa [MW]"] = (df_turb["Potencia etapa [W]"] / 1e6).round(4)

print("\n=== Resultados por etapa de la turbina Rankine ===")
display(df_turb_show[[
    "Etapa",
    "r_tip [m]",
    "r_hub [m]",
    "Área [m2]",
    "Va [m/s]",
    "Delta h [kJ/kg]",
    "Delta T [K]",
    "Relación presión etapa [-]",
    "p_salida [kPa]",
    "T_salida [°C]",
    "Potencia etapa [MW]",
    "Torque etapa [N m]"
]])


# -----------------------------
# Tabla 4: Balance de potencia
# -----------------------------

df_power = pd.DataFrame([{
    "Potencia útil de la turbina [MW]": W_turb_total_MW,
    "Potencia al eje [MW]": W_shaft_MW,
    "Potencia eléctrica estimada [MW]": W_electric_MW,
    "Trabajo específico turbina [kJ/kg]": specific_work_turb_kJ_kg,
    "Relación de presión total p4/p3 [-]": turbine_PR
}])

print("\n=== Balance de potencia de la turbina Rankine ===")
display(df_power.T)
```

# **1.4 Condensador**

CORAZA Y TUBOS.png


**Figura 1.4.** Esquemático del condensador de coraza y tubos.

## **1.4.1 Intercambiador de coraza y tubos**
El condensador se modela como un intercambiador de calor de coraza y tubos, en el cual el fluido del ciclo Rankine circula por el lado de la coraza y el agua de enfriamiento circula por el interior de los tubos. El objetivo principal del equipo es retirar calor del fluido Rankine proveniente de la turbina hasta obtener una salida completamente líquida, es decir, con una calidad de vapor igual a cero. Esta condición es necesaria porque la bomba del ciclo Rankine debe recibir líquido y no una mezcla líquido-vapor.

El modelo desarrollado corresponde a una formulación global de cero dimensiones. Por esta razón, el condensador no se divide en secciones diferenciales a lo largo de los tubos. En cambio, se utilizan áreas totales de transferencia, propiedades promedio, coeficientes globales de transferencia de calor y balances integrales de masa y energía. Con esta aproximación se calculan la transferencia de calor hacia el agua de enfriamiento, las pérdidas térmicas hacia el ambiente, la temperatura de salida del agua, la entalpía de salida del fluido Rankine y la calidad de salida del condensador.

## **1.4.2 Arquitectura del intercambiador**

La arquitectura considerada corresponde a un condensador de superficie tipo coraza y tubos. El fluido caliente es el vapor proveniente de la turbina Rankine, el cual entra a la coraza a la presión del condensador. En esta región, el vapor se encuentra cercano a condiciones de saturación y transfiere calor hacia las paredes externas de los tubos, formando una película de condensado. El fluido frío es agua de enfriamiento, que circula por el interior de los tubos en uno o varios pasos. Esta configuración permite separar físicamente el vapor del ciclo Rankine del agua de enfriamiento, evitando mezcla entre ambos fluidos.

La geometría del equipo se define mediante el número total de tubos, el número de pasos por los tubos, la longitud de cada tubo, los diámetros interno y externo, el diámetro de la coraza, el paso entre tubos y el tipo de arreglo del banco tubular. A partir de estos parámetros se calcula el área externa de transferencia de calor, que es la superficie efectiva donde ocurre la condensación del vapor. También se evalúa el área de flujo interna disponible para el agua de enfriamiento, lo cual permite estimar su velocidad, régimen de flujo, número de Reynolds, coeficiente convectivo interno y caída de presión en los tubos.

En este modelo, el diseño se verifica desde dos puntos de vista. Primero, desde el punto de vista térmico, se comprueba que la capacidad del intercambiador sea suficiente para retirar el calor necesario y alcanzar una salida líquida con calidad cero. Segundo, desde el punto de vista geométrico e hidráulico, se calcula si el haz tubular cabe dentro de la coraza, si el paso entre tubos es razonable y si la velocidad del agua de enfriamiento genera un régimen de transferencia de calor adecuado. De esta forma, el condensador no se representa únicamente como una caja negra de calor rechazado, sino como un equipo con geometría y fenómenos físicos asociados.

## **1.4.3 Modelado termofluídico del condensador**

## **1.4.3.1 Balance de masa**

La ecuación general del balance de masa para un volumen de control se expresa como:

$$
\frac{dM}{dt}
=
\sum_i \dot{m}_i
$$

donde $M$ es la masa contenida dentro del volumen de control y $\dot{m}_i$ corresponde al flujo másico de cada corriente que entra o sale del sistema. En esta formulación, los flujos de entrada se consideran positivos y los flujos de salida negativos.

Para el agua de enfriamiento, se considera flujo incompresible y sin acumulación de masa dentro de los tubos. Por tanto, el flujo másico de entrada es igual al flujo másico de salida:

$$
\dot{m}_{cool,in}
=
\dot{m}_{cool,out}
=
\dot{m}_{cool}
$$

En el lado del fluido Rankine, el modelo trabaja con un flujo másico global impuesto. Durante la simulación transitoria, antes del arranque del condensador no ingresa fluido Rankine al equipo:

$$
\dot{m}_{rankine}(t)
=
0
\qquad
t \lt t_{start}
$$

Después del tiempo de inicio, el flujo másico del fluido Rankine toma el valor nominal definido para la simulación:

$$
\dot{m}_{rankine}(t)
=
\dot{m}_{rankine,nominal}
\qquad
t\geq t_{start}
$$

De esta manera, el modelo representa una condición de arranque operativo: inicialmente el condensador no recibe fluido Rankine y luego comienza el proceso de condensación.

## **1.4.3.1.1 Hidráulica del agua refrigerante**

El agua de enfriamiento circula por el interior de los tubos y se modela como un fluido incompresible. Para determinar sus condiciones hidráulicas, primero se calcula el caudal volumétrico a partir del flujo másico y la densidad:

$$
\dot{V}_{cool}
=
\frac{\dot{m}_{cool}}{\rho_{cool}}
$$

donde $\dot{m}_{cool}$ es el flujo másico del agua de enfriamiento y $\rho_{cool}$ es su densidad.

El área transversal disponible para el flujo se calcula considerando el número total de tubos y el número de pasos del lado de tubos:

$$
A_{flow}
=
\frac{N_{tubes}}{N_{passes}}
\cdot
\frac{\pi D_i^2}{4}
$$

donde $N_{tubes}$ es el número total de tubos, $N_{passes}$ es el número de pasos del lado de tubos y $D_i$ es el diámetro interno de cada tubo.

Con esta área se obtiene la velocidad media del agua de enfriamiento dentro de los tubos:

$$
V_{cool}
=
\frac{\dot{V}_{cool}}{A_{flow}}
$$

El modelo también estima la pérdida de presión en el lado de tubos. La pérdida por fricción se calcula como:

$$
\Delta p_{fric}
=
f
\cdot
\frac{L_{total}}{D_i}
\cdot
\frac{\rho_{cool}\cdot V_{cool}^{2}}{2}
$$

La pérdida menor asociada a entrada, salida y accesorios se calcula mediante:

$$
\Delta p_{minor}
=
K
\cdot
\frac{\rho_{cool}\cdot V_{cool}^{2}}{2}
$$

Por tanto, la pérdida total de presión en los tubos es:

$$
\Delta p_{tubes}
=
\Delta p_{fric}
+
\Delta p_{minor}
$$

## **1.4.3.1.2 Hidrodinámica del vapor de agua**

El fluido Rankine ingresa al condensador por la coraza después de expandirse en la turbina. En este modelo no se resuelve el campo de velocidades del vapor dentro de la coraza ni el flujo cruzado local entre las filas del banco de tubos. En su lugar, se utiliza una descripción global basada en la presión de condensación, la temperatura de saturación, la entalpía y la calidad del fluido.

A partir de la presión del condensador, se calcula la temperatura de saturación del agua mediante una correlación tipo Antoine. Primero se convierte la presión de kPa a mmHg:

$$
p_{mmHg}
=
7.50062\cdot p_{kPa}
$$

Luego, la temperatura de saturación se obtiene como:

$$
T_{sat}
=
\frac{B}{A-\log_{10}(p_{mmHg})}
-
C
$$

donde $A$, $B$ y $C$ son constantes de la ecuación de Antoine.

La entalpía de líquido saturado se aproxima tomando como referencia el agua líquida a $0\,^{\circ}\mathrm{C}$:

$$
h_f
=
c_{p,water}
\cdot
\left(
T_{sat,K}
-
273.15
\right)
$$

El calor latente de vaporización se estima mediante una relación tipo Watson:

$$
h_{fg}
=
h_{fg,ref}
\cdot
\left[
\frac{
1-T_{sat,K}/T_c
}{
1-T_{ref}/T_c
}
\right]^{0.38}
$$

Con estas propiedades se calcula la entalpía del vapor saturado:

$$
h_g
=
h_f
+
h_{fg}
$$

La densidad aproximada del vapor se calcula mediante la ecuación de gas ideal:

$$
\rho_v
=
\frac{p_{cond}}{R_{steam}\cdot T_{sat,K}}
$$

La calidad del fluido Rankine se calcula a partir de la entalpía:

$$
x
=
\frac{h-h_f}{h_g-h_f}
$$

En el código, la calidad se limita físicamente entre cero y uno. Si la entalpía es menor o igual a $h_f$, se considera líquido saturado o subenfriado y se toma $x=0$. Si la entalpía es mayor o igual a $h_g$, se considera vapor seco o sobrecalentado y se toma $x=1$.

La condición deseada a la salida del condensador es:

$$
x_{out}
=
0
$$

Esta condición indica que el fluido Rankine sale completamente condensado y, por tanto, es apto para ingresar a la bomba.

## **1.4.3.2 Balance de energía**

La ecuación general del balance de energía para un volumen de control, considerando una temperatura media del sistema, puede expresarse como:

$$
\rho\cdot c_p\cdot \mathcal{V}
\cdot
\frac{dT}{dt}
=
\sum_i
\dot{m}_i
\cdot
c_p
\cdot
T_i
+
\sum_j
\dot{Q}_j
$$

donde $\rho$ es la densidad del fluido, $c_p$ es el calor específico a presión constante, $\mathcal{V}$ es el volumen de control, $T$ es la temperatura media del sistema, $\dot{m}_i$ es el flujo másico de cada corriente, $T_i$ es la temperatura asociada a dicha corriente y $\dot{Q}_j$ representa los flujos de calor que entran o salen del volumen de control.

En el condensador, esta ecuación se aplica de forma global para relacionar el calor cedido por el fluido Rankine, el calor recibido por el agua de enfriamiento y las pérdidas térmicas hacia el ambiente. Debido a que el modelo es global, no se resuelve la distribución espacial de temperatura dentro de cada tubo, sino que se trabaja con temperaturas de entrada, salida y valores promedio.

## **1.4.3.2.1 Agua refrigerante**

El agua de enfriamiento recibe calor desde el fluido Rankine a través de la pared de los tubos. Como el modelo no resuelve la distribución espacial de temperatura dentro de cada tubo, se utiliza un balance global entre la entrada y la salida del agua:

$$
\dot{Q}_{tubes}
=
\dot{m}_{cool}
\cdot
c_{p,cool}
\cdot
\left(
T_{cool,out}
-
T_{cool,in}
\right)
$$

A partir de este balance, la temperatura de salida del agua se calcula como:

$$
T_{cool,out}
=
T_{cool,in}
+
\frac{\dot{Q}_{tubes}}
{\dot{m}_{cool}\cdot c_{p,cool}}
$$

donde $\dot{Q}_{tubes}$ representa el calor que efectivamente recibe el agua de enfriamiento.

Para mantener una transferencia de calor físicamente posible, el modelo impone una diferencia mínima de temperatura entre el fluido Rankine en saturación y el agua de salida. Esta diferencia corresponde al pinch térmico del condensador:

$$
T_{cool,out}
\leq
T_{sat}
-
\Delta T_{pinch}
$$

Esta condición evita que el agua de enfriamiento alcance o supere la temperatura de saturación del fluido Rankine, lo cual rompería la dirección natural de la transferencia de calor.

## **1.4.3.2.2 Números adimensionales**

Para calcular el coeficiente de transferencia de calor por convección interna, primero se determina el número de Reynolds:

$$
Re
=
\frac{
\rho_{cool}
\cdot
V_{cool}
\cdot
D_i
}{
\mu_{cool}
}
$$

donde $\mu_{cool}$ es la viscosidad dinámica del agua. Este número permite identificar si el flujo interno es laminar, transicional o turbulento.

También se calcula el número de Prandtl:

$$
Pr
=
\frac{
c_{p,cool}
\cdot
\mu_{cool}
}{
k_{cool}
}
$$

donde $k_{cool}$ es la conductividad térmica del agua.

Para flujo laminar, el modelo emplea una aproximación constante del número de Nusselt:

$$
Nu
=
3.66
$$

Para flujo turbulento, se calcula primero el factor de fricción mediante una correlación tipo Gnielinski:

$$
f
=
\left(
0.79\ln(Re)
-
1.64
\right)^{-2}
$$

Luego, el número de Nusselt se calcula como:

$$
Nu
=
\frac{
\left(\frac{f}{8}\right)
\cdot
(Re-1000)
\cdot
Pr
}{
1
+
12.7
\cdot
\left(\frac{f}{8}\right)^{1/2}
\cdot
\left(
Pr^{2/3}
-
1
\right)
}
$$

Finalmente, el coeficiente de transferencia de calor por convección interna se obtiene mediante:

$$
h_i
=
\frac{
Nu
\cdot
k_{cool}
}{
D_i
}
$$

donde $h_i$ representa el coeficiente convectivo entre la pared interna del tubo y el agua de enfriamiento.

## **1.4.3.2.3 Vapor de agua**

El fluido Rankine cede calor durante el proceso de condensación. La condición objetivo de salida se define a partir de la entalpía de líquido saturado. Cuando no se considera subenfriamiento, la entalpía objetivo es:

$$
h_{out,obj}
=
h_f
$$

Si se incluye un subenfriamiento $\Delta T_{sub}$, la entalpía objetivo se corrige como:

$$
h_{out,obj}
=
h_f
-
c_{p,water}
\cdot
\Delta T_{sub}
$$

El calor necesario para llevar el fluido Rankine desde su estado de entrada hasta el estado objetivo se calcula mediante:

$$
\dot{Q}_{required}
=
\dot{m}_{rankine}
\cdot
\left(
h_{in}
-
h_{out,obj}
\right)
$$

El calor real retirado por el condensador no puede superar el calor necesario para alcanzar dicha condición objetivo. Por esta razón, se calcula como:

$$
\dot{Q}_{cond}
=
\min
\left(
\dot{Q}_{available},
\dot{Q}_{required}
\right)
$$

Una vez determinado el calor retirado, la entalpía de salida del fluido Rankine se obtiene mediante:

$$
h_{out}
=
h_{in}
-
\frac{
\dot{Q}_{cond}
}{
\dot{m}_{rankine}
}
$$

Con esta entalpía se calcula la calidad de salida:

$$
x_{out}
=
\frac{
h_{out}
-
h_f
}{
h_g
-
h_f
}
$$

La salida se considera apta para la bomba cuando:

$$
x_{out}
\leq
10^{-6}
$$

Esta tolerancia numérica representa una salida prácticamente líquida. Si la calidad de salida es mayor que este valor, el condensador no completa la condensación y la bomba recibiría una mezcla líquido-vapor.

## **1.4.3.2.4 Números adimensionales y resistencia térmica global**

La condensación del fluido Rankine ocurre sobre la superficie externa de los tubos. En el modelo, este fenómeno se representa mediante una correlación global de condensación externa, la cual depende de las propiedades del condensado, de la densidad del vapor, del calor latente, de la gravedad, del diámetro externo del tubo y de la diferencia entre la temperatura de saturación y la temperatura de pared.

El área externa total de transferencia de calor de los tubos se calcula como:

$$
A_o
=
N_{tubes}
\cdot
\pi
\cdot
D_o
\cdot
L
$$

El área interna total se calcula mediante:

$$
A_i
=
N_{tubes}
\cdot
\pi
\cdot
D_i
\cdot
L
$$

El coeficiente externo de condensación para un tubo se calcula mediante:

$$
h_{o,1}
=
0.725
\cdot
\left[
\frac{
k_l^3
\cdot
\rho_l
\cdot
(\rho_l-\rho_v)
\cdot
g
\cdot
h_{fg}
}{
\mu_l
\cdot
D_o
\cdot
(T_{sat}-T_{wall})
}
\right]^{1/4}
$$

Para representar el efecto del banco de tubos se aplica un factor correctivo:

$$
h_o
=
h_{o,1}
\cdot
\phi_{bank}
$$

donde:

$$
\phi_{bank}
=
N_{rows}^{-0.25}
$$

El coeficiente global de transferencia de calor se calcula referido al área externa de los tubos. El modelo considera la resistencia convectiva interna, el ensuciamiento interno, la conducción a través de la pared del tubo, el ensuciamiento externo y la resistencia por condensación externa:

$$
\frac{1}{U_o}
=
\frac{D_o}{D_i\cdot h_i}
+
R_{fi}
\cdot
\frac{D_o}{D_i}
+
\frac{
D_o
\cdot
\ln(D_o/D_i)
}{
2
\cdot
k_{wall}
}
+
R_{fo}
+
\frac{1}{h_o}
$$

La conductancia global del condensador se obtiene mediante:

$$
UA
=
U_o
\cdot
A_o
\cdot
F
$$

Como el fluido Rankine condensa aproximadamente a temperatura constante, las diferencias de temperatura se definen como:

$$
\Delta T_1
=
T_{sat}
-
T_{cool,out}
$$

$$
\Delta T_2
=
T_{sat}
-
T_{cool,in}
$$

La diferencia media logarítmica de temperatura se calcula como:

$$
\Delta T_{LM}
=
\frac{
\Delta T_1
-
\Delta T_2
}{
\ln
\left(
\frac{\Delta T_1}{\Delta T_2}
\right)
}
$$

Finalmente, el calor transferible hacia el agua de enfriamiento se estima mediante:

$$
\dot{Q}_{UA}
=
UA
\cdot
\Delta T_{LM}
$$

Para actualizar la temperatura de pared se utiliza una red térmica equivalente. La resistencia térmica del lado caliente se expresa como:

$$
R_{hot}
=
\frac{1}{h_o\cdot A_o}
+
\frac{R_{fo}}{A_o}
$$

La resistencia por conducción en la pared del tubo se calcula como:

$$
R_{wall}
=
\frac{
\ln(D_o/D_i)
}{
2
\cdot
\pi
\cdot
k_{wall}
\cdot
L
\cdot
N_{tubes}
}
$$

La resistencia térmica del lado frío se calcula como:

$$
R_{cold}
=
R_{wall}
+
\frac{R_{fi}}{A_i}
+
\frac{1}{h_i\cdot A_i}
$$

Con estas resistencias, la temperatura de pared se actualiza mediante:

$$
T_{wall}
=
\frac{
T_{sat}
\cdot
R_{cold}
+
T_{cool,mean}
\cdot
R_{hot}
}{
R_{hot}
+
R_{cold}
}
$$

donde la temperatura media del agua de enfriamiento es:

$$
T_{cool,mean}
=
\frac{
T_{cool,in}
+
T_{cool,out}
}{2}
$$

## **1.4.3.2.5 Convección natural con el exterior**

Además del calor transferido hacia el agua de enfriamiento, el modelo incluye pérdidas térmicas hacia el ambiente a través de la superficie externa de la coraza. El área externa aproximada de la coraza se calcula como:

$$
A_{shell}
=
\pi
\cdot
D_{shell}
\cdot
L
$$

La pérdida de calor hacia el ambiente se calcula mediante convección natural:

$$
\dot{Q}_{ambient}
=
h_{ambient}
\cdot
A_{shell}
\cdot
\left(
T_{surface}
-
T_{ambient}
\right)
$$

Para calcular el coeficiente de convección natural se emplean propiedades del aire evaluadas a la temperatura de película:

$$
T_{film}
=
\frac{
T_{surface}
+
T_{ambient}
}{2}
$$

La densidad del aire se calcula mediante gas ideal:

$$
\rho_{air}
=
\frac{
p_{ambient}
}{
R_{air}
\cdot
T_{film}
}
$$

La viscosidad cinemática del aire es:

$$
\nu_{air}
=
\frac{
\mu_{air}
}{
\rho_{air}
}
$$

El coeficiente de expansión térmica se aproxima como:

$$
\beta_{air}
=
\frac{1}{T_{film}}
$$

El número de Grashof se calcula mediante:

$$
Gr
=
\frac{
g
\cdot
\beta_{air}
\cdot
\left|T_{surface}-T_{ambient}\right|
\cdot
L_c^3
}{
\nu_{air}^2
}
$$

El número de Rayleigh se obtiene como:

$$
Ra
=
Gr
\cdot
Pr_{air}
$$

Con el número de Rayleigh se calcula el número de Nusselt para convección natural:

$$
Nu_{air}
=
\left[
0.60
+
\frac{
0.387
\cdot
Ra^{1/6}
}{
\left[
1
+
\left(
\frac{0.559}{Pr_{air}}
\right)^{9/16}
\right]^{8/27}
}
\right]^2
$$

Finalmente, el coeficiente de convección natural se calcula como:

$$
h_{ambient}
=
\frac{
Nu_{air}
\cdot
k_{air}
}{
L_c
}
$$

El calor total que puede retirar el condensador se expresa como la suma del calor transferible hacia el agua y la pérdida hacia el ambiente:

$$
\dot{Q}_{available}
=
\dot{Q}_{UA}
+
\dot{Q}_{ambient}
$$

Sin embargo, el calor real retirado se limita por el calor necesario para alcanzar la condición objetivo del fluido Rankine:

$$
\dot{Q}_{cond}
=
\min
\left(
\dot{Q}_{UA}
+
\dot{Q}_{ambient},
\dot{Q}_{required}
\right)
$$

La pérdida real al ambiente también se limita por el calor total retirado:

$$
\dot{Q}_{ambient,real}
=
\min
\left(
\dot{Q}_{ambient},
\dot{Q}_{cond}
\right)
$$

Por tanto, el calor que efectivamente recibe el agua de enfriamiento se calcula como:

$$
\dot{Q}_{tubes}
=
\dot{Q}_{cond}
-
\dot{Q}_{ambient,real}
$$

## **1.4.3.3 Implementación computacional**

Con base en el modelo matemático presentado en las secciones anteriores, se implementó un modelo computacional estacionario y transitorio para el condensador de coraza y tubos. El código utiliza una formulación global de cero dimensiones, por lo que no resuelve perfiles espaciales dentro de la coraza ni dentro de los tubos. En cambio, calcula variables promedio de operación que permiten evaluar la capacidad del condensador para retirar calor del fluido Rankine y entregar una salida líquida apta para la bomba.

Las entradas principales del modelo son:

- Presión de condensación: $p_{cond}$.
- Flujo másico del fluido Rankine: $\dot{m}_{rankine}$.
- Entalpía de entrada del fluido Rankine: $h_{in}$.
- Flujo másico del agua de enfriamiento: $\dot{m}_{cool}$.
- Temperatura de entrada del agua de enfriamiento: $T_{cool,in}$.
- Temperatura ambiente: $T_{ambient}$.
- Tiempo total de simulación: $t_{final}$.
- Paso temporal de simulación: $dt$.
- Tiempo de inicio de entrada del fluido Rankine: $t_{start}$.
- Constante de tiempo térmica del condensador: $\tau$.
- Geometría del condensador: $N_{tubes}$, $N_{passes}$, $D_i$, $D_o$, $D_{shell}$ y $L$.
- Propiedades térmicas y factores de ensuciamiento: $k_{wall}$, $R_{fi}$ y $R_{fo}$.

Las salidas principales del modelo son:

- Temperatura de salida del agua de enfriamiento: $T_{cool,out}$.
- Entalpía de salida del fluido Rankine: $h_{out}$.
- Calidad de salida del fluido Rankine: $x_{out}$.
- Calor total retirado del fluido Rankine: $\dot{Q}_{cond}$.
- Calor transferido al agua de enfriamiento: $\dot{Q}_{tubes}$.
- Pérdida de calor hacia el ambiente: $\dot{Q}_{ambient}$.
- Coeficiente global de transferencia de calor: $U_o$.
- Conductancia global del condensador: $UA$.
- Pérdida de presión en los tubos: $\Delta p_{tubes}$.
- Diagnóstico de operación del condensador.

El procedimiento computacional inicia con el cálculo de las propiedades aproximadas de saturación del fluido Rankine a la presión del condensador. A partir de estas propiedades, el código define la condición objetivo de salida como líquido saturado o ligeramente subenfriado, según la configuración del modelo. Las ecuaciones asociadas al cálculo de saturación, calidad, transferencia de calor, coeficiente global y pérdidas al ambiente fueron presentadas previamente en las secciones de balance de energía, números adimensionales y convección natural.

La condición principal de funcionamiento es que el fluido Rankine salga completamente condensado. Por esta razón, el código verifica que la calidad de salida sea prácticamente nula:

$$
x_{out}
\leq
10^{-6}
$$

Esta condición representa una salida líquida apta para la bomba del ciclo Rankine.

El modelo no impone directamente que el condensador logre la condición objetivo. En su lugar, compara el calor requerido para condensar el fluido Rankine con la capacidad térmica disponible del intercambiador. De esta forma, el calor retirado queda limitado por el calor que puede transferirse hacia el agua de enfriamiento y por las pérdidas térmicas hacia el ambiente:

$$
\dot{Q}_{cond}
\leq
\dot{Q}_{required}
$$

$$
\dot{Q}_{cond}
\leq
\dot{Q}_{UA}
+
\dot{Q}_{ambient}
$$

Para la simulación transitoria, se define un tiempo total de simulación, un paso temporal, una constante de tiempo térmica y un instante de inicio de entrada del fluido Rankine. Antes de este tiempo, se considera que no ingresa fluido Rankine al condensador:

$$
\dot{m}_{rankine}(t)
=
0
\qquad
t \lt t_{start}
$$

Después del tiempo de inicio, el flujo Rankine toma su valor nominal:

$$
\dot{m}_{rankine}(t)
=
\dot{m}_{rankine,nominal}
\qquad
t\geq t_{start}
$$

Las condiciones iniciales del arranque son:

$$
\dot{Q}(0)
=
0
$$

$$
T_{cool,out}(0)
=
T_{cool,in}
$$

La respuesta transitoria del condensador se aproxima mediante un modelo de primer orden para el calor retirado:

$$
\dot{Q}(t+\Delta t)
=
\dot{Q}(t)
+
\left[
\dot{Q}_{ss}
-
\dot{Q}(t)
\right]
\cdot
\frac{\Delta t}{\tau}
$$

donde $\dot{Q}_{ss}$ es el calor retirado en estado estacionario, $\Delta t$ es el paso temporal y $\tau$ es la constante de tiempo térmica del condensador.

En cada paso temporal, el código actualiza el calor retirado, la temperatura de salida del agua, la pérdida hacia el ambiente, la entalpía de salida del fluido Rankine y la calidad de salida. Finalmente, se generan tablas compactas y gráficas transitorias de temperatura, calor transferido y calidad, con el fin de verificar si el condensador alcanza una salida líquida apta para la bomba durante el arranque del equipo.


```python
# @title
# =============================================================================
# CONDENSADOR RANKINE - INTERCAMBIADOR DE CORAZA Y TUBOS
# MODELO GLOBAL CORREGIDO CON TABLAS COMPACTAS
# =============================================================================

import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display, Markdown

pd.options.display.float_format = "{:,.4f}".format


# =============================================================================
# 1. DATOS DE ENTRADA MODIFICABLES
# =============================================================================

g = 9.81

# -----------------------------------------------------------------------------
# Condiciones del fluido Rankine en la coraza
# -----------------------------------------------------------------------------

p_cond_Pa = 9.0e3
m_rankine_kg_s = 10.0
h_rankine_in_J_kg = 2400.0e3

subcooling_K = 0.0
modo_estricto_bomba = False

R_steam_J_kgK = 461.5

# -----------------------------------------------------------------------------
# Agua de enfriamiento por tubos
# -----------------------------------------------------------------------------

m_cool_kg_s = 650.0
T_cool_in_K = 298.15

rho_cool_kg_m3 = 997.0
mu_cool_Pa_s = 8.9e-4
cp_cool_J_kgK = 4180.0
k_cool_W_mK = 0.60

pinch_min_K = 5.0

# -----------------------------------------------------------------------------
# Geometría de coraza y tubos
# -----------------------------------------------------------------------------

N_tubos = 3000
N_pasos_tubos = 2

L_tubo_m = 10.0
D_i_tubo_m = 0.021
D_o_tubo_m = 0.025

D_coraza_m = 1.80
pitch_tubos_m = 0.032
arreglo_tubos = "triangular"

# -----------------------------------------------------------------------------
# Arquitectura física del equipo
# -----------------------------------------------------------------------------

tipo_tubos = "Rectos"
entrada_refrigerante = "Cabezal izquierdo superior"
salida_refrigerante = "Cabezal izquierdo inferior"
retorno_refrigerante = "Cabezal derecho"
entrada_rankine = "Boquilla superior de coraza"
salida_condensado = "Boquilla inferior de coraza"
venteo_no_condensables = "Sí"

# -----------------------------------------------------------------------------
# Material, ensuciamiento y correcciones
# -----------------------------------------------------------------------------

k_wall_W_mK = 16.0
R_fi_m2K_W = 0.00010
R_fo_m2K_W = 0.00010

F_correction = 0.95

# -----------------------------------------------------------------------------
# Propiedades aproximadas del condensado en la coraza
# -----------------------------------------------------------------------------

rho_L_kg_m3 = 990.0
mu_L_Pa_s = 6.5e-4
k_L_W_mK = 0.63

N_filas_verticales_estimadas = 20
phi_bank = N_filas_verticales_estimadas ** (-0.25)

# -----------------------------------------------------------------------------
# Pérdidas de presión lado tubos
# -----------------------------------------------------------------------------

K_entrada_salida_tubos = 2.0

# -----------------------------------------------------------------------------
# Pérdidas obligatorias hacia el ambiente
# -----------------------------------------------------------------------------

T_amb_K = 298.15
p_amb_Pa = 101325.0

R_air_J_kgK = 287.05
k_air_W_mK = 0.0262
mu_air_Pa_s = 1.85e-5
Pr_air = 0.71

# -----------------------------------------------------------------------------
# Configuración transitoria: arranque de entrada del fluido Rankine
# -----------------------------------------------------------------------------

t_final_s = 1200.0
dt_s = 1.0
tau_Q_s = 80.0

t_inicio_rankine_s = 300.0


# =============================================================================
# 2. FUNCIONES AUXILIARES GENERALES
# =============================================================================

def safe_div(num, den, default=0.0):
    if abs(den) <= 1e-12:
        return default
    return num / den


def saturation_temperature_water_antoine(p_kPa):
    p_mmHg = p_kPa * 7.50062

    A = 8.07131
    B = 1730.63
    C = 233.426

    if p_mmHg <= 0.0:
        raise ValueError("La presión de saturación debe ser positiva.")

    T_sat_C = B / (A - math.log10(p_mmHg)) - C
    return T_sat_C


def h_liquid_approx(T_K):
    return cp_cool_J_kgK * (T_K - 273.15)


def h_fg_watson_J_kg(T_sat_K):
    T_c_K = 647.096
    T_ref_K = 373.15
    h_fg_ref_J_kg = 2257.0e3

    if T_sat_K >= T_c_K:
        return 0.0

    Tr = T_sat_K / T_c_K
    Tr_ref = T_ref_K / T_c_K

    ratio = (1.0 - Tr) / max(1.0 - Tr_ref, 1e-12)

    return h_fg_ref_J_kg * (ratio ** 0.38)


def propiedades_saturacion_condensador(p_cond_local_Pa):
    p_kPa = p_cond_local_Pa / 1000.0

    T_sat_C = saturation_temperature_water_antoine(p_kPa)
    T_sat_K = T_sat_C + 273.15

    h_f_J_kg = h_liquid_approx(T_sat_K)
    h_fg_J_kg = h_fg_watson_J_kg(T_sat_K)
    h_g_J_kg = h_f_J_kg + h_fg_J_kg

    rho_V_kg_m3 = p_cond_local_Pa / max(R_steam_J_kgK * T_sat_K, 1e-12)

    return {
        "p_cond_Pa": p_cond_local_Pa,
        "T_sat_C": T_sat_C,
        "T_sat_K": T_sat_K,
        "h_f_J_kg": h_f_J_kg,
        "h_fg_J_kg": h_fg_J_kg,
        "h_g_J_kg": h_g_J_kg,
        "rho_V_kg_m3": rho_V_kg_m3
    }


def calidad_desde_h(h_J_kg, h_f_J_kg, h_g_J_kg):
    if h_J_kg <= h_f_J_kg:
        return 0.0

    if h_J_kg >= h_g_J_kg:
        return 1.0

    return (h_J_kg - h_f_J_kg) / max(h_g_J_kg - h_f_J_kg, 1e-12)


def estado_desde_h(h_J_kg, h_f_J_kg, h_g_J_kg):
    if h_J_kg < h_f_J_kg:
        return "Líquido subenfriado"

    if abs(h_J_kg - h_f_J_kg) <= 1e-6:
        return "Líquido saturado, x = 0"

    if h_f_J_kg < h_J_kg < h_g_J_kg:
        return "Mezcla líquido-vapor"

    return "Vapor seco o sobrecalentado"


def temperatura_rankine_desde_h(h_J_kg, h_f_J_kg, h_g_J_kg, T_sat_K):
    cp_vapor_aprox_J_kgK = 2300.0

    if h_J_kg <= h_f_J_kg:
        return 273.15 + h_J_kg / cp_cool_J_kgK

    if h_f_J_kg < h_J_kg < h_g_J_kg:
        return T_sat_K

    return T_sat_K + (h_J_kg - h_g_J_kg) / cp_vapor_aprox_J_kgK


# =============================================================================
# 3. GEOMETRÍA Y LADO DE TUBOS
# =============================================================================

def area_externa_total_tubos():
    return N_tubos * math.pi * D_o_tubo_m * L_tubo_m


def area_interna_total_tubos():
    return N_tubos * math.pi * D_i_tubo_m * L_tubo_m


def area_flujo_por_paso_tubos():
    tubos_por_paso = N_tubos / max(N_pasos_tubos, 1)
    A_un_tubo = math.pi * D_i_tubo_m**2 / 4.0
    return tubos_por_paso * A_un_tubo


def reynolds_tubos(m_cool_local_kg_s):
    Qv_m3_s = m_cool_local_kg_s / rho_cool_kg_m3
    A_flow_m2 = area_flujo_por_paso_tubos()

    V_m_s = safe_div(Qv_m3_s, A_flow_m2, default=0.0)
    Re = rho_cool_kg_m3 * V_m_s * D_i_tubo_m / max(mu_cool_Pa_s, 1e-12)

    return Re, V_m_s, A_flow_m2


def prandtl_agua():
    return cp_cool_J_kgK * mu_cool_Pa_s / max(k_cool_W_mK, 1e-12)


def factor_friccion_gnielinski(Re):
    Re = max(Re, 1.0)
    return (0.79 * math.log(Re) - 1.64) ** (-2.0)


def nusselt_interno_tubos(Re, Pr):
    Re = max(Re, 1.0)
    Pr = max(Pr, 1e-12)

    if Re < 2300.0:
        Nu = 3.66
        f = np.nan
        regimen = "Laminar"

    elif Re < 3000.0:
        Nu_lam = 3.66

        f_turb = factor_friccion_gnielinski(3000.0)
        Nu_turb_3000 = (
            (f_turb / 8.0)
            * max(3000.0 - 1000.0, 0.0)
            * Pr
        ) / (
            1.0
            + 12.7
            * math.sqrt(max(f_turb / 8.0, 1e-16))
            * (Pr ** (2.0 / 3.0) - 1.0)
        )

        w = (Re - 2300.0) / (3000.0 - 2300.0)
        Nu = (1.0 - w) * Nu_lam + w * Nu_turb_3000
        f = f_turb
        regimen = "Transición"

    else:
        f = factor_friccion_gnielinski(Re)

        Nu = (
            (f / 8.0)
            * max(Re - 1000.0, 0.0)
            * Pr
        ) / (
            1.0
            + 12.7
            * math.sqrt(max(f / 8.0, 1e-16))
            * (Pr ** (2.0 / 3.0) - 1.0)
        )

        regimen = "Turbulento"

    return Nu, f, regimen


def h_interno_tubos(m_cool_local_kg_s):
    Re, V_m_s, A_flow_m2 = reynolds_tubos(m_cool_local_kg_s)
    Pr = prandtl_agua()

    Nu, f, regimen = nusselt_interno_tubos(Re, Pr)

    hi = Nu * k_cool_W_mK / max(D_i_tubo_m, 1e-12)

    return {
        "hi_W_m2K": hi,
        "Re": Re,
        "Pr": Pr,
        "Nu": Nu,
        "f": f,
        "regimen": regimen,
        "V_m_s": V_m_s,
        "A_flow_m2": A_flow_m2
    }


def perdida_presion_tubos(m_cool_local_kg_s, f_tubos):
    Re, V_m_s, A_flow_m2 = reynolds_tubos(m_cool_local_kg_s)

    if np.isnan(f_tubos):
        if Re <= 0.0:
            f = 0.0
        else:
            f = 64.0 / Re
    else:
        f = f_tubos

    L_total_flujo_m = L_tubo_m * max(N_pasos_tubos, 1)

    dp_fric_Pa = (
        f
        * (L_total_flujo_m / max(D_i_tubo_m, 1e-12))
        * 0.5
        * rho_cool_kg_m3
        * V_m_s**2
    )

    dp_menores_Pa = (
        K_entrada_salida_tubos
        * 0.5
        * rho_cool_kg_m3
        * V_m_s**2
    )

    dp_total_Pa = dp_fric_Pa + dp_menores_Pa

    return {
        "dp_tubos_Pa": dp_total_Pa,
        "dp_fric_Pa": dp_fric_Pa,
        "dp_menores_Pa": dp_menores_Pa
    }


def verificacion_geometrica_basica():
    pitch_ratio = pitch_tubos_m / max(D_o_tubo_m, 1e-12)

    area_coraza_m2 = math.pi * D_coraza_m**2 / 4.0
    area_tubos_proyectada_m2 = N_tubos * math.pi * D_o_tubo_m**2 / 4.0

    fraccion_ocupacion = area_tubos_proyectada_m2 / max(area_coraza_m2, 1e-12)

    if arreglo_tubos.lower() == "triangular":
        area_celda = math.sqrt(3.0) * pitch_tubos_m**2 / 2.0
    else:
        area_celda = pitch_tubos_m**2

    area_requerida_arreglo_m2 = N_tubos * area_celda
    D_bundle_estimado_m = math.sqrt(4.0 * area_requerida_arreglo_m2 / math.pi)

    cabe_en_coraza = D_bundle_estimado_m < D_coraza_m
    pitch_ok = pitch_ratio >= 1.25

    return {
        "pitch_ratio": pitch_ratio,
        "area_coraza_m2": area_coraza_m2,
        "fraccion_ocupacion": fraccion_ocupacion,
        "D_bundle_estimado_m": D_bundle_estimado_m,
        "cabe_en_coraza": cabe_en_coraza,
        "pitch_ok": pitch_ok
    }


# =============================================================================
# 4. LADO CORAZA, U GLOBAL Y PÉRDIDAS AL AMBIENTE
# =============================================================================

def h_externo_condensacion(T_sat_K, h_fg_J_kg, rho_V_local_kg_m3, T_wall_K):
    deltaT = max(T_sat_K - T_wall_K, 1e-6)

    num = (
        k_L_W_mK**3
        * rho_L_kg_m3
        * max(rho_L_kg_m3 - rho_V_local_kg_m3, 1e-12)
        * g
        * max(h_fg_J_kg, 1e-12)
    )

    den = (
        max(mu_L_Pa_s, 1e-12)
        * max(D_o_tubo_m, 1e-12)
        * deltaT
    )

    ho_un_tubo = 0.725 * (num / den) ** 0.25
    ho_banco = ho_un_tubo * phi_bank

    return ho_banco, ho_un_tubo


def U_global_externo(hi, ho):
    Do = D_o_tubo_m
    Di = D_i_tubo_m

    R_total_m2K_W = (
        Do / (max(Di, 1e-12) * max(hi, 1e-12))
        + R_fi_m2K_W * Do / max(Di, 1e-12)
        + Do * math.log(max(Do / max(Di, 1e-12), 1.0 + 1e-12)) / (2.0 * max(k_wall_W_mK, 1e-12))
        + R_fo_m2K_W
        + 1.0 / max(ho, 1e-12)
    )

    return 1.0 / max(R_total_m2K_W, 1e-12)


def resistencias_para_T_wall(hi, ho):
    Ao_m2 = area_externa_total_tubos()
    Ai_m2 = area_interna_total_tubos()

    R_hot_K_W = (
        1.0 / (max(ho, 1e-12) * Ao_m2)
        + R_fo_m2K_W / max(Ao_m2, 1e-12)
    )

    R_wall_K_W = math.log(max(D_o_tubo_m / max(D_i_tubo_m, 1e-12), 1.0 + 1e-12)) / (
        2.0 * math.pi * max(k_wall_W_mK, 1e-12) * L_tubo_m * N_tubos
    )

    R_cold_K_W = (
        R_wall_K_W
        + R_fi_m2K_W / max(Ai_m2, 1e-12)
        + 1.0 / (max(hi, 1e-12) * Ai_m2)
    )

    return {
        "R_hot_K_W": R_hot_K_W,
        "R_wall_K_W": R_wall_K_W,
        "R_cold_K_W": R_cold_K_W
    }


def deltaT_LM_condensador(T_sat_K, T_cool_in_local_K, T_cool_out_local_K):
    DT1 = T_sat_K - T_cool_out_local_K
    DT2 = T_sat_K - T_cool_in_local_K

    if DT1 <= 0.0 or DT2 <= 0.0:
        return 0.0

    if abs(DT1 - DT2) <= 1e-12:
        return DT1

    return (DT1 - DT2) / math.log(DT1 / DT2)


def resolver_Q_UA(T_sat_K, T_cool_in_local_K, m_cool_local_kg_s, UA_W_K):
    C_cool = m_cool_local_kg_s * cp_cool_J_kgK

    if C_cool <= 0.0 or UA_W_K <= 0.0:
        return {
            "Q_UA_W": 0.0,
            "T_cool_out_UA_K": T_cool_in_local_K,
            "deltaT_LM_UA_K": 0.0,
            "limitacion_UA": "Sin transferencia por flujo o UA nulo"
        }

    T_cool_out_max_K = T_sat_K - pinch_min_K

    if T_cool_out_max_K <= T_cool_in_local_K:
        return {
            "Q_UA_W": 0.0,
            "T_cool_out_UA_K": T_cool_in_local_K,
            "deltaT_LM_UA_K": 0.0,
            "limitacion_UA": "Sin pinch disponible"
        }

    Q_max_pinch_W = C_cool * (T_cool_out_max_K - T_cool_in_local_K)

    def f_balance(Q_W):
        T_out_K = T_cool_in_local_K + Q_W / C_cool
        DTlm_K = deltaT_LM_condensador(
            T_sat_K,
            T_cool_in_local_K,
            T_out_K
        )
        return UA_W_K * DTlm_K - Q_W

    f0 = f_balance(0.0)
    fmax = f_balance(Q_max_pinch_W)

    if f0 <= 0.0:
        Q_W = 0.0
        limitacion = "UA insuficiente desde el inicio"

    elif fmax >= 0.0:
        Q_W = Q_max_pinch_W
        limitacion = "Limitado por pinch del agua de enfriamiento"

    else:
        q_low = 0.0
        q_high = Q_max_pinch_W

        for _ in range(100):
            q_mid = 0.5 * (q_low + q_high)
            f_mid = f_balance(q_mid)

            if abs(f_mid) < 1e-6:
                break

            if f_mid > 0.0:
                q_low = q_mid
            else:
                q_high = q_mid

        Q_W = 0.5 * (q_low + q_high)
        limitacion = "Limitado por UA-LMTD"

    T_cool_out_K = T_cool_in_local_K + Q_W / C_cool

    deltaT_LM_K = deltaT_LM_condensador(
        T_sat_K,
        T_cool_in_local_K,
        T_cool_out_K
    )

    return {
        "Q_UA_W": Q_W,
        "T_cool_out_UA_K": T_cool_out_K,
        "deltaT_LM_UA_K": deltaT_LM_K,
        "limitacion_UA": limitacion
    }


def perdida_calor_ambiente_obligatoria(T_superficie_K):
    A_shell_m2 = math.pi * D_coraza_m * L_tubo_m

    if T_superficie_K <= T_amb_K:
        return {
            "Q_amb_W": 0.0,
            "h_amb_W_m2K": 0.0,
            "Ra_air": 0.0,
            "Nu_air": 0.0,
            "A_shell_m2": A_shell_m2
        }

    T_film_air_K = 0.5 * (T_superficie_K + T_amb_K)

    rho_air_kg_m3 = p_amb_Pa / max(R_air_J_kgK * T_film_air_K, 1e-12)
    nu_air_m2_s = mu_air_Pa_s / max(rho_air_kg_m3, 1e-12)
    beta_air_1_K = 1.0 / max(T_film_air_K, 1e-12)

    L_char_m = D_coraza_m

    Gr = (
        g
        * beta_air_1_K
        * abs(T_superficie_K - T_amb_K)
        * L_char_m**3
        / max(nu_air_m2_s**2, 1e-24)
    )

    Ra = Gr * Pr_air

    Nu = (
        0.60
        + (
            0.387 * Ra ** (1.0 / 6.0)
        ) / (
            (1.0 + (0.559 / max(Pr_air, 1e-12)) ** (9.0 / 16.0)) ** (8.0 / 27.0)
        )
    ) ** 2.0

    h_amb = Nu * k_air_W_mK / max(L_char_m, 1e-12)

    Q_amb_W = h_amb * A_shell_m2 * (T_superficie_K - T_amb_K)

    return {
        "Q_amb_W": Q_amb_W,
        "h_amb_W_m2K": h_amb,
        "Ra_air": Ra,
        "Nu_air": Nu,
        "A_shell_m2": A_shell_m2
    }


# =============================================================================
# 5. MODELO ESTACIONARIO GLOBAL DEL CONDENSADOR
# =============================================================================

def resolver_estado_estacionario_condensador(
    p_cond_local_Pa=p_cond_Pa,
    m_rankine_local_kg_s=m_rankine_kg_s,
    h_rankine_in_local_J_kg=h_rankine_in_J_kg,
    T_cool_in_local_K=T_cool_in_K,
    m_cool_local_kg_s=m_cool_kg_s
):

    props = propiedades_saturacion_condensador(p_cond_local_Pa)

    T_sat_K = props["T_sat_K"]
    T_sat_C = props["T_sat_C"]
    h_f_J_kg = props["h_f_J_kg"]
    h_fg_J_kg = props["h_fg_J_kg"]
    h_g_J_kg = props["h_g_J_kg"]
    rho_V_local_kg_m3 = props["rho_V_kg_m3"]

    h_out_objetivo_J_kg = h_f_J_kg - subcooling_K * cp_cool_J_kgK

    x_in = calidad_desde_h(
        h_rankine_in_local_J_kg,
        h_f_J_kg,
        h_g_J_kg
    )

    estado_in = estado_desde_h(
        h_rankine_in_local_J_kg,
        h_f_J_kg,
        h_g_J_kg
    )

    Q_necesario_W = m_rankine_local_kg_s * max(
        h_rankine_in_local_J_kg - h_out_objetivo_J_kg,
        0.0
    )

    Ao_m2 = area_externa_total_tubos()
    Ai_m2 = area_interna_total_tubos()

    interno = h_interno_tubos(m_cool_local_kg_s)
    hi = interno["hi_W_m2K"]

    T_wall_K = T_cool_in_local_K + 0.70 * max(T_sat_K - T_cool_in_local_K, 1.0)
    T_cool_out_K = T_cool_in_local_K
    Q_cond_W = 0.0

    Q_UA_W = 0.0
    Q_amb_potencial_W = 0.0
    Q_amb_W = 0.0
    Q_tubos_W = 0.0

    ho = 0.0
    ho_un_tubo = 0.0
    Uo = 0.0
    UA_W_K = 0.0
    deltaT_LM_real_K = 0.0
    deltaT_LM_UA_K = 0.0
    limitacion_UA = ""

    for iteracion in range(100):

        ho, ho_un_tubo = h_externo_condensacion(
            T_sat_K,
            h_fg_J_kg,
            rho_V_local_kg_m3,
            T_wall_K
        )

        Uo = U_global_externo(hi, ho)
        UA_W_K = Uo * Ao_m2 * F_correction

        solucion_UA = resolver_Q_UA(
            T_sat_K,
            T_cool_in_local_K,
            m_cool_local_kg_s,
            UA_W_K
        )

        Q_UA_W = solucion_UA["Q_UA_W"]
        deltaT_LM_UA_K = solucion_UA["deltaT_LM_UA_K"]
        limitacion_UA = solucion_UA["limitacion_UA"]

        amb = perdida_calor_ambiente_obligatoria(T_sat_K)
        Q_amb_potencial_W = amb["Q_amb_W"]

        Q_total_transferible_W = Q_UA_W + Q_amb_potencial_W
        Q_cond_new_W = min(Q_total_transferible_W, Q_necesario_W)

        Q_amb_new_W = min(Q_amb_potencial_W, Q_cond_new_W)
        Q_tubos_new_W = max(Q_cond_new_W - Q_amb_new_W, 0.0)

        C_cool = m_cool_local_kg_s * cp_cool_J_kgK

        T_cool_out_new_K = T_cool_in_local_K + safe_div(
            Q_tubos_new_W,
            C_cool,
            default=0.0
        )

        T_cool_mean_K = 0.5 * (T_cool_in_local_K + T_cool_out_new_K)

        resistencias = resistencias_para_T_wall(hi, ho)

        R_hot_K_W = resistencias["R_hot_K_W"]
        R_cold_K_W = resistencias["R_cold_K_W"]

        T_wall_new_K = (
            T_sat_K * R_cold_K_W
            + T_cool_mean_K * R_hot_K_W
        ) / max(R_hot_K_W + R_cold_K_W, 1e-12)

        T_wall_new_K = min(T_wall_new_K, T_sat_K - 1e-3)
        T_wall_new_K = max(T_wall_new_K, T_cool_mean_K + 1e-3)

        error_T_wall = abs(T_wall_new_K - T_wall_K)
        error_T_out = abs(T_cool_out_new_K - T_cool_out_K)
        error_Q = abs(Q_cond_new_W - Q_cond_W) / max(abs(Q_cond_new_W), 1.0)

        T_wall_K = T_wall_new_K
        T_cool_out_K = T_cool_out_new_K
        Q_cond_W = Q_cond_new_W
        Q_amb_W = Q_amb_new_W
        Q_tubos_W = Q_tubos_new_W

        if max(error_T_wall, error_T_out, error_Q) < 1e-6:
            break

    h_out_real_J_kg = h_rankine_in_local_J_kg - safe_div(
        Q_cond_W,
        m_rankine_local_kg_s,
        default=0.0
    )

    x_out = calidad_desde_h(
        h_out_real_J_kg,
        h_f_J_kg,
        h_g_J_kg
    )

    estado_out = estado_desde_h(
        h_out_real_J_kg,
        h_f_J_kg,
        h_g_J_kg
    )

    salida_apta_bomba = x_out <= 1e-6

    if h_out_real_J_kg <= h_f_J_kg:
        T_cond_out_K = 273.15 + h_out_real_J_kg / cp_cool_J_kgK
    else:
        T_cond_out_K = T_sat_K

    T_rankine_in_K = temperatura_rankine_desde_h(
        h_rankine_in_local_J_kg,
        h_f_J_kg,
        h_g_J_kg,
        T_sat_K
    )

    T_rankine_out_K = temperatura_rankine_desde_h(
        h_out_real_J_kg,
        h_f_J_kg,
        h_g_J_kg,
        T_sat_K
    )

    deltaT_LM_real_K = deltaT_LM_condensador(
        T_sat_K,
        T_cool_in_local_K,
        T_cool_out_K
    )

    margen_termico_W = (Q_UA_W + Q_amb_potencial_W) - Q_necesario_W

    if Q_UA_W > 0.0:
        fraccion_area_efectiva = min(Q_tubos_W / Q_UA_W, 1.0)
    else:
        fraccion_area_efectiva = 0.0

    if Q_necesario_W > 0.0:
        porcentaje_objetivo = 100.0 * Q_cond_W / Q_necesario_W
    else:
        porcentaje_objetivo = 100.0

    if h_rankine_in_local_J_kg > h_out_objetivo_J_kg:
        m_rankine_max_kg_s = (Q_UA_W + Q_amb_potencial_W) / (
            h_rankine_in_local_J_kg - h_out_objetivo_J_kg
        )
    else:
        m_rankine_max_kg_s = np.inf

    dp = perdida_presion_tubos(
        m_cool_local_kg_s,
        interno["f"]
    )

    geom = verificacion_geometrica_basica()

    if salida_apta_bomba:
        diagnostico = "La salida del condensador es apta para la bomba: x_out ≈ 0."
    else:
        diagnostico = (
            "El condensador NO alcanza calidad x = 0. "
            "La bomba no debe recibir esta salida."
        )

    if modo_estricto_bomba and not salida_apta_bomba:
        raise ValueError(
            "\nCONDENSADOR INSUFICIENTE PARA GARANTIZAR x = 0.\n"
            f"Q_UA + Q_amb = {(Q_UA_W + Q_amb_potencial_W) / 1e6:.4f} MW\n"
            f"Q_necesario = {Q_necesario_W / 1e6:.4f} MW\n"
            f"Q_cond = {Q_cond_W / 1e6:.4f} MW\n"
            f"x salida = {x_out:.8f}\n"
            "La bomba no debe recibir mezcla líquido-vapor.\n"
        )

    return {
        "p_cond_Pa": p_cond_local_Pa,
        "T_sat_K": T_sat_K,
        "T_sat_C": T_sat_C,
        "T_wall_K": T_wall_K,
        "h_f_J_kg": h_f_J_kg,
        "h_fg_J_kg": h_fg_J_kg,
        "h_g_J_kg": h_g_J_kg,
        "rho_V_kg_m3": rho_V_local_kg_m3,
        "h_in_J_kg": h_rankine_in_local_J_kg,
        "h_out_objetivo_J_kg": h_out_objetivo_J_kg,
        "h_out_real_J_kg": h_out_real_J_kg,
        "T_rankine_in_K": T_rankine_in_K,
        "T_rankine_out_K": T_rankine_out_K,
        "x_in": x_in,
        "x_out": x_out,
        "estado_in": estado_in,
        "estado_out": estado_out,
        "m_rankine_kg_s": m_rankine_local_kg_s,
        "m_cool_kg_s": m_cool_local_kg_s,
        "T_cool_in_K": T_cool_in_local_K,
        "T_cool_out_K": T_cool_out_K,
        "T_cond_out_K": T_cond_out_K,
        "Q_UA_W": Q_UA_W,
        "Q_amb_W": Q_amb_W,
        "Q_amb_potencial_W": Q_amb_potencial_W,
        "Q_necesario_W": Q_necesario_W,
        "Q_cond_W": Q_cond_W,
        "Q_tubos_W": Q_tubos_W,
        "m_rankine_max_kg_s": m_rankine_max_kg_s,
        "salida_apta_bomba": salida_apta_bomba,
        "Ao_m2": Ao_m2,
        "Ai_m2": Ai_m2,
        "hi_W_m2K": hi,
        "ho_W_m2K": ho,
        "ho_un_tubo_W_m2K": ho_un_tubo,
        "Uo_W_m2K": Uo,
        "UA_W_K": UA_W_K,
        "deltaT_LM_real_K": deltaT_LM_real_K,
        "deltaT_LM_UA_K": deltaT_LM_UA_K,
        "Re_tubos": interno["Re"],
        "Pr_tubos": interno["Pr"],
        "Nu_tubos": interno["Nu"],
        "V_tubos_m_s": interno["V_m_s"],
        "dp_tubos_Pa": dp["dp_tubos_Pa"],
        "phi_bank": phi_bank,
        "pitch_ratio": geom["pitch_ratio"],
        "D_bundle_estimado_m": geom["D_bundle_estimado_m"],
        "cabe_en_coraza": geom["cabe_en_coraza"],
        "pitch_ok": geom["pitch_ok"],
        "margen_termico_W": margen_termico_W,
        "fraccion_area_efectiva": fraccion_area_efectiva,
        "porcentaje_objetivo": porcentaje_objetivo,
        "diagnostico": diagnostico,
        "iteraciones": iteracion + 1
    }


# =============================================================================
# 6. MODELO TRANSITORIO: ARRANQUE DE ENTRADA DEL FLUIDO RANKINE
# =============================================================================

def perfil_entrada_transitorio(t):
    if t < t_inicio_rankine_s:
        return {
            "h_rankine_in_J_kg": h_rankine_in_J_kg,
            "m_rankine_kg_s": 0.0,
            "T_cool_in_K": T_cool_in_K,
            "m_cool_kg_s": m_cool_kg_s,
            "hay_flujo_rankine": False
        }

    return {
        "h_rankine_in_J_kg": h_rankine_in_J_kg,
        "m_rankine_kg_s": m_rankine_kg_s,
        "T_cool_in_K": T_cool_in_K,
        "m_cool_kg_s": m_cool_kg_s,
        "hay_flujo_rankine": True
    }


def run_condenser_coraza_tubos_transitorio():
    tiempos = np.arange(0.0, t_final_s + dt_s, dt_s)

    Q_actual_W = 0.0
    resultados = []

    for t in tiempos:

        entrada_t = perfil_entrada_transitorio(t)

        if not entrada_t["hay_flujo_rankine"] or entrada_t["m_rankine_kg_s"] <= 0.0:

            Q_actual_W = Q_actual_W + (0.0 - Q_actual_W) * dt_s / max(tau_Q_s, 1e-12)

            resultados.append({
                "t [s]": t,
                "T fluido Rankine entrada [°C]": np.nan,
                "T fluido Rankine salida [°C]": np.nan,
                "T agua entrada [°C]": entrada_t["T_cool_in_K"] - 273.15,
                "T agua salida [°C]": entrada_t["T_cool_in_K"] - 273.15,
                "h entrada [kJ/kg]": np.nan,
                "h salida [kJ/kg]": np.nan,
                "h objetivo [kJ/kg]": np.nan,
                "x salida [-]": np.nan,
                "Q condensador [MW]": 0.0,
                "Q ambiente [MW]": 0.0,
                "Q hacia agua [MW]": 0.0,
                "m fluido Rankine [kg/s]": 0.0,
                "m agua [kg/s]": entrada_t["m_cool_kg_s"],
                "hay flujo Rankine": False,
                "salida apta bomba": False
            })

            continue

        ss_t = resolver_estado_estacionario_condensador(
            p_cond_local_Pa=p_cond_Pa,
            m_rankine_local_kg_s=entrada_t["m_rankine_kg_s"],
            h_rankine_in_local_J_kg=entrada_t["h_rankine_in_J_kg"],
            T_cool_in_local_K=entrada_t["T_cool_in_K"],
            m_cool_local_kg_s=entrada_t["m_cool_kg_s"]
        )

        Q_ss_W = ss_t["Q_cond_W"]

        Q_actual_W = Q_actual_W + (Q_ss_W - Q_actual_W) * dt_s / max(tau_Q_s, 1e-12)

        Q_transferible_t = ss_t["Q_UA_W"] + ss_t["Q_amb_potencial_W"]
        Q_actual_W = min(Q_actual_W, ss_t["Q_necesario_W"], Q_transferible_t)

        Q_amb_actual_W = min(ss_t["Q_amb_potencial_W"], Q_actual_W)
        Q_tubos_actual_W = max(Q_actual_W - Q_amb_actual_W, 0.0)

        h_out_actual_J_kg = entrada_t["h_rankine_in_J_kg"] - safe_div(
            Q_actual_W,
            entrada_t["m_rankine_kg_s"],
            default=0.0
        )

        C_cool_t = entrada_t["m_cool_kg_s"] * cp_cool_J_kgK

        T_cool_out_actual_K = entrada_t["T_cool_in_K"] + safe_div(
            Q_tubos_actual_W,
            C_cool_t,
            default=0.0
        )

        x_out_actual = calidad_desde_h(
            h_out_actual_J_kg,
            ss_t["h_f_J_kg"],
            ss_t["h_g_J_kg"]
        )

        T_rankine_in_K = temperatura_rankine_desde_h(
            entrada_t["h_rankine_in_J_kg"],
            ss_t["h_f_J_kg"],
            ss_t["h_g_J_kg"],
            ss_t["T_sat_K"]
        )

        T_rankine_out_K = temperatura_rankine_desde_h(
            h_out_actual_J_kg,
            ss_t["h_f_J_kg"],
            ss_t["h_g_J_kg"],
            ss_t["T_sat_K"]
        )

        salida_apta_actual = x_out_actual <= 1e-6

        resultados.append({
            "t [s]": t,
            "T fluido Rankine entrada [°C]": T_rankine_in_K - 273.15,
            "T fluido Rankine salida [°C]": T_rankine_out_K - 273.15,
            "T agua entrada [°C]": entrada_t["T_cool_in_K"] - 273.15,
            "T agua salida [°C]": T_cool_out_actual_K - 273.15,
            "h entrada [kJ/kg]": entrada_t["h_rankine_in_J_kg"] / 1000.0,
            "h salida [kJ/kg]": h_out_actual_J_kg / 1000.0,
            "h objetivo [kJ/kg]": ss_t["h_out_objetivo_J_kg"] / 1000.0,
            "x salida [-]": x_out_actual,
            "Q condensador [MW]": Q_actual_W / 1e6,
            "Q ambiente [MW]": Q_amb_actual_W / 1e6,
            "Q hacia agua [MW]": Q_tubos_actual_W / 1e6,
            "m fluido Rankine [kg/s]": entrada_t["m_rankine_kg_s"],
            "m agua [kg/s]": entrada_t["m_cool_kg_s"],
            "hay flujo Rankine": True,
            "salida apta bomba": salida_apta_actual
        })

    df_transitorio = pd.DataFrame(resultados)

    return {
        "df_transitorio": df_transitorio
    }


# =============================================================================
# 7. GRÁFICAS
# =============================================================================

def graficar_transitorio_condensador(df):

    # -------------------------------------------------------------------------
    # Figura 1. Temperaturas unificadas
    # -------------------------------------------------------------------------
    plt.figure(figsize=(9, 5))

    plt.plot(
        df["t [s]"],
        df["T fluido Rankine entrada [°C]"],
        color="tab:red",
        linestyle="-",
        linewidth=2.2,
        label="T fluido Rankine entrada"
    )

    plt.plot(
        df["t [s]"],
        df["T fluido Rankine salida [°C]"],
        color="tab:green",
        linestyle="--",
        linewidth=2.2,
        label="T fluido Rankine salida"
    )

    plt.plot(
        df["t [s]"],
        df["T agua entrada [°C]"],
        color="tab:blue",
        linestyle="-",
        linewidth=2.2,
        label="T agua entrada"
    )

    plt.plot(
        df["t [s]"],
        df["T agua salida [°C]"],
        color="tab:orange",
        linestyle="--",
        linewidth=2.2,
        label="T agua salida"
    )

    plt.axvline(
        t_inicio_rankine_s,
        color="black",
        linestyle="--",
        linewidth=2.0,
        label="Inicio entrada fluido Rankine"
    )

    plt.xlim(left=0)
    plt.ylim(bottom=20)

    plt.xlabel("Tiempo [s]")
    plt.ylabel("Temperatura [°C]")
    plt.title("Figura 1. Temperaturas de entrada y salida del condensador")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

    # -------------------------------------------------------------------------
    # Figura 2. Calor transferido
    # -------------------------------------------------------------------------
    fig, ax1 = plt.subplots(figsize=(9, 5))

    ax1.plot(
        df["t [s]"],
        df["Q hacia agua [MW]"],
        color="tab:blue",
        linestyle="-",
        linewidth=2.4,
        label="Q transferido al agua"
    )

    ax1.axvline(
        t_inicio_rankine_s,
        color="black",
        linestyle="--",
        linewidth=2.0,
        label="Inicio entrada fluido Rankine"
    )

    ax1.set_xlim(left=0)
    ax1.set_ylim(bottom=0)

    ax1.set_xlabel("Tiempo [s]")
    ax1.set_ylabel("Calor transferido al agua [MW]")
    ax1.set_title("Figura 2. Calor transferido en el condensador")
    ax1.grid(True)

    ax2 = ax1.twinx()

    ax2.plot(
        df["t [s]"],
        df["Q ambiente [MW]"] * 1000.0,
        color="tab:red",
        linestyle="--",
        linewidth=2.4,
        label="Q perdido al ambiente"
    )

    ax2.set_ylim(bottom=0)
    ax2.set_ylabel("Pérdida al ambiente [kW]")

    lineas_1, etiquetas_1 = ax1.get_legend_handles_labels()
    lineas_2, etiquetas_2 = ax2.get_legend_handles_labels()

    ax1.legend(
        lineas_1 + lineas_2,
        etiquetas_1 + etiquetas_2,
        loc="best"
    )

    plt.tight_layout()
    plt.show()

    # -------------------------------------------------------------------------
    # Figura 3. Calidad de salida
    # -------------------------------------------------------------------------
    plt.figure(figsize=(9, 5))

    plt.plot(
        df["t [s]"],
        df["x salida [-]"],
        color="tab:purple",
        linestyle="-",
        linewidth=2.4,
        label="Calidad de salida del fluido Rankine"
    )

    plt.axvline(
        t_inicio_rankine_s,
        color="black",
        linestyle="--",
        linewidth=2.0,
        label="Inicio entrada fluido Rankine"
    )

    plt.xlim(left=0)
    plt.ylim(0, 1.05)

    plt.xlabel("Tiempo [s]")
    plt.ylabel("Calidad x [-]")
    plt.title("Figura 3. Calidad de salida del fluido Rankine")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()


# =============================================================================
# 8. EJECUCIÓN DEL CONDENSADOR SOLO
# =============================================================================

resultado_estacionario = resolver_estado_estacionario_condensador()
resultado_transitorio = run_condenser_coraza_tubos_transitorio()

df_transitorio = resultado_transitorio["df_transitorio"]


# =============================================================================
# 9. IMPRESIÓN COMPACTA DE RESULTADOS
# =============================================================================

display(Markdown("# Intercambiador Condensador de Coraza y Tubos"))

df_resumen_estacionario = pd.DataFrame([{
    "p_cond [kPa]": resultado_estacionario["p_cond_Pa"] / 1000.0,
    "T_sat [°C]": resultado_estacionario["T_sat_C"],
    "T_pared [°C]": resultado_estacionario["T_wall_K"] - 273.15,

    "m_Rankine [kg/s]": resultado_estacionario["m_rankine_kg_s"],
    "m_agua [kg/s]": resultado_estacionario["m_cool_kg_s"],

    "T_Rankine_in [°C]": resultado_estacionario["T_rankine_in_K"] - 273.15,
    "T_Rankine_out [°C]": resultado_estacionario["T_rankine_out_K"] - 273.15,
    "T_agua_in [°C]": resultado_estacionario["T_cool_in_K"] - 273.15,
    "T_agua_out [°C]": resultado_estacionario["T_cool_out_K"] - 273.15,

    "h_in [kJ/kg]": resultado_estacionario["h_in_J_kg"] / 1000.0,
    "h_out [kJ/kg]": resultado_estacionario["h_out_real_J_kg"] / 1000.0,
    "h_objetivo [kJ/kg]": resultado_estacionario["h_out_objetivo_J_kg"] / 1000.0,

    "x_in [-]": resultado_estacionario["x_in"],
    "x_out [-]": resultado_estacionario["x_out"],
    "Salida apta bomba": resultado_estacionario["salida_apta_bomba"]
}])

display(Markdown("## 1. Resumen estacionario del condensador"))
display(df_resumen_estacionario.T)


df_resumen_termico = pd.DataFrame([{
    "Q_condensador [MW]": resultado_estacionario["Q_cond_W"] / 1e6,
    "Q_hacia_agua [MW]": resultado_estacionario["Q_tubos_W"] / 1e6,
    "Q_ambiente [MW]": resultado_estacionario["Q_amb_W"] / 1e6,
    "Q_ambiente [kW]": resultado_estacionario["Q_amb_W"] / 1000.0,
    "Q_ambiente/Q_condensador [%]": (
        resultado_estacionario["Q_amb_W"]
        / max(resultado_estacionario["Q_cond_W"], 1e-12)
        * 100.0
    ),
    "Q_necesario [MW]": resultado_estacionario["Q_necesario_W"] / 1e6,
    "Q_UA [MW]": resultado_estacionario["Q_UA_W"] / 1e6,

    "Uo [W/m2K]": resultado_estacionario["Uo_W_m2K"],
    "UA [W/K]": resultado_estacionario["UA_W_K"],
    "A_externa [m2]": resultado_estacionario["Ao_m2"],
    "DeltaT_LM [K]": resultado_estacionario["deltaT_LM_real_K"],

    "Margen térmico [MW]": resultado_estacionario["margen_termico_W"] / 1e6
}])

display(Markdown("## 2. Resumen térmico del condensador"))
display(df_resumen_termico.T)


df_resumen_diseno = pd.DataFrame([{
    "N_tubos [-]": N_tubos,
    "N_pasos_tubos [-]": N_pasos_tubos,
    "L_tubo [m]": L_tubo_m,
    "D_i [m]": D_i_tubo_m,
    "D_o [m]": D_o_tubo_m,
    "D_coraza [m]": D_coraza_m,

    "Re_tubos [-]": resultado_estacionario["Re_tubos"],
    "Pr_tubos [-]": resultado_estacionario["Pr_tubos"],
    "Nu_tubos [-]": resultado_estacionario["Nu_tubos"],

    "h_i [W/m2K]": resultado_estacionario["hi_W_m2K"],
    "h_o [W/m2K]": resultado_estacionario["ho_W_m2K"],
    "phi_bank [-]": resultado_estacionario["phi_bank"],

    "Delta_p_tubos [kPa]": resultado_estacionario["dp_tubos_Pa"] / 1000.0
}])

display(Markdown("## 3. Resumen de diseño del intercambiador"))
display(df_resumen_diseno.T)


columnas_transitorio_compacto = [
    "t [s]",
    "T fluido Rankine entrada [°C]",
    "T fluido Rankine salida [°C]",
    "T agua entrada [°C]",
    "T agua salida [°C]",
    "Q condensador [MW]",
    "Q hacia agua [MW]",
    "Q ambiente [MW]",
    "x salida [-]",
    "m fluido Rankine [kg/s]",
    "salida apta bomba"
]

df_transitorio_compacto = df_transitorio[columnas_transitorio_compacto].copy()

display(Markdown("## 4. Primeras 10 filas del arranque transitorio"))
display(df_transitorio_compacto.head(10).T)

display(Markdown("## 5. Últimas 10 filas del arranque transitorio"))
display(df_transitorio_compacto.tail(10).T)


df_transitorio_con_flujo = df_transitorio_compacto[
    df_transitorio_compacto["m fluido Rankine [kg/s]"] > 0
].copy()

if len(df_transitorio_con_flujo) > 0:

    df_propiedades_medias = pd.DataFrame([{
        "T_Rankine_in_media [°C]": df_transitorio_con_flujo["T fluido Rankine entrada [°C]"].mean(),
        "T_Rankine_out_media [°C]": df_transitorio_con_flujo["T fluido Rankine salida [°C]"].mean(),
        "T_agua_out_media [°C]": df_transitorio_con_flujo["T agua salida [°C]"].mean(),

        "Q_condensador_media [MW]": df_transitorio_con_flujo["Q condensador [MW]"].mean(),
        "Q_hacia_agua_media [MW]": df_transitorio_con_flujo["Q hacia agua [MW]"].mean(),
        "Q_ambiente_media [MW]": df_transitorio_con_flujo["Q ambiente [MW]"].mean(),
        "Q_ambiente_media [kW]": df_transitorio_con_flujo["Q ambiente [MW]"].mean() * 1000.0,

        "x_salida_media [-]": df_transitorio_con_flujo["x salida [-]"].mean(),
        "x_salida_final [-]": df_transitorio_con_flujo["x salida [-]"].iloc[-1],

        "m_Rankine [kg/s]": resultado_estacionario["m_rankine_kg_s"],
        "m_agua [kg/s]": resultado_estacionario["m_cool_kg_s"]
    }])

else:

    df_propiedades_medias = pd.DataFrame([{
        "Mensaje": "No hubo entrada de fluido Rankine durante la simulación."
    }])

display(Markdown("## 6. Propiedades medias calculadas en régimen con flujo Rankine"))
display(df_propiedades_medias.T)


display(Markdown("## 7. Gráficas del arranque"))
graficar_transitorio_condensador(df_transitorio)


# =============================================================================
# 10. DICCIONARIO FINAL PARA INTEGRACIÓN FUTURA AL CICLO
# =============================================================================

condenser = {
    "tipo": "coraza_y_tubos",
    "fluido_coraza": "fluido_rankine_condensando",
    "fluido_tubos": "agua_enfriamiento",
    "p_cond_Pa": resultado_estacionario["p_cond_Pa"],
    "T_sat_K": resultado_estacionario["T_sat_K"],
    "T_sat_C": resultado_estacionario["T_sat_C"],
    "T_wall_K": resultado_estacionario["T_wall_K"],
    "h_in_J_kg": resultado_estacionario["h_in_J_kg"],
    "h_out_J_kg": resultado_estacionario["h_out_real_J_kg"],
    "h_out_objetivo_J_kg": resultado_estacionario["h_out_objetivo_J_kg"],
    "T_rankine_in_K": resultado_estacionario["T_rankine_in_K"],
    "T_rankine_out_K": resultado_estacionario["T_rankine_out_K"],
    "x_in": resultado_estacionario["x_in"],
    "x_out": resultado_estacionario["x_out"],
    "m_rankine_kg_s": resultado_estacionario["m_rankine_kg_s"],
    "m_cool_kg_s": resultado_estacionario["m_cool_kg_s"],
    "T_cool_in_K": resultado_estacionario["T_cool_in_K"],
    "T_cool_out_K": resultado_estacionario["T_cool_out_K"],
    "T_cond_out_K": resultado_estacionario["T_cond_out_K"],
    "Q_UA_W": resultado_estacionario["Q_UA_W"],
    "Q_amb_W": resultado_estacionario["Q_amb_W"],
    "Q_necesario_W": resultado_estacionario["Q_necesario_W"],
    "Q_cond_W": resultado_estacionario["Q_cond_W"],
    "Q_tubos_W": resultado_estacionario["Q_tubos_W"],
    "Ao_m2": resultado_estacionario["Ao_m2"],
    "hi_W_m2K": resultado_estacionario["hi_W_m2K"],
    "ho_W_m2K": resultado_estacionario["ho_W_m2K"],
    "Uo_W_m2K": resultado_estacionario["Uo_W_m2K"],
    "UA_W_K": resultado_estacionario["UA_W_K"],
    "deltaT_LM_real_K": resultado_estacionario["deltaT_LM_real_K"],
    "Re_tubos": resultado_estacionario["Re_tubos"],
    "dp_tubos_Pa": resultado_estacionario["dp_tubos_Pa"],
    "salida_apta_bomba": resultado_estacionario["salida_apta_bomba"],
    "df_transitorio": df_transitorio
}
```

# **1.5 Caldera**

**Figura 1.5.** Esquemático del intercambiador de serpentín de la caldera.



## **1.5.1 Intercambiador de serpentín**

El intercambiador de calor de la caldera se modela como un generador de vapor con serpentín helicoidal, ubicado entre la salida de la bomba Rankine y la entrada de la turbina de vapor. Su función principal es recuperar energía térmica de los gases de escape provenientes de la turbina de gas y transferirla al fluido Rankine que circula por el interior del serpentín. De esta manera, el agua presurizada puede calentarse, evaporarse y, si la energía disponible es suficiente, alcanzar una condición de vapor sobrecalentado antes de ingresar a la turbina de vapor.

A diferencia del condensador desarrollado en la Sección 1.4, este equipo no retira calor del fluido Rankine, sino que le suministra energía. Por esta razón, el criterio de operación cambia: mientras que en el condensador se busca una calidad de salida igual a cero, en la caldera se busca una calidad de salida igual a uno y, preferiblemente, una temperatura de vapor superior a la temperatura de saturación.

## **1.5.2 Arquitectura del intercambiador**

La arquitectura considerada corresponde a un serpentín helicoidal instalado dentro de una caldera o cámara de recuperación de calor. Los gases de escape de la turbina de gas circulan por el exterior del serpentín, mientras que el fluido Rankine circula por el interior de los tubos. Esta configuración permite transferir calor desde los gases calientes hacia el agua presurizada que viene de la bomba.

La geometría del modelo se define mediante el número de serpentines en paralelo, la longitud de cada tubo, el diámetro interno, el diámetro externo, el diámetro medio de curvatura del serpentín, el diámetro de la carcasa y la longitud de la caldera. A partir de estos parámetros se calculan el área externa de transferencia de calor, el área interna de flujo, el área libre disponible para el paso de los gases y el área externa de la carcasa expuesta al ambiente.

## **1.5.3 Modelo termofluídico de la caldera**

## **1.5.3.1 Balance de masa**

El balance de masa general ya fue presentado en la Sección 1.4 para el condensador. En el caso del intercambiador caldera serpentín, se aplica la misma idea de conservación de masa para el fluido Rankine y para los gases de escape.

Para el fluido Rankine, se considera que no hay acumulación ni extracción de masa dentro del serpentín. Por tanto:

$$
\dot{m}_{rankine,in}
=
\dot{m}_{rankine,out}
=
\dot{m}_{rankine}
$$

Para los gases de escape, se considera que el flujo másico también permanece constante entre la entrada y la salida de la caldera durante la operación estacionaria:

$$
\dot{m}_{gas,in}
=
\dot{m}_{gas,out}
=
\dot{m}_{gas}
$$

Durante la simulación transitoria, el arranque del equipo se representa mediante la activación del flujo de gases calientes. Antes del tiempo de inicio no ingresan gases de escape al intercambiador:

$$
\dot{m}_{gas}(t)
=
0
\qquad
t \lt t_{start}
$$

Después del tiempo de inicio, los gases toman su valor nominal:

$$
\dot{m}_{gas}(t)
=
\dot{m}_{gas,nominal}
\qquad
t\geq t_{start}
$$

Esta condición representa un arranque operativo en el que la transferencia de calor significativa comienza cuando llegan los gases calientes provenientes de la turbina de gas.

## **1.5.3.1.1 Hidráulica del agua en el serpentín**

El fluido Rankine circula por el interior del serpentín. Inicialmente, se considera como agua líquida presurizada a la salida de la bomba. El caudal volumétrico se calcula mediante:

$$
\dot{V}_{rankine}
=
\frac{\dot{m}_{rankine}}{\rho_{water}}
$$

El área interna total disponible para el flujo se calcula considerando el número de serpentines en paralelo:

$$
A_{flow}
=
N_{serpentines}
\cdot
\frac{\pi D_i^2}{4}
$$

Con esta área se calcula la velocidad media del fluido dentro de los tubos:

$$
V_{rankine}
=
\frac{\dot{V}_{rankine}}{A_{flow}}
$$

La pérdida de presión interna se estima a partir de una suma entre pérdidas por fricción y pérdidas menores. La pérdida por fricción se calcula como:

$$
\Delta p_{fric}
=
f
\cdot
\frac{L_{tube}}{D_i}
\cdot
\frac{\rho_{water}\cdot V_{rankine}^{2}}{2}
$$

La pérdida menor se calcula mediante:

$$
\Delta p_{minor}
=
K
\cdot
\frac{\rho_{water}\cdot V_{rankine}^{2}}{2}
$$

Por tanto, la pérdida total de presión en el lado interno es:

$$
\Delta p_{internal}
=
\Delta p_{fric}
+
\Delta p_{minor}
$$

## **1.5.3.1.2 Dinámica de los gases de escape**

Los gases de escape circulan por la región externa al serpentín. Para estimar su velocidad media, se calcula primero el área libre de paso dentro de la carcasa:

$$
A_{free}
=
f_{free}
\cdot
\frac{\pi D_{shell}^{2}}{4}
$$

donde $f_{free}$ representa la fracción del área de la carcasa disponible para el flujo de gases.

La velocidad media de los gases se obtiene mediante:

$$
V_{gas}
=
\frac{\dot{m}_{gas}}
{\rho_{gas}\cdot A_{free}}
$$

Con esta velocidad se calcula el número de Reynolds externo:

$$
Re_{gas}
=
\frac{
\rho_{gas}
\cdot
V_{gas}
\cdot
D_o
}{
\mu_{gas}
}
$$

Este número permite estimar el régimen del flujo externo y calcular el coeficiente convectivo entre los gases calientes y la superficie exterior del serpentín.

## **1.5.3.2 Balance de energía**

El balance de energía general ya fue presentado en la Sección 1.4. Para la caldera, este balance se interpreta como una transferencia de energía desde los gases calientes hacia el fluido Rankine, considerando además una pérdida térmica hacia el ambiente.

El calor total retirado de los gases se reparte entre el calor transferido al fluido Rankine y el calor perdido hacia el ambiente:

$$
\dot{Q}_{gas,total}
=
\dot{Q}_{rankine}
+
\dot{Q}_{ambient}
$$

donde $\dot{Q}_{rankine}$ es el calor efectivamente recibido por el fluido Rankine y $\dot{Q}_{ambient}$ es la pérdida térmica desde la carcasa hacia el ambiente.

## **1.5.3.2.1 Vapor de agua**

En este equipo, el estado de salida del fluido Rankine no se calcula únicamente con una ecuación de calentamiento sensible, porque el fluido puede cambiar de fase. Por esta razón, el modelo utiliza la entalpía como variable principal.

La entalpía de entrada se aproxima como líquido comprimido a la presión de caldera:

$$
h_{in}
=
c_{p,water}
\cdot
\left(
T_{rankine,in}
-
273.15
\right)
+
\frac{
p_{boiler}
-
p_{ambient}
}{
\rho_{water}
}
$$

A la presión de caldera, se calcula la temperatura de saturación y las propiedades aproximadas del agua. La entalpía de vapor saturado se obtiene como:

$$
h_g
=
h_f
+
h_{fg}
$$

Si la temperatura objetivo del vapor es mayor que la temperatura de saturación, la salida deseada corresponde a vapor sobrecalentado. En ese caso, la entalpía objetivo se calcula como:

$$
h_{target}
=
h_g
+
c_{p,steam}
\cdot
\left(
T_{steam,target}
-
T_{sat}
\right)
$$

El calor requerido para llevar el fluido Rankine desde su estado de entrada hasta la condición objetivo es:

$$
\dot{Q}_{required}
=
\dot{m}_{rankine}
\cdot
\left(
h_{target}
-
h_{in}
\right)
$$

Una vez calculado el calor realmente transferido al fluido Rankine, la entalpía de salida se determina mediante:

$$
h_{out}
=
h_{in}
+
\frac{
\dot{Q}_{rankine}
}{
\dot{m}_{rankine}
}
$$

Con esta entalpía se determina el estado de salida. Si $h_{out} \lt h_f$, el fluido permanece como líquido comprimido o subenfriado. Si $h_f<h_{out} \lt h_g$, el fluido se encuentra como mezcla líquido-vapor. Si $h_{out}\geq h_g$, el fluido se encuentra como vapor seco o sobrecalentado.

La calidad de salida se calcula como:

$$
x_{out}
=
\frac{
h_{out}
-
h_f
}{
h_g
-
h_f
}
$$

Para que la salida sea apta para la turbina de vapor, se requiere que el fluido alcance una condición de vapor seco o sobrecalentado:

$$
x_{out}
=
1
$$

Además, para cumplir la condición objetivo de vapor sobrecalentado, se verifica:

$$
h_{out}
\geq
h_{target}
$$

## **1.5.3.2.2 Gases de escape**

Los gases de escape son la fuente de calor del intercambiador. Sin embargo, no se permite que se enfríen indefinidamente, porque esto podría producir condiciones no deseadas en la chimenea o en los ductos de salida. Por esta razón, el modelo impone una temperatura mínima de chimenea:

$$
T_{gas,out}
\geq
T_{stack,min}
$$

Con esta restricción, el calor máximo disponible en los gases se calcula como:

$$
\dot{Q}_{gas,available}
=
\dot{m}_{gas}
\cdot
c_{p,gas}
\cdot
\left(
T_{gas,in}
-
T_{stack,min}
\right)
$$

La temperatura de salida de los gases se actualiza mediante el balance energético:

$$
T_{gas,out}
=
T_{gas,in}
-
\frac{
\dot{Q}_{rankine}
+
\dot{Q}_{ambient}
}{
\dot{m}_{gas}
\cdot
c_{p,gas}
}
$$

Además, el modelo verifica una diferencia mínima de temperatura entre el lado caliente y el lado frío. Esta condición corresponde al pinch térmico:

$$
\Delta T_{pinch}
=
\min
\left(
T_{gas,in}
-
T_{rankine,out},
\,
T_{gas,out}
-
T_{rankine,in}
\right)
$$

La operación se considera térmicamente viable si:

$$
\Delta T_{pinch}
\geq
\Delta T_{pinch,min}
$$

## **1.5.3.2.3 Números adimensionales**

El cálculo de números adimensionales sigue la misma lógica usada en el condensador, pero adaptada a la geometría helicoidal del serpentín.

Para el flujo interno del fluido Rankine, el número de Reynolds se calcula como:

$$
Re
=
\frac{
\rho_{water}
\cdot
V_{rankine}
\cdot
D_i
}{
\mu_{water}
}
$$

El número de Prandtl se calcula mediante:

$$
Pr
=
\frac{
c_{p,water}
\cdot
\mu_{water}
}{
k_{water}
}
$$

Debido a que el tubo es curvo, se calcula adicionalmente el número de Dean:

$$
De
=
Re
\cdot
\sqrt{
\frac{D_i}{D_{coil}}
}
$$

Este número representa el efecto de la curvatura del serpentín sobre el flujo interno. En un tubo helicoidal, la curvatura genera flujos secundarios que pueden aumentar la mezcla y mejorar la transferencia de calor.

Para flujo laminar, el modelo utiliza:

$$
Nu
=
3.66
$$

Para flujo turbulento, el número de Nusselt se calcula con una corrección por curvatura:

$$
Nu
=
0.023
\cdot
Re^{0.8}
\cdot
Pr^{0.4}
\cdot
\left(
1
+
3.5
\cdot
\frac{De}{Re}
\right)
$$

El coeficiente convectivo interno se obtiene como:

$$
h_i
=
\frac{
Nu
\cdot
k_{water}
}{
D_i
}
$$

Para el lado externo de gases, el número de Nusselt se aproxima mediante:

$$
Nu_{gas}
=
0.27
\cdot
Re_{gas}^{0.63}
\cdot
Pr_{gas}^{0.36}
$$

El coeficiente convectivo externo se calcula como:

$$
h_o
=
\frac{
Nu_{gas}
\cdot
k_{gas}
}{
D_o
}
$$

La transferencia de calor entre los gases y el fluido Rankine se calcula mediante un coeficiente global referido al área externa del tubo. La formulación conserva la misma estructura de resistencias térmicas usada en el condensador, pero cambia la interpretación del lado externo: aquí no hay condensación externa, sino convección de gases calientes sobre el serpentín.

El coeficiente global se calcula como:

$$
\frac{1}{U_o}
=
\frac{D_o}{D_i\cdot h_i}
+
R_{fi}
\cdot
\frac{D_o}{D_i}
+
\frac{
D_o
\cdot
\ln(D_o/D_i)
}{
2
\cdot
k_{wall}
}
+
R_{fo}
+
\frac{1}{h_o}
$$

El área externa total de transferencia de calor es:

$$
A_o
=
N_{serpentines}
\cdot
\pi
\cdot
D_o
\cdot
L_{tube}
$$

La conductancia global del intercambiador se calcula mediante:

$$
UA
=
U_o
\cdot
A_o
\cdot
F
$$

El calor transferible por el serpentín se estima mediante el método de diferencia media logarítmica de temperatura:

$$
\dot{Q}_{UA}
=
UA
\cdot
\Delta T_{LM}
$$

Para una aproximación en contracorriente, las diferencias de temperatura se definen como:

$$
\Delta T_1
=
T_{gas,in}
-
T_{rankine,out}
$$

$$
\Delta T_2
=
T_{gas,out}
-
T_{rankine,in}
$$

Por tanto, la diferencia media logarítmica se calcula como:

$$
\Delta T_{LM}
=
\frac{
\Delta T_1
-
\Delta T_2
}{
\ln
\left(
\frac{\Delta T_1}{\Delta T_2}
\right)
}
$$

## **1.5.3.2.4 Convección natural con el exterior**

Al igual que en el condensador, el modelo incluye pérdidas térmicas obligatorias hacia el ambiente. En este caso, dichas pérdidas ocurren desde la superficie externa de la carcasa de la caldera hacia el aire circundante.

El área externa aproximada de la carcasa se calcula como:

$$
A_{shell}
=
\pi
\cdot
D_{shell}
\cdot
L_{shell}
$$

La pérdida de calor hacia el ambiente se calcula mediante convección natural:

$$
\dot{Q}_{ambient}
=
h_{ambient}
\cdot
A_{shell}
\cdot
\left(
T_{surface}
-
T_{ambient}
\right)
$$

El coeficiente $h_{ambient}$ se calcula con la misma lógica de convección natural utilizada para el condensador. La diferencia principal es que, en este equipo, la temperatura superficial de la carcasa se aproxima a partir de la temperatura media de los gases dentro de la caldera.

## **1.5.3.3 Implementación computacional**

Con base en el modelo matemático presentado en la Sección 1.5.2, se implementó un modelo computacional estacionario y transitorio para el intercambiador caldera serpentín. El código utiliza una formulación global de cero dimensiones, por lo que no resuelve perfiles espaciales dentro del serpentín ni divide el equipo en zonas separadas de economizador, evaporador y sobrecalentador. En cambio, el cambio de estado del fluido Rankine se representa mediante su entalpía, lo cual permite identificar si la salida corresponde a líquido comprimido, mezcla líquido-vapor, vapor seco o vapor sobrecalentado.

Las entradas principales del modelo son:

- Presión de caldera: $p_{boiler}$.
- Temperatura de entrada del fluido Rankine: $T_{rankine,in}$.
- Flujo másico del fluido Rankine: $\dot{m}_{rankine}$.
- Temperatura de entrada de los gases: $T_{gas,in}$.
- Flujo másico de gases: $\dot{m}_{gas}$.
- Temperatura mínima de chimenea: $T_{stack,min}$.
- Temperatura objetivo de vapor: $T_{steam,target}$.
- Parámetros geométricos del serpentín: $N_{serpentines}$, $L_{tube}$, $D_i$, $D_o$, $D_{coil}$ y $D_{shell}$.
- Propiedades térmicas y factores de ensuciamiento: $k_{wall}$, $R_{fi}$ y $R_{fo}$.

Las salidas principales del modelo son:

- Temperatura de salida del fluido Rankine: $T_{rankine,out}$.
- Entalpía de salida del fluido Rankine: $h_{out}$.
- Calidad de salida del fluido Rankine: $x_{out}$.
- Temperatura de salida de los gases: $T_{gas,out}$.
- Calor transferido al fluido Rankine: $\dot{Q}_{rankine}$.
- Calor perdido hacia el ambiente: $\dot{Q}_{ambient}$.
- Calor total retirado de los gases: $\dot{Q}_{gas,total}$.
- Coeficiente global de transferencia de calor: $U_o$.
- Conductancia global del intercambiador: $UA$.
- Pérdida de presión interna del serpentín: $\Delta p_{internal}$.
- Diagnóstico de operación del equipo.

El procedimiento computacional inicia con el cálculo de las propiedades aproximadas de saturación del agua a la presión de caldera. Luego, el código estima la entalpía de entrada del fluido Rankine, define la entalpía objetivo de salida y calcula el calor requerido para alcanzar la condición de vapor deseada. Después, evalúa las condiciones hidráulicas internas del serpentín y el flujo externo de gases para obtener los coeficientes convectivos, el coeficiente global de transferencia de calor y la conductancia global del equipo. Las ecuaciones asociadas a estos cálculos fueron presentadas previamente en las secciones de hidráulica, números adimensionales y transferencia de calor.

El modelo no impone directamente que el fluido Rankine alcance la condición objetivo. En su lugar, compara los límites físicos del sistema: el calor requerido por el fluido Rankine, el calor disponible en los gases, el calor transferible por el serpentín, la temperatura mínima de chimenea y el pinch térmico mínimo. De esta manera, el calor entregado al fluido Rankine queda determinado por la capacidad real del intercambiador y por las restricciones de operación.

La condición principal de funcionamiento es que el fluido Rankine salga como vapor apto para la turbina. Por esta razón, el código verifica que la calidad de salida alcance el valor de vapor seco:

$$
x_{out}
=
1
$$

Adicionalmente, si se define una condición objetivo de vapor sobrecalentado, el modelo verifica que la entalpía de salida sea mayor o igual a la entalpía objetivo:

$$
h_{out}
\geq
h_{target}
$$

Para la simulación transitoria, se define un tiempo total de simulación, un paso temporal, una constante de tiempo térmica y un instante de inicio de entrada de gases calientes. Antes de este tiempo, el flujo de gases se considera nulo y no hay transferencia significativa de calor hacia el fluido Rankine. Después del tiempo de inicio, los gases toman su valor nominal:

$$
\dot{m}_{gas}(t)
=
0
\qquad
t \lt t_{start}
$$

$$
\dot{m}_{gas}(t)
=
\dot{m}_{gas,nominal}
\qquad
t\geq t_{start}
$$

La evolución temporal del calor transferido se aproxima mediante una respuesta de primer orden:

$$
\dot{Q}(t+\Delta t)
=
\dot{Q}(t)
+
\left[
\dot{Q}_{ss}
-
\dot{Q}(t)
\right]
\cdot
\frac{\Delta t}{\tau}
$$

donde $\dot{Q}_{ss}$ es el calor transferido en estado estacionario, $\Delta t$ es el paso temporal y $\tau$ es la constante de tiempo térmica del intercambiador. Con el calor instantáneo, el código actualiza la temperatura de salida del fluido Rankine, la temperatura de salida de los gases, la pérdida hacia el ambiente, la entalpía de salida y la calidad de salida.


```python
# @title
# =============================================================================
# INTERCAMBIADOR CALDERA SERPENTÍN
# GENERADOR DE VAPOR CON GASES DE ESCAPE
# =============================================================================

import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display, Markdown

pd.options.display.float_format = "{:,.4f}".format


# =============================================================================
# 1. DATOS DE ENTRADA MODIFICABLES
# =============================================================================

g = 9.81

# -----------------------------------------------------------------------------
# Fluido caliente: gases de escape de la turbina de gas
# -----------------------------------------------------------------------------

T_gas_in_K = 850.0
m_gas_kg_s = 135.5

cp_gas_J_kgK = 1100.0
rho_gas_kg_m3 = 0.55
mu_gas_Pa_s = 3.5e-5
k_gas_W_mK = 0.055
Pr_gas = cp_gas_J_kgK * mu_gas_Pa_s / k_gas_W_mK

T_stack_min_K = 420.0
pinch_min_K = 20.0

# -----------------------------------------------------------------------------
# Fluido frío: agua presurizada del ciclo Rankine dentro del serpentín
# -----------------------------------------------------------------------------

p_boiler_Pa = 12.0e6
T_rankine_in_K = 320.0
m_rankine_kg_s = 10.0

T_steam_target_K = 813.15
modo_estricto_turbina = False

# -----------------------------------------------------------------------------
# Propiedades aproximadas del agua/vapor
# -----------------------------------------------------------------------------

cp_water_J_kgK = 4180.0
cp_steam_J_kgK = 2300.0
rho_water_kg_m3 = 997.0
mu_water_Pa_s = 8.9e-4
k_water_W_mK = 0.60
R_steam_J_kgK = 461.5

# -----------------------------------------------------------------------------
# Geometría del serpentín
# -----------------------------------------------------------------------------

N_serpentines = 120
L_tube_m = 60.0
D_i_tube_m = 0.035
D_o_tube_m = 0.042
D_coil_m = 1.20

D_shell_m = 4.00
L_shell_m = 10.0
shell_free_area_factor = 0.65

# -----------------------------------------------------------------------------
# Material, ensuciamiento y factor de corrección
# -----------------------------------------------------------------------------

k_wall_W_mK = 16.0
R_fi_m2K_W = 0.00010
R_fo_m2K_W = 0.00015
F_correction = 0.95

# -----------------------------------------------------------------------------
# Pérdidas de presión internas
# -----------------------------------------------------------------------------

K_minor_internal = 3.0

# -----------------------------------------------------------------------------
# Pérdidas al ambiente
# -----------------------------------------------------------------------------

T_ambient_K = 298.15
p_ambient_Pa = 101325.0

R_air_J_kgK = 287.05
k_air_W_mK = 0.0262
mu_air_Pa_s = 1.85e-5
Pr_air = 0.71

# -----------------------------------------------------------------------------
# Configuración transitoria
# -----------------------------------------------------------------------------

t_final_s = 1200.0
dt_s = 1.0
tau_Q_s = 100.0
t_start_s = 300.0

iniciar_en_regimen = False


# =============================================================================
# 2. FUNCIONES AUXILIARES GENERALES
# =============================================================================

def safe_div(num, den, default=0.0):
    if abs(den) <= 1e-12:
        return default
    return num / den


def clamp(x, xmin, xmax):
    return max(xmin, min(x, xmax))


# =============================================================================
# 3. PROPIEDADES APROXIMADAS DE SATURACIÓN DEL AGUA
# =============================================================================

def saturation_pressure_water_iapws_MPa(T_K):
    T_c = 647.096
    p_c_MPa = 22.064

    a1 = -7.85951783
    a2 = 1.84408259
    a3 = -11.7866497
    a4 = 22.6807411
    a5 = -15.9618719
    a6 = 1.80122502

    if T_K <= 273.16:
        return 0.000611657

    if T_K >= T_c:
        return p_c_MPa

    theta = 1.0 - T_K / T_c

    exponent = (
        T_c / T_K
        * (
            a1 * theta
            + a2 * theta**1.5
            + a3 * theta**3.0
            + a4 * theta**3.5
            + a5 * theta**4.0
            + a6 * theta**7.5
        )
    )

    return p_c_MPa * math.exp(exponent)


def saturation_temperature_from_pressure(p_Pa):
    p_MPa = p_Pa / 1e6

    T_low = 273.16
    T_high = 647.096

    if p_MPa <= saturation_pressure_water_iapws_MPa(T_low):
        return T_low

    if p_MPa >= saturation_pressure_water_iapws_MPa(T_high):
        return T_high

    for _ in range(100):
        T_mid = 0.5 * (T_low + T_high)
        p_mid = saturation_pressure_water_iapws_MPa(T_mid)

        if p_mid < p_MPa:
            T_low = T_mid
        else:
            T_high = T_mid

    return 0.5 * (T_low + T_high)


def h_liquid_compressed_approx(T_K, p_Pa):
    h_sensible = cp_water_J_kgK * (T_K - 273.15)
    h_pressure = (p_Pa - 101325.0) / rho_water_kg_m3

    return h_sensible + h_pressure


def h_fg_watson_J_kg(T_sat_K):
    T_c = 647.096
    T_ref = 373.15
    h_fg_ref = 2257.0e3

    if T_sat_K >= T_c:
        return 0.0

    ratio = (
        (1.0 - T_sat_K / T_c)
        /
        max(1.0 - T_ref / T_c, 1e-12)
    )

    return h_fg_ref * ratio**0.38


def propiedades_saturacion_rankine(p_Pa):
    T_sat_K = saturation_temperature_from_pressure(p_Pa)

    h_f = h_liquid_compressed_approx(T_sat_K, p_Pa)
    h_fg = h_fg_watson_J_kg(T_sat_K)
    h_g = h_f + h_fg

    rho_v = p_Pa / max(R_steam_J_kgK * T_sat_K, 1e-12)

    return {
        "T_sat_K": T_sat_K,
        "T_sat_C": T_sat_K - 273.15,
        "h_f_J_kg": h_f,
        "h_fg_J_kg": h_fg,
        "h_g_J_kg": h_g,
        "rho_v_kg_m3": rho_v
    }


def h_target_steam_from_temperature(T_target_K, props):
    T_sat_K = props["T_sat_K"]
    h_g = props["h_g_J_kg"]

    if T_target_K <= T_sat_K:
        return h_g

    return h_g + cp_steam_J_kgK * (T_target_K - T_sat_K)


def estado_rankine_desde_h(h_J_kg, props):
    h_f = props["h_f_J_kg"]
    h_g = props["h_g_J_kg"]
    T_sat_K = props["T_sat_K"]

    if h_J_kg <= h_f:
        x = 0.0
        T_K = 273.15 + h_J_kg / cp_water_J_kgK
        estado = "Líquido comprimido / subenfriado"

    elif h_f < h_J_kg < h_g:
        x = (h_J_kg - h_f) / max(h_g - h_f, 1e-12)
        T_K = T_sat_K
        estado = "Mezcla líquido-vapor"

    else:
        x = 1.0
        T_K = T_sat_K + (h_J_kg - h_g) / cp_steam_J_kgK
        estado = "Vapor seco o sobrecalentado"

    return {
        "T_K": T_K,
        "T_C": T_K - 273.15,
        "x": clamp(x, 0.0, 1.0),
        "estado": estado
    }


# =============================================================================
# 4. GEOMETRÍA DEL SERPENTÍN
# =============================================================================

def area_externa_total():
    return N_serpentines * math.pi * D_o_tube_m * L_tube_m


def area_interna_total():
    return N_serpentines * math.pi * D_i_tube_m * L_tube_m


def volumen_interno_total():
    return (
        N_serpentines
        * math.pi
        * D_i_tube_m**2
        * L_tube_m
        / 4.0
    )


def area_flujo_interno_total():
    return (
        N_serpentines
        * math.pi
        * D_i_tube_m**2
        / 4.0
    )


def area_libre_gases():
    area_shell = math.pi * D_shell_m**2 / 4.0
    return shell_free_area_factor * area_shell


def area_externa_caldera_ambiente():
    return math.pi * D_shell_m * L_shell_m


# =============================================================================
# 5. FLUJO INTERNO DEL FLUIDO RANKINE EN EL SERPENTÍN
# =============================================================================

def reynolds_serpentin(m_dot):
    Qv = m_dot / rho_water_kg_m3
    A = area_flujo_interno_total()

    V = safe_div(Qv, A)

    Re = (
        rho_water_kg_m3
        * V
        * D_i_tube_m
        / max(mu_water_Pa_s, 1e-12)
    )

    return Re, V


def prandtl_agua():
    return (
        cp_water_J_kgK
        * mu_water_Pa_s
        / max(k_water_W_mK, 1e-12)
    )


def dean_number(Re):
    return Re * math.sqrt(D_i_tube_m / max(D_coil_m, 1e-12))


def nusselt_serpentin(Re, Pr):
    De = dean_number(Re)

    if Re < 2300.0:
        Nu = 3.66
        regimen = "Laminar"
    else:
        Nu = (
            0.023
            * Re**0.8
            * Pr**0.4
            * (
                1.0
                + 3.5
                * (De / max(Re, 1e-12))
            )
        )
        regimen = "Turbulento"

    return Nu, De, regimen


def h_interno_serpentin(m_dot):
    Re, V = reynolds_serpentin(m_dot)
    Pr = prandtl_agua()

    Nu, De, regimen = nusselt_serpentin(Re, Pr)

    hi = Nu * k_water_W_mK / max(D_i_tube_m, 1e-12)

    return {
        "hi_W_m2K": hi,
        "Nu": Nu,
        "Pr": Pr,
        "Re": Re,
        "De": De,
        "V_m_s": V,
        "regimen": regimen
    }


def perdida_presion_interna(m_dot):
    interno = h_interno_serpentin(m_dot)

    Re = interno["Re"]
    V = interno["V_m_s"]

    if Re <= 0:
        f = 0.0
    elif Re < 2300:
        f = 64.0 / Re
    else:
        f = (0.79 * math.log(Re) - 1.64) ** (-2.0)

    dp_fric = (
        f
        * (L_tube_m / max(D_i_tube_m, 1e-12))
        * 0.5
        * rho_water_kg_m3
        * V**2
    )

    dp_minor = (
        K_minor_internal
        * 0.5
        * rho_water_kg_m3
        * V**2
    )

    return {
        "dp_internal_Pa": dp_fric + dp_minor,
        "dp_fric_Pa": dp_fric,
        "dp_minor_Pa": dp_minor,
        "f": f
    }


# =============================================================================
# 6. FLUJO EXTERNO DE GASES
# =============================================================================

def reynolds_externo_gases(m_gas):
    if m_gas <= 0.0:
        return 0.0, 0.0

    Qv = m_gas / max(rho_gas_kg_m3, 1e-12)
    A = area_libre_gases()

    V = safe_div(Qv, A)

    Re = (
        rho_gas_kg_m3
        * V
        * D_o_tube_m
        / max(mu_gas_Pa_s, 1e-12)
    )

    return Re, V


def h_externo_gases(m_gas):
    Re, V = reynolds_externo_gases(m_gas)

    if Re <= 0.0:
        Nu = 0.0
        ho = 0.0
    else:
        Nu = 0.27 * Re**0.63 * Pr_gas**0.36
        ho = Nu * k_gas_W_mK / max(D_o_tube_m, 1e-12)

    return {
        "ho_W_m2K": ho,
        "Nu": Nu,
        "Pr": Pr_gas,
        "Re": Re,
        "V_m_s": V
    }


# =============================================================================
# 7. COEFICIENTE GLOBAL Y LMTD
# =============================================================================

def U_global(hi, ho):
    Di = D_i_tube_m
    Do = D_o_tube_m

    if hi <= 0.0 or ho <= 0.0:
        return 0.0

    R_total = (
        Do / (Di * hi)
        + R_fi_m2K_W * Do / Di
        + Do * math.log(Do / Di) / (2.0 * k_wall_W_mK)
        + R_fo_m2K_W
        + 1.0 / ho
    )

    return 1.0 / max(R_total, 1e-12)


def deltaT_lm_counterflow(Th_in, Th_out, Tc_in, Tc_out):
    DT1 = Th_in - Tc_out
    DT2 = Th_out - Tc_in

    if DT1 <= 0.0 or DT2 <= 0.0:
        return 0.0

    if abs(DT1 - DT2) <= 1e-12:
        return DT1

    return (DT1 - DT2) / math.log(DT1 / DT2)


def pinch_minimo_real(Th_in, Th_out, Tc_in, Tc_out):
    return min(
        Th_in - Tc_out,
        Th_out - Tc_in
    )


# =============================================================================
# 8. PÉRDIDAS AL AMBIENTE
# =============================================================================

def h_ambiente_conveccion_natural(T_surface_K):
    if T_surface_K <= T_ambient_K:
        return {
            "h_ambient_W_m2K": 0.0,
            "Ra_air": 0.0,
            "Nu_air": 0.0
        }

    T_film_K = 0.5 * (T_surface_K + T_ambient_K)

    rho_air = p_ambient_Pa / max(R_air_J_kgK * T_film_K, 1e-12)
    nu_air = mu_air_Pa_s / max(rho_air, 1e-12)
    beta_air = 1.0 / max(T_film_K, 1e-12)

    L_c = D_shell_m

    Gr = (
        g
        * beta_air
        * abs(T_surface_K - T_ambient_K)
        * L_c**3
        / max(nu_air**2, 1e-24)
    )

    Ra = Gr * Pr_air

    Nu = (
        0.60
        + (
            0.387 * Ra**(1.0 / 6.0)
        )
        / (
            (
                1.0
                + (0.559 / max(Pr_air, 1e-12))**(9.0 / 16.0)
            )
            ** (8.0 / 27.0)
        )
    ) ** 2.0

    h_amb = Nu * k_air_W_mK / max(L_c, 1e-12)

    return {
        "h_ambient_W_m2K": h_amb,
        "Ra_air": Ra,
        "Nu_air": Nu
    }


def perdida_ambiente_para_Q(Q_to_rankine_W, T_gas_in_local_K, m_gas_local_kg_s):
    if m_gas_local_kg_s <= 0.0:
        return {
            "Q_ambient_W": 0.0,
            "T_surface_K": T_ambient_K,
            "h_ambient_W_m2K": 0.0,
            "Ra_air": 0.0,
            "Nu_air": 0.0
        }

    Ch = m_gas_local_kg_s * cp_gas_J_kgK

    Q_amb = 0.0
    T_surface_K = T_ambient_K

    for _ in range(40):
        T_gas_out_est = T_gas_in_local_K - (Q_to_rankine_W + Q_amb) / max(Ch, 1e-12)
        T_surface_K = 0.5 * (T_gas_in_local_K + T_gas_out_est)

        h_amb = h_ambiente_conveccion_natural(T_surface_K)
        A_shell = area_externa_caldera_ambiente()

        Q_amb_new = (
            h_amb["h_ambient_W_m2K"]
            * A_shell
            * max(T_surface_K - T_ambient_K, 0.0)
        )

        if abs(Q_amb_new - Q_amb) / max(abs(Q_amb_new), 1.0) < 1e-6:
            Q_amb = Q_amb_new
            break

        Q_amb = 0.5 * Q_amb + 0.5 * Q_amb_new

    return {
        "Q_ambient_W": max(Q_amb, 0.0),
        "T_surface_K": T_surface_K,
        "h_ambient_W_m2K": h_amb["h_ambient_W_m2K"],
        "Ra_air": h_amb["Ra_air"],
        "Nu_air": h_amb["Nu_air"]
    }


# =============================================================================
# 9. MODELO ESTACIONARIO DEL GENERADOR DE VAPOR
# =============================================================================

def evaluar_estado_para_Q(Q_to_rankine_W, datos_base):
    props = datos_base["props"]

    h_in = datos_base["h_rankine_in_J_kg"]
    m_rankine = datos_base["m_rankine_kg_s"]
    m_gas = datos_base["m_gas_kg_s"]
    T_gas_in = datos_base["T_gas_in_K"]
    T_rankine_in = datos_base["T_rankine_in_K"]
    UA = datos_base["UA_W_K"]

    Ch = m_gas * cp_gas_J_kgK

    h_out = h_in + safe_div(Q_to_rankine_W, m_rankine)
    estado_out = estado_rankine_desde_h(h_out, props)

    T_rankine_out = estado_out["T_K"]

    ambiente = perdida_ambiente_para_Q(
        Q_to_rankine_W,
        T_gas_in,
        m_gas
    )

    Q_ambient = ambiente["Q_ambient_W"]
    Q_gas_total = Q_to_rankine_W + Q_ambient

    T_gas_out = T_gas_in - safe_div(Q_gas_total, Ch)

    DTlm = deltaT_lm_counterflow(
        T_gas_in,
        T_gas_out,
        T_rankine_in,
        T_rankine_out
    )

    Q_UA = UA * DTlm

    pinch_real = pinch_minimo_real(
        T_gas_in,
        T_gas_out,
        T_rankine_in,
        T_rankine_out
    )

    return {
        "Q_to_rankine_W": Q_to_rankine_W,
        "Q_ambient_W": Q_ambient,
        "Q_gas_total_W": Q_gas_total,
        "Q_UA_W": Q_UA,
        "T_gas_out_K": T_gas_out,
        "T_rankine_out_K": T_rankine_out,
        "h_rankine_out_J_kg": h_out,
        "x_out": estado_out["x"],
        "estado_out": estado_out["estado"],
        "DTlm_K": DTlm,
        "pinch_real_K": pinch_real,
        "T_surface_K": ambiente["T_surface_K"],
        "h_ambient_W_m2K": ambiente["h_ambient_W_m2K"],
        "Ra_air": ambiente["Ra_air"],
        "Nu_air": ambiente["Nu_air"]
    }


def resolver_Q_estacionario(datos_base):
    Q_required = datos_base["Q_required_W"]
    Q_gas_available = datos_base["Q_gas_available_W"]
    UA = datos_base["UA_W_K"]

    if Q_required <= 0.0 or Q_gas_available <= 0.0 or UA <= 0.0:
        return evaluar_estado_para_Q(0.0, datos_base)

    Q_high = min(Q_required, Q_gas_available)

    def balance(Q):
        estado = evaluar_estado_para_Q(Q, datos_base)

        if estado["Q_gas_total_W"] > Q_gas_available:
            return -abs(estado["Q_gas_total_W"] - Q_gas_available)

        if estado["pinch_real_K"] < pinch_min_K:
            return -abs(pinch_min_K - estado["pinch_real_K"])

        return estado["Q_UA_W"] - Q

    f_high = balance(Q_high)

    if f_high >= 0.0:
        return evaluar_estado_para_Q(Q_high, datos_base)

    Q_low = 0.0

    for _ in range(120):
        Q_mid = 0.5 * (Q_low + Q_high)
        f_mid = balance(Q_mid)

        if abs(f_mid) < 1e-3:
            break

        if f_mid >= 0.0:
            Q_low = Q_mid
        else:
            Q_high = Q_mid

    Q_sol = 0.5 * (Q_low + Q_high)

    return evaluar_estado_para_Q(Q_sol, datos_base)


def run_boiler_serpentin_estacionario(
    T_gas_in_local_K=T_gas_in_K,
    m_gas_local_kg_s=m_gas_kg_s,
    T_rankine_in_local_K=T_rankine_in_K,
    m_rankine_local_kg_s=m_rankine_kg_s,
    p_boiler_local_Pa=p_boiler_Pa,
    T_steam_target_local_K=T_steam_target_K
):

    props = propiedades_saturacion_rankine(p_boiler_local_Pa)

    h_in = h_liquid_compressed_approx(
        T_rankine_in_local_K,
        p_boiler_local_Pa
    )

    h_target = h_target_steam_from_temperature(
        T_steam_target_local_K,
        props
    )

    Q_required = (
        m_rankine_local_kg_s
        * max(h_target - h_in, 0.0)
    )

    Q_gas_available = (
        m_gas_local_kg_s
        * cp_gas_J_kgK
        * max(T_gas_in_local_K - T_stack_min_K, 0.0)
    )

    interno = h_interno_serpentin(m_rankine_local_kg_s)
    externo = h_externo_gases(m_gas_local_kg_s)

    hi = interno["hi_W_m2K"]
    ho = externo["ho_W_m2K"]

    U = U_global(hi, ho)
    A = area_externa_total()
    UA = U * A * F_correction

    datos_base = {
        "props": props,
        "h_rankine_in_J_kg": h_in,
        "h_target_J_kg": h_target,
        "T_rankine_in_K": T_rankine_in_local_K,
        "m_rankine_kg_s": m_rankine_local_kg_s,
        "T_gas_in_K": T_gas_in_local_K,
        "m_gas_kg_s": m_gas_local_kg_s,
        "Q_required_W": Q_required,
        "Q_gas_available_W": Q_gas_available,
        "UA_W_K": UA
    }

    solucion = resolver_Q_estacionario(datos_base)

    h_out = solucion["h_rankine_out_J_kg"]

    estado_in = estado_rankine_desde_h(h_in, props)
    estado_target = estado_rankine_desde_h(h_target, props)

    cumple_vapor_objetivo = h_out >= h_target * (1.0 - 1e-6)

    cumple_vapor_turbina = (
        solucion["x_out"] >= 1.0
        and h_out >= props["h_g_J_kg"]
    )

    cumple_stack = solucion["T_gas_out_K"] >= T_stack_min_K
    cumple_pinch = solucion["pinch_real_K"] >= pinch_min_K

    dp_int = perdida_presion_interna(m_rankine_local_kg_s)

    if cumple_vapor_objetivo:
        diagnostico = "El serpentín alcanza la condición objetivo de vapor para la turbina."
    elif cumple_vapor_turbina:
        diagnostico = "El serpentín produce vapor seco, pero no alcanza completamente la temperatura objetivo."
    else:
        diagnostico = "El serpentín no alcanza vapor apto para turbina; falta área, gases o temperatura."

    if modo_estricto_turbina and not cumple_vapor_objetivo:
        raise ValueError(
            "\nEL SERPENTÍN NO ALCANZA LA CONDICIÓN OBJETIVO DE VAPOR.\n"
            f"Q requerido = {Q_required/1e6:.4f} MW\n"
            f"Q al Rankine = {solucion['Q_to_rankine_W']/1e6:.4f} MW\n"
            f"x salida = {solucion['x_out']:.6f}\n"
            f"T salida Rankine = {solucion['T_rankine_out_K'] - 273.15:.2f} °C\n"
        )

    return {
        "p_boiler_Pa": p_boiler_local_Pa,
        "T_sat_K": props["T_sat_K"],
        "T_sat_C": props["T_sat_C"],
        "h_f_J_kg": props["h_f_J_kg"],
        "h_fg_J_kg": props["h_fg_J_kg"],
        "h_g_J_kg": props["h_g_J_kg"],
        "rho_v_kg_m3": props["rho_v_kg_m3"],

        "T_gas_in_K": T_gas_in_local_K,
        "T_gas_out_K": solucion["T_gas_out_K"],
        "T_stack_min_K": T_stack_min_K,
        "m_gas_kg_s": m_gas_local_kg_s,

        "T_rankine_in_K": T_rankine_in_local_K,
        "T_rankine_out_K": solucion["T_rankine_out_K"],
        "T_steam_target_K": T_steam_target_local_K,
        "m_rankine_kg_s": m_rankine_local_kg_s,

        "h_rankine_in_J_kg": h_in,
        "h_rankine_out_J_kg": h_out,
        "h_target_J_kg": h_target,

        "estado_in": estado_in["estado"],
        "estado_out": solucion["estado_out"],
        "estado_target": estado_target["estado"],

        "x_in": estado_in["x"],
        "x_out": solucion["x_out"],

        "Q_required_W": Q_required,
        "Q_to_rankine_W": solucion["Q_to_rankine_W"],
        "Q_ambient_W": solucion["Q_ambient_W"],
        "Q_gas_total_W": solucion["Q_gas_total_W"],
        "Q_gas_available_W": Q_gas_available,
        "Q_UA_W": solucion["Q_UA_W"],

        "U_W_m2K": U,
        "UA_W_K": UA,
        "A_external_m2": A,
        "A_internal_m2": area_interna_total(),
        "DTlm_K": solucion["DTlm_K"],
        "pinch_real_K": solucion["pinch_real_K"],

        "hi_W_m2K": hi,
        "ho_W_m2K": ho,

        "Re_internal": interno["Re"],
        "De_internal": interno["De"],
        "Pr_internal": interno["Pr"],
        "Nu_internal": interno["Nu"],
        "V_internal_m_s": interno["V_m_s"],
        "regimen_internal": interno["regimen"],

        "Re_external": externo["Re"],
        "Pr_external": externo["Pr"],
        "Nu_external": externo["Nu"],
        "V_external_m_s": externo["V_m_s"],

        "h_ambient_W_m2K": solucion["h_ambient_W_m2K"],
        "T_surface_K": solucion["T_surface_K"],
        "Ra_air": solucion["Ra_air"],
        "Nu_air": solucion["Nu_air"],

        "dp_internal_Pa": dp_int["dp_internal_Pa"],

        "cumple_vapor_objetivo": cumple_vapor_objetivo,
        "cumple_vapor_turbina": cumple_vapor_turbina,
        "cumple_stack": cumple_stack,
        "cumple_pinch": cumple_pinch,
        "diagnostico": diagnostico
    }


# =============================================================================
# 10. TABLAS COMPACTAS
# =============================================================================

def crear_tablas_resultado(resultado):

    df_resumen_estacionario = pd.DataFrame([{
        "p_boiler [MPa]": resultado["p_boiler_Pa"] / 1e6,
        "T_sat [°C]": resultado["T_sat_C"],

        "m_gases [kg/s]": resultado["m_gas_kg_s"],
        "T_gases_in [°C]": resultado["T_gas_in_K"] - 273.15,
        "T_gases_out [°C]": resultado["T_gas_out_K"] - 273.15,
        "T_stack_min [°C]": resultado["T_stack_min_K"] - 273.15,

        "m_Rankine [kg/s]": resultado["m_rankine_kg_s"],
        "T_Rankine_in [°C]": resultado["T_rankine_in_K"] - 273.15,
        "T_Rankine_out [°C]": resultado["T_rankine_out_K"] - 273.15,
        "T_vapor_objetivo [°C]": resultado["T_steam_target_K"] - 273.15,

        "h_in [kJ/kg]": resultado["h_rankine_in_J_kg"] / 1000.0,
        "h_out [kJ/kg]": resultado["h_rankine_out_J_kg"] / 1000.0,
        "h_objetivo [kJ/kg]": resultado["h_target_J_kg"] / 1000.0,

        "x_in [-]": resultado["x_in"],
        "x_out [-]": resultado["x_out"],
        "Estado salida": resultado["estado_out"],
        "Salida apta turbina": resultado["cumple_vapor_turbina"],
        "Cumple vapor objetivo": resultado["cumple_vapor_objetivo"]
    }])

    df_resumen_termico = pd.DataFrame([{
        "Q_requerido [MW]": resultado["Q_required_W"] / 1e6,
        "Q_Rankine [MW]": resultado["Q_to_rankine_W"] / 1e6,
        "Q_ambiente [MW]": resultado["Q_ambient_W"] / 1e6,
        "Q_ambiente [kW]": resultado["Q_ambient_W"] / 1000.0,
        "Q_gases_total [MW]": resultado["Q_gas_total_W"] / 1e6,
        "Q_gases_disponible [MW]": resultado["Q_gas_available_W"] / 1e6,

        "U_global [W/m2K]": resultado["U_W_m2K"],
        "UA [W/K]": resultado["UA_W_K"],
        "A_externa [m2]": resultado["A_external_m2"],
        "DeltaT_LM [K]": resultado["DTlm_K"],
        "Pinch real [K]": resultado["pinch_real_K"]
    }])

    df_resumen_diseno = pd.DataFrame([{
        "N_serpentines [-]": N_serpentines,
        "L_tubo [m]": L_tube_m,
        "D_i [m]": D_i_tube_m,
        "D_o [m]": D_o_tube_m,
        "D_coil [m]": D_coil_m,
        "D_shell [m]": D_shell_m,

        "Re_interno [-]": resultado["Re_internal"],
        "Dean_interno [-]": resultado["De_internal"],
        "Pr_interno [-]": resultado["Pr_internal"],
        "Nu_interno [-]": resultado["Nu_internal"],

        "Re_gases [-]": resultado["Re_external"],
        "Nu_gases [-]": resultado["Nu_external"],

        "h_i [W/m2K]": resultado["hi_W_m2K"],
        "h_o [W/m2K]": resultado["ho_W_m2K"],
        "Delta_p_interna [kPa]": resultado["dp_internal_Pa"] / 1000.0
    }])

    df_diagnostico = pd.DataFrame([{
        "Cumple temperatura mínima chimenea": resultado["cumple_stack"],
        "Cumple pinch mínimo": resultado["cumple_pinch"],
        "Cumple vapor para turbina": resultado["cumple_vapor_turbina"],
        "Cumple vapor objetivo": resultado["cumple_vapor_objetivo"],
        "Diagnóstico": resultado["diagnostico"]
    }])

    return {
        "df_resumen_estacionario": df_resumen_estacionario,
        "df_resumen_termico": df_resumen_termico,
        "df_resumen_diseno": df_resumen_diseno,
        "df_diagnostico": df_diagnostico
    }


# =============================================================================
# 11. MODELO TRANSITORIO
# =============================================================================

def perfil_entrada_transitorio(t):
    if t < t_start_s:
        return {
            "T_gas_in_K": T_gas_in_K,
            "m_gas_kg_s": 0.0,
            "T_rankine_in_K": T_rankine_in_K,
            "m_rankine_kg_s": m_rankine_kg_s,
            "hay_gases": False
        }

    return {
        "T_gas_in_K": T_gas_in_K,
        "m_gas_kg_s": m_gas_kg_s,
        "T_rankine_in_K": T_rankine_in_K,
        "m_rankine_kg_s": m_rankine_kg_s,
        "hay_gases": True
    }


def run_boiler_serpentin_transitorio():
    tiempos = np.arange(0.0, t_final_s + dt_s, dt_s)

    resultado_ss_nominal = run_boiler_serpentin_estacionario()

    if iniciar_en_regimen:
        Q_actual = resultado_ss_nominal["Q_to_rankine_W"]
    else:
        Q_actual = 0.0

    resultados = []

    for t in tiempos:

        entrada = perfil_entrada_transitorio(t)

        if not entrada["hay_gases"] or entrada["m_gas_kg_s"] <= 0.0:

            Q_actual = Q_actual + (0.0 - Q_actual) * dt_s / max(tau_Q_s, 1e-12)

            props = propiedades_saturacion_rankine(p_boiler_Pa)

            h_in = h_liquid_compressed_approx(
                entrada["T_rankine_in_K"],
                p_boiler_Pa
            )

            estado_out = estado_rankine_desde_h(h_in, props)

            resultados.append({
                "t [s]": t,
                "T gases entrada [°C]": np.nan,
                "T gases salida [°C]": np.nan,
                "T Rankine entrada [°C]": entrada["T_rankine_in_K"] - 273.15,
                "T Rankine salida [°C]": estado_out["T_C"],
                "Q Rankine [MW]": 0.0,
                "Q ambiente [MW]": 0.0,
                "Q ambiente [kW]": 0.0,
                "Q gases total [MW]": 0.0,
                "x salida [-]": estado_out["x"],
                "m gases [kg/s]": 0.0,
                "Estado salida": estado_out["estado"],
                "Cumple vapor turbina": False
            })

            continue

        ss = run_boiler_serpentin_estacionario(
            T_gas_in_local_K=entrada["T_gas_in_K"],
            m_gas_local_kg_s=entrada["m_gas_kg_s"],
            T_rankine_in_local_K=entrada["T_rankine_in_K"],
            m_rankine_local_kg_s=entrada["m_rankine_kg_s"],
            p_boiler_local_Pa=p_boiler_Pa,
            T_steam_target_local_K=T_steam_target_K
        )

        Q_ss = ss["Q_to_rankine_W"]

        Q_actual = Q_actual + (Q_ss - Q_actual) * dt_s / max(tau_Q_s, 1e-12)

        Q_actual = min(
            Q_actual,
            ss["Q_required_W"],
            ss["Q_gas_available_W"]
        )

        props = propiedades_saturacion_rankine(p_boiler_Pa)

        h_in = h_liquid_compressed_approx(
            entrada["T_rankine_in_K"],
            p_boiler_Pa
        )

        h_out = h_in + safe_div(
            Q_actual,
            entrada["m_rankine_kg_s"]
        )

        estado_out = estado_rankine_desde_h(h_out, props)

        ambiente = perdida_ambiente_para_Q(
            Q_actual,
            entrada["T_gas_in_K"],
            entrada["m_gas_kg_s"]
        )

        Q_ambient_actual = min(
            ambiente["Q_ambient_W"],
            max(ss["Q_gas_available_W"] - Q_actual, 0.0)
        )

        Q_gas_total_actual = Q_actual + Q_ambient_actual

        T_gas_out_actual = (
            entrada["T_gas_in_K"]
            -
            safe_div(
                Q_gas_total_actual,
                entrada["m_gas_kg_s"] * cp_gas_J_kgK
            )
        )

        cumple_vapor_turbina = (
            estado_out["x"] >= 1.0
            and h_out >= props["h_g_J_kg"]
        )

        resultados.append({
            "t [s]": t,
            "T gases entrada [°C]": entrada["T_gas_in_K"] - 273.15,
            "T gases salida [°C]": T_gas_out_actual - 273.15,
            "T Rankine entrada [°C]": entrada["T_rankine_in_K"] - 273.15,
            "T Rankine salida [°C]": estado_out["T_C"],
            "Q Rankine [MW]": Q_actual / 1e6,
            "Q ambiente [MW]": Q_ambient_actual / 1e6,
            "Q ambiente [kW]": Q_ambient_actual / 1000.0,
            "Q gases total [MW]": Q_gas_total_actual / 1e6,
            "x salida [-]": estado_out["x"],
            "m gases [kg/s]": entrada["m_gas_kg_s"],
            "Estado salida": estado_out["estado"],
            "Cumple vapor turbina": cumple_vapor_turbina
        })

    return pd.DataFrame(resultados)


# =============================================================================
# 12. GRÁFICAS
# =============================================================================

def graficar_transitorio_boiler(df):

    nombre_equipo = "Intercambiador Caldera Serpentín"

    # -------------------------------------------------------------------------
    # Figura 1. Temperaturas
    # -------------------------------------------------------------------------
    plt.figure(figsize=(10, 5))

    plt.plot(
        df["t [s]"],
        df["T gases entrada [°C]"],
        color="tab:red",
        linestyle="-",
        linewidth=2.2,
        label="T gases entrada"
    )

    plt.plot(
        df["t [s]"],
        df["T gases salida [°C]"],
        color="tab:orange",
        linestyle="--",
        linewidth=2.2,
        label="T gases salida / chimenea"
    )

    plt.plot(
        df["t [s]"],
        df["T Rankine entrada [°C]"],
        color="tab:blue",
        linestyle="-",
        linewidth=2.2,
        label="T Rankine entrada"
    )

    plt.plot(
        df["t [s]"],
        df["T Rankine salida [°C]"],
        color="tab:green",
        linestyle="--",
        linewidth=2.2,
        label="T Rankine salida"
    )

    plt.axvline(
        t_start_s,
        color="black",
        linestyle="--",
        linewidth=2.0,
        label="Inicio gases calientes"
    )

    plt.xlabel("Tiempo [s]")
    plt.ylabel("Temperatura [°C]")
    plt.title(f"Figura 1. Temperaturas de entrada y salida - {nombre_equipo}")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

    # -------------------------------------------------------------------------
    # Figura 2. Calor transferido con doble eje vertical
    # -------------------------------------------------------------------------
    fig, ax1 = plt.subplots(figsize=(10, 5))

    linea_rankine, = ax1.plot(
        df["t [s]"],
        df["Q Rankine [MW]"],
        color="tab:blue",
        linestyle="-",
        linewidth=2.4,
        label="Q transferido al fluido Rankine"
    )

    linea_gases, = ax1.plot(
        df["t [s]"],
        df["Q gases total [MW]"],
        color="tab:green",
        linestyle="-",
        linewidth=2.2,
        label="Q total retirado de gases"
    )

    linea_inicio = ax1.axvline(
        t_start_s,
        color="black",
        linestyle="--",
        linewidth=2.0,
        label="Inicio gases calientes"
    )

    ax1.set_xlabel("Tiempo [s]")
    ax1.set_ylabel("Potencia térmica principal [MW]")
    ax1.grid(True)

    ax2 = ax1.twinx()

    linea_ambiente, = ax2.plot(
        df["t [s]"],
        df["Q ambiente [kW]"],
        color="tab:red",
        linestyle="--",
        linewidth=2.2,
        label="Q perdido al ambiente"
    )

    ax2.set_ylabel("Pérdida al ambiente [kW]")

    lineas = [
        linea_rankine,
        linea_gases,
        linea_ambiente,
        linea_inicio
    ]

    etiquetas = [linea.get_label() for linea in lineas]

    ax1.legend(
        lineas,
        etiquetas,
        loc="best"
    )

    plt.title(f"Figura 2. Calor transferido - {nombre_equipo}")
    fig.tight_layout()
    plt.show()

    # -------------------------------------------------------------------------
    # Figura 3. Calidad de salida
    # -------------------------------------------------------------------------
    plt.figure(figsize=(10, 5))

    plt.plot(
        df["t [s]"],
        df["x salida [-]"],
        color="tab:purple",
        linestyle="-",
        linewidth=2.4,
        label="Calidad de salida"
    )

    plt.axvline(
        t_start_s,
        color="black",
        linestyle="--",
        linewidth=2.0,
        label="Inicio gases calientes"
    )

    plt.xlabel("Tiempo [s]")
    plt.ylabel("Calidad x [-]")
    plt.ylim(0, 1.05)
    plt.title(f"Figura 3. Calidad de salida - {nombre_equipo}")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()


# =============================================================================
# 13. EJECUCIÓN DEL EQUIPO SOLO
# =============================================================================

resultado = run_boiler_serpentin_estacionario()
tablas = crear_tablas_resultado(resultado)

df_transitorio = run_boiler_serpentin_transitorio()


# =============================================================================
# 14. IMPRESIÓN COMPACTA DE RESULTADOS
# =============================================================================

display(Markdown("# Intercambiador Caldera Serpentín"))

display(Markdown("## 1. Resumen estacionario del generador de vapor"))
display(tablas["df_resumen_estacionario"].T)

display(Markdown("## 2. Resumen térmico del generador de vapor"))
display(tablas["df_resumen_termico"].T)

display(Markdown("## 3. Resumen de diseño del serpentín"))
display(tablas["df_resumen_diseno"].T)

display(Markdown("## 4. Diagnóstico de operación"))
display(tablas["df_diagnostico"].T)

columnas_transitorio = [
    "t [s]",
    "T gases entrada [°C]",
    "T gases salida [°C]",
    "T Rankine entrada [°C]",
    "T Rankine salida [°C]",
    "Q Rankine [MW]",
    "Q ambiente [MW]",
    "Q ambiente [kW]",
    "Q gases total [MW]",
    "x salida [-]",
    "m gases [kg/s]",
    "Cumple vapor turbina"
]

df_transitorio_compacto = df_transitorio[columnas_transitorio].copy()

display(Markdown("## 5. Primeras 10 filas del transitorio"))
display(df_transitorio_compacto.head(10).T)

display(Markdown("## 6. Últimas 10 filas del transitorio"))
display(df_transitorio_compacto.tail(10).T)

df_transitorio_con_gases = df_transitorio_compacto[
    df_transitorio_compacto["m gases [kg/s]"] > 0
].copy()

if len(df_transitorio_con_gases) > 0:

    df_propiedades_medias = pd.DataFrame([{
        "T_gases_out_media [°C]": df_transitorio_con_gases["T gases salida [°C]"].mean(),
        "T_Rankine_out_media [°C]": df_transitorio_con_gases["T Rankine salida [°C]"].mean(),

        "Q_Rankine_media [MW]": df_transitorio_con_gases["Q Rankine [MW]"].mean(),
        "Q_ambiente_media [MW]": df_transitorio_con_gases["Q ambiente [MW]"].mean(),
        "Q_ambiente_media [kW]": df_transitorio_con_gases["Q ambiente [kW]"].mean(),
        "Q_gases_total_media [MW]": df_transitorio_con_gases["Q gases total [MW]"].mean(),

        "x_salida_media [-]": df_transitorio_con_gases["x salida [-]"].mean(),
        "x_salida_final [-]": df_transitorio_con_gases["x salida [-]"].iloc[-1],

        "m_gases [kg/s]": resultado["m_gas_kg_s"],
        "m_Rankine [kg/s]": resultado["m_rankine_kg_s"]
    }])

else:

    df_propiedades_medias = pd.DataFrame([{
        "Mensaje": "No hubo entrada de gases calientes durante la simulación."
    }])

display(Markdown("## 7. Propiedades medias calculadas en régimen con gases calientes"))
display(df_propiedades_medias.T)

display(Markdown("## 8. Gráficas del transitorio"))
graficar_transitorio_boiler(df_transitorio)


# =============================================================================
# 15. DICCIONARIO FINAL PARA INTEGRACIÓN FUTURA AL CICLO
# =============================================================================

boiler_serpentin = {
    "tipo": "serpentin_helicoidal_en_caldera",
    "fluido_caliente": "gases_escape_turbina_gas",
    "fluido_frio": "agua_vapor_rankine",

    "p_boiler_Pa": resultado["p_boiler_Pa"],
    "T_sat_K": resultado["T_sat_K"],
    "T_sat_C": resultado["T_sat_C"],

    "m_gas_kg_s": resultado["m_gas_kg_s"],
    "T_gas_in_K": resultado["T_gas_in_K"],
    "T_gas_out_K": resultado["T_gas_out_K"],
    "T_stack_min_K": resultado["T_stack_min_K"],

    "m_rankine_kg_s": resultado["m_rankine_kg_s"],
    "T_rankine_in_K": resultado["T_rankine_in_K"],
    "T_rankine_out_K": resultado["T_rankine_out_K"],

    "h_rankine_in_J_kg": resultado["h_rankine_in_J_kg"],
    "h_rankine_out_J_kg": resultado["h_rankine_out_J_kg"],
    "h_target_J_kg": resultado["h_target_J_kg"],

    "x_in": resultado["x_in"],
    "x_out": resultado["x_out"],
    "estado_out": resultado["estado_out"],

    "Q_required_W": resultado["Q_required_W"],
    "Q_to_rankine_W": resultado["Q_to_rankine_W"],
    "Q_ambient_W": resultado["Q_ambient_W"],
    "Q_gas_total_W": resultado["Q_gas_total_W"],
    "Q_gas_available_W": resultado["Q_gas_available_W"],

    "U_W_m2K": resultado["U_W_m2K"],
    "UA_W_K": resultado["UA_W_K"],
    "A_external_m2": resultado["A_external_m2"],
    "DTlm_K": resultado["DTlm_K"],

    "hi_W_m2K": resultado["hi_W_m2K"],
    "ho_W_m2K": resultado["ho_W_m2K"],

    "dp_internal_Pa": resultado["dp_internal_Pa"],

    "cumple_vapor_objetivo": resultado["cumple_vapor_objetivo"],
    "cumple_vapor_turbina": resultado["cumple_vapor_turbina"],
    "cumple_stack": resultado["cumple_stack"],
    "cumple_pinch": resultado["cumple_pinch"],

    "df_transitorio": df_transitorio
}
```

# **2. Integración Computacional de la Planta de Cogeneración**

La integración completa de la planta se realizó conectando en un mismo código los modelos del ciclo Brayton, la bomba Rankine, el intercambiador caldera-serpentín, la turbina Rankine y el condensador de coraza y tubos. La lógica principal de la implementación consiste en que las salidas calculadas de cada equipo se usan como condiciones de entrada del siguiente, manteniendo continuidad de masa, energía y potencia en todo el sistema.

Primero se resuelve el ciclo Brayton. Este bloque entrega las condiciones de los gases de escape que alimentan térmicamente al intercambiador caldera-serpentín. En la implementación, la temperatura y el flujo másico de gases que entran a la caldera se toman directamente de la salida de la turbina Brayton:

$$
T_{gas,in}=T_{4,Brayton}
$$

$$
\dot{m}_{gas}=\dot{m}_{gas,Brayton}
$$

De esta forma, el intercambiador caldera-serpentín no utiliza condiciones independientes, sino condiciones calculadas por el ciclo de gas.

El ciclo Rankine se acopla mediante un flujo másico común que circula por la bomba, la caldera, la turbina de vapor y el condensador:

$$
\dot{m}_{pump}
=
\dot{m}_{boiler}
=
\dot{m}_{turbine}
=
\dot{m}_{cond}
=
\dot{m}_{Rankine}
$$

La bomba recibe el líquido proveniente del condensador y eleva su presión hasta la presión de caldera. Por tanto, la conexión entre condensador y bomba se define mediante:

$$
h_{pump,in}=h_{cond,out}
$$

$$
T_{pump,in}=T_{cond,out}
$$

$$
p_{pump,in}=p_{cond}
$$

A la salida de la bomba, el fluido entra al intercambiador caldera-serpentín como agua presurizada. La conexión entre la bomba y la caldera se representa como:

$$
h_{boiler,in}=h_{pump,out}
$$

$$
T_{boiler,in}=T_{pump,out}
$$

$$
p_{boiler,in}=p_{pump,out}
$$

En la caldera-serpentín, el fluido Rankine recibe calor de los gases de escape del ciclo Brayton. La salida de este equipo define las condiciones de entrada a la turbina Rankine:

$$
h_{turb,in}=h_{boiler,out}
$$

$$
T_{turb,in}=T_{boiler,out}
$$

$$
x_{turb,in}=x_{boiler,out}
$$

Para evitar estados no físicos, el código incluye una protección que permite operar la turbina Rankine únicamente cuando la caldera entrega vapor seco o sobrecalentado:

$$
x_{boiler,out}=1
$$

Si esta condición no se cumple, la turbina Rankine permanece apagada y no se calcula potencia de vapor. Esto evita que el modelo expanda líquido o mezcla líquido-vapor dentro de la turbina.

Después de la expansión, la salida de la turbina Rankine alimenta el condensador. La conexión entre estos dos equipos se realiza mediante:

$$
h_{cond,in}=h_{turb,out}
$$

$$
T_{cond,in}=T_{turb,out}
$$

$$
p_{cond,in}=p_{cond}
$$

En el condensador, el vapor expandido rechaza calor hacia el agua de enfriamiento hasta alcanzar una condición líquida. La salida del condensador se considera apta para alimentar la bomba cuando la calidad del fluido Rankine es aproximadamente cero:

$$
x_{cond,out}\approx 0
$$

Además, el código incorpora una protección física para evitar que la bomba reciba vapor o mezcla líquido-vapor. Si el condensador todavía no entrega líquido válido, la bomba conserva el último estado líquido aceptable del sistema.

La conexión general del ciclo Rankine queda representada como:

$$
\text{Condensador}
\rightarrow
\text{Bomba}
\rightarrow
\text{Caldera}
\rightarrow
\text{Turbina Rankine}
\rightarrow
\text{Condensador}
$$

Mientras que el acoplamiento entre el ciclo Brayton y el ciclo Rankine se realiza mediante los gases de escape:

$$
\text{Turbina Brayton}
\rightarrow
\text{Gases de escape}
\rightarrow
\text{Caldera}
\rightarrow
\text{Vapor Rankine}
$$

Con todos los equipos acoplados, la potencia neta del ciclo Rankine se calcula restando la potencia consumida por la bomba a la potencia eléctrica generada por la turbina de vapor:

$$
\dot{W}_{net,Rankine}
=
\dot{W}_{e,Rankine}
-
\dot{W}_{pump}
$$

La potencia neta total de la planta se obtiene sumando la potencia eléctrica del ciclo Brayton y la potencia neta del ciclo Rankine:

$$
\dot{W}_{net,plant}
=
\dot{W}_{e,Brayton}
+
\dot{W}_{net,Rankine}
$$

Finalmente, la eficiencia eléctrica global se calcula comparando la potencia neta total con la potencia térmica suministrada por el combustible:

$$
\eta_{global}
=
\frac{
\dot{W}_{net,plant}
}{
\dot{Q}_{fuel}
}
$$

Además del cálculo estacionario, el código implementa una respuesta transitoria global. En esta parte, los intercambiadores se tratan como los componentes con mayor inercia térmica, mientras que la bomba y las turbinas se recalculan en cada instante con las condiciones que reciben. Durante la simulación, el flujo de gases hacia la caldera se activa a partir del tiempo de inicio definido en el código:

$$
\dot{m}_{gas}(t)
=
0
\qquad
t \lt t_{start,boiler}
$$

$$
\dot{m}_{gas}(t)
=
\dot{m}_{gas,Brayton}
\qquad
t \geq t_{start,boiler}
$$

Con esta estructura, el código permite observar cómo la recuperación de calor de los gases de escape activa progresivamente el ciclo Rankine, cómo se habilita la turbina de vapor cuando existe vapor apto, cómo responde el condensador y cómo evoluciona la potencia neta total de la planta. En conjunto, la implementación representa una planta de cogeneración integrada bajo una formulación global cero-dimensional y cuasi-dinámica, donde las variables de flujo másico, temperatura, entalpía, calidad, calor y potencia se transmiten entre equipos de forma secuencial y físicamente consistente.


```python
# @title
# =============================================================================
# CICLO COMBINADO BRAYTON - RANKINE
# CON INTERCAMBIADOR CALDERA SERPENTÍN Y CONDENSADOR CORAZA-TUBOS
# MODELO GLOBAL ESTACIONARIO + RESPUESTA TRANSITORIA DE INTERCAMBIADORES
# =============================================================================

import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

pd.options.display.float_format = "{:,.4f}".format

# =============================================================================
# 1. ENTRADAS GLOBALES DE PLANTA
# =============================================================================

g = 9.81

# -----------------------------------------------------------------------------
# Ciclo Brayton - turbina de gas
# -----------------------------------------------------------------------------
p_amb = 101325.0
T_amb = 288.15

gamma_air = 1.4
rho_air = 1.225
cp_air = 1005.0

gamma_gas = 1.33
cp_gas = 1100.0
rho_gas = 0.55
mu_gas = 3.5e-5
k_gas = 0.055
Pr_gas = cp_gas * mu_gas / k_gas

fuel_type = "Diesel No. 2"
LHV_fuel_MJ_kg = 42.6
rho_fuel_kg_L = 0.846

m_exhaust_kg_s = 135.5
AFR_mass = 35.30352783
shaft_rpm_brayton = 6600.0

eta_inlet = 0.98
eta_combustion = 0.98
eta_combustor_pressure = 0.98
eta_compressor_stage = 0.88
eta_turbine_stage_brayton = 0.88
eta_generator_brayton = 0.97

# -----------------------------------------------------------------------------
# Ciclo Rankine
# -----------------------------------------------------------------------------
p_boiler = 12.0e6
T_steam_target = 813.15
p_condenser = 9.0e3

cp_water = 4180.0
rho_water = 997.0
mu_water = 8.9e-4
k_water = 0.60

cp_steam = 2300.0
gamma_steam_eff = 1.18
R_steam = 461.5

eta_turbine_stage_rankine = 0.86
eta_mechanical_rankine = 0.98
eta_generator_rankine = 0.97

shaft_rpm_rankine = 4000.0
n_turbine_stages_rankine = 30

# -----------------------------------------------------------------------------
# Bomba Rankine
# -----------------------------------------------------------------------------
N_pump = 5200.0
r1_pump = 0.060
r2_pump = 0.250
b1_pump = 0.030
b2_pump = 0.020
beta1_geo_deg_pump = 20.0
beta2_geo_deg_pump = 35.0
z_pump = 8
kb1_pump = 1.0
kb2_pump = 1.0
sigma_pump = 0.90
eta_h_pump = 0.78
eta_m_pump = 0.95
eta_v_pump = 0.98
eta_o_pump = eta_h_pump * eta_m_pump * eta_v_pump

delta_p_steam_boiler = 0.03

# -----------------------------------------------------------------------------
# Intercambiador caldera serpentín
# Predimensionamiento usado antes de fijar constantes:
# Q_objetivo aproximado = 62 MW, U_diseño = 40 W/m2K, DT_LM_diseño = 150 K,
# F = 0.95 => A_req ≈ 10 900 m2. Se adopta A_o ≈ 15 700 m2
# para garantizar el objetivo de vapor con margen de UA y pérdidas al ambiente.
# -----------------------------------------------------------------------------
T_stack_min = 420.0
pinch_boiler_min = 20.0

N_serpentines = 250
L_tube_boiler = 400.0
D_i_boiler = 0.042
D_o_boiler = 0.050
D_coil_boiler = 3.00
D_shell_boiler = 8.00
L_shell_boiler = 18.0
shell_free_area_factor_boiler = 0.55

k_wall_boiler = 16.0
R_fi_boiler = 0.00010
R_fo_boiler = 0.00015
F_boiler = 0.95
K_minor_boiler = 3.0

# -----------------------------------------------------------------------------
# Condensador coraza y tubos
# -----------------------------------------------------------------------------
condenser_subcooling_K = 0.0
m_cool_condenser = 1200.0
T_cool_in_condenser = 298.15
pinch_cond_min = 5.0

N_tubos_cond = 5000
N_pasos_tubos_cond = 2
L_tubo_cond = 10.0
D_i_cond = 0.021
D_o_cond = 0.025
D_coraza_cond = 2.40
pitch_tubos_cond = 0.032
arreglo_tubos_cond = "triangular"

k_wall_cond = 16.0
R_fi_cond = 0.00010
R_fo_cond = 0.00010
F_cond = 0.95

rho_L_cond = 990.0
mu_L_cond = 6.5e-4
k_L_cond = 0.63
N_filas_verticales_cond = 20
phi_bank_cond = N_filas_verticales_cond ** (-0.25)
K_entrada_salida_tubos = 2.0

# -----------------------------------------------------------------------------
# Ambiente y transitorios
# -----------------------------------------------------------------------------
T_ambient = 298.15
p_ambient = 101325.0
R_air = 287.05
k_air = 0.0262
mu_air = 1.85e-5
Pr_air = 0.71

t_final_s = 1200.0
dt_s = 1.0
t_start_boiler_s = 300.0
t_start_condenser_s = 300.0
tau_boiler_s = 100.0
tau_condenser_s = 80.0

# =============================================================================
# 2. FUNCIONES AUXILIARES
# =============================================================================

def safe_div(num, den, default=0.0):
    if abs(den) <= 1e-12:
        return default
    return num / den


def clamp(x, xmin, xmax):
    return max(xmin, min(x, xmax))


def crear_tabla(diccionario, nombre_columna="Valor"):
    return pd.DataFrame(list(diccionario.items()), columns=["Parámetro", nombre_columna])


def axial_area(r_tip, r_hub):
    return math.pi * (r_tip**2 - r_hub**2)


def area_anular_pump(r, b, kb=1.0):
    return kb * 2.0 * np.pi * r * b


def magnitud(x, y):
    return np.sqrt(x**2 + y**2)


def deg_to_rad(angle_deg):
    return np.deg2rad(angle_deg)


def rad_to_deg(angle_rad):
    return np.rad2deg(angle_rad)


def steam_density_ideal(p, T, R=R_steam):
    return p / max(R * T, 1e-12)

# =============================================================================
# 3. PROPIEDADES APROXIMADAS DE AGUA/VAPOR
# =============================================================================

def saturation_pressure_water_iapws_MPa(T_K):
    T_c = 647.096
    p_c_MPa = 22.064
    a1 = -7.85951783
    a2 = 1.84408259
    a3 = -11.7866497
    a4 = 22.6807411
    a5 = -15.9618719
    a6 = 1.80122502

    if T_K <= 273.16:
        return 0.000611657
    if T_K >= T_c:
        return p_c_MPa

    theta = 1.0 - T_K / T_c
    exponent = (T_c / T_K) * (
        a1 * theta
        + a2 * theta**1.5
        + a3 * theta**3.0
        + a4 * theta**3.5
        + a5 * theta**4.0
        + a6 * theta**7.5
    )
    return p_c_MPa * math.exp(exponent)


def saturation_temperature_from_pressure(p_Pa):
    p_MPa = p_Pa / 1e6
    T_low = 273.16
    T_high = 647.096

    if p_MPa <= saturation_pressure_water_iapws_MPa(T_low):
        return T_low
    if p_MPa >= saturation_pressure_water_iapws_MPa(T_high):
        return T_high

    for _ in range(100):
        T_mid = 0.5 * (T_low + T_high)
        p_mid = saturation_pressure_water_iapws_MPa(T_mid)
        if p_mid < p_MPa:
            T_low = T_mid
        else:
            T_high = T_mid
    return 0.5 * (T_low + T_high)


def saturation_temperature_water_antoine(p_kPa):
    p_mmHg = p_kPa * 7.50062
    A = 8.07131
    B = 1730.63
    C = 233.426
    if p_mmHg <= 0.0:
        raise ValueError("La presión de saturación debe ser positiva.")
    return B / (A - math.log10(p_mmHg)) - C


def h_liquid_compressed_approx(T_K, p_Pa=101325.0):
    h_sensible = cp_water * (T_K - 273.15)
    h_pressure = (p_Pa - 101325.0) / rho_water
    return h_sensible + h_pressure


def h_liquid_approx(T_K):
    return cp_water * (T_K - 273.15)


def h_fg_watson_J_kg(T_sat_K):
    T_c = 647.096
    T_ref = 373.15
    h_fg_ref = 2257.0e3
    if T_sat_K >= T_c:
        return 0.0
    ratio = (1.0 - T_sat_K / T_c) / max(1.0 - T_ref / T_c, 1e-12)
    return h_fg_ref * ratio**0.38


def props_rankine_saturation(p_Pa):
    T_sat_K = saturation_temperature_from_pressure(p_Pa)
    h_f = h_liquid_compressed_approx(T_sat_K, p_Pa)
    h_fg = h_fg_watson_J_kg(T_sat_K)
    h_g = h_f + h_fg
    rho_v = p_Pa / max(R_steam * T_sat_K, 1e-12)
    return {
        "T_sat_K": T_sat_K,
        "T_sat_C": T_sat_K - 273.15,
        "h_f_J_kg": h_f,
        "h_fg_J_kg": h_fg,
        "h_g_J_kg": h_g,
        "rho_v_kg_m3": rho_v,
    }


def props_cond_saturation(p_cond_Pa):
    T_sat_C = saturation_temperature_water_antoine(p_cond_Pa / 1000.0)
    T_sat_K = T_sat_C + 273.15
    h_f = h_liquid_approx(T_sat_K)
    h_fg = h_fg_watson_J_kg(T_sat_K)
    h_g = h_f + h_fg
    rho_v = p_cond_Pa / max(R_steam * T_sat_K, 1e-12)
    return {
        "T_sat_K": T_sat_K,
        "T_sat_C": T_sat_C,
        "h_f_J_kg": h_f,
        "h_fg_J_kg": h_fg,
        "h_g_J_kg": h_g,
        "rho_v_kg_m3": rho_v,
    }


def h_target_steam_from_temperature(T_target_K, p_boiler_Pa):
    props = props_rankine_saturation(p_boiler_Pa)
    if T_target_K <= props["T_sat_K"]:
        return props["h_g_J_kg"]
    return props["h_g_J_kg"] + cp_steam * (T_target_K - props["T_sat_K"])


def estado_rankine_desde_h(h_J_kg, props):
    h_f = props["h_f_J_kg"]
    h_g = props["h_g_J_kg"]
    T_sat_K = props["T_sat_K"]

    if h_J_kg <= h_f:
        x = 0.0
        T_K = 273.15 + h_J_kg / cp_water
        estado = "Líquido comprimido / subenfriado"
    elif h_f < h_J_kg < h_g:
        x = (h_J_kg - h_f) / max(h_g - h_f, 1e-12)
        T_K = T_sat_K
        estado = "Mezcla líquido-vapor"
    else:
        x = 1.0
        T_K = T_sat_K + (h_J_kg - h_g) / cp_steam
        estado = "Vapor seco o sobrecalentado"

    return {"T_K": T_K, "T_C": T_K - 273.15, "x": clamp(x, 0.0, 1.0), "estado": estado}


def calidad_desde_h(h_J_kg, h_f_J_kg, h_g_J_kg):
    if h_J_kg <= h_f_J_kg:
        return 0.0
    if h_J_kg >= h_g_J_kg:
        return 1.0
    return (h_J_kg - h_f_J_kg) / max(h_g_J_kg - h_f_J_kg, 1e-12)


def estado_desde_h_cond(h_J_kg, h_f_J_kg, h_g_J_kg):
    if h_J_kg < h_f_J_kg:
        return "Líquido subenfriado"
    if abs(h_J_kg - h_f_J_kg) <= 1e-6:
        return "Líquido saturado, x = 0"
    if h_f_J_kg < h_J_kg < h_g_J_kg:
        return "Mezcla líquido-vapor"
    return "Vapor seco o sobrecalentado"


def temperatura_desde_h_cond(h_J_kg, h_f_J_kg, h_g_J_kg, T_sat_K):
    if h_J_kg <= h_f_J_kg:
        return 273.15 + h_J_kg / cp_water
    if h_f_J_kg < h_J_kg < h_g_J_kg:
        return T_sat_K
    return T_sat_K + (h_J_kg - h_g_J_kg) / cp_steam

# =============================================================================
# 4. CICLO BRAYTON
# =============================================================================

def run_brayton_cycle():
    omega = shaft_rpm_brayton * 2.0 * math.pi / 60.0

    p1 = eta_inlet * p_amb
    T1 = T_amb

    m_fuel_kg_s = m_exhaust_kg_s / (AFR_mass + 1.0)
    m_air_kg_s = AFR_mass * m_fuel_kg_s
    m_gas_kg_s = m_air_kg_s + m_fuel_kg_s

    Q_fuel_W = m_fuel_kg_s * LHV_fuel_MJ_kg * 1e6
    Q_added_W = Q_fuel_W * eta_combustion
    dotV_air_m3_s = m_air_kg_s / rho_air

    compressor_geometry = [
        {"stage": 1, "r_tip_m": 0.720000, "r_hub_m": 0.380000, "beta1_deg": 58.750000, "beta2_deg": 41.856912},
        {"stage": 2, "r_tip_m": 0.718571, "r_hub_m": 0.387143, "beta1_deg": 58.928571, "beta2_deg": 42.214055},
        {"stage": 3, "r_tip_m": 0.717143, "r_hub_m": 0.394286, "beta1_deg": 59.107143, "beta2_deg": 42.571198},
        {"stage": 4, "r_tip_m": 0.715714, "r_hub_m": 0.401429, "beta1_deg": 59.285714, "beta2_deg": 42.928341},
        {"stage": 5, "r_tip_m": 0.714286, "r_hub_m": 0.408571, "beta1_deg": 59.464286, "beta2_deg": 43.285483},
        {"stage": 6, "r_tip_m": 0.712857, "r_hub_m": 0.415714, "beta1_deg": 59.642857, "beta2_deg": 43.642626},
        {"stage": 7, "r_tip_m": 0.711429, "r_hub_m": 0.422857, "beta1_deg": 59.821429, "beta2_deg": 43.999769},
        {"stage": 8, "r_tip_m": 0.710000, "r_hub_m": 0.430000, "beta1_deg": 60.000000, "beta2_deg": 44.356912},
        {"stage": 9, "r_tip_m": 0.708571, "r_hub_m": 0.437143, "beta1_deg": 60.178571, "beta2_deg": 44.714055},
        {"stage": 10, "r_tip_m": 0.707143, "r_hub_m": 0.444286, "beta1_deg": 60.357143, "beta2_deg": 45.071198},
        {"stage": 11, "r_tip_m": 0.705714, "r_hub_m": 0.451429, "beta1_deg": 60.535714, "beta2_deg": 45.428341},
        {"stage": 12, "r_tip_m": 0.704286, "r_hub_m": 0.458571, "beta1_deg": 60.714286, "beta2_deg": 45.785483},
        {"stage": 13, "r_tip_m": 0.702857, "r_hub_m": 0.465714, "beta1_deg": 60.892857, "beta2_deg": 46.142626},
        {"stage": 14, "r_tip_m": 0.701429, "r_hub_m": 0.472857, "beta1_deg": 61.071429, "beta2_deg": 46.499769},
        {"stage": 15, "r_tip_m": 0.700000, "r_hub_m": 0.480000, "beta1_deg": 61.250000, "beta2_deg": 46.856912},
    ]

    compressor_results = []
    p_comp_in = p1
    T_comp_in = T1
    W_comp_total_W = 0.0

    for stage_data in compressor_geometry:
        r_tip_m = stage_data["r_tip_m"]
        r_hub_m = stage_data["r_hub_m"]
        beta1_rad = math.radians(stage_data["beta1_deg"])
        beta2_rad = math.radians(stage_data["beta2_deg"])
        r_mean_m = 0.5 * (r_tip_m + r_hub_m)
        A_m2 = axial_area(r_tip_m, r_hub_m)
        Va_m_s = dotV_air_m3_s / A_m2
        U_m_s = omega * r_mean_m
        Cu1_m_s = U_m_s - Va_m_s * math.tan(beta1_rad)
        Cu2_m_s = U_m_s - Va_m_s * math.tan(beta2_rad)
        delta_h_J_kg = U_m_s * (Cu2_m_s - Cu1_m_s)
        delta_T_K = delta_h_J_kg / cp_air
        W_stage_W = m_air_kg_s * delta_h_J_kg
        pressure_ratio_stage = (1.0 + eta_compressor_stage * delta_T_K / T_comp_in) ** (gamma_air / (gamma_air - 1.0))
        p_comp_out = p_comp_in * pressure_ratio_stage
        T_comp_out = T_comp_in + delta_T_K
        compressor_results.append({
            "Etapa": stage_data["stage"],
            "p_salida [Pa]": p_comp_out,
            "T_salida [K]": T_comp_out,
            "Potencia etapa [W]": W_stage_W,
            "Torque etapa [N m]": W_stage_W / omega,
        })
        W_comp_total_W += W_stage_W
        p_comp_in = p_comp_out
        T_comp_in = T_comp_out

    p2 = p_comp_out
    T2 = T_comp_out

    T3 = (m_air_kg_s * cp_air * T2 + Q_added_W) / (m_gas_kg_s * cp_gas)
    p3 = eta_combustor_pressure * p2

    turbine_geometry = [
        {"stage": 1, "r_tip_m": 0.760000, "r_hub_m": 0.430000, "beta_l_deg": 30.000000, "beta_t_deg": 82.726736},
        {"stage": 2, "r_tip_m": 0.780000, "r_hub_m": 0.410000, "beta_l_deg": 29.000000, "beta_t_deg": 84.726736},
        {"stage": 3, "r_tip_m": 0.800000, "r_hub_m": 0.390000, "beta_l_deg": 28.000000, "beta_t_deg": 86.726736},
    ]

    turbine_results = []
    p_turb_in = p3
    T_turb_in = T3
    W_turb_total_W = 0.0

    for stage_data in turbine_geometry:
        r_tip_m = stage_data["r_tip_m"]
        r_hub_m = stage_data["r_hub_m"]
        beta_l_rad = math.radians(stage_data["beta_l_deg"])
        beta_t_rad = math.radians(stage_data["beta_t_deg"])
        r_mean_m = 0.5 * (r_tip_m + r_hub_m)
        A_m2 = axial_area(r_tip_m, r_hub_m)
        Va_m_s = dotV_air_m3_s / A_m2
        U_m_s = omega * r_mean_m
        v_t_in_m_s = U_m_s - Va_m_s * math.tan(beta_l_rad)
        v_t_out_m_s = U_m_s - Va_m_s * math.tan(beta_t_rad)
        torque_stage_Nm = r_mean_m * m_gas_kg_s * (v_t_in_m_s - v_t_out_m_s)
        W_stage_useful_W = omega * torque_stage_Nm * eta_turbine_stage_brayton
        delta_T_stage_K = W_stage_useful_W / (m_gas_kg_s * cp_gas)
        T_turb_out = T_turb_in - delta_T_stage_K
        pressure_ratio_stage = (1.0 - delta_T_stage_K / T_turb_in) ** (gamma_gas / (gamma_gas - 1.0))
        p_turb_out = p_turb_in * pressure_ratio_stage
        turbine_results.append({
            "Etapa": stage_data["stage"],
            "p_salida [Pa]": p_turb_out,
            "T_salida [K]": T_turb_out,
            "Potencia etapa [W]": W_stage_useful_W,
            "Torque etapa [N m]": torque_stage_Nm,
        })
        W_turb_total_W += W_stage_useful_W
        p_turb_in = p_turb_out
        T_turb_in = T_turb_out

    p4 = p_turb_out
    T4 = T_turb_out

    W_net_shaft_W = W_turb_total_W - W_comp_total_W
    W_electric_W = W_net_shaft_W * eta_generator_brayton
    Q_exhaust_to_ambient_W = m_gas_kg_s * cp_gas * max(T4 - T_amb, 0.0)

    return {
        "p1": p1,
        "T1": T1,
        "p2": p2,
        "T2": T2,
        "p3": p3,
        "T3": T3,
        "p4": p4,
        "T4": T4,
        "m_air_kg_s": m_air_kg_s,
        "m_fuel_kg_s": m_fuel_kg_s,
        "m_gas_kg_s": m_gas_kg_s,
        "Q_fuel_W": Q_fuel_W,
        "Q_added_W": Q_added_W,
        "W_comp_total_W": W_comp_total_W,
        "W_turb_total_W": W_turb_total_W,
        "W_net_shaft_W": W_net_shaft_W,
        "W_electric_W": W_electric_W,
        "Q_exhaust_to_ambient_W": Q_exhaust_to_ambient_W,
        "df_comp": pd.DataFrame(compressor_results),
        "df_turb": pd.DataFrame(turbine_results),
    }

# =============================================================================
# 5. PREDIMENSIONAMIENTO OPERATIVO DE FLUJO RANKINE
# =============================================================================

def estimate_rankine_mass_flow_from_exhaust(brayton):
    T_cond_out_K = saturation_temperature_water_antoine(p_condenser / 1000.0) + 273.15 - condenser_subcooling_K
    h_cond_out = h_liquid_approx(T_cond_out_K)
    p_pump_out = p_boiler / (1.0 - delta_p_steam_boiler)
    w_pump_fluid = ((p_pump_out - p_condenser) / rho_water) / eta_h_pump
    h_pump_out = h_cond_out + w_pump_fluid
    T_pump_out_K = 273.15 + h_pump_out / cp_water
    h_target = h_target_steam_from_temperature(T_steam_target, p_boiler)
    q_required_per_kg = max(h_target - h_pump_out, 1e-12)
    Q_gas_available = brayton["m_gas_kg_s"] * cp_gas * max(brayton["T4"] - T_stack_min, 0.0)
    m_rankine_design = 0.92 * Q_gas_available / q_required_per_kg
    return {
        "m_rankine_design_kg_s": m_rankine_design,
        "p_pump_out": p_pump_out,
        "T_cond_out_K": T_cond_out_K,
        "h_cond_out": h_cond_out,
        "h_pump_out_est": h_pump_out,
        "T_pump_out_est_K": T_pump_out_K,
        "h_target_J_kg": h_target,
        "q_required_per_kg": q_required_per_kg,
        "Q_gas_available_W": Q_gas_available,
    }

# =============================================================================
# 6. BOMBA RANKINE ACOPLADA
# =============================================================================

def run_rankine_pump(m_rankine_kg_s, h_in_liquid_J_kg, T_in_K, p_in, p_out):
    Q = m_rankine_kg_s / rho_water
    mdot = m_rankine_kg_s
    delta_p = p_out - p_in
    pi_p = p_out / p_in
    H_req = delta_p / (rho_water * g)
    omega = 2.0 * np.pi * N_pump / 60.0

    U1 = omega * r1_pump
    U2 = omega * r2_pump
    A1 = area_anular_pump(r1_pump, b1_pump, kb1_pump)
    A2 = area_anular_pump(r2_pump, b2_pump, kb2_pump)
    Cm1 = Q / A1
    Cm2 = Q / A2
    Cu1 = 0.0

    beta2_geo_rad = deg_to_rad(beta2_geo_deg_pump)
    Cu2_ideal = U2 - Cm2 / np.tan(beta2_geo_rad)
    Cu2_real = sigma_pump * Cu2_ideal
    delta_h_E_corr = U2 * Cu2_real - U1 * Cu1
    H_E_corr = delta_h_E_corr / g
    H_real = eta_h_pump * H_E_corr

    w_pump_fluid_J_kg = delta_p / rho_water / eta_h_pump
    h_out_J_kg = h_in_liquid_J_kg + w_pump_fluid_J_kg
    T_out_K = 273.15 + h_out_J_kg / cp_water

    P_h_req = rho_water * g * Q * H_req
    P_eje_req = P_h_req / eta_o_pump
    T_req = P_eje_req / omega

    cumple_cabeza = H_real >= H_req

    tabla_operacion = crear_tabla({
        "Flujo másico [kg/s]": mdot,
        "Caudal volumétrico [m3/s]": Q,
        "p entrada [kPa]": p_in / 1000.0,
        "p salida [MPa]": p_out / 1e6,
        "Relación de presión [-]": pi_p,
        "Cabeza requerida [m]": H_req,
        "Cabeza real estimada [m]": H_real,
        "Potencia al eje requerida [MW]": P_eje_req / 1e6,
        "Torque requerido [N m]": T_req,
        "T entrada [°C]": T_in_K - 273.15,
        "T salida [°C]": T_out_K - 273.15,
        "h entrada [kJ/kg]": h_in_liquid_J_kg / 1000.0,
        "h salida [kJ/kg]": h_out_J_kg / 1000.0,
        "¿Cumple cabeza requerida?": cumple_cabeza,
    })

    return {
        "m_rankine_kg_s": mdot,
        "p1": p_in,
        "p2": p_out,
        "T_in_K": T_in_K,
        "T_out_K": T_out_K,
        "h_in_J_kg": h_in_liquid_J_kg,
        "h_out_J_kg": h_out_J_kg,
        "P_eje_req_W": P_eje_req,
        "T_req_Nm": T_req,
        "H_req": H_req,
        "H_real": H_real,
        "cumple_cabeza": cumple_cabeza,
        "tabla_operacion": tabla_operacion,
    }

# =============================================================================
# 7. INTERCAMBIADOR CALDERA SERPENTÍN ACOPLADO
# =============================================================================

def area_externa_boiler():
    return N_serpentines * math.pi * D_o_boiler * L_tube_boiler


def area_interna_boiler():
    return N_serpentines * math.pi * D_i_boiler * L_tube_boiler


def area_flujo_interno_boiler():
    return N_serpentines * math.pi * D_i_boiler**2 / 4.0


def area_libre_gases_boiler():
    return shell_free_area_factor_boiler * math.pi * D_shell_boiler**2 / 4.0


def area_externa_caracasa_boiler():
    return math.pi * D_shell_boiler * L_shell_boiler


def h_interno_boiler(m_dot):
    Qv = m_dot / rho_water
    A = area_flujo_interno_boiler()
    V = safe_div(Qv, A)
    Re = rho_water * V * D_i_boiler / max(mu_water, 1e-12)
    Pr = cp_water * mu_water / max(k_water, 1e-12)
    De = Re * math.sqrt(D_i_boiler / max(D_coil_boiler, 1e-12))
    if Re < 2300.0:
        Nu = 3.66
        regimen = "Laminar"
    else:
        Nu = 0.023 * Re**0.8 * Pr**0.4 * (1.0 + 3.5 * De / max(Re, 1e-12))
        regimen = "Turbulento"
    hi = Nu * k_water / max(D_i_boiler, 1e-12)
    return {"hi_W_m2K": hi, "Re": Re, "Pr": Pr, "De": De, "Nu": Nu, "V_m_s": V, "regimen": regimen}


def h_externo_gases_boiler(m_gas):
    if m_gas <= 0.0:
        return {"ho_W_m2K": 0.0, "Re": 0.0, "Pr": Pr_gas, "Nu": 0.0, "V_m_s": 0.0}
    V = safe_div(m_gas / max(rho_gas, 1e-12), area_libre_gases_boiler())
    Re = rho_gas * V * D_o_boiler / max(mu_gas, 1e-12)
    Nu = 0.27 * Re**0.63 * Pr_gas**0.36
    ho = Nu * k_gas / max(D_o_boiler, 1e-12)
    return {"ho_W_m2K": ho, "Re": Re, "Pr": Pr_gas, "Nu": Nu, "V_m_s": V}


def U_global_boiler(hi, ho):
    if hi <= 0.0 or ho <= 0.0:
        return 0.0
    R_total = (
        D_o_boiler / (D_i_boiler * hi)
        + R_fi_boiler * D_o_boiler / D_i_boiler
        + D_o_boiler * math.log(D_o_boiler / D_i_boiler) / (2.0 * k_wall_boiler)
        + R_fo_boiler
        + 1.0 / ho
    )
    return 1.0 / max(R_total, 1e-12)


def deltaT_lm_counterflow(Th_in, Th_out, Tc_in, Tc_out):
    DT1 = Th_in - Tc_out
    DT2 = Th_out - Tc_in
    if DT1 <= 0.0 or DT2 <= 0.0:
        return 0.0
    if abs(DT1 - DT2) <= 1e-12:
        return DT1
    return (DT1 - DT2) / math.log(DT1 / DT2)


def pinch_real_counterflow(Th_in, Th_out, Tc_in, Tc_out):
    return min(Th_in - Tc_out, Th_out - Tc_in)


def h_ambiente_conveccion_natural(T_surface_K, D_char):
    if T_surface_K <= T_ambient:
        return {"h_amb_W_m2K": 0.0, "Ra": 0.0, "Nu": 0.0}
    T_film = 0.5 * (T_surface_K + T_ambient)
    rho_air = p_ambient / max(R_air * T_film, 1e-12)
    nu_air = mu_air / max(rho_air, 1e-12)
    beta = 1.0 / max(T_film, 1e-12)
    Gr = g * beta * abs(T_surface_K - T_ambient) * D_char**3 / max(nu_air**2, 1e-24)
    Ra = Gr * Pr_air
    Nu = (0.60 + (0.387 * Ra**(1.0 / 6.0)) / ((1.0 + (0.559 / max(Pr_air, 1e-12))**(9.0 / 16.0)) ** (8.0 / 27.0))) ** 2.0
    h_amb = Nu * k_air / max(D_char, 1e-12)
    return {"h_amb_W_m2K": h_amb, "Ra": Ra, "Nu": Nu}


def perdida_ambiente_boiler(Q_rankine_W, T_gas_in, m_gas):
    if m_gas <= 0.0:
        return {"Q_ambient_W": 0.0, "T_surface_K": T_ambient, "h_ambient_W_m2K": 0.0}
    Ch = m_gas * cp_gas
    Q_amb = 0.0
    T_surface = T_ambient
    for _ in range(40):
        T_gas_out_est = T_gas_in - (Q_rankine_W + Q_amb) / max(Ch, 1e-12)
        T_surface = 0.5 * (T_gas_in + T_gas_out_est)
        h_amb = h_ambiente_conveccion_natural(T_surface, D_shell_boiler)
        Q_new = h_amb["h_amb_W_m2K"] * area_externa_caracasa_boiler() * max(T_surface - T_ambient, 0.0)
        if abs(Q_new - Q_amb) / max(abs(Q_new), 1.0) < 1e-6:
            Q_amb = Q_new
            break
        Q_amb = 0.5 * Q_amb + 0.5 * Q_new
    return {"Q_ambient_W": max(Q_amb, 0.0), "T_surface_K": T_surface, "h_ambient_W_m2K": h_amb["h_amb_W_m2K"]}


def evaluar_boiler_para_Q(Q_rankine_W, base):
    props = base["props"]
    h_out = base["h_in"] + safe_div(Q_rankine_W, base["m_rankine"])
    estado_out = estado_rankine_desde_h(h_out, props)
    T_rankine_out = estado_out["T_K"]
    ambiente = perdida_ambiente_boiler(Q_rankine_W, base["T_gas_in"], base["m_gas"])
    Q_amb = ambiente["Q_ambient_W"]
    Q_gas_total = Q_rankine_W + Q_amb
    T_gas_out = base["T_gas_in"] - safe_div(Q_gas_total, base["m_gas"] * cp_gas)
    DTlm = deltaT_lm_counterflow(base["T_gas_in"], T_gas_out, base["T_rankine_in"], T_rankine_out)
    Q_UA = base["UA"] * DTlm
    pinch = pinch_real_counterflow(base["T_gas_in"], T_gas_out, base["T_rankine_in"], T_rankine_out)
    return {
        "Q_to_rankine_W": Q_rankine_W,
        "Q_ambient_W": Q_amb,
        "Q_gas_total_W": Q_gas_total,
        "Q_UA_W": Q_UA,
        "T_gas_out_K": T_gas_out,
        "T_rankine_out_K": T_rankine_out,
        "h_rankine_out_J_kg": h_out,
        "x_out": estado_out["x"],
        "estado_out": estado_out["estado"],
        "DTlm_K": DTlm,
        "pinch_real_K": pinch,
        "T_surface_K": ambiente["T_surface_K"],
        "h_ambient_W_m2K": ambiente["h_ambient_W_m2K"],
    }


def resolver_Q_boiler(base):
    Q_required = base["Q_required"]
    Q_gas_available = base["Q_gas_available"]
    if Q_required <= 0.0 or Q_gas_available <= 0.0 or base["UA"] <= 0.0:
        return evaluar_boiler_para_Q(0.0, base)

    Q_high = min(Q_required, Q_gas_available)

    def balance(Q):
        est = evaluar_boiler_para_Q(Q, base)
        if est["Q_gas_total_W"] > Q_gas_available:
            return -abs(est["Q_gas_total_W"] - Q_gas_available)
        if est["pinch_real_K"] < pinch_boiler_min:
            return -abs(pinch_boiler_min - est["pinch_real_K"])
        return est["Q_UA_W"] - Q

    if balance(Q_high) >= 0.0:
        return evaluar_boiler_para_Q(Q_high, base)

    Q_low = 0.0
    for _ in range(120):
        Q_mid = 0.5 * (Q_low + Q_high)
        if balance(Q_mid) >= 0.0:
            Q_low = Q_mid
        else:
            Q_high = Q_mid
    return evaluar_boiler_para_Q(0.5 * (Q_low + Q_high), base)


def run_boiler_serpentin_estacionario(brayton, pump):
    T_gas_in = brayton["T4"]
    m_gas = brayton["m_gas_kg_s"]
    T_rankine_in = pump["T_out_K"]
    h_rankine_in = pump["h_out_J_kg"]
    m_rankine = pump["m_rankine_kg_s"]

    props = props_rankine_saturation(p_boiler)
    h_target = h_target_steam_from_temperature(T_steam_target, p_boiler)
    Q_required = m_rankine * max(h_target - h_rankine_in, 0.0)
    Q_gas_available = m_gas * cp_gas * max(T_gas_in - T_stack_min, 0.0)

    interno = h_interno_boiler(m_rankine)
    externo = h_externo_gases_boiler(m_gas)
    U = U_global_boiler(interno["hi_W_m2K"], externo["ho_W_m2K"])
    A = area_externa_boiler()
    UA = U * A * F_boiler

    base = {
        "props": props,
        "h_in": h_rankine_in,
        "h_target": h_target,
        "T_rankine_in": T_rankine_in,
        "m_rankine": m_rankine,
        "T_gas_in": T_gas_in,
        "m_gas": m_gas,
        "Q_required": Q_required,
        "Q_gas_available": Q_gas_available,
        "UA": UA,
    }

    sol = resolver_Q_boiler(base)
    h_out = sol["h_rankine_out_J_kg"]

    cumple_vapor_turbina = sol["x_out"] >= 1.0 and h_out >= props["h_g_J_kg"]
    cumple_vapor_objetivo = h_out >= h_target * (1.0 - 1e-6)
    cumple_stack = sol["T_gas_out_K"] >= T_stack_min
    cumple_pinch = sol["pinch_real_K"] >= pinch_boiler_min

    if cumple_vapor_objetivo:
        diagnostico = "La caldera-serpentín alcanza la condición objetivo de vapor."
    elif cumple_vapor_turbina:
        diagnostico = "La caldera-serpentín produce vapor seco, pero no alcanza toda la temperatura objetivo."
    else:
        diagnostico = "La caldera-serpentín no produce vapor apto para turbina con estas condiciones."

    dp = perdida_presion_boiler(m_rankine)

    return {
        "p_boiler_Pa": p_boiler,
        "T_sat_K": props["T_sat_K"],
        "T_sat_C": props["T_sat_C"],
        "h_f_J_kg": props["h_f_J_kg"],
        "h_g_J_kg": props["h_g_J_kg"],
        "T_gas_in_K": T_gas_in,
        "T_gas_out_K": sol["T_gas_out_K"],
        "T_stack_min_K": T_stack_min,
        "m_gas_kg_s": m_gas,
        "T_rankine_in_K": T_rankine_in,
        "T_rankine_out_K": sol["T_rankine_out_K"],
        "T_steam_target_K": T_steam_target,
        "m_rankine_kg_s": m_rankine,
        "h_rankine_in_J_kg": h_rankine_in,
        "h_rankine_out_J_kg": h_out,
        "h_target_J_kg": h_target,
        "x_out": sol["x_out"],
        "estado_out": sol["estado_out"],
        "Q_required_W": Q_required,
        "Q_to_rankine_W": sol["Q_to_rankine_W"],
        "Q_ambient_W": sol["Q_ambient_W"],
        "Q_gas_total_W": sol["Q_gas_total_W"],
        "Q_gas_available_W": Q_gas_available,
        "Q_UA_W": sol["Q_UA_W"],
        "U_W_m2K": U,
        "UA_W_K": UA,
        "A_external_m2": A,
        "DTlm_K": sol["DTlm_K"],
        "pinch_real_K": sol["pinch_real_K"],
        "hi_W_m2K": interno["hi_W_m2K"],
        "ho_W_m2K": externo["ho_W_m2K"],
        "Re_internal": interno["Re"],
        "De_internal": interno["De"],
        "Pr_internal": interno["Pr"],
        "Nu_internal": interno["Nu"],
        "Re_external": externo["Re"],
        "Nu_external": externo["Nu"],
        "dp_internal_Pa": dp["dp_internal_Pa"],
        "cumple_vapor_turbina": cumple_vapor_turbina,
        "cumple_vapor_objetivo": cumple_vapor_objetivo,
        "cumple_stack": cumple_stack,
        "cumple_pinch": cumple_pinch,
        "diagnostico": diagnostico,
    }


def perdida_presion_boiler(m_dot):
    interno = h_interno_boiler(m_dot)
    Re = interno["Re"]
    V = interno["V_m_s"]
    if Re <= 0.0:
        f = 0.0
    elif Re < 2300.0:
        f = 64.0 / Re
    else:
        f = (0.79 * math.log(Re) - 1.64) ** (-2.0)
    dp_fric = f * (L_tube_boiler / max(D_i_boiler, 1e-12)) * 0.5 * rho_water * V**2
    dp_minor = K_minor_boiler * 0.5 * rho_water * V**2
    return {"dp_internal_Pa": dp_fric + dp_minor, "f": f}


def run_boiler_serpentin_transitorio(brayton, pump, boiler_ss):
    tiempos = np.arange(0.0, t_final_s + dt_s, dt_s)
    Q_actual = 0.0
    rows = []

    props = props_rankine_saturation(p_boiler)

    for t in tiempos:
        if t < t_start_boiler_s:
            Q_actual = Q_actual + (0.0 - Q_actual) * dt_s / max(tau_boiler_s, 1e-12)
            estado_out = estado_rankine_desde_h(pump["h_out_J_kg"], props)
            rows.append({
                "t [s]": t,
                "T gases entrada [°C]": np.nan,
                "T gases salida [°C]": np.nan,
                "T Rankine entrada [°C]": pump["T_out_K"] - 273.15,
                "T Rankine salida [°C]": estado_out["T_C"],
                "Q Rankine [MW]": 0.0,
                "Q ambiente [MW]": 0.0,
                "Q ambiente [kW]": 0.0,
                "Q gases total [MW]": 0.0,
                "x salida [-]": estado_out["x"],
                "m gases [kg/s]": 0.0,
                "m Rankine [kg/s]": pump["m_rankine_kg_s"],
                "Cumple vapor turbina": False,
            })
            continue

        Q_ss = boiler_ss["Q_to_rankine_W"]
        Q_actual = Q_actual + (Q_ss - Q_actual) * dt_s / max(tau_boiler_s, 1e-12)
        Q_actual = min(Q_actual, boiler_ss["Q_required_W"], boiler_ss["Q_gas_available_W"])

        h_out = pump["h_out_J_kg"] + Q_actual / max(pump["m_rankine_kg_s"], 1e-12)
        estado_out = estado_rankine_desde_h(h_out, props)
        ambiente = perdida_ambiente_boiler(Q_actual, brayton["T4"], brayton["m_gas_kg_s"])
        Q_amb = min(ambiente["Q_ambient_W"], max(boiler_ss["Q_gas_available_W"] - Q_actual, 0.0))
        Q_total = Q_actual + Q_amb
        T_gas_out = brayton["T4"] - Q_total / (brayton["m_gas_kg_s"] * cp_gas)
        cumple = estado_out["x"] >= 1.0 and h_out >= props["h_g_J_kg"]

        rows.append({
            "t [s]": t,
            "T gases entrada [°C]": brayton["T4"] - 273.15,
            "T gases salida [°C]": T_gas_out - 273.15,
            "T Rankine entrada [°C]": pump["T_out_K"] - 273.15,
            "T Rankine salida [°C]": estado_out["T_C"],
            "Q Rankine [MW]": Q_actual / 1e6,
            "Q ambiente [MW]": Q_amb / 1e6,
            "Q ambiente [kW]": Q_amb / 1000.0,
            "Q gases total [MW]": Q_total / 1e6,
            "x salida [-]": estado_out["x"],
            "m gases [kg/s]": brayton["m_gas_kg_s"],
            "m Rankine [kg/s]": pump["m_rankine_kg_s"],
            "Cumple vapor turbina": cumple,
        })

    return pd.DataFrame(rows)

# =============================================================================
# 8. TURBINA RANKINE ACOPLADA
# =============================================================================

def generate_rankine_turbine_geometry(n_stages):
    geometry = []
    for i in range(1, n_stages + 1):
        frac = 0.0 if n_stages == 1 else (i - 1) / (n_stages - 1)
        geometry.append({
            "stage": i,
            "r_mean_m": 0.42 + (2.65 - 0.42) * frac**1.15,
            "Va_design_m_s": 70.0 + (230.0 - 70.0) * frac**0.80,
            "beta_l_deg": 22.0 + (30.0 - 22.0) * frac,
            "eta_stage": eta_turbine_stage_rankine,
        })
    return geometry


def run_rankine_turbine(boiler):
    m_steam = boiler["m_rankine_kg_s"]
    omega = shaft_rpm_rankine * 2.0 * math.pi / 60.0
    p3 = p_boiler
    T3 = boiler["T_rankine_out_K"]
    h3 = boiler["h_rankine_out_J_kg"]

    geometry = generate_rankine_turbine_geometry(n_turbine_stages_rankine)
    p_turb_in = p3
    T_turb_in = T3
    h_turb_in = h3
    W_turb_total_W = 0.0
    W_turb_euler_total_W = 0.0
    rows = []

    for stage_data in geometry:
        stage_number = stage_data["stage"]
        r_mean_m = stage_data["r_mean_m"]
        Va_design = stage_data["Va_design_m_s"]
        beta_l_deg = stage_data["beta_l_deg"]
        eta_stage = stage_data["eta_stage"]

        p_turb_out_target = p3 * (p_condenser / p3) ** (stage_number / n_turbine_stages_rankine)
        pressure_ratio_stage = p_turb_out_target / p_turb_in
        T_turb_out_s = T_turb_in * pressure_ratio_stage ** ((gamma_steam_eff - 1.0) / gamma_steam_eff)
        delta_T_isentropic = max(T_turb_in - T_turb_out_s, 0.0)
        delta_T_actual = eta_stage * delta_T_isentropic
        T_turb_out = T_turb_in - delta_T_actual
        p_turb_out = p_turb_out_target
        delta_h_actual = cp_steam * delta_T_actual
        W_stage_useful = m_steam * delta_h_actual
        W_stage_euler = W_stage_useful / max(eta_stage, 1e-12)

        rho_steam = steam_density_ideal(p_turb_in, T_turb_in)
        dotV = m_steam / max(rho_steam, 1e-12)
        A_required = dotV / Va_design
        blade_height = A_required / (2.0 * math.pi * r_mean_m)
        r_hub = max(0.20, r_mean_m - 0.5 * blade_height)
        r_tip = r_mean_m + 0.5 * blade_height
        if r_hub <= 0.2000001:
            r_tip = math.sqrt(A_required / math.pi + r_hub**2)
            r_mean_m = 0.5 * (r_tip + r_hub)
        A_m2 = axial_area(r_tip, r_hub)
        Va = dotV / A_m2
        U = omega * r_mean_m
        beta_l_rad = math.radians(beta_l_deg)
        tan_beta_t = math.tan(beta_l_rad) + W_stage_euler / max(omega * r_mean_m * m_steam * Va, 1e-12)
        beta_t_deg = math.degrees(math.atan(tan_beta_t))
        v_t_in = U - Va * math.tan(beta_l_rad)
        v_t_out = U - Va * math.tan(math.radians(beta_t_deg))
        torque_stage = r_mean_m * m_steam * (v_t_in - v_t_out)

        rows.append({
            "Etapa": stage_number,
            "p_entrada [Pa]": p_turb_in,
            "p_salida [Pa]": p_turb_out,
            "T_entrada [K]": T_turb_in,
            "T_salida [K]": T_turb_out,
            "Delta h [J/kg]": delta_h_actual,
            "Potencia etapa [W]": W_stage_useful,
            "Torque etapa [N m]": torque_stage,
            "r_tip [m]": r_tip,
            "r_hub [m]": r_hub,
            "Área [m2]": A_m2,
        })

        W_turb_total_W += W_stage_useful
        W_turb_euler_total_W += W_stage_euler
        p_turb_in = p_turb_out
        T_turb_in = T_turb_out
        h_turb_in = h_turb_in - delta_h_actual

    p4 = p_turb_out
    T4 = T_turb_out
    h4 = h3 - W_turb_total_W / max(m_steam, 1e-12)
    W_shaft_W = W_turb_total_W * eta_mechanical_rankine
    W_electric_W = W_shaft_W * eta_generator_rankine
    T_sat_out_C = saturation_temperature_water_antoine(p4 / 1000.0)
    salida_apta_condensador = abs(p4 - p_condenser) / p_condenser <= 0.02

    return {
        "p3": p3,
        "T3": T3,
        "h3": h3,
        "p4": p4,
        "T4": T4,
        "h4": h4,
        "m_steam_kg_s": m_steam,
        "W_turb_total_W": W_turb_total_W,
        "W_shaft_W": W_shaft_W,
        "W_electric_W": W_electric_W,
        "specific_work_turb_kJ_kg": W_turb_total_W / max(m_steam, 1e-12) / 1000.0,
        "salida_apta_condensador": salida_apta_condensador,
        "T_sat_out_C": T_sat_out_C,
        "df_turb": pd.DataFrame(rows),
    }

# =============================================================================
# 9. CONDENSADOR CORAZA-TUBOS ACOPLADO
# =============================================================================

def area_externa_cond():
    return N_tubos_cond * math.pi * D_o_cond * L_tubo_cond


def area_interna_cond():
    return N_tubos_cond * math.pi * D_i_cond * L_tubo_cond


def area_flujo_por_paso_cond():
    tubos_por_paso = N_tubos_cond / max(N_pasos_tubos_cond, 1)
    return tubos_por_paso * math.pi * D_i_cond**2 / 4.0


def factor_friccion_gnielinski(Re):
    Re = max(Re, 1.0)
    return (0.79 * math.log(Re) - 1.64) ** (-2.0)


def h_interno_cond(m_cool):
    Qv = m_cool / rho_water
    A = area_flujo_por_paso_cond()
    V = safe_div(Qv, A)
    Re = rho_water * V * D_i_cond / max(mu_water, 1e-12)
    Pr = cp_water * mu_water / max(k_water, 1e-12)
    if Re < 2300.0:
        Nu = 3.66
        f = np.nan
        regimen = "Laminar"
    elif Re < 3000.0:
        Nu = 3.66
        f = factor_friccion_gnielinski(3000.0)
        regimen = "Transición"
    else:
        f = factor_friccion_gnielinski(Re)
        Nu = ((f / 8.0) * max(Re - 1000.0, 0.0) * Pr) / (
            1.0 + 12.7 * math.sqrt(max(f / 8.0, 1e-16)) * (Pr ** (2.0 / 3.0) - 1.0)
        )
        regimen = "Turbulento"
    hi = Nu * k_water / max(D_i_cond, 1e-12)
    return {"hi_W_m2K": hi, "Re": Re, "Pr": Pr, "Nu": Nu, "f": f, "V_m_s": V, "regimen": regimen}


def h_externo_condensacion(T_sat_K, h_fg, rho_v, T_wall_K):
    deltaT = max(T_sat_K - T_wall_K, 1e-6)
    num = k_L_cond**3 * rho_L_cond * max(rho_L_cond - rho_v, 1e-12) * g * max(h_fg, 1e-12)
    den = max(mu_L_cond, 1e-12) * max(D_o_cond, 1e-12) * deltaT
    ho_un_tubo = 0.725 * (num / den) ** 0.25
    return ho_un_tubo * phi_bank_cond, ho_un_tubo


def U_global_cond(hi, ho):
    R_total = (
        D_o_cond / (D_i_cond * max(hi, 1e-12))
        + R_fi_cond * D_o_cond / D_i_cond
        + D_o_cond * math.log(D_o_cond / D_i_cond) / (2.0 * k_wall_cond)
        + R_fo_cond
        + 1.0 / max(ho, 1e-12)
    )
    return 1.0 / max(R_total, 1e-12)


def deltaT_LM_condensador(T_sat_K, T_cool_in, T_cool_out):
    DT1 = T_sat_K - T_cool_out
    DT2 = T_sat_K - T_cool_in
    if DT1 <= 0.0 or DT2 <= 0.0:
        return 0.0
    if abs(DT1 - DT2) <= 1e-12:
        return DT1
    return (DT1 - DT2) / math.log(DT1 / DT2)


def resolver_Q_UA_cond(T_sat_K, T_cool_in, m_cool, UA):
    C = m_cool * cp_water
    if C <= 0.0 or UA <= 0.0:
        return {"Q_UA_W": 0.0, "T_cool_out_UA_K": T_cool_in, "deltaT_LM_UA_K": 0.0, "limitacion": "Sin flujo o UA"}
    T_out_max = T_sat_K - pinch_cond_min
    if T_out_max <= T_cool_in:
        return {"Q_UA_W": 0.0, "T_cool_out_UA_K": T_cool_in, "deltaT_LM_UA_K": 0.0, "limitacion": "Sin pinch"}
    Q_max = C * (T_out_max - T_cool_in)

    def f(Q):
        T_out = T_cool_in + Q / C
        return UA * deltaT_LM_condensador(T_sat_K, T_cool_in, T_out) - Q

    if f(Q_max) >= 0.0:
        Q = Q_max
        limit = "Limitado por pinch"
    else:
        lo, hi = 0.0, Q_max
        for _ in range(100):
            mid = 0.5 * (lo + hi)
            if f(mid) > 0.0:
                lo = mid
            else:
                hi = mid
        Q = 0.5 * (lo + hi)
        limit = "Limitado por UA-LMTD"
    T_out = T_cool_in + Q / C
    return {"Q_UA_W": Q, "T_cool_out_UA_K": T_out, "deltaT_LM_UA_K": deltaT_LM_condensador(T_sat_K, T_cool_in, T_out), "limitacion": limit}


def perdida_ambiente_cond(T_surface_K):
    A_shell = math.pi * D_coraza_cond * L_tubo_cond
    h_amb = h_ambiente_conveccion_natural(T_surface_K, D_coraza_cond)
    Q = h_amb["h_amb_W_m2K"] * A_shell * max(T_surface_K - T_ambient, 0.0)
    return {"Q_amb_W": Q, "h_amb_W_m2K": h_amb["h_amb_W_m2K"], "A_shell_m2": A_shell}


def perdida_presion_cond(m_cool, f_tubos):
    interno = h_interno_cond(m_cool)
    Re = interno["Re"]
    V = interno["V_m_s"]
    if np.isnan(f_tubos):
        f = 64.0 / max(Re, 1e-12)
    else:
        f = f_tubos
    L_total = L_tubo_cond * N_pasos_tubos_cond
    dp_fric = f * (L_total / D_i_cond) * 0.5 * rho_water * V**2
    dp_menores = K_entrada_salida_tubos * 0.5 * rho_water * V**2
    return {"dp_tubos_Pa": dp_fric + dp_menores}


def resolver_estado_estacionario_condensador(rankine_turbine):
    p_cond = p_condenser
    m_rankine = rankine_turbine["m_steam_kg_s"]
    h_in = rankine_turbine["h4"]
    T_cool_in = T_cool_in_condenser
    m_cool = m_cool_condenser

    props = props_cond_saturation(p_cond)
    T_sat = props["T_sat_K"]
    h_f = props["h_f_J_kg"]
    h_g = props["h_g_J_kg"]
    h_fg = props["h_fg_J_kg"]
    rho_v = props["rho_v_kg_m3"]
    h_out_obj = h_f - condenser_subcooling_K * cp_water
    Q_necesario = m_rankine * max(h_in - h_out_obj, 0.0)

    interno = h_interno_cond(m_cool)
    hi = interno["hi_W_m2K"]
    T_wall = T_cool_in + 0.70 * max(T_sat - T_cool_in, 1.0)
    Q_cond = 0.0
    Q_tubos = 0.0
    Q_amb = 0.0
    T_cool_out = T_cool_in

    for it in range(100):
        ho, ho_un = h_externo_condensacion(T_sat, h_fg, rho_v, T_wall)
        Uo = U_global_cond(hi, ho)
        UA = Uo * area_externa_cond() * F_cond
        sol_UA = resolver_Q_UA_cond(T_sat, T_cool_in, m_cool, UA)
        Q_UA = sol_UA["Q_UA_W"]
        amb = perdida_ambiente_cond(T_sat)
        Q_amb_pot = amb["Q_amb_W"]
        Q_new = min(Q_necesario, Q_UA + Q_amb_pot)
        Q_amb_new = min(Q_amb_pot, Q_new)
        Q_tubos_new = max(Q_new - Q_amb_new, 0.0)
        T_cool_out_new = T_cool_in + Q_tubos_new / max(m_cool * cp_water, 1e-12)
        T_cool_mean = 0.5 * (T_cool_in + T_cool_out_new)
        T_wall_new = clamp(T_cool_mean + 0.70 * max(T_sat - T_cool_mean, 1.0), T_cool_mean + 1e-3, T_sat - 1e-3)
        err = max(abs(T_wall_new - T_wall), abs(Q_new - Q_cond) / max(abs(Q_new), 1.0))
        T_wall, Q_cond, Q_amb, Q_tubos, T_cool_out = T_wall_new, Q_new, Q_amb_new, Q_tubos_new, T_cool_out_new
        if err < 1e-6:
            break

    h_out = h_in - Q_cond / max(m_rankine, 1e-12)
    x_in = calidad_desde_h(h_in, h_f, h_g)
    x_out = calidad_desde_h(h_out, h_f, h_g)
    T_rank_in = temperatura_desde_h_cond(h_in, h_f, h_g, T_sat)
    T_rank_out = temperatura_desde_h_cond(h_out, h_f, h_g, T_sat)
    salida_apta_bomba = x_out <= 1e-6
    dp = perdida_presion_cond(m_cool, interno["f"])

    diagnostico = "La salida del condensador es apta para la bomba." if salida_apta_bomba else "El condensador no alcanza x = 0; salida no apta para bomba."

    return {
        "p_cond_Pa": p_cond,
        "T_sat_K": T_sat,
        "T_sat_C": props["T_sat_C"],
        "T_wall_K": T_wall,
        "h_f_J_kg": h_f,
        "h_g_J_kg": h_g,
        "h_in_J_kg": h_in,
        "h_out_real_J_kg": h_out,
        "h_out_objetivo_J_kg": h_out_obj,
        "T_rankine_in_K": T_rank_in,
        "T_rankine_out_K": T_rank_out,
        "x_in": x_in,
        "x_out": x_out,
        "estado_out": estado_desde_h_cond(h_out, h_f, h_g),
        "m_rankine_kg_s": m_rankine,
        "m_cool_kg_s": m_cool,
        "T_cool_in_K": T_cool_in,
        "T_cool_out_K": T_cool_out,
        "T_cond_out_K": T_rank_out,
        "Q_UA_W": Q_UA,
        "Q_amb_W": Q_amb,
        "Q_amb_potencial_W": Q_amb_pot,
        "Q_necesario_W": Q_necesario,
        "Q_cond_W": Q_cond,
        "Q_tubos_W": Q_tubos,
        "salida_apta_bomba": salida_apta_bomba,
        "Ao_m2": area_externa_cond(),
        "hi_W_m2K": hi,
        "ho_W_m2K": ho,
        "Uo_W_m2K": Uo,
        "UA_W_K": UA,
        "deltaT_LM_real_K": deltaT_LM_condensador(T_sat, T_cool_in, T_cool_out),
        "Re_tubos": interno["Re"],
        "Pr_tubos": interno["Pr"],
        "Nu_tubos": interno["Nu"],
        "dp_tubos_Pa": dp["dp_tubos_Pa"],
        "diagnostico": diagnostico,
        "iteraciones": it + 1,
    }


def run_condenser_transitorio(condenser_ss):
    tiempos = np.arange(0.0, t_final_s + dt_s, dt_s)
    Q_actual = 0.0
    rows = []
    for t in tiempos:
        if t < t_start_condenser_s:
            Q_actual = Q_actual + (0.0 - Q_actual) * dt_s / max(tau_condenser_s, 1e-12)
            rows.append({
                "t [s]": t,
                "T fluido Rankine entrada [°C]": np.nan,
                "T fluido Rankine salida [°C]": np.nan,
                "T agua entrada [°C]": T_cool_in_condenser - 273.15,
                "T agua salida [°C]": T_cool_in_condenser - 273.15,
                "Q condensador [MW]": 0.0,
                "Q hacia agua [MW]": 0.0,
                "Q ambiente [MW]": 0.0,
                "x salida [-]": np.nan,
                "m fluido Rankine [kg/s]": 0.0,
                "salida apta bomba": False,
            })
            continue

        Q_ss = condenser_ss["Q_cond_W"]
        Q_actual = Q_actual + (Q_ss - Q_actual) * dt_s / max(tau_condenser_s, 1e-12)
        Q_actual = min(Q_actual, condenser_ss["Q_necesario_W"], condenser_ss["Q_UA_W"] + condenser_ss["Q_amb_potencial_W"])
        Q_amb = min(condenser_ss["Q_amb_potencial_W"], Q_actual)
        Q_tubos = max(Q_actual - Q_amb, 0.0)
        h_out = condenser_ss["h_in_J_kg"] - Q_actual / max(condenser_ss["m_rankine_kg_s"], 1e-12)
        x_out = calidad_desde_h(h_out, condenser_ss["h_f_J_kg"], condenser_ss["h_g_J_kg"])
        T_rank_out = temperatura_desde_h_cond(h_out, condenser_ss["h_f_J_kg"], condenser_ss["h_g_J_kg"], condenser_ss["T_sat_K"])
        T_cool_out = condenser_ss["T_cool_in_K"] + Q_tubos / max(condenser_ss["m_cool_kg_s"] * cp_water, 1e-12)
        rows.append({
            "t [s]": t,
            "T fluido Rankine entrada [°C]": condenser_ss["T_rankine_in_K"] - 273.15,
            "T fluido Rankine salida [°C]": T_rank_out - 273.15,
            "T agua entrada [°C]": condenser_ss["T_cool_in_K"] - 273.15,
            "T agua salida [°C]": T_cool_out - 273.15,
            "Q condensador [MW]": Q_actual / 1e6,
            "Q hacia agua [MW]": Q_tubos / 1e6,
            "Q ambiente [MW]": Q_amb / 1e6,
            "x salida [-]": x_out,
            "m fluido Rankine [kg/s]": condenser_ss["m_rankine_kg_s"],
            "salida apta bomba": x_out <= 1e-6,
        })
    return pd.DataFrame(rows)

# =============================================================================
# 10. GRÁFICAS DE TRANSITORIOS
# =============================================================================

def graficar_boiler(df):
    nombre = "Intercambiador Caldera Serpentín"
    plt.figure(figsize=(10, 5))
    plt.plot(df["t [s]"], df["T gases entrada [°C]"], linewidth=2.2, label="T gases entrada")
    plt.plot(df["t [s]"], df["T gases salida [°C]"], linestyle="--", linewidth=2.2, label="T gases salida / chimenea")
    plt.plot(df["t [s]"], df["T Rankine entrada [°C]"], linewidth=2.2, label="T Rankine entrada")
    plt.plot(df["t [s]"], df["T Rankine salida [°C]"], linestyle="--", linewidth=2.2, label="T Rankine salida")
    plt.axvline(t_start_boiler_s, color="black", linestyle="--", linewidth=2.0, label="Inicio gases calientes")
    plt.xlabel("Tiempo [s]")
    plt.ylabel("Temperatura [°C]")
    plt.title(f"Figura 1. Temperaturas de entrada y salida - {nombre}")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

    fig, ax1 = plt.subplots(figsize=(10, 5))
    l1, = ax1.plot(df["t [s]"], df["Q Rankine [MW]"], linewidth=2.4, label="Q transferido al Rankine")
    l2, = ax1.plot(df["t [s]"], df["Q gases total [MW]"], linewidth=2.2, label="Q total retirado de gases")
    l4 = ax1.axvline(t_start_boiler_s, color="black", linestyle="--", linewidth=2.0, label="Inicio gases calientes")
    ax1.set_xlabel("Tiempo [s]")
    ax1.set_ylabel("Potencia térmica principal [MW]")
    ax1.grid(True)
    ax2 = ax1.twinx()
    l3, = ax2.plot(df["t [s]"], df["Q ambiente [kW]"], linestyle="--", linewidth=2.2, label="Q perdido al ambiente")
    ax2.set_ylabel("Pérdida al ambiente [kW]")
    ax1.legend([l1, l2, l3, l4], [x.get_label() for x in [l1, l2, l3, l4]], loc="best")
    plt.title(f"Figura 2. Calor transferido - {nombre}")
    fig.tight_layout()
    plt.show()

    plt.figure(figsize=(10, 5))
    plt.plot(df["t [s]"], df["x salida [-]"], linewidth=2.4, label="Calidad de salida")
    plt.axvline(t_start_boiler_s, color="black", linestyle="--", linewidth=2.0, label="Inicio gases calientes")
    plt.xlabel("Tiempo [s]")
    plt.ylabel("Calidad x [-]")
    plt.ylim(0, 1.05)
    plt.title(f"Figura 3. Calidad de salida - {nombre}")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()


def graficar_condensador(df):
    nombre = "Intercambiador Condensador de Coraza y Tubos"
    plt.figure(figsize=(10, 5))
    plt.plot(df["t [s]"], df["T fluido Rankine entrada [°C]"], linewidth=2.2, label="T Rankine entrada")
    plt.plot(df["t [s]"], df["T fluido Rankine salida [°C]"], linestyle="--", linewidth=2.2, label="T Rankine salida")
    plt.plot(df["t [s]"], df["T agua entrada [°C]"], linewidth=2.2, label="T agua entrada")
    plt.plot(df["t [s]"], df["T agua salida [°C]"], linestyle="--", linewidth=2.2, label="T agua salida")
    plt.axvline(t_start_condenser_s, color="black", linestyle="--", linewidth=2.0, label="Inicio entrada Rankine")
    plt.xlim(left=0)
    plt.ylim(bottom=20)
    plt.xlabel("Tiempo [s]")
    plt.ylabel("Temperatura [°C]")
    plt.title(f"Figura 4. Temperaturas de entrada y salida - {nombre}")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

    fig, ax1 = plt.subplots(figsize=(10, 5))
    l1, = ax1.plot(df["t [s]"], df["Q hacia agua [MW]"], linewidth=2.4, label="Q transferido al agua")
    l2 = ax1.axvline(t_start_condenser_s, color="black", linestyle="--", linewidth=2.0, label="Inicio entrada Rankine")
    ax1.set_xlim(left=0)
    ax1.set_ylim(bottom=0)
    ax1.set_xlabel("Tiempo [s]")
    ax1.set_ylabel("Calor transferido al agua [MW]")
    ax1.grid(True)
    ax2 = ax1.twinx()
    l3, = ax2.plot(df["t [s]"], df["Q ambiente [MW]"] * 1000.0, linestyle="--", linewidth=2.4, label="Q perdido al ambiente")
    ax2.set_ylim(bottom=0)
    ax2.set_ylabel("Pérdida al ambiente [kW]")
    ax1.legend([l1, l2, l3], [x.get_label() for x in [l1, l2, l3]], loc="best")
    plt.title(f"Figura 5. Calor transferido - {nombre}")
    fig.tight_layout()
    plt.show()

    plt.figure(figsize=(10, 5))
    plt.plot(df["t [s]"], df["x salida [-]"], linewidth=2.4, label="Calidad de salida")
    plt.axvline(t_start_condenser_s, color="black", linestyle="--", linewidth=2.0, label="Inicio entrada Rankine")
    plt.xlim(left=0)
    plt.ylim(0, 1.05)
    plt.xlabel("Tiempo [s]")
    plt.ylabel("Calidad x [-]")
    plt.title(f"Figura 6. Calidad de salida - {nombre}")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

# =============================================================================
# 11. EJECUCIÓN COMPLETA DE LA PLANTA
# =============================================================================

brayton = run_brayton_cycle()
rankine_design = estimate_rankine_mass_flow_from_exhaust(brayton)

pump = run_rankine_pump(
    m_rankine_kg_s=rankine_design["m_rankine_design_kg_s"],
    h_in_liquid_J_kg=rankine_design["h_cond_out"],
    T_in_K=rankine_design["T_cond_out_K"],
    p_in=p_condenser,
    p_out=rankine_design["p_pump_out"],
)

boiler_serpentin = run_boiler_serpentin_estacionario(brayton, pump)
df_boiler_transitorio = run_boiler_serpentin_transitorio(brayton, pump, boiler_serpentin)

rankine_turbine = run_rankine_turbine(boiler_serpentin)
condenser = resolver_estado_estacionario_condensador(rankine_turbine)
df_condenser_transitorio = run_condenser_transitorio(condenser)

# Para cerrar el ciclo, la salida real calculada del condensador puede diferir ligeramente
# de la usada como estimación inicial de la bomba. Se reporta la diferencia.
error_T_cond_pump_K = condenser["T_cond_out_K"] - pump["T_in_K"]
error_h_cond_pump_kJ_kg = (condenser["h_out_real_J_kg"] - pump["h_in_J_kg"]) / 1000.0

# =============================================================================
# 12. BALANCE GLOBAL Y TABLAS DE RESULTADOS
# =============================================================================

omega_brayton = shaft_rpm_brayton * 2.0 * math.pi / 60.0
omega_rankine = shaft_rpm_rankine * 2.0 * math.pi / 60.0

W_brayton_electric_W = brayton["W_electric_W"]
W_rankine_gross_electric_W = rankine_turbine["W_electric_W"]
W_pump_W = pump["P_eje_req_W"]
W_rankine_net_electric_W = W_rankine_gross_electric_W - W_pump_W
W_total_net_electric_W = W_brayton_electric_W + W_rankine_net_electric_W
Q_fuel_W = brayton["Q_fuel_W"]
eta_brayton_electric = W_brayton_electric_W / Q_fuel_W
eta_rankine_bottoming = W_rankine_net_electric_W / max(boiler_serpentin["Q_to_rankine_W"], 1e-12)
eta_global = W_total_net_electric_W / Q_fuel_W
heat_rate_global_kJ_kWh = (Q_fuel_W / 1000.0) / max(W_total_net_electric_W / 1000.0, 1e-12) * 3600.0
fuel_m3_h = (brayton["m_fuel_kg_s"] / rho_fuel_kg_L) * 3600.0 / 1000.0
Q_stack_loss_W = brayton["m_gas_kg_s"] * cp_gas * max(boiler_serpentin["T_gas_out_K"] - T_amb, 0.0)

T_comp_brayton_Nm = brayton["W_comp_total_W"] / omega_brayton
T_turb_brayton_Nm = brayton["W_turb_total_W"] / omega_brayton
T_net_brayton_Nm = brayton["W_net_shaft_W"] / omega_brayton
T_rankine_turb_Nm = rankine_turbine["W_shaft_W"] / omega_rankine
T_pump_Nm = pump["T_req_Nm"]

w_comp_air_kJ_kg = brayton["W_comp_total_W"] / brayton["m_air_kg_s"] / 1000.0
w_turb_gas_kJ_kg = brayton["W_turb_total_W"] / brayton["m_gas_kg_s"] / 1000.0
w_pump_kJ_kg = W_pump_W / pump["m_rankine_kg_s"] / 1000.0
w_rankine_turb_kJ_kg = rankine_turbine["specific_work_turb_kJ_kg"]
w_rankine_net_kJ_kg = W_rankine_net_electric_W / pump["m_rankine_kg_s"] / 1000.0
q_boiler_kJ_kg = boiler_serpentin["Q_to_rankine_W"] / pump["m_rankine_kg_s"] / 1000.0
q_cond_kJ_kg = condenser["Q_cond_W"] / pump["m_rankine_kg_s"] / 1000.0

# -----------------------------------------------------------------------------
# Tablas compactas
# -----------------------------------------------------------------------------

df_resumen_planta = pd.DataFrame([{
    "Potencia eléctrica neta planta [MW]": W_total_net_electric_W / 1e6,
    "Potencia eléctrica Brayton [MW]": W_brayton_electric_W / 1e6,
    "Potencia eléctrica bruta Rankine [MW]": W_rankine_gross_electric_W / 1e6,
    "Potencia eléctrica neta Rankine [MW]": W_rankine_net_electric_W / 1e6,
    "Potencia consumida por bomba [MW]": W_pump_W / 1e6,
    "Potencia térmica combustible [MW]": Q_fuel_W / 1e6,
    "Calor al Rankine en caldera [MW]": boiler_serpentin["Q_to_rankine_W"] / 1e6,
    "Calor rechazado condensador [MW]": condenser["Q_cond_W"] / 1e6,
    "Pérdidas ambiente caldera [kW]": boiler_serpentin["Q_ambient_W"] / 1000.0,
    "Pérdidas ambiente condensador [kW]": condenser["Q_amb_W"] / 1000.0,
    "Eficiencia eléctrica global [%]": eta_global * 100.0,
    "Heat rate global [kJ/kWh]": heat_rate_global_kJ_kWh,
    "Consumo combustible [m3/h]": fuel_m3_h,
}])

df_resumen_equipos = pd.DataFrame([
    {"Equipo": "Compresor Brayton", "Fluido": "Aire", "Flujo másico [kg/s]": brayton["m_air_kg_s"], "Potencia producida [MW]": 0.0, "Potencia consumida [MW]": brayton["W_comp_total_W"] / 1e6, "Potencia neta [MW]": -brayton["W_comp_total_W"] / 1e6, "Trabajo específico [kJ/kg]": w_comp_air_kJ_kg, "Torque [N m]": T_comp_brayton_Nm, "Eficiencia [%]": eta_compressor_stage * 100.0},
    {"Equipo": "Cámara de combustión", "Fluido": "Aire + combustible", "Flujo másico [kg/s]": brayton["m_gas_kg_s"], "Calor asociado [MW]": brayton["Q_added_W"] / 1e6, "Trabajo específico [kJ/kg]": brayton["Q_added_W"] / brayton["m_gas_kg_s"] / 1000.0, "Eficiencia [%]": eta_combustion * 100.0},
    {"Equipo": "Turbina Brayton", "Fluido": "Gases de combustión", "Flujo másico [kg/s]": brayton["m_gas_kg_s"], "Potencia producida [MW]": brayton["W_turb_total_W"] / 1e6, "Potencia neta [MW]": brayton["W_turb_total_W"] / 1e6, "Trabajo específico [kJ/kg]": w_turb_gas_kJ_kg, "Torque [N m]": T_turb_brayton_Nm, "Eficiencia [%]": eta_turbine_stage_brayton * 100.0},
    {"Equipo": "Intercambiador Caldera Serpentín", "Fluido": "Gases / agua-vapor", "Flujo másico [kg/s]": boiler_serpentin["m_rankine_kg_s"], "Calor asociado [MW]": boiler_serpentin["Q_to_rankine_W"] / 1e6, "Trabajo específico [kJ/kg]": q_boiler_kJ_kg, "Vapor generado [kg/s]": boiler_serpentin["m_rankine_kg_s"], "Eficiencia [%]": 100.0 * boiler_serpentin["Q_to_rankine_W"] / max(boiler_serpentin["Q_gas_available_W"], 1e-12)},
    {"Equipo": "Bomba Rankine", "Fluido": "Agua líquida", "Flujo másico [kg/s]": pump["m_rankine_kg_s"], "Potencia consumida [MW]": W_pump_W / 1e6, "Potencia neta [MW]": -W_pump_W / 1e6, "Trabajo específico [kJ/kg]": w_pump_kJ_kg, "Torque [N m]": T_pump_Nm, "Eficiencia [%]": eta_o_pump * 100.0},
    {"Equipo": "Turbina Rankine", "Fluido": "Vapor de agua", "Flujo másico [kg/s]": pump["m_rankine_kg_s"], "Potencia producida [MW]": rankine_turbine["W_electric_W"] / 1e6, "Potencia neta [MW]": rankine_turbine["W_electric_W"] / 1e6, "Trabajo específico [kJ/kg]": w_rankine_turb_kJ_kg, "Torque [N m]": T_rankine_turb_Nm, "Eficiencia [%]": eta_turbine_stage_rankine * 100.0},
    {"Equipo": "Condensador de coraza y tubos", "Fluido": "Vapor / líquido", "Flujo másico [kg/s]": condenser["m_rankine_kg_s"], "Calor asociado [MW]": -condenser["Q_cond_W"] / 1e6, "Trabajo específico [kJ/kg]": q_cond_kJ_kg},
    {"Equipo": "Ciclo Rankine neto", "Fluido": "Agua-vapor", "Flujo másico [kg/s]": pump["m_rankine_kg_s"], "Potencia producida [MW]": rankine_turbine["W_electric_W"] / 1e6, "Potencia consumida [MW]": W_pump_W / 1e6, "Potencia neta [MW]": W_rankine_net_electric_W / 1e6, "Calor asociado [MW]": boiler_serpentin["Q_to_rankine_W"] / 1e6, "Trabajo específico [kJ/kg]": w_rankine_net_kJ_kg, "Eficiencia [%]": eta_rankine_bottoming * 100.0},
    {"Equipo": "Planta completa", "Fluido": "Ciclo combinado", "Potencia producida [MW]": W_brayton_electric_W / 1e6 + rankine_turbine["W_electric_W"] / 1e6, "Potencia consumida [MW]": W_pump_W / 1e6, "Potencia neta [MW]": W_total_net_electric_W / 1e6, "Calor asociado [MW]": Q_fuel_W / 1e6, "Eficiencia [%]": eta_global * 100.0},
])

df_estados_principales = pd.DataFrame([
    {"Punto": "Entrada compresor Brayton", "Fluido": "Aire", "p [kPa]": brayton["p1"] / 1000.0, "T [°C]": brayton["T1"] - 273.15, "Flujo másico [kg/s]": brayton["m_air_kg_s"]},
    {"Punto": "Salida compresor Brayton", "Fluido": "Aire comprimido", "p [kPa]": brayton["p2"] / 1000.0, "T [°C]": brayton["T2"] - 273.15, "Flujo másico [kg/s]": brayton["m_air_kg_s"]},
    {"Punto": "Salida combustión", "Fluido": "Gases calientes", "p [kPa]": brayton["p3"] / 1000.0, "T [°C]": brayton["T3"] - 273.15, "Flujo másico [kg/s]": brayton["m_gas_kg_s"]},
    {"Punto": "Escape turbina Brayton / entrada caldera serpentín", "Fluido": "Gases de escape", "p [kPa]": brayton["p4"] / 1000.0, "T [°C]": brayton["T4"] - 273.15, "Flujo másico [kg/s]": brayton["m_gas_kg_s"]},
    {"Punto": "Salida chimenea", "Fluido": "Gases de escape", "p [kPa]": brayton["p4"] / 1000.0, "T [°C]": boiler_serpentin["T_gas_out_K"] - 273.15, "Flujo másico [kg/s]": brayton["m_gas_kg_s"]},
    {"Punto": "Salida condensador / entrada bomba", "Fluido": "Agua líquida", "p [kPa]": p_condenser / 1000.0, "T [°C]": condenser["T_cond_out_K"] - 273.15, "h [kJ/kg]": condenser["h_out_real_J_kg"] / 1000.0, "x [-]": condenser["x_out"], "Flujo másico [kg/s]": pump["m_rankine_kg_s"]},
    {"Punto": "Salida bomba / entrada caldera serpentín", "Fluido": "Agua presurizada", "p [kPa]": pump["p2"] / 1000.0, "T [°C]": pump["T_out_K"] - 273.15, "h [kJ/kg]": pump["h_out_J_kg"] / 1000.0, "Flujo másico [kg/s]": pump["m_rankine_kg_s"]},
    {"Punto": "Salida caldera serpentín / entrada turbina Rankine", "Fluido": "Vapor de agua", "p [kPa]": p_boiler / 1000.0, "T [°C]": boiler_serpentin["T_rankine_out_K"] - 273.15, "h [kJ/kg]": boiler_serpentin["h_rankine_out_J_kg"] / 1000.0, "x [-]": boiler_serpentin["x_out"], "Flujo másico [kg/s]": pump["m_rankine_kg_s"]},
    {"Punto": "Salida turbina Rankine / entrada condensador", "Fluido": "Vapor expandido", "p [kPa]": rankine_turbine["p4"] / 1000.0, "T [°C]": rankine_turbine["T4"] - 273.15, "h [kJ/kg]": rankine_turbine["h4"] / 1000.0, "Flujo másico [kg/s]": pump["m_rankine_kg_s"]},
])

df_boiler_estacionario = pd.DataFrame([{
    "m_gases [kg/s]": boiler_serpentin["m_gas_kg_s"],
    "m_Rankine [kg/s]": boiler_serpentin["m_rankine_kg_s"],
    "T_gases_in [°C]": boiler_serpentin["T_gas_in_K"] - 273.15,
    "T_gases_out [°C]": boiler_serpentin["T_gas_out_K"] - 273.15,
    "T_Rankine_in [°C]": boiler_serpentin["T_rankine_in_K"] - 273.15,
    "T_Rankine_out [°C]": boiler_serpentin["T_rankine_out_K"] - 273.15,
    "h_out [kJ/kg]": boiler_serpentin["h_rankine_out_J_kg"] / 1000.0,
    "x_out [-]": boiler_serpentin["x_out"],
    "Q_Rankine [MW]": boiler_serpentin["Q_to_rankine_W"] / 1e6,
    "Q_ambiente [kW]": boiler_serpentin["Q_ambient_W"] / 1000.0,
    "U [W/m2K]": boiler_serpentin["U_W_m2K"],
    "UA [W/K]": boiler_serpentin["UA_W_K"],
    "A_externa [m2]": boiler_serpentin["A_external_m2"],
    "Cumple vapor turbina": boiler_serpentin["cumple_vapor_turbina"],
    "Cumple vapor objetivo": boiler_serpentin["cumple_vapor_objetivo"],
    "Diagnóstico": boiler_serpentin["diagnostico"],
}])

df_boiler_diseno = pd.DataFrame([{
    "N_serpentines [-]": N_serpentines,
    "L_tube [m]": L_tube_boiler,
    "D_i [m]": D_i_boiler,
    "D_o [m]": D_o_boiler,
    "D_coil [m]": D_coil_boiler,
    "D_shell [m]": D_shell_boiler,
    "L_shell [m]": L_shell_boiler,
    "Re interno [-]": boiler_serpentin["Re_internal"],
    "Dean interno [-]": boiler_serpentin["De_internal"],
    "Nu interno [-]": boiler_serpentin["Nu_internal"],
    "Re gases [-]": boiler_serpentin["Re_external"],
    "Nu gases [-]": boiler_serpentin["Nu_external"],
    "h_i [W/m2K]": boiler_serpentin["hi_W_m2K"],
    "h_o [W/m2K]": boiler_serpentin["ho_W_m2K"],
    "Delta p interna [kPa]": boiler_serpentin["dp_internal_Pa"] / 1000.0,
}])

df_condenser_estacionario = pd.DataFrame([{
    "p_cond [kPa]": condenser["p_cond_Pa"] / 1000.0,
    "T_sat [°C]": condenser["T_sat_C"],
    "m_Rankine [kg/s]": condenser["m_rankine_kg_s"],
    "m_agua [kg/s]": condenser["m_cool_kg_s"],
    "T_Rankine_in [°C]": condenser["T_rankine_in_K"] - 273.15,
    "T_Rankine_out [°C]": condenser["T_rankine_out_K"] - 273.15,
    "T_agua_in [°C]": condenser["T_cool_in_K"] - 273.15,
    "T_agua_out [°C]": condenser["T_cool_out_K"] - 273.15,
    "h_out [kJ/kg]": condenser["h_out_real_J_kg"] / 1000.0,
    "x_out [-]": condenser["x_out"],
    "Q_condensador [MW]": condenser["Q_cond_W"] / 1e6,
    "Q_hacia_agua [MW]": condenser["Q_tubos_W"] / 1e6,
    "Q_ambiente [kW]": condenser["Q_amb_W"] / 1000.0,
    "Uo [W/m2K]": condenser["Uo_W_m2K"],
    "UA [W/K]": condenser["UA_W_K"],
    "Salida apta bomba": condenser["salida_apta_bomba"],
    "Diagnóstico": condenser["diagnostico"],
}])

df_condenser_diseno = pd.DataFrame([{
    "N_tubos [-]": N_tubos_cond,
    "N_pasos [-]": N_pasos_tubos_cond,
    "L_tubo [m]": L_tubo_cond,
    "D_i [m]": D_i_cond,
    "D_o [m]": D_o_cond,
    "D_coraza [m]": D_coraza_cond,
    "Re_tubos [-]": condenser["Re_tubos"],
    "Pr_tubos [-]": condenser["Pr_tubos"],
    "Nu_tubos [-]": condenser["Nu_tubos"],
    "h_i [W/m2K]": condenser["hi_W_m2K"],
    "h_o [W/m2K]": condenser["ho_W_m2K"],
    "Delta p tubos [kPa]": condenser["dp_tubos_Pa"] / 1000.0,
}])

df_acople_ciclo = pd.DataFrame([{
    "m_Rankine diseño [kg/s]": rankine_design["m_rankine_design_kg_s"],
    "Q_gas_disponible caldera [MW]": rankine_design["Q_gas_available_W"] / 1e6,
    "q_requerido caldera estimado [kJ/kg]": rankine_design["q_required_per_kg"] / 1000.0,
    "Error T salida condensador vs entrada bomba [K]": error_T_cond_pump_K,
    "Error h salida condensador vs entrada bomba [kJ/kg]": error_h_cond_pump_kJ_kg,
    "Caldera cumple vapor turbina": boiler_serpentin["cumple_vapor_turbina"],
    "Condensador cumple salida bomba": condenser["salida_apta_bomba"],
}])

# =============================================================================
# 13. IMPRESIÓN COMPACTA DE RESULTADOS DE PLANTA
# =============================================================================
# En esta sección se muestran únicamente los resultados principales del ciclo.
# Los transitorios internos de cada intercambiador se conservan calculados en
# df_boiler_transitorio y df_condenser_transitorio, pero no se imprimen ni se
# grafican para evitar duplicar la respuesta transitoria global de planta.

# Tabla compacta de los intercambiadores dentro del ciclo estacionario.
df_resumen_intercambiadores_ciclo = pd.DataFrame([
    {
        "Equipo": "Intercambiador Caldera Serpentín",
        "Fluido caliente": "Gases de escape Brayton",
        "Fluido frío": "Agua-vapor Rankine",
        "T caliente entrada [°C]": boiler_serpentin["T_gas_in_K"] - 273.15,
        "T caliente salida [°C]": boiler_serpentin["T_gas_out_K"] - 273.15,
        "T Rankine entrada [°C]": boiler_serpentin["T_rankine_in_K"] - 273.15,
        "T Rankine salida [°C]": boiler_serpentin["T_rankine_out_K"] - 273.15,
        "Calor principal [MW]": boiler_serpentin["Q_to_rankine_W"] / 1e6,
        "Pérdida ambiente [kW]": boiler_serpentin["Q_ambient_W"] / 1000.0,
        "Calidad salida Rankine [-]": boiler_serpentin["x_out"],
        "Cumple condición": boiler_serpentin["cumple_vapor_turbina"],
    },
    {
        "Equipo": "Condensador de Coraza y Tubos",
        "Fluido caliente": "Vapor Rankine",
        "Fluido frío": "Agua de enfriamiento",
        "T caliente entrada [°C]": condenser["T_rankine_in_K"] - 273.15,
        "T caliente salida [°C]": condenser["T_rankine_out_K"] - 273.15,
        "T Rankine entrada [°C]": condenser["T_rankine_in_K"] - 273.15,
        "T Rankine salida [°C]": condenser["T_rankine_out_K"] - 273.15,
        "Calor principal [MW]": condenser["Q_cond_W"] / 1e6,
        "Pérdida ambiente [kW]": condenser["Q_amb_W"] / 1000.0,
        "Calidad salida Rankine [-]": condenser["x_out"],
        "Cumple condición": condenser["salida_apta_bomba"],
    },
])

display(Markdown("# 2.1 Resultados principales del Ciclo Combinado Brayton - Rankine"))

display(Markdown("## 2.1.1 Resumen energético de planta"))
display(df_resumen_planta.T)

display(Markdown("## 2.1.2 Resumen de equipos del ciclo"))
display(df_resumen_equipos)

display(Markdown("## 2.1.3 Estados termodinámicos principales"))
display(df_estados_principales)

display(Markdown("## 2.1.4 Acople interno del ciclo Rankine"))
display(df_acople_ciclo.T)

display(Markdown("## 2.1.5 Resumen principal de los intercambiadores integrados"))
display(df_resumen_intercambiadores_ciclo)

# Diccionario final de integración
planta_integrada = {
    "brayton": brayton,
    "rankine_design": rankine_design,
    "pump": pump,
    "boiler_serpentin": boiler_serpentin,
    "rankine_turbine": rankine_turbine,
    "condenser": condenser,
    "df_boiler_transitorio": df_boiler_transitorio,
    "df_condenser_transitorio": df_condenser_transitorio,
    "df_resumen_planta": df_resumen_planta,
    "df_resumen_equipos": df_resumen_equipos,
    "df_estados_principales": df_estados_principales,
    "df_resumen_intercambiadores_ciclo": df_resumen_intercambiadores_ciclo,
}

# =============================================================================
# 14. RESPUESTA TRANSITORIA GLOBAL PROTEGIDA DE LA PLANTA COMPLETA
# =============================================================================
# Reemplaza el bloque anterior de respuesta transitoria global por este bloque.
# La corrección principal es que ahora se agregan protecciones físicas:
# 1) La turbina Rankine solo opera si la caldera entrega vapor objetivo.
# 2) La bomba solo actualiza su entrada con condensado líquido válido.
# 3) Si el condensador aún no entrega líquido, se conserva el último inventario
#    líquido válido del pozo caliente.


def evaluar_caldera_dinamica_protegida(brayton_t, pump_t, Q_boiler_actual_W):
    """
    Evalúa la caldera en un paso temporal usando el calor dinámico acumulado.
    Las entradas provienen de la turbina Brayton y de la bomba.
    """

    props = props_rankine_saturation(p_boiler)
    h_target = h_target_steam_from_temperature(T_steam_target, p_boiler)

    h_in = pump_t["h_out_J_kg"]
    T_in = pump_t["T_out_K"]
    m_rankine = pump_t["m_rankine_kg_s"]

    T_gas_in = brayton_t["T4"]
    m_gas = brayton_t["m_gas_kg_s"]

    h_out = h_in + safe_div(Q_boiler_actual_W, m_rankine, default=0.0)
    estado_out = estado_rankine_desde_h(h_out, props)

    if m_gas > 1e-12:
        ambiente = perdida_ambiente_boiler(Q_boiler_actual_W, T_gas_in, m_gas)
        Q_amb = ambiente["Q_ambient_W"]
        Q_gas_total = Q_boiler_actual_W + Q_amb
        T_gas_out = T_gas_in - safe_div(Q_gas_total, m_gas * cp_gas, default=0.0)
    else:
        Q_amb = 0.0
        Q_gas_total = 0.0
        T_gas_out = np.nan

    cumple_vapor_turbina = estado_out["x"] >= 1.0 and h_out >= props["h_g_J_kg"]
    cumple_vapor_objetivo = (
        h_out >= h_target * 0.998
        or estado_out["T_K"] >= T_steam_target - 2.0
    )

    return {
        "props": props,
        "h_target_J_kg": h_target,
        "T_rankine_in_K": T_in,
        "T_rankine_out_K": estado_out["T_K"],
        "h_rankine_in_J_kg": h_in,
        "h_rankine_out_J_kg": h_out,
        "x_out": estado_out["x"],
        "estado_out": estado_out["estado"],
        "Q_to_rankine_W": Q_boiler_actual_W,
        "Q_ambient_W": Q_amb,
        "Q_gas_total_W": Q_gas_total,
        "T_gas_in_K": T_gas_in,
        "T_gas_out_K": T_gas_out,
        "m_gas_kg_s": m_gas,
        "m_rankine_kg_s": m_rankine,
        "cumple_vapor_turbina": cumple_vapor_turbina,
        "cumple_vapor_objetivo": cumple_vapor_objetivo,
        # Protección operativa: la turbina se habilita cuando la salida es vapor seco.
        # El sobrecalentamiento objetivo se conserva como diagnóstico, no como bloqueo duro.
        "turbina_habilitada": cumple_vapor_turbina,
    }


def evaluar_turbina_rankine_protegida(boiler_t):
    """
    Turbina Rankine con protección física.

    Correcciones aplicadas:
    1. La turbina solo opera si la caldera entrega vapor seco.
    2. La temperatura de salida de la turbina no se toma directamente del
       modelo idealizado por etapas, porque ese modelo puede dar temperaturas
       no físicas al descargar hacia el condensador.
    3. La salida se reinterpreta con propiedades de saturación a p_condenser:
       - Si h4 queda entre h_f y h_g, la salida es mezcla y T4 = T_sat.
       - Si h4 queda como líquido, se limita a h_f para evitar estados no físicos.
       - Si h4 queda como vapor sobrecalentado, T4 se calcula con cp_steam.
    """

    if not boiler_t["turbina_habilitada"]:
        return {
            "turbina_encendida": False,
            "m_steam_kg_s": 0.0,
            "m_rankine_total_kg_s": boiler_t["m_rankine_kg_s"],
            "p3": p_boiler,
            "T3": boiler_t["T_rankine_out_K"],
            "h3": boiler_t["h_rankine_out_J_kg"],
            "p4": p_condenser,
            "T4": np.nan,
            "h4": np.nan,
            "x4": np.nan,
            "W_turb_total_W": 0.0,
            "W_shaft_W": 0.0,
            "W_electric_W": 0.0,
            "specific_work_turb_kJ_kg": 0.0,
            "salida_apta_condensador": False,
        }

    m_rankine_total = boiler_t["m_rankine_kg_s"]

    boiler_para_turbina = {
        "m_rankine_kg_s": m_rankine_total,
        "T_rankine_out_K": boiler_t["T_rankine_out_K"],
        "h_rankine_out_J_kg": boiler_t["h_rankine_out_J_kg"],
    }

    turbina_base = run_rankine_turbine(boiler_para_turbina)

    props_cond = props_cond_saturation(p_condenser)
    h_f_cond = props_cond["h_f_J_kg"]
    h_g_cond = props_cond["h_g_J_kg"]
    T_sat_cond = props_cond["T_sat_K"]

    h3 = boiler_t["h_rankine_out_J_kg"]

    # Trabajo idealizado por el modelo de turbina, limitado físicamente para no
    # llevar la salida por debajo de líquido saturado a la presión del condensador.
    W_turb_total_W = turbina_base["W_turb_total_W"]
    h4_calculado = h3 - safe_div(W_turb_total_W, m_rankine_total, default=0.0)

    if h4_calculado < h_f_cond:
        h4 = h_f_cond
        W_turb_total_W = m_rankine_total * max(h3 - h4, 0.0)
    else:
        h4 = h4_calculado

    x4 = calidad_desde_h(h4, h_f_cond, h_g_cond)
    T4 = temperatura_desde_h_cond(h4, h_f_cond, h_g_cond, T_sat_cond)

    W_shaft_W = W_turb_total_W * eta_mechanical_rankine
    W_electric_W = W_shaft_W * eta_generator_rankine

    return {
        "turbina_encendida": True,
        "m_steam_kg_s": m_rankine_total,
        "m_rankine_total_kg_s": m_rankine_total,
        "p3": p_boiler,
        "T3": boiler_t["T_rankine_out_K"],
        "h3": h3,
        "p4": p_condenser,
        "T4": T4,
        "h4": h4,
        "x4": x4,
        "W_turb_total_W": W_turb_total_W,
        "W_shaft_W": W_shaft_W,
        "W_electric_W": W_electric_W,
        "specific_work_turb_kJ_kg": safe_div(W_turb_total_W, m_rankine_total, default=0.0) / 1000.0,
        "salida_apta_condensador": True,
        "T_sat_out_C": props_cond["T_sat_C"],
    }

def evaluar_condensador_dinamico_protegido(turbine_t, Q_cond_actual_W, h_liquid_valid_J_kg, T_liquid_valid_K):
    """
    Evalúa el condensador con protección física.
    Si la turbina está apagada, el condensador no recibe vapor y se conserva
    el último líquido válido para alimentar la bomba.
    """

    props = props_cond_saturation(p_condenser)
    T_sat = props["T_sat_K"]
    h_f = props["h_f_J_kg"]
    h_g = props["h_g_J_kg"]
    h_out_obj = h_f - condenser_subcooling_K * cp_water

    if (not turbine_t.get("turbina_encendida", False)) or turbine_t["m_steam_kg_s"] <= 1e-12:
        return {
            "recibe_vapor": False,
            "m_rankine_kg_s": 0.0,
            "h_in_J_kg": np.nan,
            "h_out_J_kg": h_liquid_valid_J_kg,
            "h_out_objetivo_J_kg": h_out_obj,
            "T_rankine_in_K": np.nan,
            "T_rankine_out_K": T_liquid_valid_K,
            "T_cool_in_K": T_cool_in_condenser,
            "T_cool_out_K": T_cool_in_condenser,
            "x_out": 0.0,
            "Q_cond_W": 0.0,
            "Q_tubos_W": 0.0,
            "Q_amb_W": 0.0,
            "salida_apta_bomba": True,
        }

    m_cond = turbine_t["m_steam_kg_s"]
    h_in = turbine_t["h4"]
    T_in = temperatura_desde_h_cond(h_in, h_f, h_g, T_sat)

    h_out = h_in - safe_div(Q_cond_actual_W, m_cond, default=0.0)
    h_out = max(h_out, h_out_obj)

    x_out = calidad_desde_h(h_out, h_f, h_g)
    T_out = temperatura_desde_h_cond(h_out, h_f, h_g, T_sat)

    amb = perdida_ambiente_cond(T_sat)
    Q_amb = min(amb["Q_amb_W"], max(Q_cond_actual_W, 0.0))
    Q_tubos = max(Q_cond_actual_W - Q_amb, 0.0)

    T_cool_out = T_cool_in_condenser + safe_div(
        Q_tubos,
        m_cool_condenser * cp_water,
        default=0.0
    )

    salida_apta_bomba = x_out <= 1e-3

    return {
        "recibe_vapor": True,
        "m_rankine_kg_s": m_cond,
        "h_in_J_kg": h_in,
        "h_out_J_kg": h_out,
        "h_out_objetivo_J_kg": h_out_obj,
        "T_rankine_in_K": T_in,
        "T_rankine_out_K": T_out,
        "T_cool_in_K": T_cool_in_condenser,
        "T_cool_out_K": T_cool_out,
        "x_out": x_out,
        "Q_cond_W": Q_cond_actual_W,
        "Q_tubos_W": Q_tubos,
        "Q_amb_W": Q_amb,
        "salida_apta_bomba": salida_apta_bomba,
    }


def run_planta_transitoria_global():
    """
    Simulación transitoria global 0D/cuasi-dinámica con protecciones físicas.

    La bomba siempre recibe líquido válido; no se permite que mezcla o vapor
    entren a la bomba. Los gases calientes entran a la caldera desde
    t_start_boiler_s, manteniendo la respuesta transitoria principal del sistema
    sin introducir un apagado artificial inicial.
    """

    brayton_ss = run_brayton_cycle()
    rankine_design = estimate_rankine_mass_flow_from_exhaust(brayton_ss)

    m_rankine_base = rankine_design["m_rankine_design_kg_s"]
    p_pump_out = rankine_design["p_pump_out"]

    # Inventario inicial líquido del pozo caliente del condensador.
    # Se mantiene fijo en líquido saturado/subenfriado a la presión del condensador.
    h_liquid_valid = rankine_design["h_cond_out"]
    T_liquid_valid = rankine_design["T_cond_out_K"]

    Q_boiler_actual_W = 0.0
    Q_cond_actual_W = 0.0

    rows = []
    tiempos = np.arange(0.0, t_final_s + dt_s, dt_s)

    for t in tiempos:

        gases_a_caldera = t >= t_start_boiler_s

        # ------------------------------------------------------------
        # 1. Brayton. La planta de gas está calculada desde el inicio,
        #    pero los gases solo entran a la caldera desde t_start_boiler_s.
        # ------------------------------------------------------------
        brayton_t = brayton_ss.copy()
        brayton_t["m_gas_kg_s"] = brayton_ss["m_gas_kg_s"] if gases_a_caldera else 0.0

        # ------------------------------------------------------------
        # 2. Bomba. Recibe líquido válido; no se actualiza con mezcla ni vapor.
        # ------------------------------------------------------------
        pump_t = run_rankine_pump(
            m_rankine_kg_s=m_rankine_base,
            h_in_liquid_J_kg=h_liquid_valid,
            T_in_K=T_liquid_valid,
            p_in=p_condenser,
            p_out=p_pump_out,
        )

        # ------------------------------------------------------------
        # 3. Caldera con inercia térmica.
        # ------------------------------------------------------------
        boiler_ss_t = run_boiler_serpentin_estacionario(brayton_t, pump_t)
        Q_boiler_target_W = boiler_ss_t["Q_to_rankine_W"]

        Q_boiler_actual_W = Q_boiler_actual_W + (
            Q_boiler_target_W - Q_boiler_actual_W
        ) * dt_s / max(tau_boiler_s, 1e-12)

        Q_boiler_actual_W = min(
            max(Q_boiler_actual_W, 0.0),
            boiler_ss_t["Q_required_W"],
            boiler_ss_t["Q_gas_available_W"],
        )

        boiler_t = evaluar_caldera_dinamica_protegida(
            brayton_t,
            pump_t,
            Q_boiler_actual_W
        )

        # ------------------------------------------------------------
        # 4. Turbina Rankine protegida.
        # ------------------------------------------------------------
        turbine_t = evaluar_turbina_rankine_protegida(boiler_t)

        # ------------------------------------------------------------
        # 5. Condensador con inercia térmica.
        # ------------------------------------------------------------
        if turbine_t.get("turbina_encendida", False):
            condenser_target = resolver_estado_estacionario_condensador(turbine_t)
            Q_cond_target_W = condenser_target["Q_cond_W"]
            Q_cond_required_W = condenser_target["Q_necesario_W"]
            Q_cond_capacity_W = condenser_target["Q_UA_W"] + condenser_target["Q_amb_potencial_W"]
        else:
            Q_cond_target_W = 0.0
            Q_cond_required_W = 0.0
            Q_cond_capacity_W = 0.0

        Q_cond_actual_W = Q_cond_actual_W + (
            Q_cond_target_W - Q_cond_actual_W
        ) * dt_s / max(tau_condenser_s, 1e-12)

        Q_cond_actual_W = min(
            max(Q_cond_actual_W, 0.0),
            Q_cond_required_W,
            Q_cond_capacity_W,
        )

        condenser_t = evaluar_condensador_dinamico_protegido(
            turbine_t,
            Q_cond_actual_W,
            h_liquid_valid,
            T_liquid_valid,
        )

        # Protección de bomba: solo se actualiza con condensado líquido válido.
        # Se elimina la dinámica artificial del pozo caliente; si el condensador
        # entrega líquido apto, se usa el estado objetivo líquido a p_condenser.
        if condenser_t["recibe_vapor"] and condenser_t["salida_apta_bomba"]:
            props_hotwell = props_cond_saturation(p_condenser)
            h_liquid_valid = condenser_t["h_out_objetivo_J_kg"]
            T_liquid_valid = props_hotwell["T_sat_K"] - condenser_subcooling_K

        # ------------------------------------------------------------
        # 6. Potencias y eficiencia global instantánea.
        # ------------------------------------------------------------
        W_brayton_electric_W = brayton_ss["W_electric_W"]
        W_rankine_gross_W = turbine_t["W_electric_W"]
        W_pump_W = pump_t["P_eje_req_W"]

        W_rankine_net_W = W_rankine_gross_W - W_pump_W
        W_total_net_W = W_brayton_electric_W + W_rankine_net_W

        eta_global = safe_div(W_total_net_W, brayton_ss["Q_fuel_W"], default=np.nan)
        eta_rankine = safe_div(W_rankine_net_W, max(boiler_t["Q_to_rankine_W"], 1e-12), default=np.nan)

        rows.append({
            "t [s]": t,

            # Brayton
            "T compresor entrada Brayton [°C]": brayton_ss["T1"] - 273.15,
            "T compresor salida Brayton [°C]": brayton_ss["T2"] - 273.15,
            "T combustión salida Brayton [°C]": brayton_ss["T3"] - 273.15,
            "T escape turbina Brayton [°C]": brayton_ss["T4"] - 273.15,
            "m gases a caldera [kg/s]": brayton_t["m_gas_kg_s"],

            # Bomba
            "T bomba entrada [°C]": pump_t["T_in_K"] - 273.15,
            "T bomba salida [°C]": pump_t["T_out_K"] - 273.15,
            "h bomba salida [kJ/kg]": pump_t["h_out_J_kg"] / 1000.0,

            # Caldera
            "T caldera gases entrada [°C]": boiler_t["T_gas_in_K"] - 273.15 if boiler_t["m_gas_kg_s"] > 0.0 else np.nan,
            "T chimenea [°C]": boiler_t["T_gas_out_K"] - 273.15 if boiler_t["m_gas_kg_s"] > 0.0 else np.nan,
            "T caldera Rankine entrada [°C]": boiler_t["T_rankine_in_K"] - 273.15,
            "T caldera Rankine salida [°C]": boiler_t["T_rankine_out_K"] - 273.15,
            "h caldera salida [kJ/kg]": boiler_t["h_rankine_out_J_kg"] / 1000.0,
            "x caldera salida [-]": boiler_t["x_out"],
            "Q caldera Rankine [MW]": boiler_t["Q_to_rankine_W"] / 1e6,
            "Q caldera ambiente [kW]": boiler_t["Q_ambient_W"] / 1000.0,
            "Caldera habilita turbina": boiler_t["turbina_habilitada"],

            # Turbina Rankine
            "T turbina Rankine entrada [°C]": boiler_t["T_rankine_out_K"] - 273.15,
            "T turbina Rankine salida [°C]": turbine_t["T4"] - 273.15 if turbine_t.get("turbina_encendida", False) else np.nan,
            "h turbina Rankine entrada [kJ/kg]": turbine_t["h3"] / 1000.0,
            "h turbina Rankine salida [kJ/kg]": turbine_t["h4"] / 1000.0 if turbine_t.get("turbina_encendida", False) else np.nan,
            "m vapor a turbina Rankine [kg/s]": turbine_t["m_steam_kg_s"],
            "Turbina Rankine encendida": turbine_t.get("turbina_encendida", False),

            # Condensador
            "T condensador Rankine entrada [°C]": condenser_t["T_rankine_in_K"] - 273.15 if condenser_t["recibe_vapor"] else np.nan,
            "T condensador Rankine salida [°C]": condenser_t["T_rankine_out_K"] - 273.15,
            "T agua enfriamiento entrada [°C]": condenser_t["T_cool_in_K"] - 273.15,
            "T agua enfriamiento salida [°C]": condenser_t["T_cool_out_K"] - 273.15,
            "x condensador salida [-]": condenser_t["x_out"],
            "Q condensador [MW]": condenser_t["Q_cond_W"] / 1e6,
            "Q condensador agua [MW]": condenser_t["Q_tubos_W"] / 1e6,
            "Q condensador ambiente [kW]": condenser_t["Q_amb_W"] / 1000.0,
            "Salida condensador apta bomba": condenser_t["salida_apta_bomba"],

            # Potencias
            "W Brayton eléctrica [MW]": W_brayton_electric_W / 1e6,
            "W Rankine bruta eléctrica [MW]": W_rankine_gross_W / 1e6,
            "W bomba Rankine [MW]": W_pump_W / 1e6,
            "W Rankine neta [MW]": W_rankine_net_W / 1e6,
            "W planta neta [MW]": W_total_net_W / 1e6,
            "eta Rankine instantánea [%]": eta_rankine * 100.0,
            "eta global instantánea [%]": eta_global * 100.0,
        })

    return pd.DataFrame(rows)

def graficar_planta_transitoria_global(df):
    """Gráficas principales de la respuesta dinámica global protegida."""

    # -------------------------------------------------------------------------
    # Figura 7. Temperaturas principales con doble eje vertical
    # Eje izquierdo: bomba, salida turbina Rankine y salida condensador.
    # Eje derecho: salida de caldera, porque tiene otra escala.
    # -------------------------------------------------------------------------
    fig, ax1 = plt.subplots(figsize=(11, 5))

    l1, = ax1.plot(
        df["t [s]"],
        df["T bomba salida [°C]"],
        color="tab:blue",
        linestyle="-",
        linewidth=2.3,
        label="Salida bomba / entrada caldera"
    )

    l2, = ax1.plot(
        df["t [s]"],
        df["T turbina Rankine salida [°C]"],
        color="tab:orange",
        linestyle="--",
        linewidth=2.3,
        label="Salida turbina Rankine / entrada condensador"
    )

    l3, = ax1.plot(
        df["t [s]"],
        df["T condensador Rankine salida [°C]"],
        color="tab:green",
        linestyle="--",
        linewidth=2.3,
        label="Salida condensador / entrada bomba"
    )

    l_inicio = ax1.axvline(
        t_start_boiler_s,
        color="black",
        linestyle="--",
        linewidth=1.8,
        label="Inicio gases a caldera"
    )

    ax1.set_xlabel("Tiempo [s]")
    ax1.set_ylabel("Temperaturas del ciclo Rankine [°C]")
    ax1.grid(True)

    ax2 = ax1.twinx()

    l4, = ax2.plot(
        df["t [s]"],
        df["T caldera Rankine salida [°C]"],
        color="tab:red",
        linestyle="-",
        linewidth=2.8,
        label="Salida caldera / entrada turbina Rankine"
    )

    ax2.set_ylabel("Temperatura salida caldera [°C]")

    lineas = [l1, l2, l3, l4, l_inicio]
    etiquetas = [linea.get_label() for linea in lineas]
    ax1.legend(lineas, etiquetas, loc="best")

    plt.title("Figura 7. Temperaturas principales del ciclo Rankine - Planta completa")
    fig.tight_layout()
    plt.show()

    plt.figure(figsize=(11, 5))
    plt.plot(df["t [s]"], df["T escape turbina Brayton [°C]"], color="tab:red", linewidth=2.2, label="Escape turbina Brayton / entrada caldera")
    plt.plot(df["t [s]"], df["T chimenea [°C]"], color="tab:brown", linestyle="--", linewidth=2.2, label="Salida chimenea")
    plt.axvline(t_start_boiler_s, color="black", linestyle="--", linewidth=1.8, label="Inicio gases a caldera")
    plt.xlabel("Tiempo [s]")
    plt.ylabel("Temperatura [°C]")
    plt.title("Figura 8. Temperaturas de gases - Planta completa")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

    fig, ax1 = plt.subplots(figsize=(11, 5))
    l1, = ax1.plot(df["t [s]"], df["Q caldera Rankine [MW]"], color="tab:red", linewidth=2.4, label="Q caldera al Rankine")
    l2, = ax1.plot(df["t [s]"], df["Q condensador [MW]"], color="tab:blue", linewidth=2.4, label="Q condensador retirado")
    l_inicio = ax1.axvline(t_start_boiler_s, color="black", linestyle="--", linewidth=1.8, label="Inicio gases a caldera")
    ax1.set_xlabel("Tiempo [s]")
    ax1.set_ylabel("Potencia térmica principal [MW]")
    ax1.grid(True)

    ax2 = ax1.twinx()
    l3, = ax2.plot(df["t [s]"], df["Q caldera ambiente [kW]"], color="tab:orange", linestyle="--", linewidth=2.0, label="Pérdida caldera ambiente")
    l4, = ax2.plot(df["t [s]"], df["Q condensador ambiente [kW]"], color="tab:green", linestyle="--", linewidth=2.0, label="Pérdida condensador ambiente")
    ax2.set_ylabel("Pérdidas al ambiente [kW]")
    lineas = [l1, l2, l3, l4, l_inicio]
    ax1.legend(lineas, [l.get_label() for l in lineas], loc="best")
    plt.title("Figura 9. Calores principales y pérdidas - Planta completa")
    fig.tight_layout()
    plt.show()

    plt.figure(figsize=(11, 5))
    plt.plot(df["t [s]"], df["W Brayton eléctrica [MW]"], color="tab:red", linewidth=2.2, label="Brayton eléctrica")
    plt.plot(df["t [s]"], df["W Rankine bruta eléctrica [MW]"], color="tab:blue", linewidth=2.2, label="Rankine bruta eléctrica")
    plt.plot(df["t [s]"], df["W bomba Rankine [MW]"], color="tab:green", linestyle="--", linewidth=2.2, label="Bomba Rankine")
    plt.plot(df["t [s]"], df["W planta neta [MW]"], color="tab:purple", linewidth=2.6, label="Planta neta")
    plt.axvline(t_start_boiler_s, color="black", linestyle="--", linewidth=1.8, label="Inicio gases a caldera")
    plt.xlabel("Tiempo [s]")
    plt.ylabel("Potencia [MW]")
    plt.title("Figura 10. Potencias de la planta completa")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(11, 5))
    plt.plot(df["t [s]"], df["x caldera salida [-]"], color="tab:red", linewidth=2.4, label="Calidad salida caldera")
    plt.plot(df["t [s]"], df["x condensador salida [-]"], color="tab:blue", linewidth=2.4, label="Calidad salida condensador")
    plt.axvline(t_start_boiler_s, color="black", linestyle="--", linewidth=1.8, label="Inicio gases a caldera")
    plt.xlabel("Tiempo [s]")
    plt.ylabel("Calidad x [-]")
    plt.ylim(0, 1.05)
    plt.title("Figura 11. Calidad del fluido Rankine - Planta completa")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(11, 5))
    plt.plot(df["t [s]"], df["eta global instantánea [%]"], color="tab:purple", linewidth=2.4, label="Eficiencia global instantánea")
    plt.axvline(t_start_boiler_s, color="black", linestyle="--", linewidth=1.8, label="Inicio gases a caldera")
    plt.xlabel("Tiempo [s]")
    plt.ylabel("Eficiencia [%]")
    plt.title("Figura 12. Eficiencia global transitoria")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

# =============================================================================
# 15. EJECUCIÓN DE LA RESPUESTA TRANSITORIA GLOBAL PROTEGIDA
# =============================================================================

df_planta_transitoria = run_planta_transitoria_global()

df_planta_final = pd.DataFrame([{
    "W planta neta final [MW]": df_planta_transitoria["W planta neta [MW]"].iloc[-1],
    "W Brayton final [MW]": df_planta_transitoria["W Brayton eléctrica [MW]"].iloc[-1],
    "W Rankine bruta final [MW]": df_planta_transitoria["W Rankine bruta eléctrica [MW]"].iloc[-1],
    "W bomba final [MW]": df_planta_transitoria["W bomba Rankine [MW]"].iloc[-1],
    "eta global final [%]": df_planta_transitoria["eta global instantánea [%]"].iloc[-1],
    "T salida caldera final [°C]": df_planta_transitoria["T caldera Rankine salida [°C]"].iloc[-1],
    "x salida caldera final [-]": df_planta_transitoria["x caldera salida [-]"].iloc[-1],
    "T salida condensador final [°C]": df_planta_transitoria["T condensador Rankine salida [°C]"].iloc[-1],
    "x salida condensador final [-]": df_planta_transitoria["x condensador salida [-]"].iloc[-1],
    "T chimenea final [°C]": df_planta_transitoria["T chimenea [°C]"].iloc[-1],
    "Turbina Rankine encendida final": df_planta_transitoria["Turbina Rankine encendida"].iloc[-1],
    "Salida condensador apta bomba final": df_planta_transitoria["Salida condensador apta bomba"].iloc[-1],
}])

# Resumen compacto de puntos clave de la respuesta global.
def crear_resumen_puntos_clave_global(df):
    indices = []
    etiquetas = []

    def add_idx(idx, etiqueta):
        if idx is None:
            return
        idx = int(idx)
        if idx < 0 or idx >= len(df):
            return
        if idx not in indices:
            indices.append(idx)
            etiquetas.append(etiqueta)

    add_idx(0, "Inicio simulación")

    if "Ciclo Rankine encendido" in df.columns:
        mask_rankine = df["Ciclo Rankine encendido"] == True
        if mask_rankine.any():
            add_idx(mask_rankine.idxmax(), "Inicio ciclo Rankine")

    if "m gases a caldera [kg/s]" in df.columns:
        mask_gas = df["m gases a caldera [kg/s]"] > 0
        if mask_gas.any():
            add_idx(mask_gas.idxmax(), "Entrada gases a caldera")

    if "Turbina Rankine encendida" in df.columns:
        mask_turb = df["Turbina Rankine encendida"] == True
        if mask_turb.any():
            add_idx(mask_turb.idxmax(), "Encendido turbina Rankine")

    if "Salida condensador apta bomba" in df.columns:
        mask_bomba = df["Salida condensador apta bomba"] == True
        if mask_bomba.any():
            add_idx(mask_bomba.idxmax(), "Condensador apto para bomba")

    add_idx(len(df) - 1, "Estado final")

    columnas = [
        "t [s]",
        "T caldera Rankine salida [°C]",
        "x caldera salida [-]",
        "T turbina Rankine salida [°C]",
        "T condensador Rankine salida [°C]",
        "x condensador salida [-]",
        "T chimenea [°C]",
        "Q caldera Rankine [MW]",
        "Q condensador [MW]",
        "W Brayton eléctrica [MW]",
        "W Rankine bruta eléctrica [MW]",
        "W bomba Rankine [MW]",
        "W planta neta [MW]",
        "eta global instantánea [%]",
    ]

    columnas = [c for c in columnas if c in df.columns]
    resumen = df.loc[indices, columnas].copy()
    resumen.insert(0, "Evento", etiquetas)
    return resumen


df_puntos_clave_global = crear_resumen_puntos_clave_global(df_planta_transitoria)

# Se conservan internamente df_planta_final y df_puntos_clave_global,
# pero no se muestran para mantener la salida de resultados más compacta.

display(Markdown("## 2.1.6 Gráficas de respuesta transitoria global"))
graficar_planta_transitoria_global(df_planta_transitoria)

try:
    planta_integrada["df_planta_transitoria"] = df_planta_transitoria
    planta_integrada["df_planta_final"] = df_planta_final
    planta_integrada["df_puntos_clave_global"] = df_puntos_clave_global
except NameError:
    planta_integrada = {
        "df_planta_transitoria": df_planta_transitoria,
        "df_planta_final": df_planta_final,
        "df_puntos_clave_global": df_puntos_clave_global,
    }
```

## **2.2 Análisis general de resultados del ciclo combinado Brayton–Rankine**

Las **Tablas 2.1.1** a **2.1.5** muestran el comportamiento global de la planta de ciclo combinado Brayton–Rankine y permiten verificar la distribución de potencia, calor y estados termodinámicos entre los equipos principales. La planta entrega una potencia eléctrica neta aproximada de **85.64 MW**, de los cuales **62.50 MW** provienen del ciclo Brayton y **23.14 MW** del ciclo Rankine neto. El ciclo Brayton es el bloque principal de generación, ya que la turbina de gas produce cerca de **128.57 MW**, mientras que el compresor consume aproximadamente **64.14 MW**. La cámara de combustión suministra alrededor de **155.82 MW** al flujo de gases, con una potencia térmica de combustible de **159.00 MW**.

A la salida de la turbina Brayton, los gases conservan una temperatura elevada, cercana a **596 °C**, lo que permite recuperar **61.59 MW** en el intercambiador caldera-serpentín. Con esta energía, el ciclo Rankine trabaja con un flujo másico de **21.68 kg/s** y genera vapor sobrecalentado a **12 MPa** y **540 °C**, con calidad igual a **1.0**. Después de la expansión en la turbina Rankine, el condensador retira aproximadamente **37.21 MW** y entrega una salida con calidad **0.0**, es decir, líquido apto para la bomba. Esto confirma que tanto la caldera-serpentín como el condensador cumplen correctamente su función dentro del ciclo.

El acople entre ambos ciclos mejora de forma importante el desempeño de la planta. El ciclo Rankine neto aporta **23.14 MW** adicionales aprovechando parte de la energía que, en un ciclo Brayton simple, se perdería con los gases de escape. Por esta razón, la eficiencia eléctrica global alcanza aproximadamente **53.86 %**, con un heat rate de **6684.05 kJ/kWh**. Además, la temperatura de los gases disminuye desde **596.00 °C** hasta **176.59 °C** en la chimenea, lo que evidencia una recuperación térmica significativa.

Las **Figuras 7** a **12** muestran la respuesta transitoria global de la planta. En la **Figura 7** se observa que la salida de la caldera aumenta progresivamente hasta alcanzar la condición de vapor sobrecalentado, mientras que las temperaturas asociadas al condensador, bomba y salida de turbina Rankine se mantienen cercanas a **44 °C**, coherentes con la presión fija de condensación. En la **Figura 9**, el calor transferido en la caldera se estabiliza cerca de **61.6 MW**, mientras que el calor retirado en el condensador aparece después, cuando la turbina Rankine comienza a operar.

Finalmente, las **Figuras 10** a **12** resumen el efecto de la integración. La potencia neta de la planta aumenta desde el aporte inicial del ciclo Brayton hasta aproximadamente **85.6 MW** cuando entra en operación el ciclo Rankine. La calidad de salida de la caldera llega a **1.0**, mientras que la calidad de salida del condensador cae hasta **0.0**, confirmando el cierre correcto del ciclo de vapor. En conjunto, los resultados muestran que la recuperación de calor de los gases de escape permite aumentar la potencia útil y mejorar la eficiencia global de la planta de cogeneración.


 # **3. Comparación Paramétrica de Características de la Planta de Cogeneración**

Con base en los resultados obtenidos en la **Sección 2**, se realizó un análisis paramétrico para evaluar cómo cambia el desempeño energético global de la planta de cogeneración ante variaciones geométricas y operativas. El análisis se concentró principalmente en la caldera, modificando el número de serpentines, la longitud efectiva, los diámetros característicos, la razón entre el flujo másico de gases y el flujo Rankine, la temperatura objetivo del vapor y la temperatura mínima de chimenea. Las condiciones del condensador se mantuvieron constantes, ya que en el modelo este equipo opera a presión fija y el proceso de condensación se considera aproximadamente isobárico; por tanto, sus temperaturas principales permanecen cercanas a la temperatura de saturación y no generan variaciones significativas en la eficiencia global mientras se conserve una salida líquida apta para la bomba.

Para comparar los resultados, se definió el factor relativo del parámetro como:

$$
\phi_X=
\frac{X}{X_0}
$$

También se definió la potencia neta relativa de la planta como:

$$
\overline{W}_{net}=
\frac{\dot{W}_{net,plant}}{\dot{W}_{net,plant,0}}
$$

donde valores de $\overline{W}_{net}\gt 1$ indican aumento de potencia respecto al caso base, mientras que valores de $\overline{W}_{net}\lt 1$ representan una disminución. Este análisis permite comparar de forma directa cómo las variaciones en la caldera y en las condiciones globales del ciclo afectan la recuperación de calor y la potencia neta de la planta.


```python
# @title
# =============================================================================
# 16. ANÁLISIS DE SENSIBILIDAD PARAMÉTRICA
#     CONFIGURACIÓN GENERAL Y CÁLCULO DE CURVAS
# =============================================================================

from matplotlib.lines import Line2D
from IPython.display import display, Markdown, HTML

# -----------------------------------------------------------------------------
# 16.1 Estilo general de gráficas
# -----------------------------------------------------------------------------

plt.rcParams.update({
    "font.size": 10,
    "axes.titlesize": 12,
    "axes.labelsize": 10,
    "legend.fontsize": 8.5,
    "xtick.labelsize": 9,
    "ytick.labelsize": 9,
    "axes.grid": True,
    "grid.alpha": 0.28,
    "grid.linestyle": "-",
    "axes.axisbelow": True,
})


# -----------------------------------------------------------------------------
# 16.2 Función para mostrar subtítulos centrados
# -----------------------------------------------------------------------------

def mostrar_subtitulo_figura(texto):
    display(HTML(
        f"""
        <div style="
            text-align: center;
            font-size: 16px;
            font-weight: 600;
            margin-top: 18px;
            margin-bottom: 10px;">
            {texto}
        </div>
        """
    ))


# -----------------------------------------------------------------------------
# 16.3 Valores base
# -----------------------------------------------------------------------------

SENS_PARAMS_TRACKED = [
    "N_serpentines",
    "L_tube_boiler",
    "D_i_boiler",
    "D_o_boiler",
    "D_shell_boiler",
    "T_steam_target",
    "T_stack_min",
    "m_exhaust_kg_s",
    "eta_turbine_stage_brayton",
    "eta_turbine_stage_rankine",
]

SENS_BASE_VALUES = {name: globals()[name] for name in SENS_PARAMS_TRACKED}

# Más puntos de muestreo en el mismo intervalo de variación
SENS_FACTORS = np.linspace(0.55, 1.25, 31)

SENS_INT_PARAMS = {"N_serpentines"}
SENS_ETA_PARAMS = {"eta_turbine_stage_brayton", "eta_turbine_stage_rankine"}


def sens_restore_base_values():
    """Restaura los valores base después de cada caso de sensibilidad."""
    for k, v in SENS_BASE_VALUES.items():
        globals()[k] = v


def sens_set_parameter_scaled(param_name, factor):
    """
    Modifica un parámetro con factor relativo.

    Para temperaturas se escala el valor en grados Celsius y luego se convierte
    nuevamente a Kelvin.
    """

    base = SENS_BASE_VALUES[param_name]

    if param_name in ["T_steam_target", "T_stack_min"]:
        base_C = base - 273.15
        value_C = base_C * factor
        value = value_C + 273.15

        if param_name == "T_steam_target":
            value = max(273.15 + 250.0, min(value, 273.15 + 700.0))

        if param_name == "T_stack_min":
            value = max(273.15 + 60.0, min(value, 273.15 + 350.0))

    else:
        value = base * factor

    if param_name in SENS_INT_PARAMS:
        value = int(max(1, round(value)))

    if param_name in SENS_ETA_PARAMS:
        value = clamp(value, 0.05, 0.99)

    globals()[param_name] = value
    return value


def sens_set_boiler_tube_diameter_scaled(factor):
    """
    Escala simultáneamente D_i_boiler y D_o_boiler para mantener
    una geometría físicamente consistente del tubo.
    """

    globals()["D_i_boiler"] = max(
        SENS_BASE_VALUES["D_i_boiler"] * factor,
        1e-5
    )

    globals()["D_o_boiler"] = max(
        SENS_BASE_VALUES["D_o_boiler"] * factor,
        globals()["D_i_boiler"] * 1.05
    )

    return factor


# -----------------------------------------------------------------------------
# 16.4 Evaluación estacionaria integrada para sensibilidad
# -----------------------------------------------------------------------------

def sens_run_stationary_case(m_rankine_override=None):
    """
    Evalúa la planta completa en estado estacionario para el conjunto actual
    de parámetros globales.

    Si el ciclo Rankine no es físicamente apto, se considera que la planta opera
    únicamente con el ciclo Brayton.
    """

    brayton_case = run_brayton_cycle()
    rankine_design_case = estimate_rankine_mass_flow_from_exhaust(brayton_case)

    if m_rankine_override is None:
        m_rankine_case = rankine_design_case["m_rankine_design_kg_s"]
    else:
        m_rankine_case = max(m_rankine_override, 1e-9)

    pump_case = run_rankine_pump(
        m_rankine_kg_s=m_rankine_case,
        h_in_liquid_J_kg=rankine_design_case["h_cond_out"],
        T_in_K=rankine_design_case["T_cond_out_K"],
        p_in=p_condenser,
        p_out=rankine_design_case["p_pump_out"],
    )

    boiler_case = run_boiler_serpentin_estacionario(
        brayton_case,
        pump_case
    )

    caldera_apta = bool(
        boiler_case["cumple_vapor_turbina"]
        and boiler_case["cumple_pinch"]
        and boiler_case["cumple_stack"]
    )

    if caldera_apta:
        turbine_case = run_rankine_turbine(boiler_case)
        condenser_case = resolver_estado_estacionario_condensador(turbine_case)

        condensador_apto = bool(condenser_case["salida_apta_bomba"])

        if condensador_apto:
            W_rankine_gross_W = turbine_case["W_electric_W"]
            W_pump_W = pump_case["P_eje_req_W"]
            W_rankine_net_W = W_rankine_gross_W - W_pump_W
            Q_cond_W = condenser_case["Q_cond_W"]
            x_cond_out = condenser_case["x_out"]
        else:
            W_rankine_gross_W = 0.0
            W_pump_W = 0.0
            W_rankine_net_W = 0.0
            Q_cond_W = condenser_case["Q_cond_W"]
            x_cond_out = condenser_case["x_out"]

    else:
        turbine_case = None
        condenser_case = None
        condensador_apto = False
        W_rankine_gross_W = 0.0
        W_pump_W = 0.0
        W_rankine_net_W = 0.0
        Q_cond_W = 0.0
        x_cond_out = np.nan

    W_brayton_W = brayton_case["W_electric_W"]
    W_total_W = W_brayton_W + W_rankine_net_W
    Q_fuel_W = brayton_case["Q_fuel_W"]

    eta_global_pct = 100.0 * safe_div(
        W_total_W,
        Q_fuel_W,
        default=np.nan
    )

    ciclo_apto = bool(caldera_apta and condensador_apto)

    return {
        "W_total_MW": W_total_W / 1e6,
        "W_brayton_MW": W_brayton_W / 1e6,
        "W_rankine_net_MW": W_rankine_net_W / 1e6,
        "W_rankine_gross_MW": W_rankine_gross_W / 1e6,
        "W_pump_MW": W_pump_W / 1e6,
        "eta_global_pct": eta_global_pct,
        "Q_fuel_MW": Q_fuel_W / 1e6,
        "Q_boiler_MW": boiler_case["Q_to_rankine_W"] / 1e6,
        "Q_cond_MW": Q_cond_W / 1e6,
        "m_gas_kg_s": brayton_case["m_gas_kg_s"],
        "m_rankine_kg_s": m_rankine_case,
        "R_gas_rankine": safe_div(
            brayton_case["m_gas_kg_s"],
            m_rankine_case,
            default=np.nan
        ),
        "R_cool_rankine": safe_div(
            m_cool_condenser,
            m_rankine_case,
            default=np.nan
        ),
        "x_boiler_out": boiler_case["x_out"],
        "x_cond_out": x_cond_out,
        "caldera_apta": caldera_apta,
        "condensador_apto": condensador_apto,
        "ciclo_apto": ciclo_apto,
    }


sens_restore_base_values()
SENS_BASE_CASE = sens_run_stationary_case()
SENS_W0_MW = SENS_BASE_CASE["W_total_MW"]
SENS_ETA0 = SENS_BASE_CASE["eta_global_pct"]


# -----------------------------------------------------------------------------
# 16.5 Evaluación de cada punto de sensibilidad
# -----------------------------------------------------------------------------

def sens_evaluate_point(mode, factor, param_name=None):
    """
    mode:
        scaled_param:
            Modifica directamente un parámetro global.

        boiler_ratio:
            Modifica la razón R = m_gas / m_Rankine.
            Para hacerlo, se mantiene m_gas y se cambia m_Rankine.

        boiler_tube_diameter:
            Escala simultáneamente D_i_boiler y D_o_boiler.
    """

    sens_restore_base_values()

    try:
        if mode == "scaled_param":
            valor_real = sens_set_parameter_scaled(param_name, factor)
            result = sens_run_stationary_case()

        elif mode == "boiler_tube_diameter":
            valor_real = sens_set_boiler_tube_diameter_scaled(factor)
            result = sens_run_stationary_case()

        elif mode == "boiler_ratio":
            brayton_ref = run_brayton_cycle()
            rankine_design_ref = estimate_rankine_mass_flow_from_exhaust(brayton_ref)
            m_rankine_base = rankine_design_ref["m_rankine_design_kg_s"]

            m_rankine_new = m_rankine_base / max(factor, 1e-9)
            valor_real = factor
            result = sens_run_stationary_case(m_rankine_override=m_rankine_new)

        else:
            raise ValueError(f"Modo no reconocido: {mode}")

        result["factor"] = factor
        result["valor_real"] = valor_real
        result["W_rel"] = safe_div(
            result["W_total_MW"],
            SENS_W0_MW,
            default=np.nan
        )
        result["eta_rel"] = safe_div(
            result["eta_global_pct"],
            SENS_ETA0,
            default=np.nan
        )

        return result

    except Exception as e:
        return {
            "factor": factor,
            "valor_real": np.nan,
            "W_total_MW": np.nan,
            "eta_global_pct": np.nan,
            "W_rel": np.nan,
            "eta_rel": np.nan,
            "caldera_apta": False,
            "condensador_apto": False,
            "ciclo_apto": False,
            "error": str(e),
        }

    finally:
        sens_restore_base_values()


def sens_build_curve(label, mode, param_name=None):
    rows = []

    for factor in SENS_FACTORS:
        result = sens_evaluate_point(
            mode=mode,
            factor=factor,
            param_name=param_name
        )

        rows.append({
            "Parametro": label,
            "factor": result["factor"],
            "valor_real": result["valor_real"],
            "W_rel": result["W_rel"],
            "W_total_MW": result["W_total_MW"],
            "eta_global_pct": result["eta_global_pct"],
            "caldera_apta": result["caldera_apta"],
            "condensador_apto": result["condensador_apto"],
            "ciclo_apto": result["ciclo_apto"],
        })

    return pd.DataFrame(rows)


# -----------------------------------------------------------------------------
# 16.6 Suavizado de curvas
# -----------------------------------------------------------------------------

def sens_smooth_curve(x, y, n_points=450):
    """
    Suaviza la unión entre puntos con interpolación PCHIP.
    Si scipy no está disponible, usa interpolación lineal como respaldo.
    """

    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x_valid = x[mask]
    y_valid = y[mask]

    if len(x_valid) < 2:
        return x_valid, y_valid

    x_dense = np.linspace(np.min(x_valid), np.max(x_valid), n_points)

    try:
        from scipy.interpolate import PchipInterpolator
        interpolator = PchipInterpolator(x_valid, y_valid)
        y_dense = interpolator(x_dense)
    except Exception:
        y_dense = np.interp(x_dense, x_valid, y_valid)

    return x_dense, y_dense


# -----------------------------------------------------------------------------
# 16.7 Sombreado de zonas no aptas
# -----------------------------------------------------------------------------

def sens_shade_inoperable_regions(ax, curves_dict):
    """
    Sombrea regiones donde al menos una curva del grupo no cumple condición
    operativa. La región no aparece en la leyenda.
    """

    factors = np.array(SENS_FACTORS)

    if len(factors) < 2:
        return

    dx = np.diff(factors)

    left_edges = np.zeros_like(factors)
    right_edges = np.zeros_like(factors)

    left_edges[0] = factors[0] - dx[0] / 2.0
    right_edges[-1] = factors[-1] + dx[-1] / 2.0

    for i in range(1, len(factors)):
        left_edges[i] = 0.5 * (factors[i - 1] + factors[i])

    for i in range(0, len(factors) - 1):
        right_edges[i] = 0.5 * (factors[i] + factors[i + 1])

    for i, factor in enumerate(factors):
        no_apto = False

        for _, df_curve in curves_dict.items():
            row = df_curve[np.isclose(df_curve["factor"], factor)]

            if len(row) > 0:
                if not bool(row["ciclo_apto"].iloc[0]):
                    no_apto = True

        if no_apto:
            ax.axvspan(
                left_edges[i],
                right_edges[i],
                color="#F2D6B3",
                alpha=0.22,
                zorder=0
            )


# -----------------------------------------------------------------------------
# 16.8 Gráfica con doble eje y
# -----------------------------------------------------------------------------

def sens_plot_group_dual_axis(xlabel, curves_dict):
    fig, ax1 = plt.subplots(figsize=(11.5, 5.8))
    ax2 = ax1.twinx()

    colores = [
        "tab:blue",
        "tab:green",
        "tab:red",
        "tab:purple",
        "tab:orange",
        "tab:brown",
    ]

    marcadores = ["o", "^", "s", "D", "v", "P"]

    legend_handles = []
    legend_labels = []

    sens_shade_inoperable_regions(ax1, curves_dict)

    for i, (name, df_curve) in enumerate(curves_dict.items()):
        color = colores[i % len(colores)]
        marker = marcadores[i % len(marcadores)]

        x_w, y_w = sens_smooth_curve(
            df_curve["factor"],
            df_curve["W_rel"]
        )

        x_eta, y_eta = sens_smooth_curve(
            df_curve["factor"],
            df_curve["eta_global_pct"]
        )

        ax1.plot(
            x_w,
            y_w,
            color=color,
            linewidth=2.1,
            linestyle="-"
        )

        ax1.plot(
            df_curve["factor"],
            df_curve["W_rel"],
            color=color,
            marker=marker,
            linestyle="",
            markersize=5.2,
            markerfacecolor="white",
            markeredgewidth=1.5
        )

        ax2.plot(
            x_eta,
            y_eta,
            color=color,
            linewidth=2.0,
            linestyle="--",
            alpha=0.95
        )

        ax2.plot(
            df_curve["factor"],
            df_curve["eta_global_pct"],
            color=color,
            marker=marker,
            linestyle="",
            markersize=5.0,
            markerfacecolor="white",
            markeredgewidth=1.5
        )

        legend_handles.append(
            Line2D(
                [0],
                [0],
                color=color,
                linewidth=2.1,
                marker=marker,
                markersize=5.5,
                markerfacecolor="white",
                markeredgewidth=1.5
            )
        )
        legend_labels.append(name)

    ax1.axhline(
        1.0,
        color="gray",
        linestyle=":",
        linewidth=1.1
    )

    ax1.set_xlabel(xlabel)
    ax1.set_ylabel("Potencia relativa $W/W_0$ [-]")
    ax2.set_ylabel("Eficiencia energética global [%]")

    ax1.set_xlim(SENS_FACTORS.min() - 0.03, SENS_FACTORS.max() + 0.03)

    ax1.grid(True, which="major", alpha=0.30)
    ax1.grid(True, which="minor", alpha=0.12)
    ax1.minorticks_on()

    ax1.legend(
        legend_handles,
        legend_labels,
        loc="lower center",
        bbox_to_anchor=(0.5, 1.02),
        ncol=min(len(legend_labels), 4),
        frameon=True,
        fancybox=False,
        edgecolor="0.75"
    )

    fig.tight_layout()
    plt.show()


# -----------------------------------------------------------------------------
# 16.9 Gráfica solo con eficiencia global
# -----------------------------------------------------------------------------

def sens_plot_group_eta_only(xlabel, curves_dict):
    fig, ax = plt.subplots(figsize=(11.5, 5.8))

    colores = [
        "tab:blue",
        "tab:green",
        "tab:red",
        "tab:purple",
        "tab:orange",
        "tab:brown",
    ]

    marcadores = ["o", "^", "s", "D", "v", "P"]

    legend_handles = []
    legend_labels = []

    sens_shade_inoperable_regions(ax, curves_dict)

    for i, (name, df_curve) in enumerate(curves_dict.items()):
        color = colores[i % len(colores)]
        marker = marcadores[i % len(marcadores)]

        x_eta, y_eta = sens_smooth_curve(
            df_curve["factor"],
            df_curve["eta_global_pct"]
        )

        ax.plot(
            x_eta,
            y_eta,
            color=color,
            linewidth=2.1,
            linestyle="-"
        )

        ax.plot(
            df_curve["factor"],
            df_curve["eta_global_pct"],
            color=color,
            marker=marker,
            linestyle="",
            markersize=5.2,
            markerfacecolor="white",
            markeredgewidth=1.5
        )

        legend_handles.append(
            Line2D(
                [0],
                [0],
                color=color,
                linewidth=2.1,
                marker=marker,
                markersize=5.5,
                markerfacecolor="white",
                markeredgewidth=1.5
            )
        )
        legend_labels.append(name)

    ax.set_xlabel(xlabel)
    ax.set_ylabel("Eficiencia energética global [%]")

    ax.set_xlim(SENS_FACTORS.min() - 0.03, SENS_FACTORS.max() + 0.03)

    ax.grid(True, which="major", alpha=0.30)
    ax.grid(True, which="minor", alpha=0.12)
    ax.minorticks_on()

    ax.legend(
        legend_handles,
        legend_labels,
        loc="lower center",
        bbox_to_anchor=(0.5, 1.02),
        ncol=min(len(legend_labels), 3),
        frameon=True,
        fancybox=False,
        edgecolor="0.75"
    )

    fig.tight_layout()
    plt.show()


# -----------------------------------------------------------------------------
# 16.10 Construcción de las curvas de sensibilidad
# -----------------------------------------------------------------------------

sens_boiler_geom_N = sens_build_curve(
    label=r"Número de serpentines $N_s$",
    mode="scaled_param",
    param_name="N_serpentines"
)

sens_boiler_geom_L = sens_build_curve(
    label=r"Longitud efectiva $L_t$",
    mode="scaled_param",
    param_name="L_tube_boiler"
)

sens_boiler_geom_Dtube = sens_build_curve(
    label=r"Diámetro de tubo $D_i/D_o$",
    mode="boiler_tube_diameter"
)

sens_boiler_geom_Dshell = sens_build_curve(
    label=r"Diámetro de carcasa $D_{shell}$",
    mode="scaled_param",
    param_name="D_shell_boiler"
)

sens_boiler_op_ratio = sens_build_curve(
    label=r"Razón de caudales $\dot{m}_{gas}/\dot{m}_R$",
    mode="boiler_ratio"
)

sens_boiler_op_Tsteam = sens_build_curve(
    label=r"Temperatura de vapor $T_{steam,target}$",
    mode="scaled_param",
    param_name="T_steam_target"
)

sens_boiler_op_Tstack = sens_build_curve(
    label=r"Temperatura mínima de chimenea $T_{stack,min}$",
    mode="scaled_param",
    param_name="T_stack_min"
)

sens_global_mgas = sens_build_curve(
    label=r"Flujo másico de escape $\dot{m}_{exhaust}$",
    mode="scaled_param",
    param_name="m_exhaust_kg_s"
)

sens_global_eta_brayton = sens_build_curve(
    label=r"Eficiencia turbina Brayton $\eta_{t,B}$",
    mode="scaled_param",
    param_name="eta_turbine_stage_brayton"
)

sens_global_eta_rankine = sens_build_curve(
    label=r"Eficiencia turbina Rankine $\eta_{t,R}$",
    mode="scaled_param",
    param_name="eta_turbine_stage_rankine"
)

mostrar_subtitulo_figura(
    "Figura 3.1. Eficiencia energética global contra variación de dimensiones geométricas del intercambiador."
)

sens_plot_group_dual_axis(
    xlabel="Parámetro geométrico relativo del intercambiador [-]",
    curves_dict={
        r"Número de serpentines $N_s$": sens_boiler_geom_N,
        r"Longitud efectiva $L_t$": sens_boiler_geom_L,
        r"Diámetro de tubo $D_i/D_o$": sens_boiler_geom_Dtube,
        r"Diámetro de carcasa $D_{shell}$": sens_boiler_geom_Dshell,
    }
)
```

La **Figura 3.1** muestra la sensibilidad de la potencia neta relativa y de la eficiencia energética global ante variaciones geométricas de la caldera. Se evaluaron el número de serpentines, la longitud efectiva de los tubos, el diámetro de los tubos y el diámetro de la carcasa. En general, se observa que la potencia relativa se mantiene muy cercana al caso base, con variaciones pequeñas alrededor de $\overline{W}_{net}=1$. Esto indica que, para el dimensionamiento actual, la caldera ya cuenta con una capacidad de transferencia suficiente y que aumentos adicionales en la geometría no producen mejoras significativas en el desempeño global de la planta.

Las variables más sensibles en la zona de reducción geométrica son el número de serpentines y la longitud efectiva de los tubos. Cuando estos parámetros disminuyen, la potencia relativa cae ligeramente, debido a la reducción del área de transferencia de calor disponible y, por tanto, de la capacidad de recuperación térmica desde los gases de escape. Por otro lado, el diámetro de los tubos alcanza rápidamente un comportamiento cercano al caso base, mientras que el diámetro de la carcasa tiene un efecto casi nulo en la mayor parte del intervalo analizado.


```python
# @title
mostrar_subtitulo_figura(
    "Figura 3.2. Eficiencia energética global contra variación de magnitudes operativas del intercambiador."
)

sens_plot_group_dual_axis(
    xlabel="Magnitud operativa relativa del intercambiador [-]",
    curves_dict={
        r"Razón de caudales $\dot{m}_{gas}/\dot{m}_R$": sens_boiler_op_ratio,
        r"Temperatura de vapor $T_{steam,target}$": sens_boiler_op_Tsteam,
        r"Temperatura mínima de chimenea $T_{stack,min}$": sens_boiler_op_Tstack,
    }
)
```

La **Figura 3.2** muestra que la variable operativa más sensible es la razón de caudales $\dot{m}_{gas}/\dot{m}_{R}$. Para valores bajos, la potencia relativa cae de forma importante porque el flujo Rankine es demasiado alto respecto a la energía disponible en los gases, por lo que la caldera no alcanza a calentar adecuadamente el vapor. Al aumentar esta relación, la potencia mejora rápidamente hasta una zona cercana al caso base, donde el sistema logra producir vapor apto para la turbina Rankine. Sin embargo, después de cierto punto la potencia vuelve a disminuir, ya que una razón demasiado alta implica reducir el flujo Rankine y, aunque el vapor pueda calentarse mejor, hay menos masa disponible para generar potencia en la turbina de vapor.

Respecto a las temperaturas, la temperatura objetivo del vapor $T_{steam,target}$ representa la temperatura deseada a la salida de la caldera. Su aumento puede mejorar ligeramente el desempeño cerca del punto base porque el vapor entra con mayor energía a la turbina Rankine; sin embargo, cuando esta exigencia térmica es demasiado alta, la caldera no logra cumplirla y aparece una caída fuerte de la potencia, como se observa hacia el extremo derecho. Por otro lado, la temperatura mínima de chimenea $T_{stack,min}$ limita qué tan fríos pueden salir los gases después de ceder calor. Al aumentar este límite, se reduce el calor recuperable y por eso la eficiencia tiende a disminuir suavemente. Las zonas sombreadas representan regiones no aptas de operación, asociadas principalmente a falta de energía para producir vapor adecuado o a condiciones térmicas demasiado exigentes para la caldera.


```python
# @title
mostrar_subtitulo_figura(
    "Figura 3.3. Eficiencia energética global contra variación de magnitudes operativas del ciclo de cogeneración."
)

sens_plot_group_eta_only(
    xlabel="Magnitud operativa relativa del ciclo de cogeneración [-]",
    curves_dict={
        r"Flujo másico de escape $\dot{m}_{exhaust}$": sens_global_mgas,
        r"Eficiencia turbina Brayton $\eta_{t,B}$": sens_global_eta_brayton,
        r"Eficiencia turbina Rankine $\eta_{t,R}$": sens_global_eta_rankine,
    }
)
```

La **Figura 3.3** muestra la sensibilidad de la eficiencia energética global ante variaciones operativas generales del ciclo de cogeneración. Se observa que el flujo másico de gases de escape $\dot{m}_{exhaust}$ tiene una influencia importante y casi creciente sobre la eficiencia, ya que al aumentar este flujo se incrementa la energía disponible para recuperar calor en la caldera y alimentar el ciclo Rankine. Sin embargo, en el extremo superior aparece una caída localizada, asociada a una región restringida donde el acoplamiento térmico deja de cumplir adecuadamente alguna condición de operación del ciclo.

Por otro lado, la eficiencia de la turbina Brayton $\eta_{t,B}$ es la variable más sensible, especialmente para valores bajos, donde la eficiencia global cae de forma marcada porque se reduce directamente la potencia producida por el ciclo de gas. A medida que $\eta_{t,B}$ aumenta, la eficiencia global mejora hasta estabilizarse en valores altos. En comparación, la eficiencia de la turbina Rankine $\eta_{t,R}$ produce una variación más moderada, ya que afecta solo la potencia recuperada en el ciclo de vapor. Las regiones sombreadas indican condiciones no aptas o límites de operación del modelo, por lo que el rango más conveniente corresponde a la zona cercana al caso base y antes de las caídas abruptas.


## **5. Intervenciones tecnológicas necesarias para implementar el ciclo Rankine**

### **5.1 Selección de materiales adecuados para equipos y tuberías**

La implementación del ciclo Rankine exige que tuberías, válvulas, intercambiadores y demás equipos soporten condiciones de alta presión, temperatura y posibles efectos de corrosión. Si los materiales no son adecuados, pueden presentarse fugas, deformaciones, pérdida de resistencia mecánica o deterioro acelerado de los componentes.

Por esta razón, se deben seleccionar materiales compatibles con el fluido de trabajo y con las condiciones térmicas del ciclo. Esta intervención permite aumentar la seguridad, la vida útil de los equipos y la confiabilidad operativa de la planta.

### **5.2 Sistemas de seguridad contra sobrepresión**

Durante la operación del ciclo Rankine pueden producirse incrementos de presión por acumulación de vapor, cambios bruscos de temperatura, fallas en válvulas o restricciones en las líneas. Estas condiciones pueden comprometer la integridad del generador de vapor, el intercambiador, la bomba y las tuberías.

Para controlar este problema, se deben implementar válvulas de alivio, válvulas de seguridad y sistemas de purga. Estos elementos protegen los equipos ante condiciones críticas y permiten que el ciclo opere dentro de límites seguros.

### **5.3 Evaluación de pérdidas de carga en tuberías, válvulas, accesorios e intercambiadores**

Al agregar un nuevo circuito hidráulico al ciclo Rankine, el fluido debe circular por tuberías, accesorios, válvulas e intercambiadores que generan pérdidas de carga. Estas pérdidas pueden reducir la presión disponible, aumentar el trabajo de la bomba y alterar las condiciones esperadas en la simulación.

La evaluación hidráulica permite dimensionar correctamente las tuberías, seleccionar válvulas adecuadas y escoger una bomba con la capacidad necesaria. Con esto se garantiza una circulación estable del fluido y se reduce el consumo energético asociado al bombeo.

### **5.4 Adaptación del sistema de condensación**

Un ciclo Rankine completo necesita rechazar calor para que el vapor pueda regresar al estado líquido después del proceso de expansión o transferencia de energía. Si no existe un sistema de condensación adecuado, el ciclo no puede cerrarse correctamente y la operación continua se ve afectada.

La incorporación de un condensador y un sistema de enfriamiento permite recuperar el fluido de trabajo y enviarlo nuevamente hacia la bomba. Esta intervención es necesaria para cerrar el ciclo termodinámico y mejorar la estabilidad operativa del sistema.

### **5.5 Sistema de control de nivel en el generador de vapor o intercambiador**

El nivel del fluido dentro del generador de vapor o del intercambiador puede variar por cambios de caudal, evaporación o condiciones transitorias de operación. Un nivel demasiado bajo puede generar operación en seco, mientras que un nivel excesivo puede producir arrastre de líquido o inestabilidad del proceso.

Para evitar estas situaciones, se debe implementar un sistema de control de nivel con sensores y válvulas de regulación. Esta tecnología permite mantener el fluido dentro de un rango seguro, proteger los equipos térmicos y mejorar la calidad del proceso de transferencia de energía.

### **5.6 Mantenimiento y accesibilidad de los equipos**

La instalación de nuevos equipos puede generar problemas si no se dejan espacios adecuados para inspección, limpieza, reparación o sustitución de componentes. Una mala accesibilidad aumenta los tiempos de parada y dificulta el mantenimiento de bombas, válvulas, tuberías, intercambiadores y condensadores.

Por ello, el diseño debe incluir corredores de acceso, zonas de mantenimiento y espacios libres alrededor de los equipos principales. Esta intervención mejora la disponibilidad de la planta, facilita el mantenimiento preventivo y reduce los costos asociados a paradas operativas.

### **5.7 Adecuación del terreno y del espacio disponible para nuevos equipos**

La implementación del ciclo Rankine requiere espacio físico para instalar bombas, intercambiadores, condensadores, tuberías, soportes y sistemas auxiliares. Si la planta no cuenta con una distribución adecuada, pueden generarse interferencias, dificultades de montaje y riesgos durante la operación.

La adecuación del terreno debe considerar cimentaciones, soportes estructurales, rutas de tuberías, drenajes, ventilación y zonas seguras de acceso. Esto permite integrar el ciclo de manera ordenada, segura y compatible con los equipos existentes de la planta.

### **5.8 Prevención de puntos de embotellamiento dentro del sistema**

Al incorporar nuevos equipos y líneas de flujo, pueden aparecer puntos de embotellamiento si las tuberías, válvulas o intercambiadores no tienen la capacidad suficiente para manejar el caudal requerido. Esto puede producir caídas de presión elevadas, acumulación de fluido, inestabilidad térmica y reducción del rendimiento del ciclo.

Para evitarlo, se debe analizar la capacidad hidráulica y térmica de cada componente, seleccionando diámetros, válvulas y equipos adecuados. Esta intervención permite mantener un flujo estable, reducir restricciones internas y asegurar que ningún elemento limite el desempeño global del ciclo Rankine.
